```python
j# ============================================================
# KHADIJAH TEMPERATURE-ONLY REPRODUCTION
# Stanford University | SARIMA + Prophet | Jan-Nov 2024
#
# Based on Khadijah's supplied notebook text.
#
# Necessary corrections:
# 1. Uses the actual Google Drive data path.
# 2. Corrects "f = data.copy()" to "df = data.copy()".
# 3. Uses only columns needed for temperature-only forecasting.
# 4. Aligns forecast dates explicitly with validation dates.
#
# This reproduces Khadijah's original evaluation design.
# It is NOT yet the final fair model comparison.
# ============================================================

%pip install -q pandas numpy matplotlib scikit-learn statsmodels prophet

from google.colab import drive
drive.mount("/content/drive")

import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score
)

from statsmodels.tsa.statespace.sarimax import SARIMAX
from prophet import Prophet


# ============================================================
# PATHS
# ============================================================

PROJECT = "/content/drive/MyDrive/Stanford_Weather_12M"

DATA_PATH = (
    f"{PROJECT}/data/"
    "OpenWeather19802026_StanfordOnly.csv"
)

OUTPUT_DIR = (
    f"{PROJECT}/"
    "khadijah_temp_only_reproduction"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

if not os.path.exists(DATA_PATH):
    raise FileNotFoundError(
        f"Data file was not found:\n{DATA_PATH}\n\n"
        "Confirm that the CSV is inside the project's data folder."
    )


# ============================================================
# LOAD ORIGINAL HOURLY WEATHER DATA
# ============================================================

data = pd.read_csv(DATA_PATH)
df = data.copy()  # Correction of Khadijah's original f/df typo

required_columns = [
    "dt",
    "timezone",
    "city_name",
    "temp"
]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Required columns are missing: {missing_columns}"
    )

print("SOURCE DATA")
print("-----------")
print(f"File: {DATA_PATH}")
print(f"Hourly rows: {len(df):,}")
print(f"Cities: {df['city_name'].dropna().unique()}")
print()


# ============================================================
# KHADIJAH'S TIME CONVERSION
# ============================================================

# Unix timestamp -> UTC datetime
df["dt"] = pd.to_datetime(
    pd.to_numeric(df["dt"], errors="coerce"),
    unit="s",
    utc=True
)

# Timezone offset in seconds
df["timezone_offset"] = pd.to_timedelta(
    pd.to_numeric(df["timezone"], errors="coerce"),
    unit="s"
)

# Stanford local clock time, following Khadijah's method
df["dt_local"] = (
    df["dt"] +
    df["timezone_offset"]
)

df = df.dropna(
    subset=["dt_local", "city_name", "temp"]
).copy()

df["temp"] = pd.to_numeric(
    df["temp"],
    errors="coerce"
)

df = df.dropna(subset=["temp"])

# Do not remove duplicates: preserve original handling.
# Convert local timestamps to calendar months.
df["month"] = (
    df["dt_local"]
    .dt
    .tz_localize(None)
    .dt
    .to_period("M")
)


# ============================================================
# KHADIJAH'S MONTHLY AGGREGATION
# ============================================================

monthly_summary = (
    df.groupby(
        ["city_name", "month"],
        as_index=False
    )["temp"]
    .mean()
)

monthly_summary["year"] = (
    monthly_summary["month"].dt.year
)

# Khadijah excluded the 1979 boundary period.
monthly_summary = monthly_summary[
    monthly_summary["year"] >= 1980
].copy()

# Stanford only, as required by Rebecca
stanford_df = monthly_summary[
    monthly_summary["city_name"] ==
    "Stanford University"
].copy()

if stanford_df.empty:
    raise ValueError(
        "No rows matched city_name='Stanford University'."
    )

stanford_df["month"] = (
    stanford_df["month"].dt.to_timestamp()
)

stanford_df = (
    stanford_df
    .set_index("month")
    .sort_index()
)

series = (
    stanford_df["temp"]
    .astype(float)
    .asfreq("MS")
    .dropna()
)

print("MONTHLY STANFORD DATA")
print("---------------------")
print(f"First month: {series.index.min():%Y-%m}")
print(f"Last month:  {series.index.max():%Y-%m}")
print(f"Months:      {len(series)}")
print()


# ============================================================
# KHADIJAH'S ORIGINAL VALIDATION PERIOD
# ============================================================

VAL_START = pd.Timestamp("2024-01-01")
VAL_END = pd.Timestamp("2024-11-01")

train = series[
    (series.index >= "1980-01-01") &
    (series.index < VAL_START)
].copy()

validation = series.loc[
    VAL_START:VAL_END
].copy()

if len(validation) != 11:
    raise ValueError(
        "Khadijah's validation period should contain 11 months "
        f"(January-November 2024), but {len(validation)} were found."
    )

print("ORIGINAL KHADIJAH SPLIT")
print("-----------------------")
print(
    f"Training:   {train.index.min():%Y-%m} "
    f"through {train.index.max():%Y-%m}"
)
print(
    f"Validation: {validation.index.min():%Y-%m} "
    f"through {validation.index.max():%Y-%m}"
)
print(f"Validation months: {len(validation)}")
print()


# ============================================================
# METRIC FUNCTION
# ============================================================

def calculate_metrics(actual, predicted):
    actual = np.asarray(actual, dtype=float)
    predicted = np.asarray(predicted, dtype=float)

    return {
        "RMSE_F": np.sqrt(
            mean_squared_error(actual, predicted)
        ),
        "MAE_F": mean_absolute_error(
            actual, predicted
        ),
        # Positive bias means forecast is too warm.
        "bias_F": np.mean(predicted - actual),
        "R2": r2_score(actual, predicted)
    }


# ============================================================
# MODEL 1: KHADIJAH'S FIXED SARIMA
# ============================================================

with warnings.catch_warnings():
    warnings.simplefilter("ignore")

    sarima_model = SARIMAX(
        train,
        order=(1, 1, 1),
        seasonal_order=(1, 1, 1, 12)
    )

    sarima_fit = sarima_model.fit(
        disp=False
    )

    sarima_forecast = sarima_fit.forecast(
        steps=len(validation)
    )

sarima_forecast.index = validation.index

sarima_metrics = calculate_metrics(
    validation,
    sarima_forecast
)


# ============================================================
# MODEL 2: KHADIJAH'S BASIC PROPHET
# ============================================================

prophet_train = pd.DataFrame({
    "ds": train.index,
    "y": train.values
})

prophet_validation_dates = pd.DataFrame({
    "ds": validation.index
})

with warnings.catch_warnings():
    warnings.simplefilter("ignore")

    prophet_model = Prophet(
        yearly_seasonality=True,
        weekly_seasonality=False,
        daily_seasonality=False
    )

    prophet_model.fit(prophet_train)

    prophet_output = prophet_model.predict(
        prophet_validation_dates
    )

prophet_forecast = pd.Series(
    prophet_output["yhat"].to_numpy(),
    index=validation.index,
    name="prophet_forecast_F"
)

prophet_metrics = calculate_metrics(
    validation,
    prophet_forecast
)


# ============================================================
# MODEL COMPARISON
# ============================================================

model_comparison = pd.DataFrame([
    {
        "model": "khadijah_sarima_111_111",
        "validation_start": VAL_START,
        "validation_end": VAL_END,
        "validation_months": len(validation),
        **sarima_metrics
    },
    {
        "model": "khadijah_prophet_basic",
        "validation_start": VAL_START,
        "validation_end": VAL_END,
        "validation_months": len(validation),
        **prophet_metrics
    }
]).sort_values("RMSE_F").reset_index(drop=True)

predictions = pd.DataFrame({
    "month": validation.index,
    "actual_F": validation.values,
    "sarima_forecast_F": sarima_forecast.values,
    "prophet_forecast_F": prophet_forecast.values
})

predictions["sarima_error_F"] = (
    predictions["sarima_forecast_F"] -
    predictions["actual_F"]
)

predictions["prophet_error_F"] = (
    predictions["prophet_forecast_F"] -
    predictions["actual_F"]
)


# ============================================================
# DISPLAY RESULTS
# ============================================================

print("KHADIJAH MODEL REPRODUCTION")
print("===========================")
print(
    model_comparison.to_string(
        index=False,
        float_format=lambda x: f"{x:.3f}"
    )
)

print("\nMONTHLY PREDICTIONS")
print("===================")
print(
    predictions.to_string(
        index=False,
        float_format=lambda x: f"{x:.3f}"
    )
)


# ============================================================
# PLOT
# ============================================================

plt.figure(figsize=(13, 6))

plt.plot(
    validation.index,
    validation.values,
    marker="o",
    linewidth=2.5,
    label="Actual"
)

plt.plot(
    sarima_forecast.index,
    sarima_forecast.values,
    marker="s",
    linestyle="--",
    label="Khadijah SARIMA"
)

plt.plot(
    prophet_forecast.index,
    prophet_forecast.values,
    marker="^",
    linestyle="--",
    label="Khadijah Prophet"
)

plt.title(
    "Khadijah Temperature-Only Reproduction\n"
    "Stanford University, January-November 2024"
)

plt.xlabel("Month")
plt.ylabel("Monthly Average Temperature (°F)")
plt.grid(alpha=0.3)
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()

PLOT_PATH = (
    f"{OUTPUT_DIR}/"
    "khadijah_2024_reproduction.png"
)

plt.savefig(
    PLOT_PATH,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# SAVE RESULTS
# ============================================================

monthly_summary.to_csv(
    f"{OUTPUT_DIR}/monthly_temperature_data.csv",
    index=False
)

model_comparison.to_csv(
    f"{OUTPUT_DIR}/model_comparison.csv",
    index=False
)

predictions.to_csv(
    f"{OUTPUT_DIR}/validation_predictions.csv",
    index=False
)

print("\nRESULTS SAVED IN:")
print(OUTPUT_DIR)
```

```python
# ============================================================
# FAIR 12-MONTH STABILITY COMPARISON
# Stanford temperature only
#
# Models:
#   1. Seasonal naive
#   2. SARIMA (1,1,1)(1,1,1,12)
#   3. Damped ETS
#   4. Basic additive Prophet
#
# Every model receives:
#   - identical Stanford monthly temperatures
#   - identical expanding training windows
#   - identical 12-month validation windows
#
# Source CSV is not modified.
# ============================================================

%pip install -q pandas numpy matplotlib scikit-learn statsmodels prophet

from google.colab import drive
drive.mount("/content/drive")

import os
import time
import logging
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score
)

from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.tsa.holtwinters import ExponentialSmoothing
from prophet import Prophet


# ============================================================
# SETTINGS
# ============================================================

PROJECT = "/content/drive/MyDrive/Stanford_Weather_12M"

DATA_PATH = (
    f"{PROJECT}/data/"
    "OpenWeather19802026_StanfordOnly.csv"
)

OUTPUT_DIR = (
    f"{PROJECT}/"
    "fair_12m_stability_comparison"
)

CITY = "Stanford University"

# July 2026 appears to be incomplete.
# Preserve it in the source file, but do not use it here.
LAST_COMPLETE_MONTH = pd.Timestamp("2026-06-01")

# Nine independent 12-month forecast windows.
FORECAST_ORIGINS = pd.date_range(
    start="2017-07-01",
    end="2025-07-01",
    freq="12MS"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

logging.getLogger("cmdstanpy").setLevel(logging.WARNING)
logging.getLogger("prophet").setLevel(logging.WARNING)

if not os.path.exists(DATA_PATH):
    raise FileNotFoundError(
        f"File not found:\n{DATA_PATH}"
    )


# ============================================================
# LOAD DATA
# ============================================================

raw = pd.read_csv(DATA_PATH)

required = [
    "dt",
    "timezone",
    "city_name",
    "temp"
]

missing = [
    column for column in required
    if column not in raw.columns
]

if missing:
    raise ValueError(
        f"Missing required columns: {missing}"
    )

df = raw.copy()

df = df[
    df["city_name"] == CITY
].copy()

df["dt"] = pd.to_datetime(
    pd.to_numeric(df["dt"], errors="coerce"),
    unit="s",
    utc=True
)

df["timezone_offset"] = pd.to_timedelta(
    pd.to_numeric(df["timezone"], errors="coerce"),
    unit="s"
)

# Follow the existing Joy/Khadijah local-time method.
df["dt_local"] = (
    df["dt"] +
    df["timezone_offset"]
)

df["temp"] = pd.to_numeric(
    df["temp"],
    errors="coerce"
)

df = df.dropna(
    subset=["dt_local", "temp"]
).copy()

df["month"] = (
    df["dt_local"]
    .dt
    .tz_localize(None)
    .dt
    .to_period("M")
    .dt
    .to_timestamp()
)


# ============================================================
# MONTHLY TEMPERATURE SERIES
# ============================================================

# Preserve the existing aggregation behavior:
# no duplicate removal and no source-file modification.
monthly = (
    df.groupby("month")
    .agg(
        mean_temp_F=("temp", "mean"),
        hourly_rows=("temp", "count"),
        first_timestamp=("dt_local", "min"),
        last_timestamp=("dt_local", "max")
    )
    .sort_index()
)

monthly = monthly[
    (monthly.index >= "1980-01-01") &
    (monthly.index <= LAST_COMPLETE_MONTH)
].copy()

series = (
    monthly["mean_temp_F"]
    .astype(float)
    .asfreq("MS")
)

missing_months = series[
    series.isna()
].index

if len(missing_months) > 0:
    raise ValueError(
        "Missing monthly observations were found:\n"
        f"{list(missing_months)}"
    )

print("COMMON MONTHLY DATA")
print("-------------------")
print(f"Source: {DATA_PATH}")
print(f"City: {CITY}")
print(f"First month: {series.index.min():%Y-%m}")
print(f"Last included month: {series.index.max():%Y-%m}")
print(f"Number of months: {len(series)}")
print(f"Raw hourly rows: {len(df):,}")
print()


# ============================================================
# HELPERS
# ============================================================

def calculate_metrics(actual, predicted):
    actual = np.asarray(actual, dtype=float)
    predicted = np.asarray(predicted, dtype=float)

    return {
        "RMSE_F": float(
            np.sqrt(mean_squared_error(actual, predicted))
        ),
        "MAE_F": float(
            mean_absolute_error(actual, predicted)
        ),
        # Positive bias means forecasts are too warm.
        "bias_F": float(
            np.mean(predicted - actual)
        ),
        "R2": float(
            r2_score(actual, predicted)
        )
    }


def seasonal_naive_forecast(train, test_index):
    predictions = []

    for forecast_month in test_index:
        reference_month = (
            forecast_month -
            pd.DateOffset(years=1)
        )

        if reference_month not in train.index:
            raise ValueError(
                f"Seasonal-naive reference missing: "
                f"{reference_month:%Y-%m}"
            )

        predictions.append(
            train.loc[reference_month]
        )

    return pd.Series(
        predictions,
        index=test_index,
        dtype=float
    )


def fit_sarima(train, test_index):
    with warnings.catch_warnings():
        warnings.simplefilter("ignore")

        model = SARIMAX(
            train,
            order=(1, 1, 1),
            seasonal_order=(1, 1, 1, 12),
            enforce_stationarity=False,
            enforce_invertibility=False
        )

        fitted = model.fit(disp=False)

        forecast = (
            fitted
            .get_forecast(steps=len(test_index))
            .predicted_mean
        )

    forecast.index = test_index
    return forecast.astype(float)


def fit_ets(train, test_index):
    with warnings.catch_warnings():
        warnings.simplefilter("ignore")

        model = ExponentialSmoothing(
            train,
            trend="add",
            damped_trend=True,
            seasonal="add",
            seasonal_periods=12,
            initialization_method="estimated"
        )

        fitted = model.fit(
            optimized=True,
            remove_bias=False
        )

        forecast = fitted.forecast(
            len(test_index)
        )

    forecast.index = test_index
    return forecast.astype(float)


def fit_prophet(train, test_index):
    train_frame = pd.DataFrame({
        "ds": train.index,
        "y": train.values
    })

    future_frame = pd.DataFrame({
        "ds": test_index
    })

    with warnings.catch_warnings():
        warnings.simplefilter("ignore")

        model = Prophet(
            yearly_seasonality=True,
            weekly_seasonality=False,
            daily_seasonality=False,
            seasonality_mode="additive"
        )

        model.fit(train_frame)

        forecast_frame = model.predict(
            future_frame
        )

    return pd.Series(
        forecast_frame["yhat"].to_numpy(),
        index=test_index,
        dtype=float
    )


MODEL_FUNCTIONS = {
    "seasonal_naive_12": seasonal_naive_forecast,
    "sarima_111x111_12": fit_sarima,
    "ets_add_add_damped12": fit_ets,
    "prophet_additive": fit_prophet
}


# ============================================================
# ROLLING 12-MONTH BACKTEST
# ============================================================

score_rows = []
prediction_rows = []
failure_rows = []

for origin in FORECAST_ORIGINS:

    test_index = pd.date_range(
        start=origin,
        periods=12,
        freq="MS"
    )

    train_end = (
        origin -
        pd.offsets.MonthBegin(1)
    )

    train = series.loc[:train_end].copy()
    actual = series.reindex(test_index)

    if len(train) < 120:
        raise ValueError(
            f"Insufficient training history for {origin:%Y-%m}."
        )

    if actual.isna().any():
        raise ValueError(
            f"Incomplete 12-month test window beginning "
            f"{origin:%Y-%m}."
        )

    print(
        f"\nForecast window: "
        f"{test_index.min():%Y-%m} through "
        f"{test_index.max():%Y-%m}"
    )

    print(
        f"Training through: {train.index.max():%Y-%m}"
    )

    for model_name, model_function in MODEL_FUNCTIONS.items():

        start_time = time.perf_counter()

        try:
            predicted = model_function(
                train,
                test_index
            )

            runtime_seconds = (
                time.perf_counter() -
                start_time
            )

            metrics = calculate_metrics(
                actual,
                predicted
            )

            score_rows.append({
                "origin": origin,
                "test_start": test_index.min(),
                "test_end": test_index.max(),
                "train_end": train.index.max(),
                "train_months": len(train),
                "model": model_name,
                "status": "success",
                "runtime_seconds": runtime_seconds,
                **metrics
            })

            for month in test_index:
                prediction_rows.append({
                    "origin": origin,
                    "month": month,
                    "model": model_name,
                    "actual_F": float(actual.loc[month]),
                    "forecast_F": float(predicted.loc[month]),
                    "error_F": float(
                        predicted.loc[month] -
                        actual.loc[month]
                    )
                })

            print(
                f"  {model_name:<24} "
                f"RMSE={metrics['RMSE_F']:.3f}  "
                f"MAE={metrics['MAE_F']:.3f}  "
                f"Bias={metrics['bias_F']:+.3f}"
            )

        except Exception as exc:
            runtime_seconds = (
                time.perf_counter() -
                start_time
            )

            failure_rows.append({
                "origin": origin,
                "model": model_name,
                "error_type": type(exc).__name__,
                "error_message": str(exc),
                "runtime_seconds": runtime_seconds
            })

            print(
                f"  {model_name:<24} FAILED: {exc}"
            )


# ============================================================
# RESULTS TABLES
# ============================================================

scores = pd.DataFrame(score_rows)
predictions = pd.DataFrame(prediction_rows)
failures = pd.DataFrame(failure_rows)

if scores.empty:
    raise RuntimeError(
        "Every model failed. Review the failure table."
    )

successful_scores = scores[
    scores["status"] == "success"
].copy()

# Pooled metrics use all monthly predictions together.
pooled_rows = []

for model_name, group in predictions.groupby("model"):
    metrics = calculate_metrics(
        group["actual_F"],
        group["forecast_F"]
    )

    pooled_rows.append({
        "model": model_name,
        "pooled_months": len(group),
        **metrics
    })

pooled_metrics = pd.DataFrame(pooled_rows)

# Stability metrics summarize performance across forecast years.
stability_summary = (
    successful_scores
    .groupby("model")
    .agg(
        windows_completed=("origin", "count"),
        mean_RMSE_F=("RMSE_F", "mean"),
        median_RMSE_F=("RMSE_F", "median"),
        RMSE_std_F=("RMSE_F", "std"),
        best_window_RMSE_F=("RMSE_F", "min"),
        worst_window_RMSE_F=("RMSE_F", "max"),
        mean_MAE_F=("MAE_F", "mean"),
        mean_bias_F=("bias_F", "mean"),
        worst_absolute_bias_F=(
            "bias_F",
            lambda values: np.max(np.abs(values))
        ),
        total_runtime_seconds=(
            "runtime_seconds",
            "sum"
        )
    )
    .reset_index()
)

stability_summary = stability_summary.merge(
    pooled_metrics,
    on="model",
    how="left",
    suffixes=("", "_pooled")
)

# A descriptive robustness score:
# lower average error and lower year-to-year variation are better.
# This is not used as the sole model-selection rule.
stability_summary["robustness_indicator"] = (
    stability_summary["mean_RMSE_F"] +
    stability_summary["RMSE_std_F"]
)

stability_summary = (
    stability_summary
    .sort_values([
        "mean_RMSE_F",
        "worst_window_RMSE_F",
        "RMSE_std_F"
    ])
    .reset_index(drop=True)
)


# ============================================================
# DISPLAY
# ============================================================

print("\n\nROLLING-WINDOW RESULTS")
print("======================")

display(
    successful_scores[
        [
            "test_start",
            "test_end",
            "model",
            "RMSE_F",
            "MAE_F",
            "bias_F",
            "R2",
            "runtime_seconds"
        ]
    ].sort_values(
        ["test_start", "RMSE_F"]
    ).style.format({
        "RMSE_F": "{:.3f}",
        "MAE_F": "{:.3f}",
        "bias_F": "{:+.3f}",
        "R2": "{:.3f}",
        "runtime_seconds": "{:.2f}"
    })
)

print("\nSTABILITY SUMMARY")
print("=================")

display(
    stability_summary.style.format({
        "mean_RMSE_F": "{:.3f}",
        "median_RMSE_F": "{:.3f}",
        "RMSE_std_F": "{:.3f}",
        "best_window_RMSE_F": "{:.3f}",
        "worst_window_RMSE_F": "{:.3f}",
        "mean_MAE_F": "{:.3f}",
        "mean_bias_F": "{:+.3f}",
        "worst_absolute_bias_F": "{:.3f}",
        "RMSE_F": "{:.3f}",
        "MAE_F": "{:.3f}",
        "bias_F": "{:+.3f}",
        "R2": "{:.3f}",
        "robustness_indicator": "{:.3f}",
        "total_runtime_seconds": "{:.2f}"
    })
)

if not failures.empty:
    print("\nMODEL FAILURES")
    print("==============")
    display(failures)
else:
    print("\nAll models completed every forecast window.")


# ============================================================
# PLOT: RMSE BY FORECAST WINDOW
# ============================================================

plt.figure(figsize=(14, 7))

for model_name, group in (
    successful_scores
    .sort_values("test_start")
    .groupby("model")
):
    plt.plot(
        group["test_start"],
        group["RMSE_F"],
        marker="o",
        linewidth=2,
        label=model_name
    )

plt.title(
    "Stanford Temperature Forecast Stability\n"
    "12-Month RMSE by Forecast Window"
)

plt.xlabel("Beginning of 12-month forecast window")
plt.ylabel("RMSE (°F)")
plt.grid(alpha=0.3)
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()

rmse_plot_path = (
    f"{OUTPUT_DIR}/"
    "rmse_by_forecast_window.png"
)

plt.savefig(
    rmse_plot_path,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# PLOT: MEAN AND WORST RMSE
# ============================================================

plot_summary = (
    stability_summary
    .sort_values("mean_RMSE_F")
    .copy()
)

x = np.arange(len(plot_summary))
width = 0.36

plt.figure(figsize=(12, 6))

plt.bar(
    x - width / 2,
    plot_summary["mean_RMSE_F"],
    width,
    label="Mean 12-month RMSE"
)

plt.bar(
    x + width / 2,
    plot_summary["worst_window_RMSE_F"],
    width,
    label="Worst-window RMSE"
)

plt.xticks(
    x,
    plot_summary["model"],
    rotation=20,
    ha="right"
)

plt.ylabel("RMSE (°F)")
plt.title("Average Accuracy and Worst-Year Risk")
plt.grid(axis="y", alpha=0.3)
plt.legend()
plt.tight_layout()

summary_plot_path = (
    f"{OUTPUT_DIR}/"
    "mean_and_worst_rmse.png"
)

plt.savefig(
    summary_plot_path,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# SAVE ALL RESULTS
# ============================================================

monthly.reset_index().to_csv(
    f"{OUTPUT_DIR}/monthly_data_used.csv",
    index=False
)

successful_scores.to_csv(
    f"{OUTPUT_DIR}/window_model_scores.csv",
    index=False
)

predictions.to_csv(
    f"{OUTPUT_DIR}/all_backtest_predictions.csv",
    index=False
)

stability_summary.to_csv(
    f"{OUTPUT_DIR}/stability_summary.csv",
    index=False
)

failures.to_csv(
    f"{OUTPUT_DIR}/model_failures.csv",
    index=False
)

print("\nFILES SAVED IN:")
print(OUTPUT_DIR)

print("\nNEXT:")
print(
    "Send the STABILITY SUMMARY table. "
    "We will use mean error, worst-window error, variability, "
    "bias and failures to choose the stable 12-month model."
)
```

```python
# ============================================================
# SARIMA RESIDUAL DIAGNOSTICS
#
# Data used:
#   Stanford monthly average temperature
#   January 1980 through June 2026
#
# Diagnostics:
#   1. Out-of-sample rolling-backtest residuals
#   2. Final fitted-model standardized residuals
#   3. Ljung-Box autocorrelation tests
#   4. Monthly bias
#   5. Residual time series, histogram, Q-Q and ACF plots
# ============================================================

%pip install -q pandas numpy matplotlib scipy statsmodels

from google.colab import drive
drive.mount("/content/drive")

import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy import stats

from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.stats.diagnostic import acorr_ljungbox
from statsmodels.stats.stattools import (
    durbin_watson,
    jarque_bera
)
from statsmodels.graphics.tsaplots import plot_acf
from statsmodels.graphics.gofplots import qqplot


# ============================================================
# PATHS
# ============================================================

PROJECT = "/content/drive/MyDrive/Stanford_Weather_12M"

MONTHLY_PATH = (
    f"{PROJECT}/"
    "fair_12m_stability_comparison/"
    "monthly_data_used.csv"
)

BACKTEST_PATH = (
    f"{PROJECT}/"
    "final_12m_challenger_comparison/"
    "all_backtest_predictions.csv"
)

OUTPUT_DIR = (
    f"{PROJECT}/"
    "sarima_residual_diagnostics"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

for path in [MONTHLY_PATH, BACKTEST_PATH]:
    if not os.path.exists(path):
        raise FileNotFoundError(
            f"Required file not found:\n{path}"
        )


# ============================================================
# LOAD MONTHLY DATA
# ============================================================

monthly = pd.read_csv(
    MONTHLY_PATH,
    parse_dates=["month"]
)

series = (
    monthly
    .set_index("month")["mean_temp_F"]
    .astype(float)
    .sort_index()
    .asfreq("MS")
)

# Explicit final production-training period.
series = series.loc[
    "1980-01-01":"2026-06-01"
].copy()

if series.isna().any():
    raise ValueError(
        "Missing months were found in the final training series."
    )

print("FINAL TRAINING DATA")
print("===================")
print(f"First month: {series.index.min():%Y-%m}")
print(f"Last month:  {series.index.max():%Y-%m}")
print(f"Months:      {len(series)}")
print()


# ============================================================
# LOAD OUT-OF-SAMPLE BACKTEST RESIDUALS
# ============================================================

backtests = pd.read_csv(
    BACKTEST_PATH,
    parse_dates=["origin", "month"]
)

sarima_backtests = (
    backtests[
        backtests["model"] ==
        "sarima_111x111_12"
    ]
    .sort_values("month")
    .copy()
)

if len(sarima_backtests) != 108:
    print(
        "Warning: Expected 108 SARIMA backtest predictions, "
        f"but found {len(sarima_backtests)}."
    )

# Residual convention:
# residual = actual - forecast
# Positive residual = model forecast was too cold.
sarima_backtests["residual_F"] = (
    sarima_backtests["actual_F"] -
    sarima_backtests["forecast_F"]
)

sarima_backtests["forecast_bias_F"] = (
    sarima_backtests["forecast_F"] -
    sarima_backtests["actual_F"]
)

sarima_backtests["calendar_month"] = (
    sarima_backtests["month"].dt.month
)

sarima_backtests["month_name"] = (
    sarima_backtests["month"].dt.month_name()
)


# ============================================================
# FIT FINAL SARIMA THROUGH JUNE 2026
# ============================================================

with warnings.catch_warnings():
    warnings.simplefilter("ignore")

    final_model = SARIMAX(
        series,
        order=(1, 1, 1),
        seasonal_order=(1, 1, 1, 12),
        enforce_stationarity=False,
        enforce_invertibility=False
    )

    final_fit = final_model.fit(
        disp=False,
        maxiter=200
    )


# ============================================================
# EXTRACT STANDARDIZED ONE-STEP-AHEAD RESIDUALS
# ============================================================

standardized_errors = np.asarray(
    final_fit
    .filter_results
    .standardized_forecasts_error[0],
    dtype=float
)

# Early observations are affected by model initialization and
# seasonal differencing. Exclude a conservative burn-in period.
model_burn = int(
    getattr(final_fit, "loglikelihood_burn", 0)
)

burn_in = max(
    model_burn,
    24
)

standardized_residuals = pd.Series(
    standardized_errors,
    index=series.index,
    name="standardized_residual"
).iloc[burn_in:]

standardized_residuals = (
    standardized_residuals
    .replace([np.inf, -np.inf], np.nan)
    .dropna()
)

print("FITTED MODEL")
print("============")
print("SARIMA(1,1,1)(1,1,1,12)")
print(f"Converged: {final_fit.mle_retvals.get('converged', 'unknown')}")
print(f"AIC: {final_fit.aic:.3f}")
print(f"BIC: {final_fit.bic:.3f}")
print(f"Burn-in observations excluded: {burn_in}")
print(
    "Usable standardized residuals: "
    f"{len(standardized_residuals)}"
)
print()


# ============================================================
# OUT-OF-SAMPLE ERROR SUMMARY
# ============================================================

out_resid = (
    sarima_backtests["residual_F"]
    .astype(float)
)

out_rmse = float(
    np.sqrt(
        np.mean(
            np.square(out_resid)
        )
    )
)

out_mae = float(
    np.mean(
        np.abs(out_resid)
    )
)

# Forecast bias uses forecast - actual.
out_forecast_bias = float(
    sarima_backtests[
        "forecast_bias_F"
    ].mean()
)

out_residual_sd = float(
    out_resid.std(ddof=1)
)

outlier_threshold = (
    3.0 * out_residual_sd
)

outlier_count = int(
    (
        np.abs(
            out_resid -
            out_resid.mean()
        ) > outlier_threshold
    ).sum()
)

out_summary = pd.DataFrame([{
    "diagnostic_set":
        "rolling_12_month_backtests",
    "observations":
        len(out_resid),
    "start_month":
        sarima_backtests["month"].min(),
    "end_month":
        sarima_backtests["month"].max(),
    "RMSE_F":
        out_rmse,
    "MAE_F":
        out_mae,
    "forecast_bias_F":
        out_forecast_bias,
    "residual_mean_F":
        float(out_resid.mean()),
    "residual_std_F":
        out_residual_sd,
    "minimum_residual_F":
        float(out_resid.min()),
    "maximum_residual_F":
        float(out_resid.max()),
    "three_sigma_outliers":
        outlier_count
}])


# ============================================================
# LJUNG-BOX TESTS
# ============================================================

# Tests whether residual autocorrelation remains.
# Desired result: p-values generally greater than 0.05.
backtest_ljung_box = acorr_ljungbox(
    out_resid,
    lags=[6, 12, 18, 24],
    return_df=True
)

backtest_ljung_box.index.name = "lag"
backtest_ljung_box = (
    backtest_ljung_box
    .reset_index()
)

backtest_ljung_box[
    "residual_source"
] = "out_of_sample_backtest"


fitted_ljung_box = acorr_ljungbox(
    standardized_residuals,
    lags=[6, 12, 18, 24],
    return_df=True
)

fitted_ljung_box.index.name = "lag"
fitted_ljung_box = (
    fitted_ljung_box
    .reset_index()
)

fitted_ljung_box[
    "residual_source"
] = "fitted_standardized_residuals"

ljung_box_results = pd.concat(
    [
        backtest_ljung_box,
        fitted_ljung_box
    ],
    ignore_index=True
)


# ============================================================
# ADDITIONAL TESTS
# ============================================================

jb_stat, jb_pvalue, skewness, kurtosis = (
    jarque_bera(
        standardized_residuals
    )
)

additional_tests = pd.DataFrame([
    {
        "test": "Durbin-Watson",
        "statistic": durbin_watson(
            standardized_residuals
        ),
        "p_value": np.nan,
        "interpretation":
            "A value near 2 suggests little lag-1 autocorrelation."
    },
    {
        "test": "Jarque-Bera normality",
        "statistic": jb_stat,
        "p_value": jb_pvalue,
        "interpretation":
            "p > 0.05 supports approximate normality; "
            "normality is helpful but not strictly required."
    },
    {
        "test": "Residual skewness",
        "statistic": skewness,
        "p_value": np.nan,
        "interpretation":
            "A value near 0 indicates symmetric residuals."
    },
    {
        "test": "Residual kurtosis",
        "statistic": kurtosis,
        "p_value": np.nan,
        "interpretation":
            "A normal distribution has kurtosis near 3."
    }
])


# ============================================================
# ERROR BY CALENDAR MONTH
# ============================================================

month_order = [
    "January",
    "February",
    "March",
    "April",
    "May",
    "June",
    "July",
    "August",
    "September",
    "October",
    "November",
    "December"
]

monthly_error_summary = (
    sarima_backtests
    .groupby(
        ["calendar_month", "month_name"]
    )
    .agg(
        observations=("residual_F", "count"),
        RMSE_F=(
            "residual_F",
            lambda values: float(
                np.sqrt(
                    np.mean(
                        np.square(values)
                    )
                )
            )
        ),
        MAE_F=(
            "residual_F",
            lambda values: float(
                np.mean(
                    np.abs(values)
                )
            )
        ),
        forecast_bias_F=(
            "forecast_bias_F",
            "mean"
        ),
        mean_residual_F=(
            "residual_F",
            "mean"
        )
    )
    .reset_index()
    .sort_values("calendar_month")
)


# ============================================================
# PRINT RESULTS
# ============================================================

print("OUT-OF-SAMPLE RESIDUAL SUMMARY")
print("==============================")
display(
    out_summary.style.format({
        "RMSE_F": "{:.3f}",
        "MAE_F": "{:.3f}",
        "forecast_bias_F": "{:+.3f}",
        "residual_mean_F": "{:+.3f}",
        "residual_std_F": "{:.3f}",
        "minimum_residual_F": "{:+.3f}",
        "maximum_residual_F": "{:+.3f}"
    })
)

print("\nLJUNG-BOX AUTOCORRELATION TESTS")
print("===============================")
display(
    ljung_box_results.style.format({
        "lb_stat": "{:.3f}",
        "lb_pvalue": "{:.4f}"
    })
)

print("\nADDITIONAL FITTED-RESIDUAL TESTS")
print("================================")
display(
    additional_tests.style.format({
        "statistic": "{:.4f}",
        "p_value": "{:.4f}"
    })
)

print("\nERROR BY CALENDAR MONTH")
print("=======================")
display(
    monthly_error_summary.style.format({
        "RMSE_F": "{:.3f}",
        "MAE_F": "{:.3f}",
        "forecast_bias_F": "{:+.3f}",
        "mean_residual_F": "{:+.3f}"
    })
)


# ============================================================
# PLOT 1: OUT-OF-SAMPLE RESIDUALS OVER TIME
# ============================================================

plt.figure(figsize=(14, 5))

plt.plot(
    sarima_backtests["month"],
    sarima_backtests["residual_F"],
    marker="o",
    markersize=3,
    linewidth=1.2
)

plt.axhline(
    0,
    color="black",
    linestyle="--",
    linewidth=1
)

plt.axhline(
    out_resid.mean(),
    color="red",
    linestyle=":",
    label=(
        f"Mean residual = "
        f"{out_resid.mean():+.2f}°F"
    )
)

plt.title(
    "SARIMA Out-of-Sample Residuals\n"
    "Residual = Actual − Forecast"
)

plt.xlabel("Month")
plt.ylabel("Residual (°F)")
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/"
    "out_of_sample_residuals_over_time.png",
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# PLOT 2: HISTOGRAM
# ============================================================

plt.figure(figsize=(9, 5))

plt.hist(
    out_resid,
    bins=18,
    edgecolor="black",
    alpha=0.75
)

plt.axvline(
    0,
    color="black",
    linestyle="--"
)

plt.axvline(
    out_resid.mean(),
    color="red",
    linestyle=":",
    label=(
        f"Mean = "
        f"{out_resid.mean():+.2f}°F"
    )
)

plt.title(
    "Distribution of Out-of-Sample "
    "Forecast Residuals"
)

plt.xlabel("Actual − Forecast (°F)")
plt.ylabel("Frequency")
plt.legend()
plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/"
    "out_of_sample_residual_histogram.png",
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# PLOT 3: BACKTEST RESIDUAL ACF
# ============================================================

fig, ax = plt.subplots(
    figsize=(11, 5)
)

plot_acf(
    out_resid,
    lags=24,
    zero=False,
    ax=ax
)

ax.set_title(
    "Autocorrelation of Out-of-Sample "
    "SARIMA Residuals"
)

ax.set_xlabel("Lag in months")
ax.set_ylabel("Autocorrelation")
plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/"
    "out_of_sample_residual_acf.png",
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# PLOT 4: STANDARDIZED FITTED RESIDUALS
# ============================================================

fig, axes = plt.subplots(
    2,
    2,
    figsize=(14, 10)
)

axes[0, 0].plot(
    standardized_residuals.index,
    standardized_residuals.values,
    linewidth=0.9
)

axes[0, 0].axhline(
    0,
    color="black",
    linestyle="--"
)

axes[0, 0].set_title(
    "Standardized Fitted Residuals"
)

axes[0, 0].set_xlabel("Month")
axes[0, 0].set_ylabel("Standardized residual")
axes[0, 0].grid(alpha=0.3)


axes[0, 1].hist(
    standardized_residuals,
    bins=25,
    edgecolor="black",
    alpha=0.75
)

axes[0, 1].set_title(
    "Standardized Residual Distribution"
)

axes[0, 1].set_xlabel(
    "Standardized residual"
)

axes[0, 1].set_ylabel("Frequency")


plot_acf(
    standardized_residuals,
    lags=36,
    zero=False,
    ax=axes[1, 0]
)

axes[1, 0].set_title(
    "Standardized Residual ACF"
)

axes[1, 0].set_xlabel(
    "Lag in months"
)


qqplot(
    standardized_residuals,
    line="45",
    fit=True,
    ax=axes[1, 1]
)

axes[1, 1].set_title(
    "Standardized Residual Q-Q Plot"
)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/"
    "fitted_residual_diagnostics.png",
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# PLOT 5: BIAS BY CALENDAR MONTH
# ============================================================

plt.figure(figsize=(12, 5))

bar_colors = np.where(
    monthly_error_summary[
        "forecast_bias_F"
    ] >= 0,
    "tomato",
    "steelblue"
)

plt.bar(
    monthly_error_summary["month_name"],
    monthly_error_summary["forecast_bias_F"],
    color=bar_colors
)

plt.axhline(
    0,
    color="black",
    linestyle="--"
)

plt.title(
    "SARIMA Forecast Bias by Calendar Month\n"
    "Positive = forecast too warm"
)

plt.xlabel("Calendar month")
plt.ylabel("Mean forecast bias (°F)")
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/"
    "forecast_bias_by_calendar_month.png",
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# SAVE OUTPUTS
# ============================================================

sarima_backtests.to_csv(
    f"{OUTPUT_DIR}/"
    "sarima_backtest_residuals.csv",
    index=False
)

standardized_residuals.rename_axis(
    "month"
).reset_index().to_csv(
    f"{OUTPUT_DIR}/"
    "sarima_fitted_standardized_residuals.csv",
    index=False
)

out_summary.to_csv(
    f"{OUTPUT_DIR}/"
    "out_of_sample_residual_summary.csv",
    index=False
)

ljung_box_results.to_csv(
    f"{OUTPUT_DIR}/"
    "ljung_box_tests.csv",
    index=False
)

additional_tests.to_csv(
    f"{OUTPUT_DIR}/"
    "additional_residual_tests.csv",
    index=False
)

monthly_error_summary.to_csv(
    f"{OUTPUT_DIR}/"
    "error_by_calendar_month.csv",
    index=False
)

print("\nDIAGNOSTICS SAVED IN:")
print(OUTPUT_DIR)

print("\nSEND ME THESE THREE OUTPUTS:")
print("1. OUT-OF-SAMPLE RESIDUAL SUMMARY")
print("2. LJUNG-BOX AUTOCORRELATION TESTS")
print("3. ERROR BY CALENDAR MONTH")
```

```python
# ============================================================
# SARIMA PREDICTION-INTERVAL CALIBRATION
#
# Compares:
#   1. Native SARIMA 80% and 95% intervals
#   2. Empirically calibrated 80% and 95% intervals
#
# Empirical validation uses leave-one-window-out calibration:
# the errors from a forecast window are never used to
# calibrate that same window.
#
# Final calibration widths use all 108 historical
# out-of-sample SARIMA errors.
# ============================================================

%pip install -q pandas numpy matplotlib statsmodels

from google.colab import drive
drive.mount("/content/drive")

import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from statsmodels.tsa.statespace.sarimax import SARIMAX


# ============================================================
# PATHS
# ============================================================

PROJECT = "/content/drive/MyDrive/Stanford_Weather_12M"

MONTHLY_PATH = (
    f"{PROJECT}/"
    "fair_12m_stability_comparison/"
    "monthly_data_used.csv"
)

BACKTEST_PATH = (
    f"{PROJECT}/"
    "final_12m_challenger_comparison/"
    "all_backtest_predictions.csv"
)

OUTPUT_DIR = (
    f"{PROJECT}/"
    "sarima_interval_calibration"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

for path in [MONTHLY_PATH, BACKTEST_PATH]:
    if not os.path.exists(path):
        raise FileNotFoundError(
            f"Required file not found:\n{path}"
        )


# ============================================================
# LOAD MONTHLY SERIES
# ============================================================

monthly = pd.read_csv(
    MONTHLY_PATH,
    parse_dates=["month"]
)

series = (
    monthly
    .set_index("month")["mean_temp_F"]
    .astype(float)
    .sort_index()
    .asfreq("MS")
    .loc["1980-01-01":"2026-06-01"]
)

if series.isna().any():
    raise ValueError(
        "Missing months were found in the temperature series."
    )


# ============================================================
# LOAD SAVED SARIMA BACKTEST PREDICTIONS
# ============================================================

all_predictions = pd.read_csv(
    BACKTEST_PATH,
    parse_dates=["origin", "month"]
)

sarima_predictions = (
    all_predictions[
        all_predictions["model"] ==
        "sarima_111x111_12"
    ]
    .sort_values(
        ["origin", "month"]
    )
    .copy()
)

if len(sarima_predictions) != 108:
    raise ValueError(
        "Expected 108 SARIMA predictions "
        f"(9 windows × 12 months), but found "
        f"{len(sarima_predictions)}."
    )

# Make the residual definition explicit.
sarima_predictions["residual_F"] = (
    sarima_predictions["actual_F"] -
    sarima_predictions["forecast_F"]
)

sarima_predictions["absolute_error_F"] = (
    sarima_predictions["residual_F"].abs()
)

# Forecast horizon: 1 through 12 within every window.
sarima_predictions["forecast_horizon"] = (
    sarima_predictions
    .groupby("origin")
    .cumcount() + 1
)


# ============================================================
# RECREATE NATIVE SARIMA INTERVALS
# ============================================================

native_rows = []

origins = (
    sarima_predictions["origin"]
    .drop_duplicates()
    .sort_values()
)

print("RECREATING NATIVE SARIMA INTERVALS")
print("==================================")

for origin in origins:

    origin = pd.Timestamp(origin)

    test_index = pd.date_range(
        start=origin,
        periods=12,
        freq="MS"
    )

    train_end = (
        origin -
        pd.offsets.MonthBegin(1)
    )

    train = series.loc[:train_end].copy()
    actual = series.reindex(test_index)

    if actual.isna().any():
        raise ValueError(
            f"Incomplete test data for origin {origin:%Y-%m}."
        )

    with warnings.catch_warnings():
        warnings.simplefilter("ignore")

        model = SARIMAX(
            train,
            order=(1, 1, 1),
            seasonal_order=(1, 1, 1, 12),
            enforce_stationarity=False,
            enforce_invertibility=False
        )

        fitted = model.fit(
            disp=False,
            maxiter=200
        )

        forecast_result = fitted.get_forecast(
            steps=12
        )

        point_forecast = (
            forecast_result.predicted_mean
        )

        interval_80 = (
            forecast_result.conf_int(
                alpha=0.20
            )
        )

        interval_95 = (
            forecast_result.conf_int(
                alpha=0.05
            )
        )

    point_forecast.index = test_index
    interval_80.index = test_index
    interval_95.index = test_index

    for horizon, month in enumerate(
        test_index,
        start=1
    ):
        native_rows.append({
            "origin": origin,
            "month": month,
            "forecast_horizon": horizon,
            "actual_F": float(
                actual.loc[month]
            ),
            "forecast_F": float(
                point_forecast.loc[month]
            ),
            "native_80_lower_F": float(
                interval_80.loc[month].iloc[0]
            ),
            "native_80_upper_F": float(
                interval_80.loc[month].iloc[1]
            ),
            "native_95_lower_F": float(
                interval_95.loc[month].iloc[0]
            ),
            "native_95_upper_F": float(
                interval_95.loc[month].iloc[1]
            )
        })

    print(
        f"Completed {test_index.min():%Y-%m} "
        f"through {test_index.max():%Y-%m}"
    )

native_intervals = pd.DataFrame(
    native_rows
)


# ============================================================
# NATIVE INTERVAL COVERAGE
# ============================================================

native_intervals["covered_80"] = (
    (
        native_intervals["actual_F"] >=
        native_intervals["native_80_lower_F"]
    ) &
    (
        native_intervals["actual_F"] <=
        native_intervals["native_80_upper_F"]
    )
)

native_intervals["covered_95"] = (
    (
        native_intervals["actual_F"] >=
        native_intervals["native_95_lower_F"]
    ) &
    (
        native_intervals["actual_F"] <=
        native_intervals["native_95_upper_F"]
    )
)

native_intervals["width_80_F"] = (
    native_intervals["native_80_upper_F"] -
    native_intervals["native_80_lower_F"]
)

native_intervals["width_95_F"] = (
    native_intervals["native_95_upper_F"] -
    native_intervals["native_95_lower_F"]
)


# ============================================================
# FINITE-SAMPLE CONFORMAL QUANTILE
# ============================================================

def conformal_quantile(
    absolute_errors,
    coverage
):
    """
    Finite-sample split-conformal quantile.

    For desired coverage 1-alpha:
        k = ceil((n + 1) * (1-alpha))

    The kth ordered absolute residual becomes the symmetric
    interval half-width.
    """

    values = np.sort(
        np.asarray(
            absolute_errors,
            dtype=float
        )
    )

    values = values[
        np.isfinite(values)
    ]

    n = len(values)

    if n == 0:
        raise ValueError(
            "No calibration residuals were supplied."
        )

    rank = int(
        np.ceil(
            (n + 1) * coverage
        )
    )

    rank = min(
        max(rank, 1),
        n
    )

    return float(
        values[rank - 1]
    )


# ============================================================
# LEAVE-ONE-WINDOW-OUT EMPIRICAL VALIDATION
# ============================================================

empirical_rows = []

for held_out_origin in origins:

    held_out_origin = pd.Timestamp(
        held_out_origin
    )

    calibration = sarima_predictions[
        sarima_predictions["origin"] !=
        held_out_origin
    ].copy()

    evaluation = sarima_predictions[
        sarima_predictions["origin"] ==
        held_out_origin
    ].copy()

    # Pool residuals across horizons because eight residuals
    # per horizon are too few for reliable 95% calibration.
    half_width_80 = conformal_quantile(
        calibration["absolute_error_F"],
        coverage=0.80
    )

    half_width_95 = conformal_quantile(
        calibration["absolute_error_F"],
        coverage=0.95
    )

    for _, row in evaluation.iterrows():

        lower_80 = (
            row["forecast_F"] -
            half_width_80
        )

        upper_80 = (
            row["forecast_F"] +
            half_width_80
        )

        lower_95 = (
            row["forecast_F"] -
            half_width_95
        )

        upper_95 = (
            row["forecast_F"] +
            half_width_95
        )

        empirical_rows.append({
            "origin": held_out_origin,
            "month": row["month"],
            "forecast_horizon":
                row["forecast_horizon"],
            "actual_F":
                row["actual_F"],
            "forecast_F":
                row["forecast_F"],
            "calibration_observations":
                len(calibration),
            "empirical_80_half_width_F":
                half_width_80,
            "empirical_80_lower_F":
                lower_80,
            "empirical_80_upper_F":
                upper_80,
            "empirical_95_half_width_F":
                half_width_95,
            "empirical_95_lower_F":
                lower_95,
            "empirical_95_upper_F":
                upper_95,
            "covered_80":
                lower_80 <=
                row["actual_F"] <=
                upper_80,
            "covered_95":
                lower_95 <=
                row["actual_F"] <=
                upper_95
        })

empirical_intervals = pd.DataFrame(
    empirical_rows
)

empirical_intervals["width_80_F"] = (
    2 *
    empirical_intervals[
        "empirical_80_half_width_F"
    ]
)

empirical_intervals["width_95_F"] = (
    2 *
    empirical_intervals[
        "empirical_95_half_width_F"
    ]
)


# ============================================================
# COVERAGE SUMMARY
# ============================================================

coverage_summary = pd.DataFrame([
    {
        "method": "native_sarima",
        "target_coverage": 0.80,
        "observed_coverage": float(
            native_intervals[
                "covered_80"
            ].mean()
        ),
        "covered_months": int(
            native_intervals[
                "covered_80"
            ].sum()
        ),
        "total_months": len(
            native_intervals
        ),
        "mean_interval_width_F": float(
            native_intervals[
                "width_80_F"
            ].mean()
        )
    },
    {
        "method": "native_sarima",
        "target_coverage": 0.95,
        "observed_coverage": float(
            native_intervals[
                "covered_95"
            ].mean()
        ),
        "covered_months": int(
            native_intervals[
                "covered_95"
            ].sum()
        ),
        "total_months": len(
            native_intervals
        ),
        "mean_interval_width_F": float(
            native_intervals[
                "width_95_F"
            ].mean()
        )
    },
    {
        "method":
            "empirical_leave_one_window_out",
        "target_coverage": 0.80,
        "observed_coverage": float(
            empirical_intervals[
                "covered_80"
            ].mean()
        ),
        "covered_months": int(
            empirical_intervals[
                "covered_80"
            ].sum()
        ),
        "total_months": len(
            empirical_intervals
        ),
        "mean_interval_width_F": float(
            empirical_intervals[
                "width_80_F"
            ].mean()
        )
    },
    {
        "method":
            "empirical_leave_one_window_out",
        "target_coverage": 0.95,
        "observed_coverage": float(
            empirical_intervals[
                "covered_95"
            ].mean()
        ),
        "covered_months": int(
            empirical_intervals[
                "covered_95"
            ].sum()
        ),
        "total_months": len(
            empirical_intervals
        ),
        "mean_interval_width_F": float(
            empirical_intervals[
                "width_95_F"
            ].mean()
        )
    }
])


# ============================================================
# COVERAGE BY FORECAST HORIZON
# ============================================================

native_by_horizon = (
    native_intervals
    .groupby("forecast_horizon")
    .agg(
        observations=("actual_F", "count"),
        coverage_80=("covered_80", "mean"),
        coverage_95=("covered_95", "mean"),
        mean_width_80_F=("width_80_F", "mean"),
        mean_width_95_F=("width_95_F", "mean")
    )
    .reset_index()
)

native_by_horizon["method"] = (
    "native_sarima"
)

empirical_by_horizon = (
    empirical_intervals
    .groupby("forecast_horizon")
    .agg(
        observations=("actual_F", "count"),
        coverage_80=("covered_80", "mean"),
        coverage_95=("covered_95", "mean"),
        mean_width_80_F=("width_80_F", "mean"),
        mean_width_95_F=("width_95_F", "mean")
    )
    .reset_index()
)

empirical_by_horizon["method"] = (
    "empirical_leave_one_window_out"
)

coverage_by_horizon = pd.concat(
    [
        native_by_horizon,
        empirical_by_horizon
    ],
    ignore_index=True
)


# ============================================================
# FINAL EMPIRICAL CALIBRATION WIDTHS
# ============================================================

# These use all 108 historical out-of-sample residuals.
# They will be applied to the final July 2026-June 2027
# point forecasts if empirical intervals are selected.

final_half_width_80 = conformal_quantile(
    sarima_predictions["absolute_error_F"],
    coverage=0.80
)

final_half_width_95 = conformal_quantile(
    sarima_predictions["absolute_error_F"],
    coverage=0.95
)

final_calibration = pd.DataFrame([
    {
        "model": "sarima_111x111_12",
        "calibration_method":
            "pooled_absolute_out_of_sample_errors",
        "calibration_observations":
            len(sarima_predictions),
        "target_coverage":
            0.80,
        "half_width_F":
            final_half_width_80,
        "full_width_F":
            2 * final_half_width_80
    },
    {
        "model": "sarima_111x111_12",
        "calibration_method":
            "pooled_absolute_out_of_sample_errors",
        "calibration_observations":
            len(sarima_predictions),
        "target_coverage":
            0.95,
        "half_width_F":
            final_half_width_95,
        "full_width_F":
            2 * final_half_width_95
    }
])


# ============================================================
# DISPLAY RESULTS
# ============================================================

print("\nINTERVAL COVERAGE SUMMARY")
print("=========================")

display(
    coverage_summary.style.format({
        "target_coverage": "{:.0%}",
        "observed_coverage": "{:.1%}",
        "mean_interval_width_F": "{:.3f}"
    })
)

print("\nCOVERAGE BY FORECAST HORIZON")
print("============================")

display(
    coverage_by_horizon.style.format({
        "coverage_80": "{:.1%}",
        "coverage_95": "{:.1%}",
        "mean_width_80_F": "{:.3f}",
        "mean_width_95_F": "{:.3f}"
    })
)

print("\nFINAL EMPIRICAL CALIBRATION WIDTHS")
print("==================================")

display(
    final_calibration.style.format({
        "target_coverage": "{:.0%}",
        "half_width_F": "{:.3f}",
        "full_width_F": "{:.3f}"
    })
)


# ============================================================
# PLOT COVERAGE COMPARISON
# ============================================================

plot_data = coverage_summary.copy()

plot_data["label"] = (
    plot_data["method"] +
    " " +
    (
        plot_data["target_coverage"] *
        100
    ).astype(int).astype(str) +
    "%"
)

x = np.arange(
    len(plot_data)
)

plt.figure(figsize=(11, 6))

bars = plt.bar(
    x,
    plot_data["observed_coverage"],
    color=[
        "steelblue",
        "royalblue",
        "darkorange",
        "orangered"
    ]
)

plt.scatter(
    x,
    plot_data["target_coverage"],
    color="black",
    marker="D",
    s=55,
    label="Target coverage",
    zorder=3
)

plt.xticks(
    x,
    plot_data["label"],
    rotation=20,
    ha="right"
)

plt.ylim(0, 1.05)
plt.ylabel("Observed historical coverage")

plt.title(
    "SARIMA Prediction-Interval Calibration\n"
    "108 Out-of-Sample Monthly Forecasts"
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.legend()
plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/"
    "interval_coverage_comparison.png",
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# PLOT COVERAGE BY HORIZON
# ============================================================

plt.figure(figsize=(13, 6))

for method, group in (
    coverage_by_horizon
    .groupby("method")
):
    plt.plot(
        group["forecast_horizon"],
        group["coverage_80"],
        marker="o",
        label=f"{method}: 80%"
    )

    plt.plot(
        group["forecast_horizon"],
        group["coverage_95"],
        marker="s",
        linestyle="--",
        label=f"{method}: 95%"
    )

plt.axhline(
    0.80,
    color="gray",
    linestyle=":",
    label="80% target"
)

plt.axhline(
    0.95,
    color="black",
    linestyle=":",
    label="95% target"
)

plt.xticks(
    range(1, 13)
)

plt.ylim(0, 1.05)

plt.xlabel("Forecast horizon in months")
plt.ylabel("Observed coverage")

plt.title(
    "Prediction-Interval Coverage "
    "by Forecast Horizon"
)

plt.grid(alpha=0.3)
plt.legend(
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/"
    "coverage_by_forecast_horizon.png",
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# SAVE RESULTS
# ============================================================

native_intervals.to_csv(
    f"{OUTPUT_DIR}/"
    "native_interval_backtests.csv",
    index=False
)

empirical_intervals.to_csv(
    f"{OUTPUT_DIR}/"
    "empirical_interval_backtests.csv",
    index=False
)

coverage_summary.to_csv(
    f"{OUTPUT_DIR}/"
    "interval_coverage_summary.csv",
    index=False
)

coverage_by_horizon.to_csv(
    f"{OUTPUT_DIR}/"
    "coverage_by_horizon.csv",
    index=False
)

final_calibration.to_csv(
    f"{OUTPUT_DIR}/"
    "final_empirical_calibration.csv",
    index=False
)

print("\nRESULTS SAVED IN:")
print(OUTPUT_DIR)

print("\nSEND ME:")
print("1. INTERVAL COVERAGE SUMMARY")
print("2. FINAL EMPIRICAL CALIBRATION WIDTHS")
```

```python
# ============================================================
# FINAL PRODUCTION RUN — STANFORD 12-MONTH TEMPERATURE MODEL
#
# Selected model: SARIMA(1,1,1)(1,1,1,12)
# Training data:  January 1980 through June 2026
# Forecast:       July 2026 through June 2027
#
# Official uncertainty intervals:
# Native SARIMA 80% and 95% prediction intervals
# ============================================================

from google.colab import drive
drive.mount("/content/drive")

# ============================================================
# IMPORTS
# ============================================================

import os
import sys
import json
import hashlib
import platform
import warnings
from datetime import datetime, timezone

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels

from statsmodels.tsa.statespace.sarimax import SARIMAX
from IPython.display import display


# ============================================================
# PATHS
# ============================================================

PROJECT = "/content/drive/MyDrive/Stanford_Weather_12M"

MONTHLY_FILE = (
    f"{PROJECT}/fair_12m_stability_comparison/"
    "monthly_data_used.csv"
)

RAW_FILE = (
    f"{PROJECT}/data/"
    "OpenWeather19802026_StanfordOnly.csv"
)

OUTPUT_DIR = (
    f"{PROJECT}/final_12m_production_forecast"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

FORECAST_FILE = f"{OUTPUT_DIR}/official_forecast_Jul2026_Jun2027.csv"
TRAINING_FILE = f"{OUTPUT_DIR}/final_training_series.csv"
MODEL_FILE = f"{OUTPUT_DIR}/sarima_111x111_12_fitted_model.pkl"
PARAMETERS_FILE = f"{OUTPUT_DIR}/fitted_model_parameters.csv"
METADATA_FILE = f"{OUTPUT_DIR}/reproducibility_metadata.json"
SUMMARY_FILE = f"{OUTPUT_DIR}/model_summary.txt"
PLOT_FILE = f"{OUTPUT_DIR}/official_forecast_plot.png"
REQUIREMENTS_FILE = f"{OUTPUT_DIR}/package_versions.txt"


# ============================================================
# HELPER FUNCTIONS
# ============================================================

def file_sha256(path, chunk_size=1024 * 1024):
    """Calculate a SHA-256 fingerprint without loading entire file."""
    if not os.path.exists(path):
        return None

    digest = hashlib.sha256()

    with open(path, "rb") as file:
        while True:
            chunk = file.read(chunk_size)

            if not chunk:
                break

            digest.update(chunk)

    return digest.hexdigest()


def find_column(columns, possible_names):
    """Find a column regardless of capitalization."""
    lookup = {str(col).lower().strip(): col for col in columns}

    for name in possible_names:
        if name.lower() in lookup:
            return lookup[name.lower()]

    return None


# ============================================================
# LOAD THE AUDITED MONTHLY DATA
# ============================================================

if not os.path.exists(MONTHLY_FILE):
    raise FileNotFoundError(
        f"Audited monthly file was not found:\n{MONTHLY_FILE}\n\n"
        "Run the fair stability-comparison cell first."
    )

monthly = pd.read_csv(MONTHLY_FILE)

month_column = find_column(
    monthly.columns,
    ["month", "date", "ds"]
)

temperature_column = find_column(
    monthly.columns,
    [
        "mean_temp_F",
        "temp",
        "temperature",
        "temperature_F",
        "y"
    ]
)

if month_column is None:
    raise ValueError(
        f"Could not identify the month column. "
        f"Available columns: {monthly.columns.tolist()}"
    )

if temperature_column is None:
    raise ValueError(
        f"Could not identify the temperature column. "
        f"Available columns: {monthly.columns.tolist()}"
    )

monthly[month_column] = pd.to_datetime(
    monthly[month_column],
    errors="coerce"
)

monthly[temperature_column] = pd.to_numeric(
    monthly[temperature_column],
    errors="coerce"
)

monthly = (
    monthly[[month_column, temperature_column]]
    .rename(
        columns={
            month_column: "month",
            temperature_column: "mean_temp_F"
        }
    )
    .dropna()
    .sort_values("month")
    .drop_duplicates(subset="month", keep="last")
    .reset_index(drop=True)
)

# Retain only the previously approved final training period.
# July 2026 is not used because it was incomplete in the source.
monthly = monthly[
    (monthly["month"] >= "1980-01-01") &
    (monthly["month"] <= "2026-06-01")
].copy()

series = (
    monthly
    .set_index("month")["mean_temp_F"]
    .astype(float)
    .asfreq("MS")
)


# ============================================================
# REQUIRED DATA VALIDATION
# ============================================================

expected_index = pd.date_range(
    start="1980-01-01",
    end="2026-06-01",
    freq="MS"
)

if len(series) != 558:
    raise ValueError(
        f"Expected 558 monthly observations, but found {len(series)}."
    )

if not series.index.equals(expected_index):
    missing_months = expected_index.difference(series.index)
    extra_months = series.index.difference(expected_index)

    raise ValueError(
        "The monthly sequence is not the expected continuous "
        "January 1980–June 2026 series.\n"
        f"Missing months: {missing_months.tolist()}\n"
        f"Extra months: {extra_months.tolist()}"
    )

if series.isna().any():
    missing = series[series.isna()].index.strftime("%Y-%m").tolist()

    raise ValueError(
        f"Missing temperatures were found for: {missing}"
    )

if not np.isfinite(series).all():
    raise ValueError("The training series contains non-finite values.")


print("FINAL TRAINING DATA")
print("===================")
print(f"Source:       {MONTHLY_FILE}")
print(f"First month:  {series.index.min():%Y-%m}")
print(f"Last month:   {series.index.max():%Y-%m}")
print(f"Observations: {len(series)}")
print(f"Missing:      {int(series.isna().sum())}")
print()


# ============================================================
# FIT THE SELECTED PRODUCTION MODEL
# ============================================================

ORDER = (1, 1, 1)
SEASONAL_ORDER = (1, 1, 1, 12)

with warnings.catch_warnings():
    warnings.simplefilter("ignore")

    final_model = SARIMAX(
        series,
        order=ORDER,
        seasonal_order=SEASONAL_ORDER,
        trend=None,
        enforce_stationarity=False,
        enforce_invertibility=False
    )

    final_fit = final_model.fit(
        disp=False,
        maxiter=200
    )

converged = bool(
    final_fit.mle_retvals.get("converged", False)
)

if not converged:
    raise RuntimeError(
        "The final SARIMA optimization did not converge. "
        "Do not publish the forecast until this is investigated."
    )


print("FITTED PRODUCTION MODEL")
print("=======================")
print(f"Model:     SARIMA{ORDER}{SEASONAL_ORDER}")
print(f"Converged: {converged}")
print(f"AIC:       {final_fit.aic:.3f}")
print(f"BIC:       {final_fit.bic:.3f}")
print()


# ============================================================
# CREATE THE OFFICIAL 12-MONTH FORECAST
# ============================================================

forecast_result = final_fit.get_forecast(steps=12)

point_forecast = np.asarray(
    forecast_result.predicted_mean,
    dtype=float
)

interval_80 = np.asarray(
    forecast_result.conf_int(alpha=0.20),
    dtype=float
)

interval_95 = np.asarray(
    forecast_result.conf_int(alpha=0.05),
    dtype=float
)

forecast_months = pd.date_range(
    start="2026-07-01",
    periods=12,
    freq="MS"
)

forecast = pd.DataFrame({
    "month": forecast_months,
    "mean_temp_F": point_forecast,

    # Official validated native SARIMA intervals
    "lower_80_F": interval_80[:, 0],
    "upper_80_F": interval_80[:, 1],
    "lower_95_F": interval_95[:, 0],
    "upper_95_F": interval_95[:, 1],

    "model": "SARIMA(1,1,1)(1,1,1,12)",
    "trained_through": pd.Timestamp("2026-06-01"),
    "interval_method": "native_sarima"
})

# Sensitivity intervals based on the 108 historical out-of-sample
# forecast errors. These are retained for comparison, but are not
# the official intervals.
EMPIRICAL_80_HALF_WIDTH = 2.420
EMPIRICAL_95_HALF_WIDTH = 4.071

forecast["sensitivity_lower_80_F"] = (
    forecast["mean_temp_F"] - EMPIRICAL_80_HALF_WIDTH
)

forecast["sensitivity_upper_80_F"] = (
    forecast["mean_temp_F"] + EMPIRICAL_80_HALF_WIDTH
)

forecast["sensitivity_lower_95_F"] = (
    forecast["mean_temp_F"] - EMPIRICAL_95_HALF_WIDTH
)

forecast["sensitivity_upper_95_F"] = (
    forecast["mean_temp_F"] + EMPIRICAL_95_HALF_WIDTH
)


print("OFFICIAL JULY 2026–JUNE 2027 FORECAST")
print("======================================")

display(
    forecast[
        [
            "month",
            "mean_temp_F",
            "lower_80_F",
            "upper_80_F",
            "lower_95_F",
            "upper_95_F"
        ]
    ].round(3)
)


# ============================================================
# SAVE TRAINING DATA, FORECAST AND MODEL
# ============================================================

training_output = (
    series
    .rename("mean_temp_F")
    .reset_index()
    .rename(columns={"index": "month"})
)

training_output.to_csv(
    TRAINING_FILE,
    index=False
)

forecast.to_csv(
    FORECAST_FILE,
    index=False
)

# Statsmodels fitted-results object.
# Loading this file later requires a compatible statsmodels version.
final_fit.save(
    MODEL_FILE,
    remove_data=False
)

parameters = pd.DataFrame({
    "parameter": final_fit.params.index,
    "estimate": final_fit.params.values
})

parameters.to_csv(
    PARAMETERS_FILE,
    index=False
)

with open(SUMMARY_FILE, "w") as file:
    file.write(final_fit.summary().as_text())


# ============================================================
# FORECAST PLOT
# ============================================================

plot_history = series.loc["2016-01-01":]

fig, ax = plt.subplots(figsize=(14, 7))

ax.plot(
    plot_history.index,
    plot_history.values,
    color="black",
    linewidth=1.7,
    label="Observed monthly average"
)

ax.plot(
    forecast["month"],
    forecast["mean_temp_F"],
    color="#B1040E",
    linewidth=2.3,
    marker="o",
    label="Official SARIMA forecast"
)

ax.fill_between(
    forecast["month"],
    forecast["lower_95_F"],
    forecast["upper_95_F"],
    color="#B1040E",
    alpha=0.13,
    label="95% prediction interval"
)

ax.fill_between(
    forecast["month"],
    forecast["lower_80_F"],
    forecast["upper_80_F"],
    color="#B1040E",
    alpha=0.25,
    label="80% prediction interval"
)

ax.axvline(
    pd.Timestamp("2026-06-01"),
    color="gray",
    linestyle="--",
    linewidth=1.3,
    label="Training cutoff"
)

ax.set_title(
    "Stanford Monthly Average Temperature Forecast\n"
    "SARIMA(1,1,1)(1,1,1,12)"
)

ax.set_xlabel("Month")
ax.set_ylabel("Monthly average temperature (°F)")
ax.grid(alpha=0.25)
ax.legend(loc="best")

fig.tight_layout()
fig.savefig(
    PLOT_FILE,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# REPRODUCIBILITY METADATA
# ============================================================

generated_utc = datetime.now(timezone.utc).isoformat()

metadata = {
    "project": "Stanford 12-month temperature forecasting",
    "artifact_status": "production_candidate_for_review",

    "model": {
        "model_class": "SARIMAX",
        "display_name": "SARIMA(1,1,1)(1,1,1,12)",
        "order": list(ORDER),
        "seasonal_order": list(SEASONAL_ORDER),
        "trend": None,
        "enforce_stationarity": False,
        "enforce_invertibility": False,
        "converged": converged,
        "aic": float(final_fit.aic),
        "bic": float(final_fit.bic),
        "parameters": {
            str(name): float(value)
            for name, value in final_fit.params.items()
        }
    },

    "training_data": {
        "city": "Stanford University",
        "target": "monthly mean temperature in degrees Fahrenheit",
        "first_month": "1980-01-01",
        "last_month": "2026-06-01",
        "observations": int(len(series)),
        "monthly_frequency": "MS",
        "monthly_file": MONTHLY_FILE,
        "monthly_file_sha256": file_sha256(MONTHLY_FILE),
        "raw_hourly_file": RAW_FILE,
        "raw_hourly_file_sha256": file_sha256(RAW_FILE),
        "exclusion_note": (
            "July 2026 source observations were excluded because "
            "the month was incomplete. The raw source was not modified."
        )
    },

    "forecast": {
        "forecast_start": "2026-07-01",
        "forecast_end": "2027-06-01",
        "horizon_months": 12,
        "official_interval_method": "native SARIMA",
        "official_interval_levels": ["80%", "95%"],
        "sensitivity_interval_method": (
            "pooled absolute errors from 108 rolling "
            "out-of-sample monthly forecasts"
        ),
        "empirical_80_half_width_F": EMPIRICAL_80_HALF_WIDTH,
        "empirical_95_half_width_F": EMPIRICAL_95_HALF_WIDTH
    },

    "historical_validation": {
        "rolling_windows": 9,
        "forecast_months_evaluated": 108,
        "window_start": "2017-07-01",
        "window_end": "2026-06-01",
        "mean_RMSE_F": 1.953,
        "median_RMSE_F": 1.860,
        "RMSE_std_F": 0.528,
        "worst_window_RMSE_F": 2.789,
        "pooled_RMSE_F": 2.016,
        "pooled_MAE_F": 1.564,
        "pooled_bias_F": -0.001,
        "native_80_observed_coverage": 0.843,
        "native_95_observed_coverage": 0.954
    },

    "software": {
        "python": platform.python_version(),
        "pandas": pd.__version__,
        "numpy": np.__version__,
        "statsmodels": statsmodels.__version__,
        "platform": platform.platform()
    },

    "generated_at_utc": generated_utc,

    "output_files": {
        "forecast": FORECAST_FILE,
        "training_series": TRAINING_FILE,
        "fitted_model": MODEL_FILE,
        "model_parameters": PARAMETERS_FILE,
        "model_summary": SUMMARY_FILE,
        "forecast_plot": PLOT_FILE
    }
}

with open(METADATA_FILE, "w") as file:
    json.dump(
        metadata,
        file,
        indent=2
    )

package_versions = [
    f"python=={platform.python_version()}",
    f"numpy=={np.__version__}",
    f"pandas=={pd.__version__}",
    f"statsmodels=={statsmodels.__version__}",
    f"matplotlib=={plt.matplotlib.__version__}"
]

with open(REQUIREMENTS_FILE, "w") as file:
    file.write("\n".join(package_versions) + "\n")


# ============================================================
# FINAL VERIFICATION
# ============================================================

required_outputs = [
    FORECAST_FILE,
    TRAINING_FILE,
    MODEL_FILE,
    PARAMETERS_FILE,
    METADATA_FILE,
    SUMMARY_FILE,
    PLOT_FILE,
    REQUIREMENTS_FILE
]

missing_outputs = [
    path for path in required_outputs
    if not os.path.exists(path)
]

if missing_outputs:
    raise RuntimeError(
        f"Some final files were not created:\n{missing_outputs}"
    )


print()
print("FINAL PRODUCTION ARTIFACTS SAVED")
print("================================")
print(f"Directory:       {OUTPUT_DIR}")
print(f"Forecast:        {FORECAST_FILE}")
print(f"Fitted model:    {MODEL_FILE}")
print(f"Training data:   {TRAINING_FILE}")
print(f"Parameters:      {PARAMETERS_FILE}")
print(f"Metadata:        {METADATA_FILE}")
print(f"Model summary:   {SUMMARY_FILE}")
print(f"Forecast plot:   {PLOT_FILE}")
print(f"Package versions:{REQUIREMENTS_FILE}")
print()
print("The model is now packaged as a production candidate for review.")
```

```python
# ============================================================
# EXACT MULTI-STATION FILE
# ============================================================

FULL_DATA_FILE = (
    "/content/drive/MyDrive/Stanford_Weather_12M/"
    "OpenWeatherMap_Data_5_stations.csv"
)

if not os.path.exists(FULL_DATA_FILE):
    raise FileNotFoundError(
        f"File not found:\n{FULL_DATA_FILE}"
    )

print("MULTI-STATION SOURCE FILE")
print("=========================")
print(FULL_DATA_FILE)
print(f"File size: {os.path.getsize(FULL_DATA_FILE) / 1024**2:,.1f} MB")
print()


# ============================================================
# LOAD ONLY COLUMNS NEEDED FOR THIS TEST
# ============================================================

available_columns = pd.read_csv(
    FULL_DATA_FILE,
    nrows=0
).columns.tolist()

print("AVAILABLE COLUMNS")
print("=================")
print(available_columns)
print()

required_columns = [
    "city_name",
    "temp"
]

if "dt" in available_columns:
    required_columns.append("dt")
elif "dt_iso" in available_columns:
    required_columns.append("dt_iso")
else:
    raise ValueError(
        "The file has neither a 'dt' nor a 'dt_iso' column."
    )

if "timezone" in available_columns:
    required_columns.append("timezone")

missing_columns = [
    column for column in ["city_name", "temp"]
    if column not in available_columns
]

if missing_columns:
    raise ValueError(
        f"Required columns are missing: {missing_columns}"
    )

raw = pd.read_csv(
    FULL_DATA_FILE,
    usecols=required_columns,
    low_memory=False
)

raw["city_name"] = (
    raw["city_name"]
    .astype(str)
    .str.strip()
)

raw["temp"] = pd.to_numeric(
    raw["temp"],
    errors="coerce"
)


# ============================================================
# CONVERT TIME
# ============================================================

if "dt" in raw.columns:

    raw["dt_utc"] = pd.to_datetime(
        pd.to_numeric(raw["dt"], errors="coerce"),
        unit="s",
        utc=True,
        errors="coerce"
    )

    if "timezone" in raw.columns:
        timezone_seconds = pd.to_numeric(
            raw["timezone"],
            errors="coerce"
        ).fillna(0)

        raw["dt_local"] = (
            raw["dt_utc"].dt.tz_localize(None)
            + pd.to_timedelta(timezone_seconds, unit="s")
        )

    else:
        raw["dt_local"] = (
            raw["dt_utc"].dt.tz_localize(None)
        )

else:

    raw["dt_local"] = pd.to_datetime(
        raw["dt_iso"]
        .astype(str)
        .str.replace(
            " +0000 UTC",
            "",
            regex=False
        ),
        errors="coerce"
    )


raw = raw.dropna(
    subset=["city_name", "temp", "dt_local"]
)


# ============================================================
# PRINT EVERY CITY/STATION
# ============================================================

city_inventory = (
    raw
    .groupby("city_name")
    .agg(
        hourly_rows=("temp", "size"),
        first_observation=("dt_local", "min"),
        last_observation=("dt_local", "max"),
        missing_temperature_rows=("temp", lambda x: x.isna().sum()),
        latitude=(
            "temp",
            lambda x: np.nan
        )
    )
    .reset_index()
    .drop(columns="latitude")
    .sort_values("city_name")
    .reset_index(drop=True)
)

print("ALL CITIES/STATIONS IN THE FILE")
print("================================")
display(city_inventory)

print()
print(f"Number of cities/stations: {raw['city_name'].nunique()}")
print()

for number, city in enumerate(
    sorted(raw["city_name"].unique()),
    start=1
):
    print(f"{number}. {city}")


# ============================================================
# IDENTIFY STANFORD, MENLO PARK AND PALO ALTO
# ============================================================

all_cities = sorted(raw["city_name"].unique())

stanford_candidates = [
    city for city in all_cities
    if "stanford" in city.lower()
]

menlo_candidates = [
    city for city in all_cities
    if "menlo" in city.lower()
]

palo_alto_candidates = [
    city for city in all_cities
    if "palo" in city.lower()
]

print()
print("AUTOMATIC LOCATION MATCHES")
print("==========================")
print(f"Stanford matches:  {stanford_candidates}")
print(f"Menlo matches:     {menlo_candidates}")
print(f"Palo Alto matches: {palo_alto_candidates}")

if not stanford_candidates:
    raise ValueError(
        "No Stanford station was identified."
    )

STANFORD_NAME = stanford_candidates[0]

MENLO_NAME = (
    menlo_candidates[0]
    if menlo_candidates
    else None
)

PALO_ALTO_NAME = (
    palo_alto_candidates[0]
    if palo_alto_candidates
    else None
)

print()
print("SELECTED LOCATIONS")
print("==================")
print(f"Stanford:  {STANFORD_NAME}")
print(f"Menlo:     {MENLO_NAME}")
print(f"Palo Alto: {PALO_ALTO_NAME}")
```

```python
# ============================================================
# TEST WHETHER NEIGHBORING STATIONS IMPROVE STANFORD UNIVERSITY
#
# Requires the previous inventory cell to have run, creating `raw`.
#
# Target: Stanford University
# Challengers:
#   1. Stanford-only SARIMA
#   2. SARIMAX + forecasted Menlo Park
#   3. SARIMAX + forecasted Palo Alto
#   4. SARIMAX + forecasted Redwood City
#   5. SARIMAX + forecasted Stanford station
#   6. SARIMAX + forecasted regional average
#
# No actual future neighboring-city temperatures are supplied.
# Each external temperature series is itself forecast first.
# ============================================================

import os
import time
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from statsmodels.tsa.statespace.sarimax import SARIMAX
from IPython.display import display


# ============================================================
# CONFIGURATION
# ============================================================

PROJECT = "/content/drive/MyDrive/Stanford_Weather_12M"

OUTPUT_DIR = (
    f"{PROJECT}/neighboring_city_challenger_test"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

TARGET_CITY = "Stanford University"

NEIGHBORING_CITIES = [
    "Menlo Park",
    "Palo Alto",
    "Redwood City",
    "Stanford"
]

ORDER = (1, 1, 1)
SEASONAL_ORDER = (1, 1, 1, 12)

DATA_START = pd.Timestamp("1980-01-01")

# December 2024 is partial, so exclude it.
DATA_END = pd.Timestamp("2024-11-01")


# ============================================================
# CREATE MONTHLY TEMPERATURE DATA
# ============================================================

raw["month"] = (
    raw["dt_local"]
    .dt.to_period("M")
    .dt.to_timestamp()
)

monthly_long = (
    raw
    .groupby(
        ["city_name", "month"],
        as_index=False
    )["temp"]
    .agg(
        mean_temp_F="mean",
        hourly_observations="count"
    )
)

monthly_temperature = (
    monthly_long
    .pivot(
        index="month",
        columns="city_name",
        values="mean_temp_F"
    )
    .sort_index()
)

monthly_counts = (
    monthly_long
    .pivot(
        index="month",
        columns="city_name",
        values="hourly_observations"
    )
    .sort_index()
)

expected_index = pd.date_range(
    DATA_START,
    DATA_END,
    freq="MS"
)

monthly_temperature = (
    monthly_temperature
    .reindex(expected_index)
)

monthly_counts = (
    monthly_counts
    .reindex(expected_index)
)

monthly_temperature.index.name = "month"
monthly_counts.index.name = "month"


# ============================================================
# VERIFY LOCATIONS
# ============================================================

required_cities = [
    TARGET_CITY,
    *NEIGHBORING_CITIES
]

missing_cities = [
    city for city in required_cities
    if city not in monthly_temperature.columns
]

if missing_cities:
    raise ValueError(
        f"Missing required stations: {missing_cities}"
    )

print("MONTHLY DATA USED FOR LOCATION TEST")
print("===================================")
print(f"Target station: {TARGET_CITY}")
print(f"First month:    {monthly_temperature.index.min():%Y-%m}")
print(f"Last month:     {monthly_temperature.index.max():%Y-%m}")
print(f"Possible months:{len(monthly_temperature)}")
print()


coverage_rows = []

for city in required_cities:
    coverage_rows.append({
        "city": city,
        "months_available":
            int(monthly_temperature[city].notna().sum()),
        "missing_months":
            int(monthly_temperature[city].isna().sum()),
        "coverage_percent":
            100 * monthly_temperature[city].notna().mean(),
        "minimum_hourly_observations_in_month":
            monthly_counts[city].min()
    })

coverage = pd.DataFrame(coverage_rows)

print("MONTHLY COVERAGE")
print("================")
display(coverage.round(2))


# ============================================================
# REGIONAL AVERAGE
#
# This uses only the four non-target stations.
# ============================================================

monthly_temperature["Regional average"] = (
    monthly_temperature[NEIGHBORING_CITIES]
    .mean(axis=1, skipna=True)
)

external_series = {
    "forecasted_Menlo_Park":
        monthly_temperature["Menlo Park"],

    "forecasted_Palo_Alto":
        monthly_temperature["Palo Alto"],

    "forecasted_Redwood_City":
        monthly_temperature["Redwood City"],

    "forecasted_Stanford_station":
        monthly_temperature["Stanford"],

    "forecasted_regional_average":
        monthly_temperature["Regional average"]
}

target = monthly_temperature[TARGET_CITY].astype(float)


# ============================================================
# MODEL FUNCTIONS
# ============================================================

def prepare_training_series(series):
    """
    Fill occasional internal missing training observations.
    This function is applied separately inside each historical
    training period and cannot use observations from the test period.
    """

    series = series.astype(float).copy()

    series = series.interpolate(
        method="time",
        limit_direction="both"
    )

    return series


def fit_univariate_forecast(
    training_series,
    steps=12
):
    """Fit fixed SARIMA and produce a 12-month forecast."""

    training_series = prepare_training_series(
        training_series
    )

    if training_series.isna().any():
        raise ValueError(
            "Training series still contains missing values."
        )

    with warnings.catch_warnings():
        warnings.simplefilter("ignore")

        model = SARIMAX(
            training_series,
            order=ORDER,
            seasonal_order=SEASONAL_ORDER,
            trend=None,
            enforce_stationarity=False,
            enforce_invertibility=False
        )

        fit = model.fit(
            disp=False,
            maxiter=200
        )

    if not fit.mle_retvals.get("converged", False):
        raise RuntimeError(
            "Univariate SARIMA failed to converge."
        )

    return np.asarray(
        fit.get_forecast(
            steps=steps
        ).predicted_mean,
        dtype=float
    )


def fit_external_location_challenger(
    target_train,
    external_train,
    steps=12
):
    """
    First forecast the neighboring location.

    Then use that forecast—not its actual future values—as the
    future external input to the Stanford University model.
    """

    common_training = pd.concat(
        [
            target_train.rename("target"),
            external_train.rename("external")
        ],
        axis=1
    )

    common_training["target"] = (
        prepare_training_series(
            common_training["target"]
        )
    )

    common_training["external"] = (
        prepare_training_series(
            common_training["external"]
        )
    )

    common_training = common_training.dropna()

    if len(common_training) < 120:
        raise ValueError(
            "Insufficient common training history."
        )

    # Forecast the neighboring station first.
    external_forecast = fit_univariate_forecast(
        common_training["external"],
        steps=steps
    )

    future_index = pd.date_range(
        common_training.index[-1]
        + pd.offsets.MonthBegin(1),
        periods=steps,
        freq="MS"
    )

    future_external = pd.DataFrame(
        {
            "external": external_forecast
        },
        index=future_index
    )

    with warnings.catch_warnings():
        warnings.simplefilter("ignore")

        stanford_model = SARIMAX(
            common_training["target"],
            exog=common_training[["external"]],
            order=ORDER,
            seasonal_order=SEASONAL_ORDER,
            trend=None,
            enforce_stationarity=False,
            enforce_invertibility=False
        )

        stanford_fit = stanford_model.fit(
            disp=False,
            maxiter=200
        )

    if not stanford_fit.mle_retvals.get(
        "converged",
        False
    ):
        raise RuntimeError(
            "Stanford SARIMAX challenger failed to converge."
        )

    stanford_forecast = (
        stanford_fit
        .get_forecast(
            steps=steps,
            exog=future_external
        )
        .predicted_mean
    )

    return np.asarray(
        stanford_forecast,
        dtype=float
    )


def error_metrics(actual, forecast):

    actual = np.asarray(actual, dtype=float)
    forecast = np.asarray(forecast, dtype=float)

    errors = forecast - actual

    return {
        "RMSE_F": np.sqrt(np.mean(errors ** 2)),
        "MAE_F": np.mean(np.abs(errors)),
        "bias_F": np.mean(errors)
    }


# ============================================================
# ROLLING TEST WINDOWS
#
# Nine 12-month windows, all ending before the partial
# December 2024 observations.
# ============================================================

TEST_STARTS = pd.date_range(
    start="2015-07-01",
    end="2023-07-01",
    freq="12MS"
)

results = []
predictions = []

print()
print("ROLLING LOCATION CHALLENGER TEST")
print("================================")

for test_start in TEST_STARTS:

    test_end = (
        test_start
        + pd.DateOffset(months=11)
    )

    training_end = (
        test_start
        - pd.offsets.MonthBegin(1)
    )

    target_train = target.loc[:training_end]
    actual = target.loc[test_start:test_end]

    if len(actual) != 12 or actual.isna().any():
        print(
            f"Skipping {test_start:%Y-%m}: "
            "target test data incomplete."
        )
        continue

    print()
    print(
        f"Test window: {test_start:%Y-%m} "
        f"through {test_end:%Y-%m}"
    )

    # --------------------------------------------------------
    # Stanford University-only baseline
    # --------------------------------------------------------

    try:
        start_time = time.time()

        baseline_forecast = fit_univariate_forecast(
            target_train,
            steps=12
        )

        metrics = error_metrics(
            actual,
            baseline_forecast
        )

        results.append({
            "test_start": test_start,
            "test_end": test_end,
            "model": "Stanford_University_only",
            **metrics,
            "runtime_seconds":
                time.time() - start_time,
            "status": "completed"
        })

        for horizon, (
            month,
            actual_value,
            forecast_value
        ) in enumerate(
            zip(
                actual.index,
                actual.values,
                baseline_forecast
            ),
            start=1
        ):
            predictions.append({
                "test_start": test_start,
                "month": month,
                "forecast_horizon": horizon,
                "model": "Stanford_University_only",
                "actual_F": actual_value,
                "forecast_F": forecast_value,
                "error_F":
                    forecast_value - actual_value
            })

        print(
            "  Stanford University only: "
            f"RMSE={metrics['RMSE_F']:.3f}, "
            f"MAE={metrics['MAE_F']:.3f}, "
            f"Bias={metrics['bias_F']:+.3f}"
        )

    except Exception as error:
        results.append({
            "test_start": test_start,
            "test_end": test_end,
            "model": "Stanford_University_only",
            "RMSE_F": np.nan,
            "MAE_F": np.nan,
            "bias_F": np.nan,
            "runtime_seconds": np.nan,
            "status": f"failed: {error}"
        })

        print(
            f"  Stanford University only FAILED: {error}"
        )

    # --------------------------------------------------------
    # Neighboring-location challengers
    # --------------------------------------------------------

    for location_name, location_series in external_series.items():

        try:
            start_time = time.time()

            external_train = (
                location_series.loc[:training_end]
            )

            challenger_forecast = (
                fit_external_location_challenger(
                    target_train=target_train,
                    external_train=external_train,
                    steps=12
                )
            )

            metrics = error_metrics(
                actual,
                challenger_forecast
            )

            results.append({
                "test_start": test_start,
                "test_end": test_end,
                "model": location_name,
                **metrics,
                "runtime_seconds":
                    time.time() - start_time,
                "status": "completed"
            })

            for horizon, (
                month,
                actual_value,
                forecast_value
            ) in enumerate(
                zip(
                    actual.index,
                    actual.values,
                    challenger_forecast
                ),
                start=1
            ):
                predictions.append({
                    "test_start": test_start,
                    "month": month,
                    "forecast_horizon": horizon,
                    "model": location_name,
                    "actual_F": actual_value,
                    "forecast_F": forecast_value,
                    "error_F":
                        forecast_value - actual_value
                })

            print(
                f"  {location_name}: "
                f"RMSE={metrics['RMSE_F']:.3f}, "
                f"MAE={metrics['MAE_F']:.3f}, "
                f"Bias={metrics['bias_F']:+.3f}"
            )

        except Exception as error:
            results.append({
                "test_start": test_start,
                "test_end": test_end,
                "model": location_name,
                "RMSE_F": np.nan,
                "MAE_F": np.nan,
                "bias_F": np.nan,
                "runtime_seconds":
                    time.time() - start_time,
                "status": f"failed: {error}"
            })

            print(
                f"  {location_name} FAILED: {error}"
            )


results = pd.DataFrame(results)
predictions = pd.DataFrame(predictions)


# ============================================================
# STABILITY SUMMARY
# ============================================================

completed_results = results[
    results["status"] == "completed"
].copy()

summary_rows = []

for model_name, model_results in (
    completed_results.groupby("model")
):

    model_predictions = predictions[
        predictions["model"] == model_name
    ]

    pooled_errors = (
        model_predictions["error_F"]
        .to_numpy()
    )

    summary_rows.append({
        "model": model_name,
        "windows_completed": len(model_results),
        "mean_RMSE_F":
            model_results["RMSE_F"].mean(),
        "median_RMSE_F":
            model_results["RMSE_F"].median(),
        "RMSE_std_F":
            model_results["RMSE_F"].std(ddof=0),
        "best_window_RMSE_F":
            model_results["RMSE_F"].min(),
        "worst_window_RMSE_F":
            model_results["RMSE_F"].max(),
        "mean_MAE_F":
            model_results["MAE_F"].mean(),
        "mean_bias_F":
            model_results["bias_F"].mean(),
        "pooled_months": len(pooled_errors),
        "pooled_RMSE_F":
            np.sqrt(np.mean(pooled_errors ** 2)),
        "pooled_MAE_F":
            np.mean(np.abs(pooled_errors)),
        "pooled_bias_F":
            np.mean(pooled_errors)
    })

summary = (
    pd.DataFrame(summary_rows)
    .sort_values("pooled_RMSE_F")
    .reset_index(drop=True)
)


# ============================================================
# COMPARE AGAINST STANFORD UNIVERSITY-ONLY BASELINE
# ============================================================

baseline = summary[
    summary["model"]
    == "Stanford_University_only"
].iloc[0]

comparison_rows = []

for _, challenger in summary.iterrows():

    if challenger["model"] == "Stanford_University_only":
        continue

    pooled_improvement = (
        baseline["pooled_RMSE_F"]
        - challenger["pooled_RMSE_F"]
    )

    improvement_percent = (
        100
        * pooled_improvement
        / baseline["pooled_RMSE_F"]
    )

    # Conservative rule:
    # require at least 0.10°F lower pooled RMSE,
    # no worse worst-window performance,
    # and all nine windows completed.
    meaningful_improvement = (
        challenger["windows_completed"] == 9
        and pooled_improvement >= 0.10
        and challenger["worst_window_RMSE_F"]
            <= baseline["worst_window_RMSE_F"]
    )

    comparison_rows.append({
        "challenger": challenger["model"],
        "baseline_pooled_RMSE_F":
            baseline["pooled_RMSE_F"],
        "challenger_pooled_RMSE_F":
            challenger["pooled_RMSE_F"],
        "RMSE_improvement_F":
            pooled_improvement,
        "RMSE_improvement_percent":
            improvement_percent,
        "baseline_worst_RMSE_F":
            baseline["worst_window_RMSE_F"],
        "challenger_worst_RMSE_F":
            challenger["worst_window_RMSE_F"],
        "meaningful_stable_improvement":
            meaningful_improvement
    })

comparison = (
    pd.DataFrame(comparison_rows)
    .sort_values(
        "challenger_pooled_RMSE_F"
    )
    .reset_index(drop=True)
)


# ============================================================
# DISPLAY RESULTS
# ============================================================

print()
print("LOCATION CHALLENGER STABILITY SUMMARY")
print("=====================================")
display(summary.round(3))

print()
print("IMPROVEMENT OVER STANFORD-ONLY")
print("==============================")
display(comparison.round(3))


# ============================================================
# DECISION
# ============================================================

qualified = comparison[
    comparison["meaningful_stable_improvement"]
]

print()
print("AUTOMATIC DECISION")
print("==================")

if qualified.empty:

    print(
        "Neighboring locations did not provide a sufficiently "
        "large and stable improvement."
    )

    print(
        "Recommendation: retain the Stanford University-only "
        "SARIMA model."
    )

else:

    winner = qualified.iloc[0]

    print(
        f"Best qualifying challenger: "
        f"{winner['challenger']}"
    )

    print(
        f"Pooled RMSE improvement: "
        f"{winner['RMSE_improvement_F']:.3f} °F "
        f"({winner['RMSE_improvement_percent']:.1f}%)"
    )

    print(
        "Recommendation: present this challenger to Rebecca, "
        "but obtain approval before changing the Stanford-only "
        "production model."
    )


# ============================================================
# PLOT RESULTS
# ============================================================

plot_data = summary.copy()

x = np.arange(len(plot_data))
width = 0.36

fig, ax = plt.subplots(figsize=(13, 6))

ax.bar(
    x - width / 2,
    plot_data["pooled_RMSE_F"],
    width,
    label="Pooled RMSE"
)

ax.bar(
    x + width / 2,
    plot_data["worst_window_RMSE_F"],
    width,
    label="Worst-window RMSE"
)

ax.set_xticks(x)

ax.set_xticklabels(
    plot_data["model"],
    rotation=25,
    ha="right"
)

ax.set_ylabel("Forecast error (°F)")

ax.set_title(
    "Does Neighboring-Station Weather Improve "
    "Stanford University Forecasts?"
)

ax.grid(
    axis="y",
    alpha=0.25
)

ax.legend()

fig.tight_layout()

PLOT_FILE = (
    f"{OUTPUT_DIR}/neighboring_station_comparison.png"
)

fig.savefig(
    PLOT_FILE,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# SAVE OUTPUTS
# ============================================================

coverage.to_csv(
    f"{OUTPUT_DIR}/station_coverage.csv",
    index=False
)

results.to_csv(
    f"{OUTPUT_DIR}/rolling_window_results.csv",
    index=False
)

predictions.to_csv(
    f"{OUTPUT_DIR}/monthly_predictions.csv",
    index=False
)

summary.to_csv(
    f"{OUTPUT_DIR}/stability_summary.csv",
    index=False
)

comparison.to_csv(
    f"{OUTPUT_DIR}/comparison_to_stanford_only.csv",
    index=False
)

print()
print("RESULTS SAVED IN:")
print(OUTPUT_DIR)

print()
print("SEND ME THESE TWO TABLES:")
print("1. LOCATION CHALLENGER STABILITY SUMMARY")
print("2. IMPROVEMENT OVER STANFORD-ONLY")
```

```python
# ============================================================
# FAIR MATCHED-WINDOW LOCATION COMPARISON
#
# Each challenger is compared only on months/windows for which
# both it and the Stanford University-only baseline succeeded.
# ============================================================

import os
import numpy as np
import pandas as pd
from IPython.display import display

PROJECT = "/content/drive/MyDrive/Stanford_Weather_12M"

OUTPUT_DIR = (
    f"{PROJECT}/neighboring_city_challenger_test"
)

PREDICTIONS_FILE = (
    f"{OUTPUT_DIR}/monthly_predictions.csv"
)

predictions = pd.read_csv(
    PREDICTIONS_FILE,
    parse_dates=["test_start", "month"]
)

BASELINE = "Stanford_University_only"

baseline_predictions = (
    predictions[
        predictions["model"] == BASELINE
    ]
    [
        [
            "test_start",
            "month",
            "actual_F",
            "forecast_F",
            "error_F"
        ]
    ]
    .rename(
        columns={
            "forecast_F": "baseline_forecast_F",
            "error_F": "baseline_error_F"
        }
    )
)

challenger_names = sorted(
    predictions.loc[
        predictions["model"] != BASELINE,
        "model"
    ].unique()
)

comparison_rows = []
window_rows = []

for challenger_name in challenger_names:

    challenger_predictions = (
        predictions[
            predictions["model"] == challenger_name
        ]
        [
            [
                "test_start",
                "month",
                "forecast_F",
                "error_F"
            ]
        ]
        .rename(
            columns={
                "forecast_F": "challenger_forecast_F",
                "error_F": "challenger_error_F"
            }
        )
    )

    matched = baseline_predictions.merge(
        challenger_predictions,
        on=["test_start", "month"],
        how="inner"
    )

    # Keep only complete 12-month windows for both models.
    complete_windows = (
        matched
        .groupby("test_start")
        .size()
    )

    complete_windows = complete_windows[
        complete_windows == 12
    ].index

    matched = matched[
        matched["test_start"].isin(complete_windows)
    ].copy()

    if matched.empty:
        continue

    baseline_rmse = np.sqrt(
        np.mean(
            matched["baseline_error_F"] ** 2
        )
    )

    challenger_rmse = np.sqrt(
        np.mean(
            matched["challenger_error_F"] ** 2
        )
    )

    baseline_mae = np.mean(
        np.abs(matched["baseline_error_F"])
    )

    challenger_mae = np.mean(
        np.abs(matched["challenger_error_F"])
    )

    baseline_bias = np.mean(
        matched["baseline_error_F"]
    )

    challenger_bias = np.mean(
        matched["challenger_error_F"]
    )

    baseline_window_rmse = (
        matched
        .groupby("test_start")["baseline_error_F"]
        .apply(
            lambda errors:
                np.sqrt(np.mean(errors ** 2))
        )
    )

    challenger_window_rmse = (
        matched
        .groupby("test_start")["challenger_error_F"]
        .apply(
            lambda errors:
                np.sqrt(np.mean(errors ** 2))
        )
    )

    window_comparison = pd.DataFrame({
        "test_start": baseline_window_rmse.index,
        "baseline_RMSE_F":
            baseline_window_rmse.values,
        "challenger_RMSE_F":
            challenger_window_rmse.values
    })

    window_comparison["challenger"] = (
        challenger_name
    )

    window_comparison["RMSE_improvement_F"] = (
        window_comparison["baseline_RMSE_F"]
        - window_comparison["challenger_RMSE_F"]
    )

    window_comparison["winner"] = np.where(
        window_comparison["challenger_RMSE_F"]
        < window_comparison["baseline_RMSE_F"],
        challenger_name,
        BASELINE
    )

    window_rows.append(window_comparison)

    improvement_F = (
        baseline_rmse - challenger_rmse
    )

    improvement_percent = (
        100 * improvement_F / baseline_rmse
    )

    challenger_wins = int(
        (
            challenger_window_rmse
            < baseline_window_rmse
        ).sum()
    )

    baseline_wins = int(
        (
            baseline_window_rmse
            < challenger_window_rmse
        ).sum()
    )

    # Conservative operational requirement.
    meaningful_improvement = (
        improvement_F >= 0.10
        and challenger_window_rmse.max()
            <= baseline_window_rmse.max()
    )

    comparison_rows.append({
        "challenger": challenger_name,
        "matched_windows": len(complete_windows),
        "matched_months": len(matched),

        "baseline_pooled_RMSE_F":
            baseline_rmse,

        "challenger_pooled_RMSE_F":
            challenger_rmse,

        "RMSE_improvement_F":
            improvement_F,

        "RMSE_improvement_percent":
            improvement_percent,

        "baseline_pooled_MAE_F":
            baseline_mae,

        "challenger_pooled_MAE_F":
            challenger_mae,

        "baseline_bias_F":
            baseline_bias,

        "challenger_bias_F":
            challenger_bias,

        "baseline_worst_window_RMSE_F":
            baseline_window_rmse.max(),

        "challenger_worst_window_RMSE_F":
            challenger_window_rmse.max(),

        "challenger_window_wins":
            challenger_wins,

        "baseline_window_wins":
            baseline_wins,

        "meaningful_stable_improvement":
            meaningful_improvement
    })


matched_summary = (
    pd.DataFrame(comparison_rows)
    .sort_values(
        "challenger_pooled_RMSE_F"
    )
    .reset_index(drop=True)
)

matched_window_results = pd.concat(
    window_rows,
    ignore_index=True
)

print("FAIR MATCHED-WINDOW COMPARISON")
print("==============================")

display(
    matched_summary.round(3)
)

print()
print("FINAL LOCATION DECISION")
print("=======================")

qualified = matched_summary[
    matched_summary[
        "meaningful_stable_improvement"
    ]
]

if qualified.empty:
    print(
        "No neighboring-location challenger provides a "
        "meaningful and stable improvement."
    )
    print(
        "Retain the Stanford University-only SARIMA model."
    )
else:
    print(
        "At least one challenger meets the predefined "
        "improvement requirement."
    )

matched_summary.to_csv(
    f"{OUTPUT_DIR}/fair_matched_window_summary.csv",
    index=False
)

matched_window_results.to_csv(
    f"{OUTPUT_DIR}/fair_matched_window_details.csv",
    index=False
)

print()
print("Saved:")
print(
    f"{OUTPUT_DIR}/fair_matched_window_summary.csv"
)
```

```python
# ============================================================
# FINAL MULTIVARIATE CHALLENGER
#
# Stanford University target plus four separately forecasted
# external station-temperature predictors.
# ============================================================

ALL_EXTERNAL_CITIES = [
    "Menlo Park",
    "Palo Alto",
    "Redwood City",
    "Stanford"
]


def fit_all_stations_challenger(
    target_train,
    external_train,
    steps=12
):
    """
    Forecast each neighboring station separately and supply all
    four forecasts to the Stanford University SARIMAX model.
    """

    joined = pd.concat(
        [
            target_train.rename("target"),
            external_train
        ],
        axis=1
    )

    joined["target"] = (
        joined["target"]
        .astype(float)
        .interpolate(
            method="time",
            limit_direction="both"
        )
    )

    for city in ALL_EXTERNAL_CITIES:
        joined[city] = (
            joined[city]
            .astype(float)
            .interpolate(
                method="time",
                limit_direction="both"
            )
        )

    joined = joined.dropna()

    if len(joined) < 120:
        raise ValueError(
            "Insufficient common training history."
        )

    future_index = pd.date_range(
        joined.index[-1] + pd.offsets.MonthBegin(1),
        periods=steps,
        freq="MS"
    )

    future_external = pd.DataFrame(
        index=future_index
    )

    # Forecast each neighboring station independently.
    for city in ALL_EXTERNAL_CITIES:

        city_forecast = fit_univariate_forecast(
            joined[city],
            steps=steps
        )

        future_external[city] = city_forecast

    with warnings.catch_warnings():
        warnings.simplefilter("ignore")

        model = SARIMAX(
            joined["target"],
            exog=joined[ALL_EXTERNAL_CITIES],
            order=ORDER,
            seasonal_order=SEASONAL_ORDER,
            trend=None,
            enforce_stationarity=False,
            enforce_invertibility=False
        )

        fit = model.fit(
            disp=False,
            maxiter=500
        )

    if not fit.mle_retvals.get("converged", False):
        raise RuntimeError(
            "All-stations SARIMAX did not converge."
        )

    forecast = (
        fit
        .get_forecast(
            steps=steps,
            exog=future_external[ALL_EXTERNAL_CITIES]
        )
        .predicted_mean
    )

    return np.asarray(
        forecast,
        dtype=float
    )


# ============================================================
# RUN THE SAME ROLLING WINDOWS
# ============================================================

all_station_results = []
all_station_predictions = []

print("ALL-STATIONS-SEPARATELY CHALLENGER")
print("==================================")

for test_start in TEST_STARTS:

    test_end = test_start + pd.DateOffset(months=11)
    training_end = test_start - pd.offsets.MonthBegin(1)

    target_train = target.loc[:training_end]

    external_train = (
        monthly_temperature[
            ALL_EXTERNAL_CITIES
        ]
        .loc[:training_end]
    )

    actual = target.loc[test_start:test_end]

    if len(actual) != 12 or actual.isna().any():
        continue

    try:
        forecast = fit_all_stations_challenger(
            target_train=target_train,
            external_train=external_train,
            steps=12
        )

        metrics = error_metrics(
            actual,
            forecast
        )

        all_station_results.append({
            "test_start": test_start,
            "test_end": test_end,
            "model":
                "Stanford_plus_all_stations_separately",
            **metrics,
            "status": "completed"
        })

        for horizon, (
            month,
            actual_value,
            forecast_value
        ) in enumerate(
            zip(
                actual.index,
                actual.values,
                forecast
            ),
            start=1
        ):
            all_station_predictions.append({
                "test_start": test_start,
                "month": month,
                "forecast_horizon": horizon,
                "model":
                    "Stanford_plus_all_stations_separately",
                "actual_F": actual_value,
                "forecast_F": forecast_value,
                "error_F":
                    forecast_value - actual_value
            })

        print(
            f"{test_start:%Y-%m} through "
            f"{test_end:%Y-%m}: "
            f"RMSE={metrics['RMSE_F']:.3f}, "
            f"MAE={metrics['MAE_F']:.3f}, "
            f"Bias={metrics['bias_F']:+.3f}"
        )

    except Exception as error:

        all_station_results.append({
            "test_start": test_start,
            "test_end": test_end,
            "model":
                "Stanford_plus_all_stations_separately",
            "RMSE_F": np.nan,
            "MAE_F": np.nan,
            "bias_F": np.nan,
            "status": f"failed: {error}"
        })

        print(
            f"{test_start:%Y-%m} through "
            f"{test_end:%Y-%m}: FAILED — {error}"
        )


all_station_results = pd.DataFrame(
    all_station_results
)

all_station_predictions = pd.DataFrame(
    all_station_predictions
)


# ============================================================
# FAIRLY MATCH AGAINST BASELINE COMPLETED WINDOWS
# ============================================================

baseline_predictions = (
    predictions[
        predictions["model"]
        == "Stanford_University_only"
    ]
    [
        [
            "test_start",
            "month",
            "actual_F",
            "error_F"
        ]
    ]
    .rename(
        columns={
            "error_F": "baseline_error_F"
        }
    )
)

all_station_matched = (
    baseline_predictions
    .merge(
        all_station_predictions[
            [
                "test_start",
                "month",
                "error_F"
            ]
        ].rename(
            columns={
                "error_F":
                    "all_stations_error_F"
            }
        ),
        on=["test_start", "month"],
        how="inner"
    )
)

complete_windows = (
    all_station_matched
    .groupby("test_start")
    .size()
)

complete_windows = complete_windows[
    complete_windows == 12
].index

all_station_matched = all_station_matched[
    all_station_matched["test_start"].isin(
        complete_windows
    )
].copy()

if all_station_matched.empty:
    raise ValueError(
        "No complete matched test windows were available."
    )

baseline_rmse = np.sqrt(
    np.mean(
        all_station_matched[
            "baseline_error_F"
        ] ** 2
    )
)

all_stations_rmse = np.sqrt(
    np.mean(
        all_station_matched[
            "all_stations_error_F"
        ] ** 2
    )
)

baseline_mae = np.mean(
    np.abs(
        all_station_matched[
            "baseline_error_F"
        ]
    )
)

all_stations_mae = np.mean(
    np.abs(
        all_station_matched[
            "all_stations_error_F"
        ]
    )
)

baseline_bias = np.mean(
    all_station_matched[
        "baseline_error_F"
    ]
)

all_stations_bias = np.mean(
    all_station_matched[
        "all_stations_error_F"
    ]
)

baseline_window_rmse = (
    all_station_matched
    .groupby("test_start")[
        "baseline_error_F"
    ]
    .apply(
        lambda errors:
            np.sqrt(np.mean(errors ** 2))
    )
)

all_stations_window_rmse = (
    all_station_matched
    .groupby("test_start")[
        "all_stations_error_F"
    ]
    .apply(
        lambda errors:
            np.sqrt(np.mean(errors ** 2))
    )
)

improvement = (
    baseline_rmse - all_stations_rmse
)

result = pd.DataFrame([{
    "challenger":
        "Stanford_plus_all_stations_separately",

    "matched_windows":
        len(complete_windows),

    "matched_months":
        len(all_station_matched),

    "baseline_pooled_RMSE_F":
        baseline_rmse,

    "challenger_pooled_RMSE_F":
        all_stations_rmse,

    "RMSE_improvement_F":
        improvement,

    "RMSE_improvement_percent":
        100 * improvement / baseline_rmse,

    "baseline_pooled_MAE_F":
        baseline_mae,

    "challenger_pooled_MAE_F":
        all_stations_mae,

    "baseline_bias_F":
        baseline_bias,

    "challenger_bias_F":
        all_stations_bias,

    "baseline_worst_window_RMSE_F":
        baseline_window_rmse.max(),

    "challenger_worst_window_RMSE_F":
        all_stations_window_rmse.max(),

    "challenger_window_wins":
        int(
            (
                all_stations_window_rmse
                < baseline_window_rmse
            ).sum()
        ),

    "baseline_window_wins":
        int(
            (
                baseline_window_rmse
                < all_stations_window_rmse
            ).sum()
        ),

    "meaningful_stable_improvement":
        bool(
            improvement >= 0.10
            and all_stations_window_rmse.max()
                <= baseline_window_rmse.max()
        )
}])

print()
print("ALL-STATIONS FAIR MATCHED COMPARISON")
print("====================================")
display(result.round(3))

result.to_csv(
    f"{OUTPUT_DIR}/all_stations_matched_comparison.csv",
    index=False
)

all_station_results.to_csv(
    f"{OUTPUT_DIR}/all_stations_rolling_results.csv",
    index=False
)

all_station_predictions.to_csv(
    f"{OUTPUT_DIR}/all_stations_monthly_predictions.csv",
    index=False
)

print()
print("RESULT SAVED IN:")
print(
    f"{OUTPUT_DIR}/all_stations_matched_comparison.csv"
)
```

```python
# ============================================================
# 26-MONTH FORECAST-HORIZON STABILITY TEST
#
# Final operational origin:
# Training through June 2026
# Full forecast: July 2026–August 2028 (26 months)
# Requested slice: September 2027–August 2028
#                  = horizons 15–26
#
# Historical backtests use June forecast origins so every test
# has the same calendar alignment as the final requested period.
# ============================================================

from google.colab import drive
drive.mount("/content/drive")

!pip install -q prophet

# ============================================================
# IMPORTS
# ============================================================

import os
import time
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.tsa.holtwinters import ExponentialSmoothing
from prophet import Prophet
from IPython.display import display


# ============================================================
# CONFIGURATION
# ============================================================

PROJECT = "/content/drive/MyDrive/Stanford_Weather_12M"

MONTHLY_FILE = (
    f"{PROJECT}/fair_12m_stability_comparison/"
    "monthly_data_used.csv"
)

OUTPUT_DIR = (
    f"{PROJECT}/forecast_horizon_26m_comparison"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

FORECAST_HORIZON = 26
REQUESTED_HORIZONS = list(range(15, 27))

SARIMA_ORDER = (1, 1, 1)
SARIMA_SEASONAL_ORDER = (1, 1, 1, 12)

# Nine historical forecast origins, all at the end of June.
# Example:
# Train through June 2015 -> forecast July 2015–August 2017.
TRAINING_END_DATES = pd.date_range(
    start="2015-06-01",
    end="2023-06-01",
    freq="12MS"
)

MODEL_NAMES = [
    "sarima_111x111_12",
    "prophet_additive",
    "ets_add_add_damped12",
    "seasonal_trend_15y",
    "climatology_15y",
    "seasonal_naive_12"
]


# ============================================================
# LOAD AUDITED MONTHLY STANFORD DATA
# ============================================================

if not os.path.exists(MONTHLY_FILE):
    raise FileNotFoundError(
        f"Monthly file not found:\n{MONTHLY_FILE}"
    )

monthly = pd.read_csv(MONTHLY_FILE)

month_candidates = [
    column for column in monthly.columns
    if column.lower() in {"month", "date", "ds"}
]

temperature_candidates = [
    column for column in monthly.columns
    if column.lower() in {
        "mean_temp_f",
        "temp",
        "temperature",
        "temperature_f",
        "y"
    }
]

if not month_candidates:
    raise ValueError(
        f"Could not identify the month column. "
        f"Columns: {monthly.columns.tolist()}"
    )

if not temperature_candidates:
    raise ValueError(
        f"Could not identify the temperature column. "
        f"Columns: {monthly.columns.tolist()}"
    )

month_column = month_candidates[0]
temperature_column = temperature_candidates[0]

monthly[month_column] = pd.to_datetime(
    monthly[month_column],
    errors="coerce"
)

monthly[temperature_column] = pd.to_numeric(
    monthly[temperature_column],
    errors="coerce"
)

monthly = (
    monthly[
        [month_column, temperature_column]
    ]
    .rename(
        columns={
            month_column: "month",
            temperature_column: "mean_temp_F"
        }
    )
    .dropna()
    .drop_duplicates(
        subset="month",
        keep="last"
    )
    .sort_values("month")
)

monthly = monthly[
    (monthly["month"] >= "1980-01-01") &
    (monthly["month"] <= "2026-06-01")
].copy()

series = (
    monthly
    .set_index("month")["mean_temp_F"]
    .astype(float)
    .asfreq("MS")
)

expected_index = pd.date_range(
    "1980-01-01",
    "2026-06-01",
    freq="MS"
)

if not series.index.equals(expected_index):
    raise ValueError(
        "The monthly dates are not the expected continuous "
        "January 1980–June 2026 sequence."
    )

if series.isna().any():
    raise ValueError(
        "The monthly training series contains missing values."
    )

print("MONTHLY DATA")
print("============")
print(f"Source:      {MONTHLY_FILE}")
print(f"First month: {series.index.min():%Y-%m}")
print(f"Last month:  {series.index.max():%Y-%m}")
print(f"Months:      {len(series)}")
print()


# ============================================================
# FORECAST FUNCTIONS
# ============================================================

def forecast_sarima(train, steps):

    with warnings.catch_warnings():
        warnings.simplefilter("ignore")

        model = SARIMAX(
            train,
            order=SARIMA_ORDER,
            seasonal_order=SARIMA_SEASONAL_ORDER,
            trend=None,
            enforce_stationarity=False,
            enforce_invertibility=False
        )

        fit = model.fit(
            disp=False,
            maxiter=500
        )

    forecast = fit.get_forecast(
        steps=steps
    ).predicted_mean

    converged = bool(
        fit.mle_retvals.get("converged", False)
    )

    return np.asarray(forecast), converged


def forecast_ets(train, steps):

    with warnings.catch_warnings():
        warnings.simplefilter("ignore")

        model = ExponentialSmoothing(
            train,
            trend="add",
            damped_trend=True,
            seasonal="add",
            seasonal_periods=12,
            initialization_method="estimated"
        )

        fit = model.fit(
            optimized=True,
            remove_bias=False
        )

    return np.asarray(
        fit.forecast(steps),
        dtype=float
    )


def forecast_prophet(train, steps):

    prophet_data = pd.DataFrame({
        "ds": train.index,
        "y": train.values
    })

    with warnings.catch_warnings():
        warnings.simplefilter("ignore")

        model = Prophet(
            growth="linear",
            yearly_seasonality=True,
            weekly_seasonality=False,
            daily_seasonality=False,
            seasonality_mode="additive",
            changepoint_prior_scale=0.05,
            interval_width=0.95
        )

        model.fit(prophet_data)

    future = pd.DataFrame({
        "ds": pd.date_range(
            train.index[-1] + pd.offsets.MonthBegin(1),
            periods=steps,
            freq="MS"
        )
    })

    forecast = model.predict(future)

    return forecast["yhat"].to_numpy(dtype=float)


def forecast_seasonal_naive(train, steps):

    history = list(
        train.astype(float).values
    )

    forecast = []

    for _ in range(steps):
        prediction = history[-12]
        forecast.append(prediction)
        history.append(prediction)

    return np.asarray(forecast, dtype=float)


def forecast_climatology_15y(train, steps):
    """
    Forecast each future calendar month using that month's
    average over the most recent 15 available years.
    """

    future_months = pd.date_range(
        train.index[-1] + pd.offsets.MonthBegin(1),
        periods=steps,
        freq="MS"
    )

    forecast = []

    for future_month in future_months:

        same_month_history = train[
            train.index.month == future_month.month
        ].tail(15)

        if len(same_month_history) < 5:
            raise ValueError(
                "Insufficient history for climatology."
            )

        forecast.append(
            same_month_history.mean()
        )

    return np.asarray(forecast, dtype=float)


def forecast_seasonal_trend_15y(train, steps):
    """
    For each calendar month, fit a linear trend to the latest
    15 observations for that calendar month.
    """

    future_months = pd.date_range(
        train.index[-1] + pd.offsets.MonthBegin(1),
        periods=steps,
        freq="MS"
    )

    forecast = []

    for future_month in future_months:

        same_month_history = train[
            train.index.month == future_month.month
        ].tail(15)

        if len(same_month_history) < 5:
            raise ValueError(
                "Insufficient history for seasonal trend."
            )

        historical_years = (
            same_month_history.index.year
            .to_numpy(dtype=float)
        )

        historical_values = (
            same_month_history
            .to_numpy(dtype=float)
        )

        slope, intercept = np.polyfit(
            historical_years,
            historical_values,
            deg=1
        )

        prediction = (
            intercept
            + slope * future_month.year
        )

        forecast.append(prediction)

    return np.asarray(forecast, dtype=float)


def produce_forecast(model_name, train, steps):

    if model_name == "sarima_111x111_12":
        forecast, converged = forecast_sarima(
            train,
            steps
        )
        return forecast, converged

    if model_name == "prophet_additive":
        return forecast_prophet(
            train,
            steps
        ), True

    if model_name == "ets_add_add_damped12":
        return forecast_ets(
            train,
            steps
        ), True

    if model_name == "seasonal_trend_15y":
        return forecast_seasonal_trend_15y(
            train,
            steps
        ), True

    if model_name == "climatology_15y":
        return forecast_climatology_15y(
            train,
            steps
        ), True

    if model_name == "seasonal_naive_12":
        return forecast_seasonal_naive(
            train,
            steps
        ), True

    raise ValueError(
        f"Unknown model: {model_name}"
    )


# ============================================================
# METRIC FUNCTIONS
# ============================================================

def calculate_metrics(actual, forecast):

    actual = np.asarray(actual, dtype=float)
    forecast = np.asarray(forecast, dtype=float)

    error = forecast - actual

    return {
        "RMSE_F": np.sqrt(
            np.mean(error ** 2)
        ),
        "MAE_F": np.mean(
            np.abs(error)
        ),
        "bias_F": np.mean(error)
    }


def metrics_for_horizon_range(
    window_predictions,
    first_horizon,
    last_horizon
):

    subset = window_predictions[
        window_predictions["forecast_horizon"]
        .between(first_horizon, last_horizon)
    ]

    return calculate_metrics(
        subset["actual_F"],
        subset["forecast_F"]
    )


# ============================================================
# RUN HISTORICAL 26-MONTH BACKTESTS
# ============================================================

window_results = []
monthly_predictions = []

print("26-MONTH ROLLING BACKTESTS")
print("==========================")

for training_end in TRAINING_END_DATES:

    test_start = (
        training_end
        + pd.offsets.MonthBegin(1)
    )

    test_end = (
        test_start
        + pd.DateOffset(
            months=FORECAST_HORIZON - 1
        )
    )

    train = series.loc[:training_end]
    actual = series.loc[test_start:test_end]

    if len(actual) != FORECAST_HORIZON:
        print(
            f"Skipping training end {training_end:%Y-%m}: "
            "26 complete future months unavailable."
        )
        continue

    print()
    print(
        f"Training through {training_end:%Y-%m}; "
        f"testing {test_start:%Y-%m}–{test_end:%Y-%m}"
    )

    for model_name in MODEL_NAMES:

        start_time = time.time()

        try:
            forecast, converged = produce_forecast(
                model_name,
                train,
                FORECAST_HORIZON
            )

            if len(forecast) != FORECAST_HORIZON:
                raise ValueError(
                    "Forecast did not contain 26 months."
                )

            window_prediction = pd.DataFrame({
                "training_end": training_end,
                "test_start": test_start,
                "test_end": test_end,
                "month": actual.index,
                "forecast_horizon":
                    np.arange(
                        1,
                        FORECAST_HORIZON + 1
                    ),
                "model": model_name,
                "actual_F": actual.values,
                "forecast_F": forecast,
                "error_F":
                    forecast - actual.values
            })

            monthly_predictions.append(
                window_prediction
            )

            full_metrics = calculate_metrics(
                actual.values,
                forecast
            )

            h1_12 = metrics_for_horizon_range(
                window_prediction,
                1,
                12
            )

            h13_26 = metrics_for_horizon_range(
                window_prediction,
                13,
                26
            )

            h15_26 = metrics_for_horizon_range(
                window_prediction,
                15,
                26
            )

            window_results.append({
                "training_end": training_end,
                "test_start": test_start,
                "test_end": test_end,
                "model": model_name,
                "converged": converged,
                "status": "completed",

                "full_26_RMSE_F":
                    full_metrics["RMSE_F"],
                "full_26_MAE_F":
                    full_metrics["MAE_F"],
                "full_26_bias_F":
                    full_metrics["bias_F"],

                "h1_12_RMSE_F":
                    h1_12["RMSE_F"],
                "h1_12_MAE_F":
                    h1_12["MAE_F"],
                "h1_12_bias_F":
                    h1_12["bias_F"],

                "h13_26_RMSE_F":
                    h13_26["RMSE_F"],
                "h13_26_MAE_F":
                    h13_26["MAE_F"],
                "h13_26_bias_F":
                    h13_26["bias_F"],

                "h15_26_RMSE_F":
                    h15_26["RMSE_F"],
                "h15_26_MAE_F":
                    h15_26["MAE_F"],
                "h15_26_bias_F":
                    h15_26["bias_F"],

                "runtime_seconds":
                    time.time() - start_time
            })

            convergence_note = (
                ""
                if converged
                else " [optimizer warning]"
            )

            print(
                f"  {model_name:<25}"
                f"Full RMSE={full_metrics['RMSE_F']:.3f}  "
                f"H15–26 RMSE={h15_26['RMSE_F']:.3f}"
                f"{convergence_note}"
            )

        except Exception as error:

            window_results.append({
                "training_end": training_end,
                "test_start": test_start,
                "test_end": test_end,
                "model": model_name,
                "converged": False,
                "status": f"failed: {error}",
                "runtime_seconds":
                    time.time() - start_time
            })

            print(
                f"  {model_name:<25}FAILED: {error}"
            )


window_results = pd.DataFrame(
    window_results
)

monthly_predictions = pd.concat(
    monthly_predictions,
    ignore_index=True
)


# ============================================================
# STABILITY SUMMARY
# ============================================================

completed = window_results[
    window_results["status"] == "completed"
].copy()

summary_rows = []

for model_name, model_windows in (
    completed.groupby("model")
):

    model_predictions = monthly_predictions[
        monthly_predictions["model"] == model_name
    ]

    full_metrics = calculate_metrics(
        model_predictions["actual_F"],
        model_predictions["forecast_F"]
    )

    requested_predictions = model_predictions[
        model_predictions["forecast_horizon"]
        .between(15, 26)
    ]

    requested_metrics = calculate_metrics(
        requested_predictions["actual_F"],
        requested_predictions["forecast_F"]
    )

    summary_rows.append({
        "model": model_name,
        "windows_completed":
            len(model_windows),

        "optimizer_warnings":
            int((~model_windows["converged"]).sum()),

        "mean_full_26_RMSE_F":
            model_windows["full_26_RMSE_F"].mean(),

        "full_26_RMSE_std_F":
            model_windows["full_26_RMSE_F"].std(ddof=0),

        "worst_full_26_RMSE_F":
            model_windows["full_26_RMSE_F"].max(),

        "pooled_full_26_RMSE_F":
            full_metrics["RMSE_F"],

        "pooled_full_26_MAE_F":
            full_metrics["MAE_F"],

        "pooled_full_26_bias_F":
            full_metrics["bias_F"],

        "mean_h15_26_RMSE_F":
            model_windows["h15_26_RMSE_F"].mean(),

        "h15_26_RMSE_std_F":
            model_windows["h15_26_RMSE_F"].std(ddof=0),

        "worst_h15_26_RMSE_F":
            model_windows["h15_26_RMSE_F"].max(),

        "pooled_h15_26_RMSE_F":
            requested_metrics["RMSE_F"],

        "pooled_h15_26_MAE_F":
            requested_metrics["MAE_F"],

        "pooled_h15_26_bias_F":
            requested_metrics["bias_F"]
    })

stability_summary = (
    pd.DataFrame(summary_rows)
    .sort_values(
        [
            "pooled_h15_26_RMSE_F",
            "worst_h15_26_RMSE_F"
        ]
    )
    .reset_index(drop=True)
)


# ============================================================
# PERFORMANCE BY FORECAST HORIZON
# ============================================================

horizon_rows = []

for (
    model_name,
    forecast_horizon
), group in monthly_predictions.groupby(
    ["model", "forecast_horizon"]
):

    metrics = calculate_metrics(
        group["actual_F"],
        group["forecast_F"]
    )

    horizon_rows.append({
        "model": model_name,
        "forecast_horizon":
            forecast_horizon,
        "observations": len(group),
        **metrics
    })

horizon_summary = pd.DataFrame(
    horizon_rows
)


# ============================================================
# DISPLAY RESULTS
# ============================================================

print()
print("26-MONTH STABILITY SUMMARY")
print("==========================")
display(
    stability_summary.round(3)
)

print()
print("PRIMARY DECISION COLUMNS — HORIZONS 15–26")
print("==========================================")

display(
    stability_summary[
        [
            "model",
            "windows_completed",
            "optimizer_warnings",
            "mean_h15_26_RMSE_F",
            "h15_26_RMSE_std_F",
            "worst_h15_26_RMSE_F",
            "pooled_h15_26_RMSE_F",
            "pooled_h15_26_MAE_F",
            "pooled_h15_26_bias_F"
        ]
    ].round(3)
)


# ============================================================
# PLOT MODEL PERFORMANCE FOR HORIZONS 15–26
# ============================================================

plot_data = stability_summary.copy()

x = np.arange(len(plot_data))
width = 0.36

fig, ax = plt.subplots(figsize=(13, 6))

ax.bar(
    x - width / 2,
    plot_data["pooled_h15_26_RMSE_F"],
    width,
    label="Pooled RMSE, horizons 15–26"
)

ax.bar(
    x + width / 2,
    plot_data["worst_h15_26_RMSE_F"],
    width,
    label="Worst-window RMSE, horizons 15–26"
)

ax.set_xticks(x)

ax.set_xticklabels(
    plot_data["model"],
    rotation=25,
    ha="right"
)

ax.set_ylabel("Forecast error (°F)")

ax.set_title(
    "Model Stability for Forecast Horizons 15–26 Months"
)

ax.grid(
    axis="y",
    alpha=0.25
)

ax.legend()

fig.tight_layout()

SUMMARY_PLOT = (
    f"{OUTPUT_DIR}/h15_26_model_comparison.png"
)

fig.savefig(
    SUMMARY_PLOT,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# PLOT ERROR BY FORECAST HORIZON
# ============================================================

fig, ax = plt.subplots(figsize=(14, 7))

for model_name in MODEL_NAMES:

    model_horizons = horizon_summary[
        horizon_summary["model"] == model_name
    ].sort_values("forecast_horizon")

    ax.plot(
        model_horizons["forecast_horizon"],
        model_horizons["RMSE_F"],
        marker="o",
        linewidth=1.8,
        label=model_name
    )

ax.axvspan(
    15,
    26,
    alpha=0.12,
    color="green",
    label="Requested horizons 15–26"
)

ax.set_xlabel("Forecast horizon in months")
ax.set_ylabel("Pooled RMSE (°F)")

ax.set_title(
    "Forecast Accuracy as the Horizon Increases"
)

ax.set_xticks(
    range(1, 27)
)

ax.grid(alpha=0.25)
ax.legend(bbox_to_anchor=(1.02, 1), loc="upper left")

fig.tight_layout()

HORIZON_PLOT = (
    f"{OUTPUT_DIR}/rmse_by_forecast_horizon.png"
)

fig.savefig(
    HORIZON_PLOT,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# SAVE RESULTS
# ============================================================

window_results.to_csv(
    f"{OUTPUT_DIR}/rolling_26m_window_results.csv",
    index=False
)

monthly_predictions.to_csv(
    f"{OUTPUT_DIR}/monthly_26m_predictions.csv",
    index=False
)

stability_summary.to_csv(
    f"{OUTPUT_DIR}/26m_stability_summary.csv",
    index=False
)

horizon_summary.to_csv(
    f"{OUTPUT_DIR}/performance_by_horizon.csv",
    index=False
)

print()
print("RESULTS SAVED IN:")
print(OUTPUT_DIR)

print()
print("SEND ME:")
print("1. 26-MONTH STABILITY SUMMARY")
print("2. PRIMARY DECISION COLUMNS — HORIZONS 15–26")
```

```python
# ============================================================
# FINAL 26-MONTH ENSEMBLE CHALLENGER
#
# Tests fixed, pre-specified ensembles:
# 1. 50% SARIMA + 50% climatology
# 2. 50% SARIMA + 50% ETS
# 3. Equal average of SARIMA + climatology + ETS
#
# No weights are tuned on the test data.
# ============================================================

import os
import numpy as np
import pandas as pd
from IPython.display import display

PROJECT = "/content/drive/MyDrive/Stanford_Weather_12M"

OUTPUT_DIR = (
    f"{PROJECT}/forecast_horizon_26m_comparison"
)

PREDICTIONS_FILE = (
    f"{OUTPUT_DIR}/monthly_26m_predictions.csv"
)

predictions = pd.read_csv(
    PREDICTIONS_FILE,
    parse_dates=[
        "training_end",
        "test_start",
        "test_end",
        "month"
    ]
)

required_models = [
    "sarima_111x111_12",
    "climatology_15y",
    "ets_add_add_damped12"
]

available_models = predictions["model"].unique()

missing_models = [
    model for model in required_models
    if model not in available_models
]

if missing_models:
    raise ValueError(
        f"Missing required predictions: {missing_models}"
    )


# ============================================================
# ALIGN THE THREE MODELS ON IDENTICAL MONTHS
# ============================================================

key_columns = [
    "training_end",
    "test_start",
    "test_end",
    "month",
    "forecast_horizon",
    "actual_F"
]

wide = (
    predictions[
        predictions["model"].isin(required_models)
    ]
    .pivot_table(
        index=key_columns,
        columns="model",
        values="forecast_F",
        aggfunc="first"
    )
    .reset_index()
    .dropna()
)

# Fixed ensemble forecasts
wide["ensemble_sarima_climatology"] = (
    0.50 * wide["sarima_111x111_12"]
    + 0.50 * wide["climatology_15y"]
)

wide["ensemble_sarima_ets"] = (
    0.50 * wide["sarima_111x111_12"]
    + 0.50 * wide["ets_add_add_damped12"]
)

wide["ensemble_sarima_climatology_ets"] = (
    (
        wide["sarima_111x111_12"]
        + wide["climatology_15y"]
        + wide["ets_add_add_damped12"]
    )
    / 3
)


# ============================================================
# CREATE LONG-FORM PREDICTIONS
# ============================================================

comparison_models = {
    "sarima_111x111_12":
        wide["sarima_111x111_12"],

    "climatology_15y":
        wide["climatology_15y"],

    "ets_add_add_damped12":
        wide["ets_add_add_damped12"],

    "ensemble_sarima_climatology":
        wide["ensemble_sarima_climatology"],

    "ensemble_sarima_ets":
        wide["ensemble_sarima_ets"],

    "ensemble_sarima_climatology_ets":
        wide["ensemble_sarima_climatology_ets"]
}

long_rows = []

for model_name, forecast_values in comparison_models.items():

    model_data = wide[key_columns].copy()

    model_data["model"] = model_name
    model_data["forecast_F"] = forecast_values
    model_data["error_F"] = (
        model_data["forecast_F"]
        - model_data["actual_F"]
    )

    long_rows.append(model_data)

ensemble_predictions = pd.concat(
    long_rows,
    ignore_index=True
)


# ============================================================
# METRIC FUNCTION
# ============================================================

def summarize_errors(data):

    errors = data["error_F"].to_numpy()

    return {
        "RMSE_F":
            np.sqrt(np.mean(errors ** 2)),
        "MAE_F":
            np.mean(np.abs(errors)),
        "bias_F":
            np.mean(errors)
    }


# ============================================================
# HORIZONS 15–26 SUMMARY
# ============================================================

requested = ensemble_predictions[
    ensemble_predictions["forecast_horizon"]
    .between(15, 26)
].copy()

summary_rows = []

for model_name, model_data in requested.groupby("model"):

    pooled_metrics = summarize_errors(
        model_data
    )

    window_metrics = (
        model_data
        .groupby("test_start")
        .apply(
            lambda group: pd.Series(
                summarize_errors(group)
            ),
            include_groups=False
        )
        .reset_index()
    )

    summary_rows.append({
        "model": model_name,
        "windows_completed":
            window_metrics["test_start"].nunique(),

        "months_evaluated":
            len(model_data),

        "mean_window_RMSE_F":
            window_metrics["RMSE_F"].mean(),

        "RMSE_std_F":
            window_metrics["RMSE_F"].std(ddof=0),

        "worst_window_RMSE_F":
            window_metrics["RMSE_F"].max(),

        "pooled_RMSE_F":
            pooled_metrics["RMSE_F"],

        "pooled_MAE_F":
            pooled_metrics["MAE_F"],

        "pooled_bias_F":
            pooled_metrics["bias_F"]
    })

ensemble_summary = (
    pd.DataFrame(summary_rows)
    .sort_values(
        [
            "pooled_RMSE_F",
            "worst_window_RMSE_F"
        ]
    )
    .reset_index(drop=True)
)


# ============================================================
# WINDOW WINS
# ============================================================

window_rmse = (
    requested
    .groupby(
        ["test_start", "model"]
    )
    .apply(
        lambda group:
            np.sqrt(
                np.mean(group["error_F"] ** 2)
            ),
        include_groups=False
    )
    .unstack("model")
)

window_winners = (
    window_rmse
    .idxmin(axis=1)
    .value_counts()
    .rename_axis("model")
    .reset_index(name="window_wins")
)

ensemble_summary = ensemble_summary.merge(
    window_winners,
    on="model",
    how="left"
)

ensemble_summary["window_wins"] = (
    ensemble_summary["window_wins"]
    .fillna(0)
    .astype(int)
)


# ============================================================
# DISPLAY
# ============================================================

print("FINAL H15–26 ENSEMBLE CHALLENGER")
print("================================")

display(
    ensemble_summary.round(3)
)

best_model = ensemble_summary.iloc[0]

print()
print("PRELIMINARY WINNER")
print("==================")
print(f"Model: {best_model['model']}")
print(
    f"Pooled RMSE: "
    f"{best_model['pooled_RMSE_F']:.3f} °F"
)
print(
    f"Worst-window RMSE: "
    f"{best_model['worst_window_RMSE_F']:.3f} °F"
)
print(
    f"Bias: "
    f"{best_model['pooled_bias_F']:+.3f} °F"
)


# ============================================================
# SAVE
# ============================================================

ensemble_summary.to_csv(
    f"{OUTPUT_DIR}/h15_26_ensemble_summary.csv",
    index=False
)

ensemble_predictions.to_csv(
    f"{OUTPUT_DIR}/h15_26_ensemble_predictions.csv",
    index=False
)

window_rmse.to_csv(
    f"{OUTPUT_DIR}/h15_26_ensemble_window_rmse.csv"
)

print()
print("RESULTS SAVED IN:")
print(OUTPUT_DIR)
print()
print("SEND ME THE FINAL H15–26 ENSEMBLE CHALLENGER TABLE.")
```

```python
# ============================================================
# FINAL HORIZONS 15–26 PRODUCTION RUN
#
# Selected model:
#   50% SARIMA(1,1,1)(1,1,1,12)
#   50% 15-year calendar-month climatology
#
# Training data:
#   January 1980–June 2026
#
# Full calculation horizon:
#   July 2026–August 2028 (26 months)
#
# Official output:
#   September 2027–August 2028
#   = forecast horizons 15–26
#
# Intervals:
#   Empirical/conformal-style intervals calibrated from
#   108 historical out-of-sample errors.
# ============================================================

from google.colab import drive
drive.mount("/content/drive")

# ============================================================
# IMPORTS
# ============================================================

import os
import json
import hashlib
import platform
import warnings
from datetime import datetime, timezone

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels

from statsmodels.tsa.statespace.sarimax import SARIMAX
from IPython.display import display


# ============================================================
# PATHS
# ============================================================

PROJECT = "/content/drive/MyDrive/Stanford_Weather_12M"

MONTHLY_FILE = (
    f"{PROJECT}/fair_12m_stability_comparison/"
    "monthly_data_used.csv"
)

BACKTEST_PREDICTIONS_FILE = (
    f"{PROJECT}/forecast_horizon_26m_comparison/"
    "h15_26_ensemble_predictions.csv"
)

OUTPUT_DIR = (
    f"{PROJECT}/final_h15_26_ensemble_forecast"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

FORECAST_FILE = (
    f"{OUTPUT_DIR}/"
    "official_forecast_Sep2027_Aug2028.csv"
)

FULL_FORECAST_FILE = (
    f"{OUTPUT_DIR}/"
    "full_26_month_component_forecasts.csv"
)

TRAINING_FILE = (
    f"{OUTPUT_DIR}/final_training_series.csv"
)

MODEL_FILE = (
    f"{OUTPUT_DIR}/"
    "sarima_component_fitted_model.pkl"
)

INTERVAL_VALIDATION_FILE = (
    f"{OUTPUT_DIR}/"
    "interval_validation_summary.csv"
)

INTERVAL_WIDTH_FILE = (
    f"{OUTPUT_DIR}/"
    "final_interval_widths.csv"
)

CALIBRATION_ERRORS_FILE = (
    f"{OUTPUT_DIR}/"
    "ensemble_calibration_errors.csv"
)

METADATA_FILE = (
    f"{OUTPUT_DIR}/"
    "reproducibility_metadata.json"
)

PLOT_FILE = (
    f"{OUTPUT_DIR}/"
    "official_forecast_Sep2027_Aug2028.png"
)

MODEL_SUMMARY_FILE = (
    f"{OUTPUT_DIR}/"
    "sarima_component_summary.txt"
)

PACKAGE_FILE = (
    f"{OUTPUT_DIR}/package_versions.txt"
)


# ============================================================
# CONFIGURATION
# ============================================================

SARIMA_ORDER = (1, 1, 1)
SARIMA_SEASONAL_ORDER = (1, 1, 1, 12)

SARIMA_WEIGHT = 0.50
CLIMATOLOGY_WEIGHT = 0.50

FULL_HORIZON = 26

REQUESTED_FIRST_HORIZON = 15
REQUESTED_LAST_HORIZON = 26

EXPECTED_TRAINING_START = pd.Timestamp("1980-01-01")
EXPECTED_TRAINING_END = pd.Timestamp("2026-06-01")

EXPECTED_OUTPUT_START = pd.Timestamp("2027-09-01")
EXPECTED_OUTPUT_END = pd.Timestamp("2028-08-01")


# ============================================================
# HELPER FUNCTIONS
# ============================================================

def file_sha256(path, chunk_size=1024 * 1024):

    if not os.path.exists(path):
        return None

    digest = hashlib.sha256()

    with open(path, "rb") as file:
        while True:
            chunk = file.read(chunk_size)

            if not chunk:
                break

            digest.update(chunk)

    return digest.hexdigest()


def find_column(columns, candidates):

    column_lookup = {
        str(column).lower().strip(): column
        for column in columns
    }

    for candidate in candidates:
        if candidate.lower() in column_lookup:
            return column_lookup[candidate.lower()]

    return None


def conformal_half_width(
    absolute_errors,
    target_coverage
):
    """
    Finite-sample empirical/conformal quantile.

    For n calibration observations, select order statistic:
        ceil((n + 1) * target coverage)

    The result is capped at the largest available error.
    """

    errors = np.sort(
        np.asarray(
            absolute_errors,
            dtype=float
        )
    )

    errors = errors[
        np.isfinite(errors)
    ]

    n = len(errors)

    if n == 0:
        raise ValueError(
            "No valid calibration errors were supplied."
        )

    rank = int(
        np.ceil(
            (n + 1) * target_coverage
        )
    )

    rank = min(
        max(rank, 1),
        n
    )

    return float(
        errors[rank - 1]
    )


def interval_metrics(
    actual,
    lower,
    upper
):

    actual = np.asarray(actual, dtype=float)
    lower = np.asarray(lower, dtype=float)
    upper = np.asarray(upper, dtype=float)

    covered = (
        (actual >= lower)
        & (actual <= upper)
    )

    return {
        "observed_coverage":
            float(np.mean(covered)),

        "covered_months":
            int(np.sum(covered)),

        "total_months":
            int(len(actual)),

        "mean_interval_width_F":
            float(np.mean(upper - lower))
    }


# ============================================================
# LOAD OUT-OF-SAMPLE ENSEMBLE PREDICTIONS
# ============================================================

if not os.path.exists(
    BACKTEST_PREDICTIONS_FILE
):
    raise FileNotFoundError(
        "Ensemble backtest predictions were not found:\n"
        f"{BACKTEST_PREDICTIONS_FILE}"
    )

backtests = pd.read_csv(
    BACKTEST_PREDICTIONS_FILE,
    parse_dates=[
        "training_end",
        "test_start",
        "test_end",
        "month"
    ]
)

required_columns = {
    "training_end",
    "test_start",
    "month",
    "forecast_horizon",
    "model",
    "actual_F",
    "forecast_F",
    "error_F"
}

missing_columns = (
    required_columns
    - set(backtests.columns)
)

if missing_columns:
    raise ValueError(
        "Backtest file is missing columns: "
        f"{sorted(missing_columns)}"
    )

SELECTED_MODEL_NAME = (
    "ensemble_sarima_climatology"
)

ensemble_backtests = backtests[
    (
        backtests["model"]
        == SELECTED_MODEL_NAME
    )
    &
    (
        backtests["forecast_horizon"]
        .between(
            REQUESTED_FIRST_HORIZON,
            REQUESTED_LAST_HORIZON
        )
    )
].copy()

ensemble_backtests = (
    ensemble_backtests
    .sort_values(
        [
            "test_start",
            "forecast_horizon"
        ]
    )
    .reset_index(drop=True)
)

if len(ensemble_backtests) != 108:
    raise ValueError(
        "Expected 108 out-of-sample ensemble predictions "
        f"for horizons 15–26, but found "
        f"{len(ensemble_backtests)}."
    )

if (
    ensemble_backtests["test_start"]
    .nunique()
    != 9
):
    raise ValueError(
        "Expected nine historical forecast windows."
    )

ensemble_backtests["absolute_error_F"] = (
    ensemble_backtests["error_F"].abs()
)

print("ENSEMBLE CALIBRATION DATA")
print("=========================")
print(
    f"Model:       {SELECTED_MODEL_NAME}"
)
print(
    f"Windows:     "
    f"{ensemble_backtests['test_start'].nunique()}"
)
print(
    f"Predictions: {len(ensemble_backtests)}"
)
print(
    f"Horizons:    "
    f"{ensemble_backtests['forecast_horizon'].min()}–"
    f"{ensemble_backtests['forecast_horizon'].max()}"
)
print()


# ============================================================
# LEAVE-ONE-WINDOW-OUT INTERVAL VALIDATION
#
# For every historical 26-month forecast:
# - remove its 12 requested months;
# - calculate interval widths from the other eight windows;
# - evaluate the held-out window.
#
# This prevents each window's own errors from determining its
# prediction intervals.
# ============================================================

validation_rows = []
validation_predictions = []

historical_windows = sorted(
    ensemble_backtests["test_start"].unique()
)

for held_out_start in historical_windows:

    calibration = ensemble_backtests[
        ensemble_backtests["test_start"]
        != held_out_start
    ].copy()

    held_out = ensemble_backtests[
        ensemble_backtests["test_start"]
        == held_out_start
    ].copy()

    half_width_80 = conformal_half_width(
        calibration["absolute_error_F"],
        target_coverage=0.80
    )

    half_width_95 = conformal_half_width(
        calibration["absolute_error_F"],
        target_coverage=0.95
    )

    held_out["lower_80_F"] = (
        held_out["forecast_F"]
        - half_width_80
    )

    held_out["upper_80_F"] = (
        held_out["forecast_F"]
        + half_width_80
    )

    held_out["lower_95_F"] = (
        held_out["forecast_F"]
        - half_width_95
    )

    held_out["upper_95_F"] = (
        held_out["forecast_F"]
        + half_width_95
    )

    held_out["calibration_80_half_width_F"] = (
        half_width_80
    )

    held_out["calibration_95_half_width_F"] = (
        half_width_95
    )

    validation_predictions.append(
        held_out
    )

    metrics_80 = interval_metrics(
        held_out["actual_F"],
        held_out["lower_80_F"],
        held_out["upper_80_F"]
    )

    metrics_95 = interval_metrics(
        held_out["actual_F"],
        held_out["lower_95_F"],
        held_out["upper_95_F"]
    )

    validation_rows.append({
        "held_out_test_start":
            pd.Timestamp(held_out_start),

        "target_coverage":
            "80%",

        "calibration_observations":
            len(calibration),

        "half_width_F":
            half_width_80,

        **metrics_80
    })

    validation_rows.append({
        "held_out_test_start":
            pd.Timestamp(held_out_start),

        "target_coverage":
            "95%",

        "calibration_observations":
            len(calibration),

        "half_width_F":
            half_width_95,

        **metrics_95
    })


validation_predictions = pd.concat(
    validation_predictions,
    ignore_index=True
)

window_validation = pd.DataFrame(
    validation_rows
)


# ============================================================
# AGGREGATE LEAVE-ONE-WINDOW-OUT COVERAGE
# ============================================================

aggregate_validation_rows = []

for coverage_label in ["80%", "95%"]:

    if coverage_label == "80%":
        lower_column = "lower_80_F"
        upper_column = "upper_80_F"
        target_value = 0.80
    else:
        lower_column = "lower_95_F"
        upper_column = "upper_95_F"
        target_value = 0.95

    metrics = interval_metrics(
        validation_predictions["actual_F"],
        validation_predictions[lower_column],
        validation_predictions[upper_column]
    )

    aggregate_validation_rows.append({
        "method":
            "empirical_leave_one_window_out",

        "target_coverage":
            coverage_label,

        "target_coverage_numeric":
            target_value,

        **metrics
    })

aggregate_validation = pd.DataFrame(
    aggregate_validation_rows
)

print("LEAVE-ONE-WINDOW-OUT INTERVAL VALIDATION")
print("========================================")

display(
    aggregate_validation.round(3)
)


# ============================================================
# COVERAGE BY FORECAST HORIZON
# ============================================================

horizon_coverage_rows = []

for horizon, group in (
    validation_predictions
    .groupby("forecast_horizon")
):

    metrics_80 = interval_metrics(
        group["actual_F"],
        group["lower_80_F"],
        group["upper_80_F"]
    )

    metrics_95 = interval_metrics(
        group["actual_F"],
        group["lower_95_F"],
        group["upper_95_F"]
    )

    horizon_coverage_rows.append({
        "forecast_horizon": horizon,
        "observations": len(group),
        "coverage_80":
            metrics_80["observed_coverage"],
        "coverage_95":
            metrics_95["observed_coverage"],
        "mean_width_80_F":
            metrics_80["mean_interval_width_F"],
        "mean_width_95_F":
            metrics_95["mean_interval_width_F"]
    })

horizon_coverage = pd.DataFrame(
    horizon_coverage_rows
)

print()
print("INTERVAL COVERAGE BY HORIZON")
print("============================")

display(
    horizon_coverage.round(3)
)


# ============================================================
# FINAL CALIBRATION WIDTHS USING ALL 108 ERRORS
# ============================================================

FINAL_80_HALF_WIDTH = conformal_half_width(
    ensemble_backtests["absolute_error_F"],
    target_coverage=0.80
)

FINAL_95_HALF_WIDTH = conformal_half_width(
    ensemble_backtests["absolute_error_F"],
    target_coverage=0.95
)

final_widths = pd.DataFrame([
    {
        "model":
            SELECTED_MODEL_NAME,
        "forecast_horizons":
            "15–26",
        "calibration_method":
            "pooled_absolute_out_of_sample_errors",
        "calibration_observations":
            len(ensemble_backtests),
        "target_coverage":
            "80%",
        "half_width_F":
            FINAL_80_HALF_WIDTH,
        "full_width_F":
            2 * FINAL_80_HALF_WIDTH
    },
    {
        "model":
            SELECTED_MODEL_NAME,
        "forecast_horizons":
            "15–26",
        "calibration_method":
            "pooled_absolute_out_of_sample_errors",
        "calibration_observations":
            len(ensemble_backtests),
        "target_coverage":
            "95%",
        "half_width_F":
            FINAL_95_HALF_WIDTH,
        "full_width_F":
            2 * FINAL_95_HALF_WIDTH
    }
])

print()
print("FINAL EMPIRICAL INTERVAL WIDTHS")
print("===============================")

display(
    final_widths.round(3)
)


# ============================================================
# LOAD FINAL TRAINING SERIES
# ============================================================

if not os.path.exists(MONTHLY_FILE):
    raise FileNotFoundError(
        f"Monthly data not found:\n{MONTHLY_FILE}"
    )

monthly = pd.read_csv(
    MONTHLY_FILE
)

month_column = find_column(
    monthly.columns,
    ["month", "date", "ds"]
)

temperature_column = find_column(
    monthly.columns,
    [
        "mean_temp_F",
        "temp",
        "temperature",
        "temperature_F",
        "y"
    ]
)

if month_column is None:
    raise ValueError(
        "Could not identify month column."
    )

if temperature_column is None:
    raise ValueError(
        "Could not identify temperature column."
    )

monthly[month_column] = pd.to_datetime(
    monthly[month_column],
    errors="coerce"
)

monthly[temperature_column] = pd.to_numeric(
    monthly[temperature_column],
    errors="coerce"
)

monthly = (
    monthly[
        [month_column, temperature_column]
    ]
    .rename(
        columns={
            month_column: "month",
            temperature_column: "mean_temp_F"
        }
    )
    .dropna()
    .drop_duplicates(
        subset="month",
        keep="last"
    )
    .sort_values("month")
)

monthly = monthly[
    (
        monthly["month"]
        >= EXPECTED_TRAINING_START
    )
    &
    (
        monthly["month"]
        <= EXPECTED_TRAINING_END
    )
].copy()

series = (
    monthly
    .set_index("month")["mean_temp_F"]
    .astype(float)
    .asfreq("MS")
)

expected_index = pd.date_range(
    EXPECTED_TRAINING_START,
    EXPECTED_TRAINING_END,
    freq="MS"
)

if not series.index.equals(expected_index):
    raise ValueError(
        "Training dates are not the expected continuous "
        "January 1980–June 2026 sequence."
    )

if len(series) != 558:
    raise ValueError(
        f"Expected 558 training months; found {len(series)}."
    )

if series.isna().any():
    raise ValueError(
        "Training series contains missing values."
    )

print()
print("FINAL TRAINING DATA")
print("===================")
print(f"First month: {series.index.min():%Y-%m}")
print(f"Last month:  {series.index.max():%Y-%m}")
print(f"Months:      {len(series)}")
print()


# ============================================================
# FIT FINAL SARIMA COMPONENT
# ============================================================

with warnings.catch_warnings():
    warnings.simplefilter("ignore")

    final_sarima_model = SARIMAX(
        series,
        order=SARIMA_ORDER,
        seasonal_order=SARIMA_SEASONAL_ORDER,
        trend=None,
        enforce_stationarity=False,
        enforce_invertibility=False
    )

    final_sarima_fit = (
        final_sarima_model.fit(
            disp=False,
            maxiter=500
        )
    )

sarima_converged = bool(
    final_sarima_fit
    .mle_retvals
    .get("converged", False)
)

if not sarima_converged:
    raise RuntimeError(
        "Final SARIMA component did not converge."
    )

sarima_forecast = np.asarray(
    final_sarima_fit
    .get_forecast(
        steps=FULL_HORIZON
    )
    .predicted_mean,
    dtype=float
)


# ============================================================
# CREATE FINAL 15-YEAR CLIMATOLOGY COMPONENT
# ============================================================

future_months = pd.date_range(
    start="2026-07-01",
    periods=FULL_HORIZON,
    freq="MS"
)

climatology_forecast = []

for future_month in future_months:

    same_calendar_month = series[
        series.index.month
        == future_month.month
    ].tail(15)

    if len(same_calendar_month) != 15:
        raise ValueError(
            f"Expected 15 historical values for "
            f"calendar month {future_month.month}; "
            f"found {len(same_calendar_month)}."
        )

    climatology_forecast.append(
        float(same_calendar_month.mean())
    )

climatology_forecast = np.asarray(
    climatology_forecast,
    dtype=float
)


# ============================================================
# COMBINE COMPONENT FORECASTS
# ============================================================

ensemble_forecast = (
    SARIMA_WEIGHT * sarima_forecast
    + CLIMATOLOGY_WEIGHT * climatology_forecast
)

full_forecast = pd.DataFrame({
    "month": future_months,
    "forecast_horizon":
        np.arange(1, FULL_HORIZON + 1),
    "sarima_forecast_F":
        sarima_forecast,
    "climatology_15y_forecast_F":
        climatology_forecast,
    "ensemble_forecast_F":
        ensemble_forecast,
    "sarima_weight":
        SARIMA_WEIGHT,
    "climatology_weight":
        CLIMATOLOGY_WEIGHT,
    "trained_through":
        EXPECTED_TRAINING_END
})


# ============================================================
# EXTRACT HORIZONS 15–26
# ============================================================

official_forecast = full_forecast[
    full_forecast["forecast_horizon"]
    .between(
        REQUESTED_FIRST_HORIZON,
        REQUESTED_LAST_HORIZON
    )
].copy()

if len(official_forecast) != 12:
    raise ValueError(
        "Official forecast does not contain 12 months."
    )

if (
    official_forecast["month"].min()
    != EXPECTED_OUTPUT_START
):
    raise ValueError(
        "Official forecast does not begin in September 2027."
    )

if (
    official_forecast["month"].max()
    != EXPECTED_OUTPUT_END
):
    raise ValueError(
        "Official forecast does not end in August 2028."
    )


# ============================================================
# ADD CALIBRATED INTERVALS
# ============================================================

official_forecast["lower_80_F"] = (
    official_forecast["ensemble_forecast_F"]
    - FINAL_80_HALF_WIDTH
)

official_forecast["upper_80_F"] = (
    official_forecast["ensemble_forecast_F"]
    + FINAL_80_HALF_WIDTH
)

official_forecast["lower_95_F"] = (
    official_forecast["ensemble_forecast_F"]
    - FINAL_95_HALF_WIDTH
)

official_forecast["upper_95_F"] = (
    official_forecast["ensemble_forecast_F"]
    + FINAL_95_HALF_WIDTH
)

official_forecast["model"] = (
    "50% SARIMA + 50% 15-year climatology"
)

official_forecast["interval_method"] = (
    "empirical pooled out-of-sample absolute errors"
)

official_forecast = official_forecast[
    [
        "month",
        "forecast_horizon",
        "ensemble_forecast_F",
        "lower_80_F",
        "upper_80_F",
        "lower_95_F",
        "upper_95_F",
        "sarima_forecast_F",
        "climatology_15y_forecast_F",
        "sarima_weight",
        "climatology_weight",
        "model",
        "trained_through",
        "interval_method"
    ]
]


# ============================================================
# DISPLAY OFFICIAL FORECAST
# ============================================================

print("OFFICIAL SEPTEMBER 2027–AUGUST 2028 FORECAST")
print("=============================================")

display(
    official_forecast[
        [
            "month",
            "forecast_horizon",
            "ensemble_forecast_F",
            "lower_80_F",
            "upper_80_F",
            "lower_95_F",
            "upper_95_F"
        ]
    ].round(3)
)


# ============================================================
# SAVE PRODUCTION ARTIFACTS
# ============================================================

training_output = (
    series
    .rename("mean_temp_F")
    .reset_index()
    .rename(columns={"index": "month"})
)

training_output.to_csv(
    TRAINING_FILE,
    index=False
)

full_forecast.to_csv(
    FULL_FORECAST_FILE,
    index=False
)

official_forecast.to_csv(
    FORECAST_FILE,
    index=False
)

ensemble_backtests.to_csv(
    CALIBRATION_ERRORS_FILE,
    index=False
)

aggregate_validation.to_csv(
    INTERVAL_VALIDATION_FILE,
    index=False
)

final_widths.to_csv(
    INTERVAL_WIDTH_FILE,
    index=False
)

horizon_coverage.to_csv(
    f"{OUTPUT_DIR}/interval_coverage_by_horizon.csv",
    index=False
)

window_validation.to_csv(
    f"{OUTPUT_DIR}/interval_validation_by_window.csv",
    index=False
)

validation_predictions.to_csv(
    f"{OUTPUT_DIR}/interval_validation_predictions.csv",
    index=False
)

final_sarima_fit.save(
    MODEL_FILE,
    remove_data=False
)

with open(
    MODEL_SUMMARY_FILE,
    "w"
) as file:
    file.write(
        final_sarima_fit
        .summary()
        .as_text()
    )


# ============================================================
# FORECAST PLOT
# ============================================================

plot_history = series.loc["2018-01-01":]

fig, ax = plt.subplots(
    figsize=(14, 7)
)

ax.plot(
    plot_history.index,
    plot_history.values,
    color="black",
    linewidth=1.6,
    label="Observed Stanford temperature"
)

ax.plot(
    official_forecast["month"],
    official_forecast["ensemble_forecast_F"],
    color="#B1040E",
    marker="o",
    linewidth=2.3,
    label="SARIMA–climatology ensemble"
)

ax.fill_between(
    official_forecast["month"],
    official_forecast["lower_95_F"],
    official_forecast["upper_95_F"],
    color="#B1040E",
    alpha=0.13,
    label="Empirical 95% interval"
)

ax.fill_between(
    official_forecast["month"],
    official_forecast["lower_80_F"],
    official_forecast["upper_80_F"],
    color="#B1040E",
    alpha=0.27,
    label="Empirical 80% interval"
)

ax.axvline(
    EXPECTED_TRAINING_END,
    color="gray",
    linestyle="--",
    linewidth=1.3,
    label="Training cutoff"
)

ax.set_title(
    "Stanford Monthly Average Temperature Forecast\n"
    "September 2027–August 2028"
)

ax.set_xlabel("Month")
ax.set_ylabel(
    "Monthly average temperature (°F)"
)

ax.grid(alpha=0.25)
ax.legend(loc="best")

fig.tight_layout()

fig.savefig(
    PLOT_FILE,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# REPRODUCIBILITY METADATA
# ============================================================

metadata = {
    "project":
        "Stanford medium-horizon temperature forecast",

    "artifact_status":
        "production_candidate_for_review",

    "target":
        "Stanford University monthly mean temperature in Fahrenheit",

    "selected_model": {
        "name":
            "50% SARIMA + 50% 15-year climatology",

        "sarima_order":
            list(SARIMA_ORDER),

        "sarima_seasonal_order":
            list(SARIMA_SEASONAL_ORDER),

        "sarima_weight":
            SARIMA_WEIGHT,

        "climatology_weight":
            CLIMATOLOGY_WEIGHT,

        "climatology_lookback_years":
            15,

        "sarima_converged":
            sarima_converged,

        "sarima_aic":
            float(final_sarima_fit.aic),

        "sarima_bic":
            float(final_sarima_fit.bic)
    },

    "training_data": {
        "first_month":
            str(EXPECTED_TRAINING_START.date()),

        "last_month":
            str(EXPECTED_TRAINING_END.date()),

        "observations":
            int(len(series)),

        "monthly_data_file":
            MONTHLY_FILE,

        "monthly_data_sha256":
            file_sha256(MONTHLY_FILE)
    },

    "forecast": {
        "calculation_horizon_months":
            FULL_HORIZON,

        "official_first_horizon":
            REQUESTED_FIRST_HORIZON,

        "official_last_horizon":
            REQUESTED_LAST_HORIZON,

        "official_start":
            str(EXPECTED_OUTPUT_START.date()),

        "official_end":
            str(EXPECTED_OUTPUT_END.date()),

        "official_months":
            12
    },

    "historical_validation": {
        "rolling_windows":
            9,

        "out_of_sample_months":
            108,

        "pooled_RMSE_F":
            1.819,

        "pooled_MAE_F":
            1.445,

        "pooled_bias_F":
            -0.183,

        "worst_window_RMSE_F":
            2.398,

        "interval_validation_method":
            "leave-one-forecast-window-out",

        "final_interval_calibration_method":
            "finite-sample quantile of 108 pooled absolute out-of-sample errors",

        "interval_80_half_width_F":
            FINAL_80_HALF_WIDTH,

        "interval_95_half_width_F":
            FINAL_95_HALF_WIDTH
    },

    "software": {
        "python":
            platform.python_version(),

        "numpy":
            np.__version__,

        "pandas":
            pd.__version__,

        "statsmodels":
            statsmodels.__version__,

        "matplotlib":
            plt.matplotlib.__version__
    },

    "generated_at_utc":
        datetime.now(timezone.utc).isoformat(),

    "output_files": {
        "official_forecast":
            FORECAST_FILE,

        "full_component_forecast":
            FULL_FORECAST_FILE,

        "sarima_component_model":
            MODEL_FILE,

        "training_series":
            TRAINING_FILE,

        "interval_validation":
            INTERVAL_VALIDATION_FILE,

        "interval_widths":
            INTERVAL_WIDTH_FILE,

        "forecast_plot":
            PLOT_FILE
    }
}

with open(
    METADATA_FILE,
    "w"
) as file:
    json.dump(
        metadata,
        file,
        indent=2
    )

package_versions = [
    f"python=={platform.python_version()}",
    f"numpy=={np.__version__}",
    f"pandas=={pd.__version__}",
    f"statsmodels=={statsmodels.__version__}",
    f"matplotlib=={plt.matplotlib.__version__}"
]

with open(
    PACKAGE_FILE,
    "w"
) as file:
    file.write(
        "\n".join(package_versions) + "\n"
    )


# ============================================================
# FINAL FILE VERIFICATION
# ============================================================

required_outputs = [
    FORECAST_FILE,
    FULL_FORECAST_FILE,
    TRAINING_FILE,
    MODEL_FILE,
    INTERVAL_VALIDATION_FILE,
    INTERVAL_WIDTH_FILE,
    CALIBRATION_ERRORS_FILE,
    METADATA_FILE,
    PLOT_FILE,
    MODEL_SUMMARY_FILE,
    PACKAGE_FILE
]

missing_outputs = [
    path for path in required_outputs
    if not os.path.exists(path)
]

if missing_outputs:
    raise RuntimeError(
        "Some production files were not created:\n"
        f"{missing_outputs}"
    )


print()
print("FINAL PRODUCTION ARTIFACTS SAVED")
print("================================")
print(f"Directory:          {OUTPUT_DIR}")
print(f"Official forecast:  {FORECAST_FILE}")
print(f"Full components:    {FULL_FORECAST_FILE}")
print(f"SARIMA component:   {MODEL_FILE}")
print(f"Interval validation:{INTERVAL_VALIDATION_FILE}")
print(f"Interval widths:    {INTERVAL_WIDTH_FILE}")
print(f"Metadata:           {METADATA_FILE}")
print(f"Forecast plot:      {PLOT_FILE}")

print()
print(
    "The September 2027–August 2028 ensemble forecast "
    "is now packaged as a production candidate for review."
)
```
