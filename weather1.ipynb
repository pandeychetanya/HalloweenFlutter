```python
# ============================================================
# FOUNDATIONAL ANALYSIS 1
# UNDERSTAND ONE RAW MONTH OF STANFORD TEMPERATURE DATA
#
# Based on the timestamp-processing and aggregation approach
# used in Joy's and Khadijah's supplied code.
#
# This cell does NOT:
# - remove duplicate observations
# - interpolate missing observations
# - change the source file
# - fit a forecasting model
# ============================================================

from google.colab import drive
drive.mount("/content/drive")

# ============================================================
# IMPORTS
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display


# ============================================================
# SETTINGS
# ============================================================

PROJECT = "/content/drive/MyDrive/Stanford_Weather_12M"

DATA_FILE = (
    f"{PROJECT}/data/"
    "OpenWeather19802026_StanfordOnly.csv"
)

CITY = "Stanford University"

# Choose a complete historical month.
# January avoids daylight-saving-time clock changes,
# which makes the initial hourly audit easier to understand.
ANALYSIS_MONTH = "2024-01"


# ============================================================
# CHECK THAT THE FILE EXISTS
# ============================================================

if not os.path.exists(DATA_FILE):
    raise FileNotFoundError(
        f"Could not find the source file:\n{DATA_FILE}\n\n"
        "Change DATA_FILE if your CSV is stored elsewhere."
    )


# ============================================================
# LOAD THE RAW DATA
# ============================================================

raw = pd.read_csv(DATA_FILE)

print("SOURCE FILE")
print("===========")
print(DATA_FILE)
print(f"Rows: {len(raw):,}")
print(f"Columns: {len(raw.columns)}")
print()

print("AVAILABLE COLUMNS")
print("=================")
print(raw.columns.tolist())
print()


# ============================================================
# VERIFY THE REQUIRED COLUMNS
# ============================================================

required_columns = [
    "dt",
    "timezone",
    "city_name",
    "temp"
]

missing_columns = [
    column
    for column in required_columns
    if column not in raw.columns
]

if missing_columns:
    raise ValueError(
        f"Required columns are missing: {missing_columns}"
    )


# ============================================================
# KEEP STANFORD UNIVERSITY
# ============================================================

df = raw[
    raw["city_name"] == CITY
].copy()

if df.empty:
    raise ValueError(
        f"No observations were found for '{CITY}'.\n"
        f"Available city names:\n"
        f"{raw['city_name'].dropna().unique().tolist()}"
    )


# ============================================================
# CONVERT THE TIMESTAMPS
#
# This follows Joy and Khadijah's approach:
# 1. Interpret dt as Unix seconds in UTC.
# 2. Add the timezone offset supplied by OpenWeather.
# 3. Remove timezone awareness after obtaining local time.
# ============================================================

df["dt_utc"] = pd.to_datetime(
    pd.to_numeric(
        df["dt"],
        errors="coerce"
    ),
    unit="s",
    utc=True
)

df["timezone_offset"] = pd.to_timedelta(
    pd.to_numeric(
        df["timezone"],
        errors="coerce"
    ),
    unit="s"
)

df["dt_local"] = (
    df["dt_utc"] +
    df["timezone_offset"]
)

# Keep local time as an ordinary timestamp.
df["dt_local"] = (
    df["dt_local"]
    .dt
    .tz_localize(None)
)

df["temp"] = pd.to_numeric(
    df["temp"],
    errors="coerce"
)

df = (
    df.dropna(
        subset=[
            "dt_utc",
            "dt_local",
            "temp"
        ]
    )
    .sort_values("dt_local")
    .reset_index(drop=True)
)


# ============================================================
# SELECT ONE COMPLETE LOCAL CALENDAR MONTH
# ============================================================

month_start = pd.Timestamp(
    f"{ANALYSIS_MONTH}-01"
)

next_month_start = (
    month_start +
    pd.offsets.MonthBegin(1)
)

one_month = df[
    (df["dt_local"] >= month_start) &
    (df["dt_local"] < next_month_start)
].copy()

if one_month.empty:
    raise ValueError(
        f"No data were found for {ANALYSIS_MONTH}."
    )


# ============================================================
# VERIFY THAT THE OBSERVATIONS ARE HOURLY
# ============================================================

one_month["time_difference"] = (
    one_month["dt_local"]
    .diff()
)

time_difference_counts = (
    one_month["time_difference"]
    .value_counts()
    .rename_axis("time_difference")
    .reset_index(name="number_of_occurrences")
)

one_hour = pd.Timedelta(hours=1)

hourly_step_percentage = (
    one_month["time_difference"]
    .dropna()
    .eq(one_hour)
    .mean()
    * 100
)


# ============================================================
# CHECK EXPECTED HOURS, DUPLICATES AND MISSING TIMESTAMPS
# ============================================================

expected_timestamps = pd.date_range(
    start=month_start,
    end=next_month_start,
    freq="h",
    inclusive="left"
)

observed_timestamps = pd.DatetimeIndex(
    one_month["dt_local"]
)

missing_timestamps = (
    expected_timestamps
    .difference(observed_timestamps)
)

duplicate_timestamp_count = int(
    one_month["dt_local"]
    .duplicated(keep=False)
    .sum()
)


# ============================================================
# CALCULATE THE MONTHLY MEAN
#
# This is the foundational version of the monthly aggregation
# used by both Joy and Khadijah.
# ============================================================

monthly_mean_temperature = float(
    one_month["temp"].mean()
)

monthly_minimum_temperature = float(
    one_month["temp"].min()
)

monthly_maximum_temperature = float(
    one_month["temp"].max()
)


# ============================================================
# PRINT THE AUDIT RESULTS
# ============================================================

print("ONE RAW MONTH")
print("=============")
print(f"City:                 {CITY}")
print(f"Local calendar month: {ANALYSIS_MONTH}")
print(
    "First local timestamp:",
    one_month["dt_local"].min()
)
print(
    "Last local timestamp: ",
    one_month["dt_local"].max()
)
print(f"Observed rows:         {len(one_month):,}")
print(f"Expected hourly rows:  {len(expected_timestamps):,}")
print(
    "Missing temperatures:",
    int(one_month["temp"].isna().sum())
)
print(
    "Duplicate timestamp rows:",
    duplicate_timestamp_count
)
print(
    "Missing hourly timestamps:",
    len(missing_timestamps)
)
print(
    "Consecutive differences equal to one hour:",
    f"{hourly_step_percentage:.2f}%"
)
print()

print("TEMPERATURE SUMMARY")
print("===================")
print(
    f"Monthly mean: {monthly_mean_temperature:.3f} °F"
)
print(
    f"Monthly minimum hourly value: "
    f"{monthly_minimum_temperature:.3f} °F"
)
print(
    f"Monthly maximum hourly value: "
    f"{monthly_maximum_temperature:.3f} °F"
)
print()

print("MOST COMMON TIME DIFFERENCES")
print("============================")
display(
    time_difference_counts.head(10)
)

print("FIRST 48 RAW OBSERVATIONS")
print("=========================")

columns_to_display = [
    "dt",
    "dt_iso",
    "timezone",
    "dt_utc",
    "dt_local",
    "city_name",
    "temp"
]

columns_to_display = [
    column
    for column in columns_to_display
    if column in one_month.columns
]

display(
    one_month[
        columns_to_display
    ].head(48)
)


# ============================================================
# DISPLAY MISSING OR DUPLICATE TIMES, IF ANY
# ============================================================

if len(missing_timestamps) > 0:
    print("FIRST MISSING HOURLY TIMESTAMPS")
    print("===============================")

    display(
        pd.DataFrame({
            "missing_dt_local":
                missing_timestamps[:48]
        })
    )
else:
    print(
        "No missing hourly timestamps were found "
        "in this month."
    )

print()

duplicates = one_month[
    one_month["dt_local"]
    .duplicated(keep=False)
].copy()

if not duplicates.empty:
    print("DUPLICATE TIMESTAMP OBSERVATIONS")
    print("================================")

    display(
        duplicates[
            columns_to_display
        ].head(48)
    )
else:
    print(
        "No duplicate local timestamps were found "
        "in this month."
    )


# ============================================================
# PLOT 1: HOURLY TEMPERATURE THROUGH THE MONTH
# ============================================================

plt.figure(
    figsize=(14, 5)
)

plt.plot(
    one_month["dt_local"],
    one_month["temp"],
    linewidth=0.8,
    color="#2F75B5"
)

plt.axhline(
    monthly_mean_temperature,
    color="#C00000",
    linestyle="--",
    linewidth=2,
    label=(
        "Monthly mean = "
        f"{monthly_mean_temperature:.2f} °F"
    )
)

plt.title(
    f"Stanford University Hourly Temperature — "
    f"{month_start:%B %Y}"
)

plt.xlabel(
    "Local Stanford time"
)

plt.ylabel(
    "Temperature (°F)"
)

plt.grid(
    alpha=0.25
)

plt.legend()

plt.tight_layout()
plt.show()


# ============================================================
# PLOT 2: DISTRIBUTION OF HOURLY TEMPERATURES
# ============================================================

plt.figure(
    figsize=(9, 5)
)

plt.hist(
    one_month["temp"],
    bins=25,
    color="#5B9BD5",
    edgecolor="white"
)

plt.axvline(
    monthly_mean_temperature,
    color="#C00000",
    linestyle="--",
    linewidth=2,
    label=(
        "Monthly mean = "
        f"{monthly_mean_temperature:.2f} °F"
    )
)

plt.title(
    f"Distribution of Hourly Temperatures — "
    f"{month_start:%B %Y}"
)

plt.xlabel(
    "Temperature (°F)"
)

plt.ylabel(
    "Number of hourly observations"
)

plt.grid(
    axis="y",
    alpha=0.25
)

plt.legend()

plt.tight_layout()
plt.show()


# ============================================================
# DAILY SUMMARY WITHIN THE MONTH
#
# This is for understanding the raw month only.
# The forecasting models still use monthly mean temperature.
# ============================================================

daily_summary = (
    one_month
    .set_index("dt_local")["temp"]
    .resample("D")
    .agg(
        daily_mean_F="mean",
        daily_min_F="min",
        daily_max_F="max",
        hourly_rows="count"
    )
)

print("DAILY SUMMARY WITHIN THE SELECTED MONTH")
print("=======================================")

display(
    daily_summary.round(3)
)


# ============================================================
# FINAL MONTHLY VALUE
# ============================================================

monthly_result = pd.DataFrame({
    "city_name": [CITY],
    "month": [month_start],
    "hourly_rows": [len(one_month)],
    "mean_temp_F": [monthly_mean_temperature],
    "minimum_hourly_temp_F": [
        monthly_minimum_temperature
    ],
    "maximum_hourly_temp_F": [
        monthly_maximum_temperature
    ],
    "missing_hourly_timestamps": [
        len(missing_timestamps)
    ],
    "duplicate_timestamp_rows": [
        duplicate_timestamp_count
    ]
})

print("FINAL ONE-MONTH RESULT")
print("======================")

display(
    monthly_result.round(3)
)
```

```python
# ============================================================
# FOUNDATIONAL ANALYSIS 1B
# INVESTIGATE DUPLICATE TIMESTAMPS IN THE SELECTED MONTH
#
# Do not remove or modify anything yet.
# This cell only diagnoses the duplicate observations.
# ============================================================

# Count how many rows occur at every local timestamp.
timestamp_counts = (
    one_month
    .groupby("dt_local")
    .size()
    .rename("rows_at_timestamp")
    .sort_values(ascending=False)
)

duplicate_timestamp_counts = (
    timestamp_counts[
        timestamp_counts > 1
    ]
)

number_of_unique_timestamps = (
    one_month["dt_local"]
    .nunique()
)

number_of_duplicated_timestamps = (
    len(duplicate_timestamp_counts)
)

# Extra rows beyond one observation per timestamp.
number_of_extra_rows = int(
    (
        duplicate_timestamp_counts - 1
    ).sum()
)


print("TIMESTAMP COUNT EXPLANATION")
print("===========================")

print(
    f"Total rows:                  "
    f"{len(one_month):,}"
)

print(
    f"Unique local timestamps:     "
    f"{number_of_unique_timestamps:,}"
)

print(
    f"Expected hourly timestamps:  "
    f"{len(expected_timestamps):,}"
)

print(
    f"Duplicated timestamps:        "
    f"{number_of_duplicated_timestamps:,}"
)

print(
    f"Extra rows from duplicates:   "
    f"{number_of_extra_rows:,}"
)

print(
    "Rows belonging to duplicate groups: "
    f"{duplicate_timestamp_count:,}"
)

print()


# ============================================================
# DISPLAY THE DUPLICATED TIMESTAMPS
# ============================================================

print("DUPLICATED TIMESTAMPS")
print("=====================")

display(
    duplicate_timestamp_counts
    .rename_axis("dt_local")
    .reset_index()
)


# ============================================================
# DISPLAY EVERY ROW INVOLVED IN A DUPLICATE
# ============================================================

duplicate_rows = (
    one_month[
        one_month["dt_local"]
        .isin(
            duplicate_timestamp_counts.index
        )
    ]
    .copy()
    .sort_values(
        [
            "dt_local",
            "temp"
        ]
    )
)

duplicate_display_columns = [
    "dt",
    "dt_iso",
    "timezone",
    "dt_utc",
    "dt_local",
    "city_name",
    "lat",
    "lon",
    "temp",
    "feels_like",
    "temp_min",
    "temp_max",
    "dew_point",
    "humidity",
    "pressure",
    "wind_speed",
    "wind_deg"
]

duplicate_display_columns = [
    column
    for column in duplicate_display_columns
    if column in duplicate_rows.columns
]

display(
    duplicate_rows[
        duplicate_display_columns
    ]
)


# ============================================================
# CHECK WHETHER DUPLICATED TIMESTAMPS HAVE DIFFERENT TEMPERATURES
# ============================================================

duplicate_temperature_summary = (
    duplicate_rows
    .groupby("dt_local")
    .agg(
        rows=("temp", "size"),
        first_temp_F=("temp", "first"),
        last_temp_F=("temp", "last"),
        minimum_temp_F=("temp", "min"),
        maximum_temp_F=("temp", "max"),
        unique_temperature_values=(
            "temp",
            "nunique"
        )
    )
)

duplicate_temperature_summary[
    "temperature_range_F"
] = (
    duplicate_temperature_summary[
        "maximum_temp_F"
    ]
    -
    duplicate_temperature_summary[
        "minimum_temp_F"
    ]
)

print("TEMPERATURES WITHIN DUPLICATED HOURS")
print("====================================")

display(
    duplicate_temperature_summary
    .round(3)
)


# ============================================================
# CHECK EXACT FULL-ROW DUPLICATES
# ============================================================

# Exclude the diagnostic column created in the previous cell.
comparison_columns = [
    column
    for column in one_month.columns
    if column != "time_difference"
]

exact_duplicate_mask = (
    one_month[
        comparison_columns
    ]
    .duplicated(
        keep=False
    )
)

exact_duplicate_rows = (
    one_month[
        exact_duplicate_mask
    ]
    .copy()
)

print("EXACT FULL-ROW DUPLICATES")
print("=========================")

print(
    f"Rows that are exact duplicates across "
    f"all compared columns: "
    f"{len(exact_duplicate_rows):,}"
)

if not exact_duplicate_rows.empty:
    display(
        exact_duplicate_rows[
            duplicate_display_columns
        ]
        .sort_values("dt_local")
    )
else:
    print(
        "No exact full-row duplicates were found. "
        "The rows share timestamps but differ in "
        "at least one field."
    )

print()


# ============================================================
# COMPARE POSSIBLE MONTHLY MEAN CALCULATIONS
#
# 1. Existing Joy/Khadijah behavior:
#    average every source row.
#
# 2. Keep first:
#    one observation per timestamp.
#
# 3. Keep last:
#    one observation per timestamp.
#
# 4. Average within timestamp:
#    average duplicated observations first,
#    then give every hour equal weight.
# ============================================================

mean_using_all_rows = float(
    one_month["temp"].mean()
)

one_per_timestamp_first = (
    one_month
    .drop_duplicates(
        subset="dt_local",
        keep="first"
    )
)

mean_keep_first = float(
    one_per_timestamp_first[
        "temp"
    ].mean()
)

one_per_timestamp_last = (
    one_month
    .drop_duplicates(
        subset="dt_local",
        keep="last"
    )
)

mean_keep_last = float(
    one_per_timestamp_last[
        "temp"
    ].mean()
)

hourly_equal_weight_series = (
    one_month
    .groupby("dt_local")["temp"]
    .mean()
)

mean_equal_weight_per_hour = float(
    hourly_equal_weight_series.mean()
)


mean_comparison = pd.DataFrame({
    "method": [
        "All source rows",
        "One row per timestamp — keep first",
        "One row per timestamp — keep last",
        "Average duplicates within hour first"
    ],
    "observations_used": [
        len(one_month),
        len(one_per_timestamp_first),
        len(one_per_timestamp_last),
        len(hourly_equal_weight_series)
    ],
    "january_mean_temp_F": [
        mean_using_all_rows,
        mean_keep_first,
        mean_keep_last,
        mean_equal_weight_per_hour
    ]
})

mean_comparison[
    "difference_from_existing_F"
] = (
    mean_comparison[
        "january_mean_temp_F"
    ]
    -
    mean_using_all_rows
)


print("EFFECT OF DUPLICATES ON THE MONTHLY MEAN")
print("========================================")

display(
    mean_comparison.round(6)
)
```

```python
# ============================================================
# AUDIT DUPLICATE TIMESTAMPS AND MONTHLY TEMPERATURE AGGREGATION
# January 1980 through June 2026
#
# Baseline:
#   Joy/Khadijah pattern = average every source row
#
# Comparison:
#   First average duplicate records within each local hour,
#   then average those equally weighted hourly temperatures
# ============================================================

from google.colab import drive
drive.mount("/content/drive")

from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display


# ============================================================
# 1. FILE SETTINGS
# ============================================================

PROJECT = Path("/content/drive/MyDrive/Stanford_Weather_12M")

# Change only this line if your file is stored elsewhere.
DATA_FILE = PROJECT / "data" / "OpenWeather19802026_StanfordOnly.csv"

OUTPUT_DIR = PROJECT / "hourly_to_monthly_aggregation_audit"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

CITY = "Stanford University"
START_MONTH = pd.Timestamp("1980-01-01")
END_MONTH = pd.Timestamp("2026-06-01")


# ============================================================
# 2. LOAD ONLY THE COLUMNS NEEDED FOR THIS AUDIT
# ============================================================

df = pd.read_csv(
    DATA_FILE,
    usecols=["dt", "dt_iso", "timezone", "city_name", "temp"]
)

df["temp"] = pd.to_numeric(df["temp"], errors="coerce")
df["timezone"] = pd.to_numeric(df["timezone"], errors="coerce")


# ============================================================
# 3. RECREATE LOCAL TIME
#
# This follows Rebecca/Khadijah's timestamp approach:
# Unix timestamp -> UTC datetime -> add timezone offset
# ============================================================

df["dt_utc"] = pd.to_datetime(
    df["dt"],
    unit="s",
    utc=True,
    errors="coerce"
)

df["timezone_offset"] = pd.to_timedelta(
    df["timezone"],
    unit="s"
)

# Remove timezone information after applying the local offset.
# This gives ordinary Stanford local timestamps for grouping.
df["dt_local"] = (
    df["dt_utc"] + df["timezone_offset"]
).dt.tz_localize(None)

df = df.loc[
    (df["city_name"] == CITY)
    & df["dt_local"].notna()
].copy()

df["month"] = (
    df["dt_local"]
    .dt.to_period("M")
    .dt.to_timestamp()
)

df = df.loc[
    (df["month"] >= START_MONTH)
    & (df["month"] <= END_MONTH)
].copy()

df = df.sort_values(["dt_local"]).reset_index(drop=True)


# ============================================================
# 4. MEASURE ROWS, UNIQUE HOURS AND DUPLICATE TIMESTAMPS
# ============================================================

monthly_counts = (
    df.groupby("month")
      .agg(
          source_rows=("dt_local", "size"),
          unique_hourly_timestamps=("dt_local", "nunique"),
          nonmissing_temperature_rows=("temp", "count"),
          missing_temperature_rows=("temp", lambda x: x.isna().sum())
      )
)

monthly_counts["extra_rows_from_duplicates"] = (
    monthly_counts["source_rows"]
    - monthly_counts["unique_hourly_timestamps"]
)

# Number of source rows belonging to duplicated timestamp groups.
duplicate_mask = df.duplicated(
    subset=["dt_local"],
    keep=False
)

duplicate_rows_by_month = (
    df.loc[duplicate_mask]
      .groupby("month")
      .size()
      .rename("rows_in_duplicate_timestamp_groups")
)

monthly_counts = monthly_counts.join(
    duplicate_rows_by_month,
    how="left"
)

monthly_counts["rows_in_duplicate_timestamp_groups"] = (
    monthly_counts["rows_in_duplicate_timestamp_groups"]
    .fillna(0)
    .astype(int)
)


# ============================================================
# 5. JOY/KHADIJAH MONTHLY MEAN
#
# Their pattern groups all source records by month and takes
# the average temperature. Duplicate hours therefore receive
# additional weight if they appear more than once.
# ============================================================

all_rows_mean = (
    df.groupby("month")["temp"]
      .mean()
      .rename("mean_temp_all_rows_F")
)


# ============================================================
# 6. EQUAL-HOUR MONTHLY MEAN
#
# First average records sharing the same local hourly timestamp.
# Then give every unique hour exactly one vote in the month.
# ============================================================

hourly_equal_weight = (
    df.groupby(["month", "dt_local"], as_index=False)
      .agg(
          hourly_temp_F=("temp", "mean"),
          rows_at_timestamp=("temp", "size"),
          unique_temperature_values=("temp", "nunique")
      )
)

equal_hour_mean = (
    hourly_equal_weight
    .groupby("month")["hourly_temp_F"]
    .mean()
    .rename("mean_temp_equal_hour_F")
)


# ============================================================
# 7. CHECK WHETHER DUPLICATE HOURS CONTAIN DIFFERENT TEMPERATURES
# ============================================================

duplicate_hour_details = hourly_equal_weight.loc[
    hourly_equal_weight["rows_at_timestamp"] > 1
].copy()

different_temp_duplicate_hours = (
    duplicate_hour_details
    .groupby("month")
    .apply(
        lambda group: int(
            (group["unique_temperature_values"] > 1).sum()
        ),
        include_groups=False
    )
    .rename("duplicate_hours_with_different_temperatures")
)


# ============================================================
# 8. BUILD THE MONTH-BY-MONTH COMPARISON
# ============================================================

comparison = (
    monthly_counts
    .join(all_rows_mean)
    .join(equal_hour_mean)
    .join(different_temp_duplicate_hours, how="left")
    .reset_index()
)

comparison["duplicate_hours_with_different_temperatures"] = (
    comparison["duplicate_hours_with_different_temperatures"]
    .fillna(0)
    .astype(int)
)

comparison["equal_hour_minus_all_rows_F"] = (
    comparison["mean_temp_equal_hour_F"]
    - comparison["mean_temp_all_rows_F"]
)

comparison["absolute_difference_F"] = (
    comparison["equal_hour_minus_all_rows_F"].abs()
)

comparison["duplicate_row_percent"] = (
    100
    * comparison["extra_rows_from_duplicates"]
    / comparison["source_rows"]
)

comparison["year_month"] = (
    comparison["month"].dt.strftime("%Y-%m")
)


# ============================================================
# 9. VERIFY THE REQUESTED DATE RANGE
# ============================================================

expected_months = pd.date_range(
    START_MONTH,
    END_MONTH,
    freq="MS"
)

missing_calendar_months = expected_months.difference(
    pd.DatetimeIndex(comparison["month"])
)

print("DATA RANGE")
print("==========")
print(f"Source file:          {DATA_FILE}")
print(f"City:                 {CITY}")
print(f"First included month: {comparison['month'].min():%Y-%m}")
print(f"Last included month:  {comparison['month'].max():%Y-%m}")
print(f"Months found:         {len(comparison)}")
print(f"Months expected:      {len(expected_months)}")
print(f"Missing months:       {len(missing_calendar_months)}")

if len(missing_calendar_months) > 0:
    print(
        "Missing calendar months:",
        [month.strftime("%Y-%m") for month in missing_calendar_months]
    )


# ============================================================
# 10. OVERALL AUDIT SUMMARY
# ============================================================

months_with_duplicates = (
    comparison["extra_rows_from_duplicates"] > 0
).sum()

months_with_changed_mean = (
    comparison["absolute_difference_F"] > 1e-12
).sum()

summary = pd.DataFrame({
    "measure": [
        "Months audited",
        "Months containing repeated timestamps",
        "Months whose mean changes after equal-hour weighting",
        "Total extra rows from repeated timestamps",
        "Duplicate hours with different temperature values",
        "Mean absolute monthly difference (°F)",
        "Median absolute monthly difference (°F)",
        "Largest absolute monthly difference (°F)",
        "Average signed difference (°F)"
    ],
    "value": [
        len(comparison),
        months_with_duplicates,
        months_with_changed_mean,
        int(comparison["extra_rows_from_duplicates"].sum()),
        int(
            comparison[
                "duplicate_hours_with_different_temperatures"
            ].sum()
        ),
        comparison["absolute_difference_F"].mean(),
        comparison["absolute_difference_F"].median(),
        comparison["absolute_difference_F"].max(),
        comparison["equal_hour_minus_all_rows_F"].mean()
    ]
})

print("\nOVERALL AGGREGATION AUDIT")
print("=========================")
display(summary)


# ============================================================
# 11. SHOW MONTHS MOST AFFECTED BY DUPLICATES
# ============================================================

most_affected = (
    comparison.loc[
        comparison["extra_rows_from_duplicates"] > 0,
        [
            "year_month",
            "source_rows",
            "unique_hourly_timestamps",
            "extra_rows_from_duplicates",
            "duplicate_row_percent",
            "duplicate_hours_with_different_temperatures",
            "mean_temp_all_rows_F",
            "mean_temp_equal_hour_F",
            "equal_hour_minus_all_rows_F",
            "absolute_difference_F"
        ]
    ]
    .sort_values(
        "absolute_difference_F",
        ascending=False
    )
    .reset_index(drop=True)
)

print("\n20 MONTHS WITH THE LARGEST AGGREGATION DIFFERENCES")
print("==================================================")
display(most_affected.head(20).round(6))


# ============================================================
# 12. REDISPLAY JANUARY 2024 AS A CONSISTENCY CHECK
# ============================================================

january_2024 = comparison.loc[
    comparison["month"] == pd.Timestamp("2024-01-01"),
    [
        "year_month",
        "source_rows",
        "unique_hourly_timestamps",
        "extra_rows_from_duplicates",
        "rows_in_duplicate_timestamp_groups",
        "mean_temp_all_rows_F",
        "mean_temp_equal_hour_F",
        "equal_hour_minus_all_rows_F"
    ]
]

print("\nJANUARY 2024 CONSISTENCY CHECK")
print("==============================")
display(january_2024.round(6))


# ============================================================
# 13. PLOT THE EFFECT ACROSS TIME
# ============================================================

fig, axes = plt.subplots(
    2,
    1,
    figsize=(14, 8),
    sharex=True
)

axes[0].plot(
    comparison["month"],
    comparison["equal_hour_minus_all_rows_F"],
    color="darkblue",
    linewidth=1
)

axes[0].axhline(
    0,
    color="black",
    linewidth=1
)

axes[0].set_title(
    "Effect of Equal-Hour Weighting on Monthly Mean Temperature"
)
axes[0].set_ylabel(
    "Equal-hour mean − all-row mean (°F)"
)
axes[0].grid(alpha=0.25)

axes[1].bar(
    comparison["month"],
    comparison["extra_rows_from_duplicates"],
    width=25,
    color="darkorange"
)

axes[1].set_title(
    "Extra Source Rows Caused by Repeated Hourly Timestamps"
)
axes[1].set_ylabel("Extra rows")
axes[1].set_xlabel("Month")
axes[1].grid(axis="y", alpha=0.25)

plt.tight_layout()

plot_file = OUTPUT_DIR / "duplicate_timestamp_effect_by_month.png"
plt.savefig(plot_file, dpi=200, bbox_inches="tight")
plt.show()


# ============================================================
# 14. SAVE RESULTS
# ============================================================

comparison_file = OUTPUT_DIR / "monthly_aggregation_comparison.csv"
summary_file = OUTPUT_DIR / "aggregation_audit_summary.csv"
affected_file = OUTPUT_DIR / "months_with_duplicate_timestamps.csv"
duplicate_details_file = OUTPUT_DIR / "duplicate_hour_details.csv"

comparison.to_csv(comparison_file, index=False)
summary.to_csv(summary_file, index=False)
most_affected.to_csv(affected_file, index=False)
duplicate_hour_details.to_csv(
    duplicate_details_file,
    index=False
)

print("\nFILES SAVED")
print("===========")
print(f"Full monthly comparison: {comparison_file}")
print(f"Overall summary:         {summary_file}")
print(f"Affected months:         {affected_file}")
print(f"Duplicate-hour details:  {duplicate_details_file}")
print(f"Plot:                    {plot_file}")

print("\nSEND ME THESE OUTPUTS:")
print("1. OVERALL AGGREGATION AUDIT")
print("2. 20 MONTHS WITH THE LARGEST AGGREGATION DIFFERENCES")
print("3. JANUARY 2024 CONSISTENCY CHECK")
```

```python
# ============================================================
# LEVEL 3: EXAMINE TWO HISTORICAL YEARS
# Stanford monthly mean temperature: 2023–2024
#
# Important:
# This retains all source rows, including repeated timestamps,
# to remain consistent with Joy and Khadijah's aggregation.
# ============================================================

from google.colab import drive
drive.mount("/content/drive")

from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display


# ============================================================
# 1. FILE SETTINGS
# ============================================================

PROJECT = Path("/content/drive/MyDrive/Stanford_Weather_12M")
DATA_FILE = PROJECT / "data" / "OpenWeather19802026_StanfordOnly.csv"

OUTPUT_DIR = PROJECT / "foundational_analysis" / "level_3_historical_year"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

CITY = "Stanford University"


# ============================================================
# 2. LOAD THE HOURLY SOURCE DATA
# ============================================================

df = pd.read_csv(
    DATA_FILE,
    usecols=["dt", "dt_iso", "timezone", "city_name", "temp"]
)

df["temp"] = pd.to_numeric(df["temp"], errors="coerce")
df["timezone"] = pd.to_numeric(df["timezone"], errors="coerce")

df = df.loc[df["city_name"] == CITY].copy()


# ============================================================
# 3. CONVERT UTC TIMESTAMP TO STANFORD LOCAL TIME
#
# This follows the timestamp approach in the existing work:
# Unix time -> UTC -> add the supplied timezone offset.
# ============================================================

df["dt_utc"] = pd.to_datetime(
    df["dt"],
    unit="s",
    utc=True,
    errors="coerce"
)

df["timezone_offset"] = pd.to_timedelta(
    df["timezone"],
    unit="s"
)

df["dt_local"] = (
    df["dt_utc"] + df["timezone_offset"]
).dt.tz_localize(None)

df = df.loc[df["dt_local"].notna()].copy()

df["month"] = (
    df["dt_local"]
    .dt.to_period("M")
    .dt.to_timestamp()
)


# ============================================================
# 4. CREATE MONTHLY MEAN TEMPERATURE
#
# Joy/Khadijah pattern:
# Average all source rows within each month.
# No duplicate timestamps are removed or averaged first.
# ============================================================

monthly = (
    df.groupby("month")
      .agg(
          mean_temp_F=("temp", "mean"),
          source_rows=("temp", "size"),
          valid_temp_rows=("temp", "count"),
          missing_temp_rows=("temp", lambda x: x.isna().sum()),
          minimum_temp_F=("temp", "min"),
          maximum_temp_F=("temp", "max")
      )
      .sort_index()
)

# This is the temperature series used for plotting and modeling.
series = monthly["mean_temp_F"].copy()


# ============================================================
# 5. SELECT TWO COMPLETE HISTORICAL YEARS
# ============================================================

two_years = monthly.loc[
    "2023-01-01":"2024-12-01"
].copy()

expected_months = pd.date_range(
    "2023-01-01",
    "2024-12-01",
    freq="MS"
)

missing_months = expected_months.difference(two_years.index)

print("TWO-YEAR DATA CHECK")
print("===================")
print(f"City:                 {CITY}")
print(f"Aggregation:          Mean of all source rows")
print(f"First month:          {two_years.index.min():%Y-%m}")
print(f"Last month:           {two_years.index.max():%Y-%m}")
print(f"Expected months:      {len(expected_months)}")
print(f"Months found:         {len(two_years)}")
print(f"Missing observations: {len(missing_months)}")

if len(missing_months) > 0:
    print(
        "Missing months:",
        [month.strftime("%Y-%m") for month in missing_months]
    )
else:
    print("All 24 monthly observations are present.")


# ============================================================
# 6. DISPLAY THE MONTHLY DATA
# ============================================================

display_table = two_years.reset_index()

display_table["year"] = display_table["month"].dt.year
display_table["month_name"] = display_table["month"].dt.month_name()

display_table = display_table[
    [
        "month",
        "year",
        "month_name",
        "mean_temp_F",
        "minimum_temp_F",
        "maximum_temp_F",
        "source_rows",
        "valid_temp_rows",
        "missing_temp_rows"
    ]
]

print("\nMONTHLY DATA: 2023–2024")
print("=======================")

display(
    display_table.round(
        {
            "mean_temp_F": 3,
            "minimum_temp_F": 2,
            "maximum_temp_F": 2
        }
    )
)


# ============================================================
# 7. PLOT THE CONTINUOUS TWO-YEAR TEMPERATURE CYCLE
# ============================================================

fig, ax = plt.subplots(figsize=(14, 5))

ax.plot(
    two_years.index,
    two_years["mean_temp_F"],
    marker="o",
    markersize=6,
    linewidth=2,
    color="darkred"
)

# Mark the boundary between the two years.
ax.axvline(
    pd.Timestamp("2024-01-01"),
    color="gray",
    linestyle="--",
    alpha=0.8,
    label="Start of 2024"
)

# Label each monthly point.
for month, value in two_years["mean_temp_F"].items():
    ax.annotate(
        f"{value:.1f}",
        (month, value),
        xytext=(0, 7),
        textcoords="offset points",
        ha="center",
        fontsize=8
    )

ax.set_title(
    "Stanford Monthly Mean Temperature: jan 12023–dec 2024\n"
    "All source rows retained"
)
ax.set_xlabel("Month")
ax.set_ylabel("Monthly mean temperature (°F)")
ax.grid(alpha=0.25)
ax.legend()

plt.tight_layout()

continuous_plot_file = (
    OUTPUT_DIR / "stanford_monthly_temperature_2023_2024.png"
)

plt.savefig(
    continuous_plot_file,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 8. COMPARE THE SAME CALENDAR MONTH ACROSS THE TWO YEARS
# ============================================================

comparison = (
    display_table
    .pivot(
        index="month_name",
        columns="year",
        values="mean_temp_F"
    )
)

calendar_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

comparison = comparison.reindex(calendar_order)

comparison["2024_minus_2023_F"] = (
    comparison[2024] - comparison[2023]
)

comparison["absolute_difference_F"] = (
    comparison["2024_minus_2023_F"].abs()
)

print("\nSAME-MONTH COMPARISON")
print("=====================")
print(
    "Positive difference: 2024 was warmer than 2023.\n"
    "Negative difference: 2024 was cooler than 2023."
)

display(comparison.round(3))


# ============================================================
# 9. PLOT 2023 AND 2024 ON THE SAME CALENDAR-MONTH AXIS
# ============================================================

fig, ax = plt.subplots(figsize=(13, 5))

month_numbers = np.arange(1, 13)

ax.plot(
    month_numbers,
    comparison[2023],
    marker="o",
    linewidth=2,
    label="2023",
    color="steelblue"
)

ax.plot(
    month_numbers,
    comparison[2024],
    marker="o",
    linewidth=2,
    label="2024",
    color="darkorange"
)

ax.set_xticks(month_numbers)
ax.set_xticklabels(
    [month[:3] for month in calendar_order]
)

ax.set_title(
    "Does the Annual Temperature Cycle Repeat?\n"
    "Comparison of the Same Calendar Months in 2023 and 2024"
)
ax.set_xlabel("Calendar month")
ax.set_ylabel("Monthly mean temperature (°F)")
ax.grid(alpha=0.25)
ax.legend()

plt.tight_layout()

comparison_plot_file = (
    OUTPUT_DIR / "same_month_comparison_2023_vs_2024.png"
)

plt.savefig(
    comparison_plot_file,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 10. IDENTIFY THE LARGEST YEAR-TO-YEAR DIFFERENCES
#
# These are not automatically data errors. They are months that
# deserve visual inspection because their temperatures differed
# substantially between the two years.
# ============================================================

largest_differences = (
    comparison
    .sort_values(
        "absolute_difference_F",
        ascending=False
    )
    .reset_index()
)

print("\nMONTHS WITH THE LARGEST 2023–2024 DIFFERENCES")
print("=============================================")
display(largest_differences.head(5).round(3))


# ============================================================
# 11. SIMPLE SEASONALITY SUMMARY
# ============================================================

coldest_month = two_years["mean_temp_F"].idxmin()
warmest_month = two_years["mean_temp_F"].idxmax()

correlation_2023_2024 = comparison[[2023, 2024]].corr().iloc[0, 1]

seasonality_summary = pd.DataFrame({
    "check": [
        "Coldest observed month",
        "Coldest monthly mean (°F)",
        "Warmest observed month",
        "Warmest monthly mean (°F)",
        "Correlation between 2023 and 2024 monthly patterns",
        "Largest same-month difference (°F)",
        "Month with largest same-month difference",
        "Missing monthly observations"
    ],
    "result": [
        coldest_month.strftime("%Y-%m"),
        round(two_years.loc[coldest_month, "mean_temp_F"], 3),
        warmest_month.strftime("%Y-%m"),
        round(two_years.loc[warmest_month, "mean_temp_F"], 3),
        round(correlation_2023_2024, 3),
        round(comparison["absolute_difference_F"].max(), 3),
        comparison["absolute_difference_F"].idxmax(),
        len(missing_months)
    ]
})

print("\nSEASONALITY SUMMARY")
print("===================")
display(seasonality_summary)


# ============================================================
# 12. SAVE THE RESULTS
# ============================================================

monthly_file = OUTPUT_DIR / "monthly_temperature_2023_2024.csv"
comparison_file = OUTPUT_DIR / "same_month_comparison_2023_2024.csv"
summary_file = OUTPUT_DIR / "seasonality_summary_2023_2024.csv"

display_table.to_csv(monthly_file, index=False)
comparison.reset_index().to_csv(comparison_file, index=False)
seasonality_summary.to_csv(summary_file, index=False)

print("\nFILES SAVED")
print("===========")
print(f"Monthly data:       {monthly_file}")
print(f"Year comparison:    {comparison_file}")
print(f"Seasonality summary:{summary_file}")
print(f"Continuous plot:    {continuous_plot_file}")
print(f"Comparison plot:    {comparison_plot_file}")

print("\nSEND ME:")
print("1. SAME-MONTH COMPARISON")
print("2. MONTHS WITH THE LARGEST 2023–2024 DIFFERENCES")
print("3. SEASONALITY SUMMARY")
print("4. The two plots, if anything looks unusual")
```

```python
# ============================================================
# DOES THE ANNUAL TEMPERATURE CYCLE REPEAT?
# Compare 10 complete years: 2015–2024
#
# Uses all source rows, including repeated timestamps,
# consistent with Joy and Khadijah's aggregation.
# ============================================================

from google.colab import drive
drive.mount("/content/drive")

from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display


# ============================================================
# 1. SETTINGS
# ============================================================

PROJECT = Path("/content/drive/MyDrive/Stanford_Weather_12M")
DATA_FILE = PROJECT / "data" / "OpenWeather19802026_StanfordOnly.csv"

OUTPUT_DIR = (
    PROJECT
    / "foundational_analysis"
    / "annual_cycle_2015_2024"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

CITY = "Stanford University"
START_YEAR = 2015
END_YEAR = 2024


# ============================================================
# 2. LOAD HOURLY TEMPERATURE DATA
# ============================================================

df = pd.read_csv(
    DATA_FILE,
    usecols=["dt", "timezone", "city_name", "temp"]
)

df["temp"] = pd.to_numeric(df["temp"], errors="coerce")
df["timezone"] = pd.to_numeric(
    df["timezone"],
    errors="coerce"
)

df = df.loc[df["city_name"] == CITY].copy()


# ============================================================
# 3. CREATE STANFORD LOCAL TIME
# ============================================================

df["dt_utc"] = pd.to_datetime(
    df["dt"],
    unit="s",
    utc=True,
    errors="coerce"
)

df["timezone_offset"] = pd.to_timedelta(
    df["timezone"],
    unit="s"
)

df["dt_local"] = (
    df["dt_utc"] + df["timezone_offset"]
).dt.tz_localize(None)

df = df.loc[
    df["dt_local"].notna()
    & df["temp"].notna()
].copy()

df["month"] = (
    df["dt_local"]
    .dt.to_period("M")
    .dt.to_timestamp()
)


# ============================================================
# 4. CREATE MONTHLY MEAN TEMPERATURE
#
# This deliberately averages all rows in each month,
# following Joy and Khadijah.
# ============================================================

monthly = (
    df.groupby("month")
      .agg(
          mean_temp_F=("temp", "mean"),
          source_rows=("temp", "size")
      )
      .sort_index()
)

monthly_10y = monthly.loc[
    f"{START_YEAR}-01-01":
    f"{END_YEAR}-12-01"
].copy()

monthly_10y["year"] = monthly_10y.index.year
monthly_10y["month_number"] = monthly_10y.index.month
monthly_10y["month_name"] = monthly_10y.index.month_name()


# ============================================================
# 5. CHECK THAT ALL 120 MONTHS ARE PRESENT
# ============================================================

expected_months = pd.date_range(
    f"{START_YEAR}-01-01",
    f"{END_YEAR}-12-01",
    freq="MS"
)

missing_months = expected_months.difference(
    monthly_10y.index
)

print("TEN-YEAR DATA CHECK")
print("===================")
print(f"City:             {CITY}")
print(f"Period:           {START_YEAR}-01 through {END_YEAR}-12")
print(f"Expected months:  {len(expected_months)}")
print(f"Months found:     {len(monthly_10y)}")
print(f"Missing months:   {len(missing_months)}")

if len(missing_months) > 0:
    print(
        "Missing:",
        [month.strftime("%Y-%m") for month in missing_months]
    )
else:
    print("All 120 monthly observations are present.")


# ============================================================
# 6. CREATE YEAR × CALENDAR-MONTH TABLE
# ============================================================

annual_cycles = (
    monthly_10y
    .pivot(
        index="month_number",
        columns="year",
        values="mean_temp_F"
    )
    .reindex(range(1, 13))
)

month_names = [
    "Jan", "Feb", "Mar", "Apr",
    "May", "Jun", "Jul", "Aug",
    "Sep", "Oct", "Nov", "Dec"
]

annual_cycles.index = month_names
annual_cycles.index.name = "calendar_month"

print("\nMONTHLY TEMPERATURE BY YEAR")
print("===========================")
display(annual_cycles.round(2))


# ============================================================
# 7. PLOT ALL TEN ANNUAL CYCLES
# ============================================================

fig, ax = plt.subplots(figsize=(14, 7))

colors = plt.cm.viridis(
    np.linspace(0.05, 0.95, END_YEAR - START_YEAR + 1)
)

for color, year in zip(
    colors,
    range(START_YEAR, END_YEAR + 1)
):
    ax.plot(
        range(1, 13),
        annual_cycles[year],
        marker="o",
        linewidth=1.7,
        alpha=0.85,
        color=color,
        label=str(year)
    )

ax.set_xticks(range(1, 13))
ax.set_xticklabels(month_names)

ax.set_title(
    "Does Stanford's Annual Temperature Cycle Repeat?\n"
    f"Monthly Mean Temperature, {START_YEAR}–{END_YEAR}"
)
ax.set_xlabel("Calendar month")
ax.set_ylabel("Monthly mean temperature (°F)")
ax.grid(alpha=0.25)

ax.legend(
    title="Year",
    ncol=2,
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.tight_layout()

cycles_plot_file = (
    OUTPUT_DIR / "annual_temperature_cycles_2015_2024.png"
)

plt.savefig(
    cycles_plot_file,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 8. CALCULATE THE 10-YEAR AVERAGE SEASONAL CYCLE
# ============================================================

seasonal_summary = pd.DataFrame({
    "calendar_month": month_names,
    "ten_year_mean_F": annual_cycles.mean(axis=1).values,
    "minimum_year_value_F": annual_cycles.min(axis=1).values,
    "maximum_year_value_F": annual_cycles.max(axis=1).values,
    "standard_deviation_F": annual_cycles.std(axis=1).values,
    "range_across_years_F": (
        annual_cycles.max(axis=1)
        - annual_cycles.min(axis=1)
    ).values
})

print("\nTEN-YEAR CALENDAR-MONTH SUMMARY")
print("===============================")
display(seasonal_summary.round(3))


# ============================================================
# 9. PLOT THE AVERAGE CYCLE WITH OBSERVED YEAR-TO-YEAR RANGE
# ============================================================

fig, ax = plt.subplots(figsize=(13, 6))

x = np.arange(1, 13)

ax.fill_between(
    x,
    seasonal_summary["minimum_year_value_F"],
    seasonal_summary["maximum_year_value_F"],
    color="lightsteelblue",
    alpha=0.45,
    label="Observed minimum–maximum, 2015–2024"
)

ax.plot(
    x,
    seasonal_summary["ten_year_mean_F"],
    marker="o",
    markersize=7,
    linewidth=3,
    color="darkblue",
    label="10-year calendar-month average"
)

ax.set_xticks(x)
ax.set_xticklabels(month_names)

ax.set_title(
    "Stanford's Typical Annual Temperature Cycle\n"
    "Average and Year-to-Year Range, 2015–2024"
)
ax.set_xlabel("Calendar month")
ax.set_ylabel("Monthly mean temperature (°F)")
ax.grid(alpha=0.25)
ax.legend()

plt.tight_layout()

average_plot_file = (
    OUTPUT_DIR / "average_annual_cycle_with_range.png"
)

plt.savefig(
    average_plot_file,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 10. CORRELATION BETWEEN ANNUAL SHAPES
#
# High positive correlations mean the winter-to-summer-to-winter
# shape is similar across years, even if one year is warmer.
# ============================================================

year_correlations = annual_cycles.corr()

pairwise_values = []

for first_year in year_correlations.columns:
    for second_year in year_correlations.columns:
        if first_year < second_year:
            pairwise_values.append({
                "first_year": first_year,
                "second_year": second_year,
                "cycle_correlation":
                    year_correlations.loc[first_year, second_year]
            })

pairwise_correlations = pd.DataFrame(pairwise_values)

print("\nYEAR-TO-YEAR ANNUAL-CYCLE CORRELATIONS")
print("======================================")
display(year_correlations.round(3))

print("\nPAIRWISE CORRELATION SUMMARY")
print("============================")
print(
    f"Mean correlation:    "
    f"{pairwise_correlations['cycle_correlation'].mean():.3f}"
)
print(
    f"Median correlation:  "
    f"{pairwise_correlations['cycle_correlation'].median():.3f}"
)
print(
    f"Lowest correlation:  "
    f"{pairwise_correlations['cycle_correlation'].min():.3f}"
)
print(
    f"Highest correlation: "
    f"{pairwise_correlations['cycle_correlation'].max():.3f}"
)

lowest_pair = pairwise_correlations.loc[
    pairwise_correlations["cycle_correlation"].idxmin()
]

print(
    "Least similar pair: "
    f"{int(lowest_pair['first_year'])} and "
    f"{int(lowest_pair['second_year'])}"
)


# ============================================================
# 11. FIND THE MOST UNUSUAL MONTH-YEAR OBSERVATIONS
#
# Deviation means:
# observed temperature minus that calendar month's 10-year mean.
# ============================================================

long_form = (
    annual_cycles
    .reset_index()
    .melt(
        id_vars="calendar_month",
        var_name="year",
        value_name="mean_temp_F"
    )
)

month_average_map = (
    seasonal_summary
    .set_index("calendar_month")["ten_year_mean_F"]
)

long_form["calendar_month_average_F"] = (
    long_form["calendar_month"]
    .map(month_average_map)
)

long_form["deviation_from_month_average_F"] = (
    long_form["mean_temp_F"]
    - long_form["calendar_month_average_F"]
)

long_form["absolute_deviation_F"] = (
    long_form["deviation_from_month_average_F"].abs()
)

unusual_months = (
    long_form
    .sort_values(
        "absolute_deviation_F",
        ascending=False
    )
    .reset_index(drop=True)
)

print("\n10 MOST UNUSUAL MONTH-YEAR OBSERVATIONS")
print("=======================================")
display(unusual_months.head(10).round(3))


# ============================================================
# 12. BASIC CONCLUSION METRICS
# ============================================================

average_monthly_std = (
    seasonal_summary["standard_deviation_F"].mean()
)

largest_monthly_range_row = seasonal_summary.loc[
    seasonal_summary["range_across_years_F"].idxmax()
]

overall_summary = pd.DataFrame({
    "measure": [
        "Years examined",
        "Monthly observations",
        "Missing monthly observations",
        "Mean pairwise annual-cycle correlation",
        "Median pairwise annual-cycle correlation",
        "Average calendar-month standard deviation (°F)",
        "Calendar month with largest year-to-year range",
        "Largest year-to-year range (°F)"
    ],
    "value": [
        END_YEAR - START_YEAR + 1,
        len(monthly_10y),
        len(missing_months),
        pairwise_correlations["cycle_correlation"].mean(),
        pairwise_correlations["cycle_correlation"].median(),
        average_monthly_std,
        largest_monthly_range_row["calendar_month"],
        largest_monthly_range_row["range_across_years_F"]
    ]
})

print("\nANNUAL-CYCLE SUMMARY")
print("====================")
display(overall_summary)


# ============================================================
# 13. SAVE RESULTS
# ============================================================

annual_cycles_file = OUTPUT_DIR / "annual_cycles_2015_2024.csv"
seasonal_summary_file = OUTPUT_DIR / "calendar_month_summary.csv"
correlation_file = OUTPUT_DIR / "annual_cycle_correlations.csv"
unusual_months_file = OUTPUT_DIR / "unusual_months.csv"
overall_summary_file = OUTPUT_DIR / "annual_cycle_summary.csv"

annual_cycles.to_csv(annual_cycles_file)
seasonal_summary.to_csv(
    seasonal_summary_file,
    index=False
)
year_correlations.to_csv(correlation_file)
unusual_months.to_csv(
    unusual_months_file,
    index=False
)
overall_summary.to_csv(
    overall_summary_file,
    index=False
)

print("\nFILES SAVED")
print("===========")
print(f"Annual cycles:       {annual_cycles_file}")
print(f"Monthly summary:     {seasonal_summary_file}")
print(f"Year correlations:   {correlation_file}")
print(f"Unusual months:      {unusual_months_file}")
print(f"Overall summary:     {overall_summary_file}")
print(f"Annual cycles plot:  {cycles_plot_file}")
print(f"Average cycle plot:  {average_plot_file}")

print("\nSEND ME:")
print("1. TEN-YEAR CALENDAR-MONTH SUMMARY")
print("2. PAIRWISE CORRELATION SUMMARY")
print("3. 10 MOST UNUSUAL MONTH-YEAR OBSERVATIONS")
print("4. ANNUAL-CYCLE SUMMARY")
print("5. Both plots")
```

```python
# ============================================================
# LEVEL 4: ONE SARIMA MODEL, ONE HISTORICAL FORECAST WINDOW
#
# Train through: June 2023
# Forecast:      July 2023 through June 2024
# Target:        Stanford monthly mean temperature
#
# All source rows are retained, consistent with Joy/Khadijah.
# ============================================================

from google.colab import drive
drive.mount("/content/drive")

from pathlib import Path
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display
from statsmodels.tsa.statespace.sarimax import SARIMAX


# ------------------------------------------------------------
# 1. File locations
# ------------------------------------------------------------

PROJECT = Path("/content/drive/MyDrive/Stanford_Weather_12M")
DATA_FILE = PROJECT / "data" / "OpenWeather19802026_StanfordOnly.csv"

OUTPUT_DIR = (
    PROJECT
    / "foundational_analysis"
    / "one_sarima_window"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)


# ------------------------------------------------------------
# 2. Load Stanford hourly temperature data
# ------------------------------------------------------------

df = pd.read_csv(
    DATA_FILE,
    usecols=["dt", "timezone", "city_name", "temp"]
)

df = df.loc[
    df["city_name"] == "Stanford University"
].copy()

df["temp"] = pd.to_numeric(df["temp"], errors="coerce")
df["timezone"] = pd.to_numeric(df["timezone"], errors="coerce")


# ------------------------------------------------------------
# 3. Convert UTC timestamps to Stanford local time
# ------------------------------------------------------------

df["dt_utc"] = pd.to_datetime(
    df["dt"],
    unit="s",
    utc=True,
    errors="coerce"
)

df["timezone_offset"] = pd.to_timedelta(
    df["timezone"],
    unit="s"
)

df["dt_local"] = (
    df["dt_utc"] + df["timezone_offset"]
).dt.tz_localize(None)

df = df.dropna(subset=["dt_local", "temp"])

df["month"] = (
    df["dt_local"]
    .dt.to_period("M")
    .dt.to_timestamp()
)


# ------------------------------------------------------------
# 4. Convert hourly rows to monthly mean temperature
#
# This matches the existing Joy/Khadijah approach:
# average every source row without removing duplicates.
# ------------------------------------------------------------

series = (
    df.groupby("month")["temp"]
      .mean()
      .sort_index()
      .rename("mean_temp_F")
      .loc["1980-01-01":"2026-06-01"]
      .asfreq("MS")
)

print("MONTHLY SERIES")
print("==============")
print(f"First month: {series.index.min():%Y-%m}")
print(f"Last month:  {series.index.max():%Y-%m}")
print(f"Months:      {len(series)}")
print(f"Missing:     {series.isna().sum()}")

if series.isna().any():
    raise ValueError("The monthly temperature series contains missing values.")


# ------------------------------------------------------------
# 5. Define one historical train/test window
# ------------------------------------------------------------

train_end = pd.Timestamp("2023-06-01")

test_months = pd.date_range(
    start="2023-07-01",
    periods=12,
    freq="MS"
)

train = series.loc[:train_end]
actual = series.reindex(test_months)

print("\nONE HISTORICAL FORECAST WINDOW")
print("==============================")
print(f"Training: {train.index.min():%Y-%m} through "
      f"{train.index.max():%Y-%m}")
print(f"Testing:  {test_months.min():%Y-%m} through "
      f"{test_months.max():%Y-%m}")
print(f"Training months: {len(train)}")
print(f"Testing months:  {len(actual)}")


# ------------------------------------------------------------
# 6. Fit SARIMA and forecast 12 months
#
# This is the same SARIMA structure used in the earlier work:
# SARIMA(1,1,1)(1,1,1,12)
# ------------------------------------------------------------

with warnings.catch_warnings():
    warnings.simplefilter("ignore")

    model = SARIMAX(
        train,
        order=(1, 1, 1),
        seasonal_order=(1, 1, 1, 12),
        enforce_stationarity=False,
        enforce_invertibility=False
    )

    fitted_model = model.fit(disp=False)

forecast = fitted_model.get_forecast(
    steps=12
).predicted_mean

forecast.index = test_months

print("\nMODEL FIT")
print("=========")
print("Model:     SARIMA(1,1,1)(1,1,1,12)")
print(f"Converged: {fitted_model.mle_retvals.get('converged')}")
print(f"AIC:       {fitted_model.aic:.3f}")
print(f"BIC:       {fitted_model.bic:.3f}")


# ------------------------------------------------------------
# 7. Compare the forecast with actual temperatures
#
# Error = forecast - actual
# Positive error: forecast was too warm
# Negative error: forecast was too cold
# ------------------------------------------------------------

one_window = pd.DataFrame({
    "actual_F": actual,
    "forecast_F": forecast
})

one_window.index.name = "month"

one_window["error_F"] = (
    one_window["forecast_F"]
    - one_window["actual_F"]
)

one_window["absolute_error_F"] = (
    one_window["error_F"].abs()
)

one_window["forecast_horizon"] = np.arange(1, 13)

one_window["direction"] = np.select(
    [
        one_window["error_F"] > 0,
        one_window["error_F"] < 0
    ],
    [
        "Too warm",
        "Too cold"
    ],
    default="Exact"
)

print("\nMONTH-BY-MONTH RESULTS")
print("======================")
print("Error = forecast − actual")

display(
    one_window[
        [
            "forecast_horizon",
            "actual_F",
            "forecast_F",
            "error_F",
            "absolute_error_F",
            "direction"
        ]
    ].round(3)
)


# ------------------------------------------------------------
# 8. Calculate overall error metrics
# ------------------------------------------------------------

rmse = np.sqrt(
    np.mean(one_window["error_F"] ** 2)
)

mae = np.mean(
    np.abs(one_window["error_F"])
)

bias = np.mean(
    one_window["error_F"]
)

largest_error_month = (
    one_window["absolute_error_F"].idxmax()
)

best_month = (
    one_window["absolute_error_F"].idxmin()
)

print("\nERROR METRICS")
print("=============")
print(f"RMSE: {rmse:.3f} °F")
print(f"MAE:  {mae:.3f} °F")
print(f"Bias: {bias:+.3f} °F")

print(
    f"\nBest-forecast month: "
    f"{best_month:%Y-%m}, "
    f"error={one_window.loc[best_month, 'error_F']:+.3f} °F"
)

print(
    f"Largest-error month: "
    f"{largest_error_month:%Y-%m}, "
    f"error={one_window.loc[largest_error_month, 'error_F']:+.3f} °F"
)

if bias > 0:
    print(
        f"The model was too warm by approximately "
        f"{bias:.3f} °F on average."
    )
elif bias < 0:
    print(
        f"The model was too cold by approximately "
        f"{abs(bias):.3f} °F on average."
    )
else:
    print("The model had no average warm/cold bias.")


# ------------------------------------------------------------
# 9. Check whether errors increase farther into the forecast
# ------------------------------------------------------------

first_half_mae = (
    one_window.iloc[:6]["absolute_error_F"].mean()
)

second_half_mae = (
    one_window.iloc[6:]["absolute_error_F"].mean()
)

horizon_error_correlation = (
    one_window["forecast_horizon"]
    .corr(one_window["absolute_error_F"])
)

print("\nFORECAST-HORIZON CHECK")
print("======================")
print(f"MAE, horizons 1–6:  {first_half_mae:.3f} °F")
print(f"MAE, horizons 7–12: {second_half_mae:.3f} °F")
print(
    "Correlation between forecast horizon and absolute error: "
    f"{horizon_error_correlation:+.3f}"
)

if second_half_mae > first_half_mae:
    print(
        "For this window, the later six months were less accurate."
    )
else:
    print(
        "For this window, errors did not increase in the later "
        "six months."
    )


# ------------------------------------------------------------
# 10. Check whether SARIMA reproduced the seasonal shape
#
# This correlation compares the rise and fall of the actual
# and forecast temperature curves.
# ------------------------------------------------------------

seasonal_shape_correlation = (
    one_window["actual_F"]
    .corr(one_window["forecast_F"])
)

print("\nSEASONAL-PATTERN CHECK")
print("======================")
print(
    "Actual-versus-forecast correlation: "
    f"{seasonal_shape_correlation:.3f}"
)

if seasonal_shape_correlation >= 0.90:
    print(
        "The forecast reproduced the seasonal pattern very well."
    )
elif seasonal_shape_correlation >= 0.70:
    print(
        "The forecast reproduced the broad seasonal pattern."
    )
else:
    print(
        "The forecast did not reproduce the seasonal pattern strongly."
    )


# ------------------------------------------------------------
# 11. Plot actual versus forecast
# ------------------------------------------------------------

fig, ax = plt.subplots(figsize=(13, 5))

ax.plot(
    one_window.index,
    one_window["actual_F"],
    marker="o",
    linewidth=2.5,
    color="black",
    label="Actual"
)

ax.plot(
    one_window.index,
    one_window["forecast_F"],
    marker="o",
    linewidth=2.5,
    linestyle="--",
    color="darkred",
    label="SARIMA forecast"
)

ax.set_title(
    "One SARIMA Forecast Window\n"
    "Train Through June 2023; Forecast July 2023–June 2024"
)

ax.set_xlabel("Month")
ax.set_ylabel("Monthly mean temperature (°F)")
ax.grid(alpha=0.25)
ax.legend()

plt.tight_layout()

forecast_plot = (
    OUTPUT_DIR
    / "actual_vs_sarima_Jul2023_Jun2024.png"
)

plt.savefig(
    forecast_plot,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 12. Plot the monthly forecast errors
# ------------------------------------------------------------

fig, ax = plt.subplots(figsize=(13, 4))

colors = np.where(
    one_window["error_F"] >= 0,
    "darkorange",
    "steelblue"
)

ax.bar(
    one_window.index,
    one_window["error_F"],
    color=colors,
    width=20
)

ax.axhline(
    0,
    color="black",
    linewidth=1
)

ax.set_title(
    "SARIMA Forecast Errors by Month\n"
    "Orange = Too Warm; Blue = Too Cold"
)

ax.set_xlabel("Month")
ax.set_ylabel("Forecast error (°F)")
ax.grid(axis="y", alpha=0.25)

plt.tight_layout()

error_plot = (
    OUTPUT_DIR
    / "sarima_errors_Jul2023_Jun2024.png"
)

plt.savefig(
    error_plot,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 13. Save the results
# ------------------------------------------------------------

results_file = (
    OUTPUT_DIR
    / "one_window_results_Jul2023_Jun2024.csv"
)

one_window.reset_index().to_csv(
    results_file,
    index=False
)

summary = pd.DataFrame({
    "model": ["sarima_111x111_12"],
    "train_end": [train_end],
    "test_start": [test_months.min()],
    "test_end": [test_months.max()],
    "RMSE_F": [rmse],
    "MAE_F": [mae],
    "bias_F": [bias],
    "first_6_month_MAE_F": [first_half_mae],
    "last_6_month_MAE_F": [second_half_mae],
    "horizon_error_correlation": [horizon_error_correlation],
    "seasonal_shape_correlation": [seasonal_shape_correlation]
})

summary_file = (
    OUTPUT_DIR
    / "one_window_summary_Jul2023_Jun2024.csv"
)

summary.to_csv(summary_file, index=False)

print("\nFILES SAVED")
print("===========")
print(f"Monthly results: {results_file}")
print(f"Summary:         {summary_file}")
print(f"Forecast plot:   {forecast_plot}")
print(f"Error plot:      {error_plot}")

print("\nSEND ME:")
print("1. MONTH-BY-MONTH RESULTS")
print("2. ERROR METRICS")
print("3. FORECAST-HORIZON CHECK")
print("4. SEASONAL-PATTERN CHECK")
print("5. Both plots")
```

```python
# ============================================================
# LEVEL 5: COMPARE ALL MODELS IN ONE FORECAST WINDOW
#
# Train through: June 2023
# Forecast:      July 2023 through June 2024
#
# Models:
#   1. SARIMA(1,1,1)(1,1,1,12)
#   2. Prophet with additive yearly seasonality
#   3. Damped additive ETS
#   4. Seasonal naive baseline
#
# All models receive exactly the same training data and are
# evaluated on exactly the same 12 months.
#
# Monthly aggregation retains all source rows, consistent with
# Joy and Khadijah's original processing.
# ============================================================


# ------------------------------------------------------------
# 0. Install Prophet if necessary
# ------------------------------------------------------------

!pip install -q prophet


# ------------------------------------------------------------
# 1. Imports
# ------------------------------------------------------------

from google.colab import drive
drive.mount("/content/drive")

from pathlib import Path
import time
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display
from prophet import Prophet
from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.tsa.holtwinters import ExponentialSmoothing


# ------------------------------------------------------------
# 2. Settings
# ------------------------------------------------------------

PROJECT = Path("/content/drive/MyDrive/Stanford_Weather_12M")
DATA_FILE = PROJECT / "data" / "OpenWeather19802026_StanfordOnly.csv"

OUTPUT_DIR = (
    PROJECT
    / "foundational_analysis"
    / "level_5_single_window_model_comparison"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

CITY = "Stanford University"

TRAIN_END = pd.Timestamp("2023-06-01")
TEST_START = pd.Timestamp("2023-07-01")
FORECAST_MONTHS = 12


# ------------------------------------------------------------
# 3. Load hourly Stanford temperature
# ------------------------------------------------------------

df = pd.read_csv(
    DATA_FILE,
    usecols=["dt", "timezone", "city_name", "temp"]
)

df = df.loc[
    df["city_name"] == CITY
].copy()

df["temp"] = pd.to_numeric(
    df["temp"],
    errors="coerce"
)

df["timezone"] = pd.to_numeric(
    df["timezone"],
    errors="coerce"
)


# ------------------------------------------------------------
# 4. Convert UTC time to Stanford local time
# ------------------------------------------------------------

df["dt_utc"] = pd.to_datetime(
    df["dt"],
    unit="s",
    utc=True,
    errors="coerce"
)

df["timezone_offset"] = pd.to_timedelta(
    df["timezone"],
    unit="s"
)

df["dt_local"] = (
    df["dt_utc"] + df["timezone_offset"]
).dt.tz_localize(None)

df = df.dropna(
    subset=["dt_local", "temp"]
)

df["month"] = (
    df["dt_local"]
    .dt.to_period("M")
    .dt.to_timestamp()
)


# ------------------------------------------------------------
# 5. Create monthly mean temperature
#
# All source rows are included, matching Joy/Khadijah.
# ------------------------------------------------------------

series = (
    df.groupby("month")["temp"]
      .mean()
      .sort_index()
      .rename("mean_temp_F")
      .loc["1980-01-01":"2026-06-01"]
      .asfreq("MS")
)

if series.isna().any():
    raise ValueError(
        "The monthly series contains missing observations."
    )


# ------------------------------------------------------------
# 6. Create the common training and test window
# ------------------------------------------------------------

train = series.loc[:TRAIN_END].copy()

test_months = pd.date_range(
    start=TEST_START,
    periods=FORECAST_MONTHS,
    freq="MS"
)

actual = series.reindex(test_months)

if actual.isna().any():
    raise ValueError(
        "The historical test window contains missing observations."
    )

print("COMMON FORECAST WINDOW")
print("======================")
print(f"City:              {CITY}")
print("Aggregation:       Mean of all source rows")
print(
    f"Training:          {train.index.min():%Y-%m} "
    f"through {train.index.max():%Y-%m}"
)
print(
    f"Testing:           {test_months.min():%Y-%m} "
    f"through {test_months.max():%Y-%m}"
)
print(f"Training months:   {len(train)}")
print(f"Forecast months:   {len(test_months)}")
print(f"Missing train:     {train.isna().sum()}")
print(f"Missing test:      {actual.isna().sum()}")


# ============================================================
# 7. MODEL FUNCTIONS
# ============================================================


# ------------------------------------------------------------
# Model 1: SARIMA
# ------------------------------------------------------------

def fit_sarima(training_series, forecast_dates):
    """
    Fit SARIMA(1,1,1)(1,1,1,12) and return forecasts.
    """

    with warnings.catch_warnings():
        warnings.simplefilter("ignore")

        model = SARIMAX(
            training_series,
            order=(1, 1, 1),
            seasonal_order=(1, 1, 1, 12),
            enforce_stationarity=False,
            enforce_invertibility=False
        )

        fitted = model.fit(disp=False)

    prediction = fitted.get_forecast(
        steps=len(forecast_dates)
    ).predicted_mean

    prediction.index = forecast_dates

    return prediction.astype(float)


# ------------------------------------------------------------
# Model 2: Prophet
# ------------------------------------------------------------

def fit_prophet(training_series, forecast_dates):
    """
    Fit additive Prophet with yearly seasonality.

    Weekly and daily seasonality are disabled because each
    observation represents one month.
    """

    prophet_train = pd.DataFrame({
        "ds": training_series.index,
        "y": training_series.values
    })

    model = Prophet(
        growth="linear",
        seasonality_mode="additive",
        yearly_seasonality=True,
        weekly_seasonality=False,
        daily_seasonality=False,
        interval_width=0.95
    )

    model.fit(prophet_train)

    future = pd.DataFrame({
        "ds": forecast_dates
    })

    prophet_output = model.predict(future)

    prediction = pd.Series(
        prophet_output["yhat"].to_numpy(),
        index=forecast_dates,
        name="forecast_F"
    )

    return prediction.astype(float)


# ------------------------------------------------------------
# Model 3: ETS
# ------------------------------------------------------------

def fit_ets(training_series, forecast_dates):
    """
    Fit additive Exponential Smoothing with:
      - additive trend;
      - damped trend;
      - additive 12-month seasonality.
    """

    with warnings.catch_warnings():
        warnings.simplefilter("ignore")

        model = ExponentialSmoothing(
            training_series,
            trend="add",
            damped_trend=True,
            seasonal="add",
            seasonal_periods=12,
            initialization_method="estimated"
        )

        fitted = model.fit(
            optimized=True
        )

    prediction = fitted.forecast(
        len(forecast_dates)
    )

    prediction.index = forecast_dates

    return prediction.astype(float)


# ------------------------------------------------------------
# Model 4: Seasonal naive
# ------------------------------------------------------------

def seasonal_naive_forecast(
    training_series,
    forecast_dates
):
    """
    Forecast each month using the temperature observed
    12 months earlier.

    For example:
      July 2023 forecast = July 2022 actual temperature.
    """

    predictions = []

    for forecast_month in forecast_dates:
        previous_year_month = (
            forecast_month - pd.DateOffset(years=1)
        )

        predictions.append(
            training_series.loc[previous_year_month]
        )

    return pd.Series(
        predictions,
        index=forecast_dates,
        name="forecast_F",
        dtype=float
    )


# ============================================================
# 8. RUN ALL MODELS ON THE SAME WINDOW
# ============================================================

models = {
    "sarima_111x111_12": fit_sarima,
    "prophet_additive": fit_prophet,
    "ets_add_add_damped12": fit_ets,
    "seasonal_naive_12": seasonal_naive_forecast
}

predictions = pd.DataFrame(
    index=test_months
)

predictions.index.name = "month"
predictions["actual_F"] = actual

results = []

for model_name, model_function in models.items():

    print(f"\nRunning {model_name}...")

    start_time = time.perf_counter()

    try:
        prediction = model_function(
            train,
            test_months
        )

        runtime = time.perf_counter() - start_time

        prediction = prediction.reindex(test_months)

        if prediction.isna().any():
            raise ValueError(
                "The model returned missing forecasts."
            )

        predictions[model_name] = prediction

        error = prediction - actual

        rmse = np.sqrt(
            np.mean(error ** 2)
        )

        mae = np.mean(
            np.abs(error)
        )

        bias = np.mean(error)

        seasonal_shape_correlation = (
            actual.corr(prediction)
        )

        largest_error_month = (
            np.abs(error).idxmax()
        )

        results.append({
            "model": model_name,
            "status": "completed",
            "RMSE_F": rmse,
            "MAE_F": mae,
            "bias_F": bias,
            "seasonal_shape_correlation":
                seasonal_shape_correlation,
            "largest_error_month":
                largest_error_month.strftime("%Y-%m"),
            "largest_absolute_error_F":
                np.abs(error).max(),
            "runtime_seconds": runtime
        })

        print(
            f"  RMSE={rmse:.3f}, "
            f"MAE={mae:.3f}, "
            f"Bias={bias:+.3f}"
        )

    except Exception as error_message:
        runtime = time.perf_counter() - start_time

        results.append({
            "model": model_name,
            "status": f"failed: {error_message}",
            "RMSE_F": np.nan,
            "MAE_F": np.nan,
            "bias_F": np.nan,
            "seasonal_shape_correlation": np.nan,
            "largest_error_month": None,
            "largest_absolute_error_F": np.nan,
            "runtime_seconds": runtime
        })

        print(f"  FAILED: {error_message}")


# ============================================================
# 9. MODEL COMPARISON TABLE
# ============================================================

results = pd.DataFrame(results)

results = results.sort_values(
    "RMSE_F",
    ascending=True,
    na_position="last"
).reset_index(drop=True)

results["RMSE_rank"] = (
    results["RMSE_F"]
    .rank(method="min")
)

print("\nSINGLE-WINDOW MODEL COMPARISON")
print("==============================")
display(results.round(3))


# ============================================================
# 10. MONTH-BY-MONTH FORECAST TABLE
# ============================================================

monthly_comparison = predictions.copy()

for model_name in models:
    if model_name in monthly_comparison.columns:
        monthly_comparison[
            f"{model_name}_error_F"
        ] = (
            monthly_comparison[model_name]
            - monthly_comparison["actual_F"]
        )

print("\nMONTH-BY-MONTH FORECASTS")
print("========================")
print("Error columns use forecast − actual.")

display(monthly_comparison.round(3))


# ============================================================
# 11. IDENTIFY THE SINGLE-WINDOW WINNER
# ============================================================

completed_results = results.loc[
    results["status"] == "completed"
].copy()

if completed_results.empty:
    raise RuntimeError(
        "No model completed the forecast successfully."
    )

winner = completed_results.iloc[0]

print("\nSINGLE-WINDOW WINNER")
print("====================")
print(f"Model: {winner['model']}")
print(f"RMSE:  {winner['RMSE_F']:.3f} °F")
print(f"MAE:   {winner['MAE_F']:.3f} °F")
print(f"Bias:  {winner['bias_F']:+.3f} °F")

print(
    "\nImportant: this model won one historical window only. "
    "It is not yet the stable-model winner."
)


# ============================================================
# 12. PLOT ALL MODEL FORECASTS
# ============================================================

fig, ax = plt.subplots(figsize=(14, 6))

ax.plot(
    predictions.index,
    predictions["actual_F"],
    marker="o",
    markersize=8,
    linewidth=3,
    color="black",
    label="Actual"
)

plot_styles = {
    "sarima_111x111_12": {
        "color": "darkred",
        "linestyle": "--"
    },
    "prophet_additive": {
        "color": "darkgreen",
        "linestyle": "--"
    },
    "ets_add_add_damped12": {
        "color": "darkorange",
        "linestyle": "--"
    },
    "seasonal_naive_12": {
        "color": "steelblue",
        "linestyle": ":"
    }
}

for model_name, style in plot_styles.items():

    if model_name not in predictions.columns:
        continue

    ax.plot(
        predictions.index,
        predictions[model_name],
        marker="o",
        linewidth=2,
        color=style["color"],
        linestyle=style["linestyle"],
        label=model_name
    )

ax.set_title(
    "Single-Window Forecast Comparison\n"
    "Train Through June 2023; Forecast July 2023–June 2024"
)

ax.set_xlabel("Month")
ax.set_ylabel("Monthly mean temperature (°F)")
ax.grid(alpha=0.25)
ax.legend()

plt.tight_layout()

forecast_plot = (
    OUTPUT_DIR
    / "single_window_all_model_forecasts.png"
)

plt.savefig(
    forecast_plot,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 13. PLOT ABSOLUTE ERROR BY MONTH
# ============================================================

absolute_errors = pd.DataFrame(
    index=test_months
)

for model_name in models:
    if model_name in predictions.columns:
        absolute_errors[model_name] = (
            predictions[model_name]
            - predictions["actual_F"]
        ).abs()

fig, ax = plt.subplots(figsize=(14, 6))

absolute_errors.plot(
    marker="o",
    linewidth=2,
    ax=ax
)

ax.set_title(
    "Absolute Forecast Error by Month\n"
    "Lower Values Are Better"
)

ax.set_xlabel("Month")
ax.set_ylabel("Absolute error (°F)")
ax.grid(alpha=0.25)

plt.tight_layout()

error_plot = (
    OUTPUT_DIR
    / "single_window_absolute_errors.png"
)

plt.savefig(
    error_plot,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 14. SAVE RESULTS
# ============================================================

results_file = (
    OUTPUT_DIR
    / "single_window_model_comparison.csv"
)

predictions_file = (
    OUTPUT_DIR
    / "single_window_monthly_predictions.csv"
)

errors_file = (
    OUTPUT_DIR
    / "single_window_absolute_errors.csv"
)

results.to_csv(
    results_file,
    index=False
)

monthly_comparison.reset_index().to_csv(
    predictions_file,
    index=False
)

absolute_errors.reset_index().to_csv(
    errors_file,
    index=False
)

print("\nFILES SAVED")
print("===========")
print(f"Model comparison:    {results_file}")
print(f"Monthly predictions: {predictions_file}")
print(f"Absolute errors:     {errors_file}")
print(f"Forecast plot:       {forecast_plot}")
print(f"Error plot:          {error_plot}")

print("\nSEND ME:")
print("1. SINGLE-WINDOW MODEL COMPARISON")
print("2. SINGLE-WINDOW WINNER")
print("3. MONTH-BY-MONTH FORECASTS")
print("4. Both plots")
```

```python
# ============================================================
# LEVEL 6: RUN SARIMA ACROSS ALL NINE HISTORICAL WINDOWS
#
# Model:
#   SARIMA(1,1,1)(1,1,1,12)
#
# Forecast windows:
#   July 2017–June 2018
#   July 2018–June 2019
#   ...
#   July 2025–June 2026
#
# Purpose:
# Determine whether the same SARIMA model performs consistently
# across different historical periods.
#
# All source rows are retained, consistent with Joy/Khadijah.
# ============================================================


# ------------------------------------------------------------
# 1. Imports
# ------------------------------------------------------------

from google.colab import drive
drive.mount("/content/drive")

from pathlib import Path
import json
import time
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display
from statsmodels.tsa.statespace.sarimax import SARIMAX


# ------------------------------------------------------------
# 2. Settings
# ------------------------------------------------------------

PROJECT = Path("/content/drive/MyDrive/Stanford_Weather_12M")
DATA_FILE = PROJECT / "data" / "OpenWeather19802026_StanfordOnly.csv"

OUTPUT_DIR = (
    PROJECT
    / "foundational_analysis"
    / "level_6_sarima_all_windows"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

CITY = "Stanford University"
FORECAST_HORIZON = 12

FORECAST_ORIGINS = pd.to_datetime([
    "2017-07-01",
    "2018-07-01",
    "2019-07-01",
    "2020-07-01",
    "2021-07-01",
    "2022-07-01",
    "2023-07-01",
    "2024-07-01",
    "2025-07-01"
])

MODEL_NAME = "sarima_111x111_12"

MODEL_PARAMETERS = {
    "order": (1, 1, 1),
    "seasonal_order": (1, 1, 1, 12),
    "enforce_stationarity": False,
    "enforce_invertibility": False
}


# ------------------------------------------------------------
# 3. Load Stanford hourly temperature data
# ------------------------------------------------------------

df = pd.read_csv(
    DATA_FILE,
    usecols=["dt", "timezone", "city_name", "temp"]
)

df = df.loc[
    df["city_name"] == CITY
].copy()

df["temp"] = pd.to_numeric(
    df["temp"],
    errors="coerce"
)

df["timezone"] = pd.to_numeric(
    df["timezone"],
    errors="coerce"
)


# ------------------------------------------------------------
# 4. Convert UTC timestamps to Stanford local time
# ------------------------------------------------------------

df["dt_utc"] = pd.to_datetime(
    df["dt"],
    unit="s",
    utc=True,
    errors="coerce"
)

df["timezone_offset"] = pd.to_timedelta(
    df["timezone"],
    unit="s"
)

df["dt_local"] = (
    df["dt_utc"] + df["timezone_offset"]
).dt.tz_localize(None)

df = df.dropna(
    subset=["dt_local", "temp"]
)

df["month"] = (
    df["dt_local"]
    .dt.to_period("M")
    .dt.to_timestamp()
)


# ------------------------------------------------------------
# 5. Create the monthly mean temperature series
#
# This follows Joy and Khadijah:
# all source rows are included in each monthly average.
# ------------------------------------------------------------

series = (
    df.groupby("month")["temp"]
      .mean()
      .sort_index()
      .rename("mean_temp_F")
      .loc["1980-01-01":"2026-06-01"]
      .asfreq("MS")
)

print("MONTHLY DATA")
print("============")
print(f"City:        {CITY}")
print("Aggregation: Mean of all source rows")
print(f"First month: {series.index.min():%Y-%m}")
print(f"Last month:  {series.index.max():%Y-%m}")
print(f"Months:      {len(series)}")
print(f"Missing:     {series.isna().sum()}")

if series.isna().any():
    raise ValueError(
        "The monthly series contains missing observations."
    )


# ------------------------------------------------------------
# 6. Define the SARIMA forecast function
#
# Returns:
#   - forecast
#   - convergence status
#   - fitting warnings
#   - model information
# ------------------------------------------------------------

def fit_sarima(training_series, forecast_dates):
    """
    Fit SARIMA(1,1,1)(1,1,1,12) using only the supplied
    training series and forecast the requested future months.
    """

    start_time = time.perf_counter()

    with warnings.catch_warnings(record=True) as caught_warnings:
        warnings.simplefilter("always")

        model = SARIMAX(
            training_series,
            **MODEL_PARAMETERS
        )

        fitted_model = model.fit(
            disp=False
        )

    runtime_seconds = (
        time.perf_counter() - start_time
    )

    forecast = (
        fitted_model
        .get_forecast(steps=len(forecast_dates))
        .predicted_mean
    )

    forecast.index = forecast_dates
    forecast.name = "forecast_F"

    converged = bool(
        fitted_model
        .mle_retvals
        .get("converged", False)
    )

    warning_messages = [
        str(warning.message)
        for warning in caught_warnings
    ]

    model_information = {
        "converged": converged,
        "warnings": warning_messages,
        "warning_count": len(warning_messages),
        "runtime_seconds": runtime_seconds,
        "AIC": fitted_model.aic,
        "BIC": fitted_model.bic
    }

    return forecast.astype(float), model_information


# ------------------------------------------------------------
# 7. Run the same model across all nine forecast windows
# ------------------------------------------------------------

window_results = []
monthly_results = []

print("\nRUNNING NINE 12-MONTH SARIMA BACKTESTS")
print("======================================")

for window_number, origin in enumerate(
    FORECAST_ORIGINS,
    start=1
):

    test_months = pd.date_range(
        start=origin,
        periods=FORECAST_HORIZON,
        freq="MS"
    )

    train_end = (
        origin - pd.offsets.MonthBegin(1)
    )

    train = series.loc[:train_end].copy()
    actual = series.reindex(test_months)

    test_end = test_months[-1]

    print(
        f"\nWindow {window_number}: "
        f"{origin:%Y-%m} through {test_end:%Y-%m}"
    )
    print(f"Training through: {train_end:%Y-%m}")

    # Confirm that the window is valid before fitting.
    if train.empty:
        print("FAILED: Training data are empty.")

        window_results.append({
            "window_number": window_number,
            "test_start": origin,
            "test_end": test_end,
            "train_end": train_end,
            "training_months": 0,
            "model": MODEL_NAME,
            "status": "failed",
            "converged": False,
            "warning_count": np.nan,
            "warning_messages": "Training data are empty",
            "RMSE_F": np.nan,
            "MAE_F": np.nan,
            "bias_F": np.nan,
            "worst_month": None,
            "worst_absolute_error_F": np.nan,
            "runtime_seconds": np.nan,
            "AIC": np.nan,
            "BIC": np.nan
        })

        continue

    if train.isna().any():
        print("FAILED: Training data contain missing values.")

        window_results.append({
            "window_number": window_number,
            "test_start": origin,
            "test_end": test_end,
            "train_end": train_end,
            "training_months": len(train),
            "model": MODEL_NAME,
            "status": "failed",
            "converged": False,
            "warning_count": np.nan,
            "warning_messages":
                "Training data contain missing values",
            "RMSE_F": np.nan,
            "MAE_F": np.nan,
            "bias_F": np.nan,
            "worst_month": None,
            "worst_absolute_error_F": np.nan,
            "runtime_seconds": np.nan,
            "AIC": np.nan,
            "BIC": np.nan
        })

        continue

    if actual.isna().any():
        print("FAILED: Test data contain missing values.")

        window_results.append({
            "window_number": window_number,
            "test_start": origin,
            "test_end": test_end,
            "train_end": train_end,
            "training_months": len(train),
            "model": MODEL_NAME,
            "status": "failed",
            "converged": False,
            "warning_count": np.nan,
            "warning_messages":
                "Test data contain missing values",
            "RMSE_F": np.nan,
            "MAE_F": np.nan,
            "bias_F": np.nan,
            "worst_month": None,
            "worst_absolute_error_F": np.nan,
            "runtime_seconds": np.nan,
            "AIC": np.nan,
            "BIC": np.nan
        })

        continue

    try:
        forecast, model_info = fit_sarima(
            train,
            test_months
        )

        error = forecast - actual
        absolute_error = error.abs()

        rmse = np.sqrt(
            np.mean(error ** 2)
        )

        mae = np.mean(
            absolute_error
        )

        bias = np.mean(error)

        worst_month = (
            absolute_error.idxmax()
        )

        status = (
            "completed"
            if model_info["converged"]
            else "completed_not_converged"
        )

        window_results.append({
            "window_number": window_number,
            "test_start": origin,
            "test_end": test_end,
            "train_end": train_end,
            "training_months": len(train),
            "model": MODEL_NAME,
            "status": status,
            "converged": model_info["converged"],
            "warning_count": model_info["warning_count"],
            "warning_messages": " | ".join(
                model_info["warnings"]
            ),
            "RMSE_F": rmse,
            "MAE_F": mae,
            "bias_F": bias,
            "worst_month": worst_month,
            "worst_absolute_error_F":
                absolute_error.loc[worst_month],
            "runtime_seconds":
                model_info["runtime_seconds"],
            "AIC": model_info["AIC"],
            "BIC": model_info["BIC"]
        })

        # Save every monthly prediction and error.
        for horizon, month in enumerate(
            test_months,
            start=1
        ):
            monthly_results.append({
                "window_number": window_number,
                "test_start": origin,
                "test_end": test_end,
                "train_end": train_end,
                "forecast_horizon": horizon,
                "month": month,
                "actual_F": actual.loc[month],
                "forecast_F": forecast.loc[month],
                "error_F": error.loc[month],
                "absolute_error_F":
                    absolute_error.loc[month],
                "squared_error":
                    error.loc[month] ** 2,
                "converged":
                    model_info["converged"]
            })

        print(
            f"RMSE={rmse:.3f} °F, "
            f"MAE={mae:.3f} °F, "
            f"Bias={bias:+.3f} °F, "
            f"Converged={model_info['converged']}"
        )

    except Exception as error_message:

        print(f"FAILED: {error_message}")

        window_results.append({
            "window_number": window_number,
            "test_start": origin,
            "test_end": test_end,
            "train_end": train_end,
            "training_months": len(train),
            "model": MODEL_NAME,
            "status": "failed",
            "converged": False,
            "warning_count": np.nan,
            "warning_messages": str(error_message),
            "RMSE_F": np.nan,
            "MAE_F": np.nan,
            "bias_F": np.nan,
            "worst_month": None,
            "worst_absolute_error_F": np.nan,
            "runtime_seconds": np.nan,
            "AIC": np.nan,
            "BIC": np.nan
        })


# ------------------------------------------------------------
# 8. Create result tables
# ------------------------------------------------------------

window_results = pd.DataFrame(
    window_results
)

monthly_results = pd.DataFrame(
    monthly_results
)

print("\nWINDOW-BY-WINDOW SARIMA RESULTS")
print("===============================")

display(
    window_results[
        [
            "window_number",
            "test_start",
            "test_end",
            "training_months",
            "status",
            "converged",
            "warning_count",
            "RMSE_F",
            "MAE_F",
            "bias_F",
            "worst_month",
            "worst_absolute_error_F",
            "runtime_seconds"
        ]
    ].round(3)
)


# ------------------------------------------------------------
# 9. Calculate overall stability metrics
# ------------------------------------------------------------

successful_windows = window_results.loc[
    window_results["RMSE_F"].notna()
].copy()

if successful_windows.empty:
    raise RuntimeError(
        "SARIMA did not complete any historical windows."
    )

pooled_error = monthly_results["error_F"]

pooled_rmse = np.sqrt(
    np.mean(pooled_error ** 2)
)

pooled_mae = np.mean(
    np.abs(pooled_error)
)

pooled_bias = np.mean(
    pooled_error
)

stability_summary = pd.DataFrame({
    "model": [MODEL_NAME],
    "windows_attempted": [len(FORECAST_ORIGINS)],
    "windows_completed": [len(successful_windows)],
    "windows_converged": [
        int(successful_windows["converged"].sum())
    ],
    "total_months_evaluated": [
        len(monthly_results)
    ],
    "mean_window_RMSE_F": [
        successful_windows["RMSE_F"].mean()
    ],
    "median_window_RMSE_F": [
        successful_windows["RMSE_F"].median()
    ],
    "RMSE_std_F": [
        successful_windows["RMSE_F"].std()
    ],
    "best_window_RMSE_F": [
        successful_windows["RMSE_F"].min()
    ],
    "worst_window_RMSE_F": [
        successful_windows["RMSE_F"].max()
    ],
    "mean_window_MAE_F": [
        successful_windows["MAE_F"].mean()
    ],
    "mean_window_bias_F": [
        successful_windows["bias_F"].mean()
    ],
    "worst_absolute_window_bias_F": [
        successful_windows["bias_F"].abs().max()
    ],
    "pooled_RMSE_F": [pooled_rmse],
    "pooled_MAE_F": [pooled_mae],
    "pooled_bias_F": [pooled_bias],
    "total_warning_count": [
        successful_windows["warning_count"].sum()
    ]
})

print("\nSARIMA STABILITY SUMMARY")
print("========================")
display(stability_summary.round(3))


# ------------------------------------------------------------
# 10. Identify the best and worst forecast windows
# ------------------------------------------------------------

best_window = successful_windows.loc[
    successful_windows["RMSE_F"].idxmin()
]

worst_window = successful_windows.loc[
    successful_windows["RMSE_F"].idxmax()
]

print("\nBEST AND WORST WINDOWS")
print("======================")

print(
    f"Best window:  "
    f"{best_window['test_start']:%Y-%m} through "
    f"{best_window['test_end']:%Y-%m}; "
    f"RMSE={best_window['RMSE_F']:.3f} °F"
)

print(
    f"Worst window: "
    f"{worst_window['test_start']:%Y-%m} through "
    f"{worst_window['test_end']:%Y-%m}; "
    f"RMSE={worst_window['RMSE_F']:.3f} °F"
)


# ------------------------------------------------------------
# 11. Plot RMSE for every historical window
# ------------------------------------------------------------

plot_data = successful_windows.copy()

plot_data["window_label"] = (
    plot_data["test_start"].dt.strftime("%Y-%m")
    + "\nto\n"
    + plot_data["test_end"].dt.strftime("%Y-%m")
)

fig, ax = plt.subplots(figsize=(14, 6))

ax.plot(
    plot_data["window_label"],
    plot_data["RMSE_F"],
    marker="o",
    markersize=8,
    linewidth=2.5,
    color="darkred"
)

# Add the mean RMSE as a reference line.
mean_window_rmse = (
    plot_data["RMSE_F"].mean()
)

ax.axhline(
    mean_window_rmse,
    color="gray",
    linestyle="--",
    linewidth=1.5,
    label=f"Mean RMSE = {mean_window_rmse:.3f} °F"
)

# Write each RMSE above its point.
for position, value in enumerate(
    plot_data["RMSE_F"]
):
    ax.annotate(
        f"{value:.2f}",
        (position, value),
        xytext=(0, 8),
        textcoords="offset points",
        ha="center"
    )

ax.set_title(
    "SARIMA Stability Across Nine Historical Forecast Windows\n"
    "Each Point Represents a Separate 12-Month Forecast"
)

ax.set_xlabel("Historical forecast window")
ax.set_ylabel("RMSE (°F)")
ax.grid(alpha=0.25)
ax.legend()

plt.xticks(rotation=35, ha="right")
plt.tight_layout()

rmse_plot_file = (
    OUTPUT_DIR
    / "sarima_rmse_across_nine_windows.png"
)

plt.savefig(
    rmse_plot_file,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 12. Plot bias for every historical window
# ------------------------------------------------------------

fig, ax = plt.subplots(figsize=(14, 5))

bias_colors = np.where(
    plot_data["bias_F"] >= 0,
    "darkorange",
    "steelblue"
)

ax.bar(
    plot_data["window_label"],
    plot_data["bias_F"],
    color=bias_colors
)

ax.axhline(
    0,
    color="black",
    linewidth=1
)

ax.set_title(
    "SARIMA Bias Across Historical Forecast Windows\n"
    "Orange = Too Warm; Blue = Too Cold"
)

ax.set_xlabel("Historical forecast window")
ax.set_ylabel("Mean forecast error (°F)")
ax.grid(axis="y", alpha=0.25)

plt.xticks(rotation=35, ha="right")
plt.tight_layout()

bias_plot_file = (
    OUTPUT_DIR
    / "sarima_bias_across_nine_windows.png"
)

plt.savefig(
    bias_plot_file,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 13. Save model parameters and results
# ------------------------------------------------------------

window_results_file = (
    OUTPUT_DIR
    / "sarima_window_metrics.csv"
)

monthly_results_file = (
    OUTPUT_DIR
    / "sarima_monthly_predictions_and_errors.csv"
)

stability_summary_file = (
    OUTPUT_DIR
    / "sarima_stability_summary.csv"
)

parameters_file = (
    OUTPUT_DIR
    / "sarima_parameters.json"
)

window_results.to_csv(
    window_results_file,
    index=False
)

monthly_results.to_csv(
    monthly_results_file,
    index=False
)

stability_summary.to_csv(
    stability_summary_file,
    index=False
)

parameters_to_save = {
    "model_name": MODEL_NAME,
    "order": list(MODEL_PARAMETERS["order"]),
    "seasonal_order": list(
        MODEL_PARAMETERS["seasonal_order"]
    ),
    "enforce_stationarity":
        MODEL_PARAMETERS["enforce_stationarity"],
    "enforce_invertibility":
        MODEL_PARAMETERS["enforce_invertibility"],
    "forecast_horizon_months":
        FORECAST_HORIZON,
    "forecast_origins": [
        date.strftime("%Y-%m-%d")
        for date in FORECAST_ORIGINS
    ],
    "temperature_aggregation":
        "mean of all source rows per local calendar month",
    "duplicate_timestamp_policy":
        "retained to match Joy/Khadijah"
}

with open(parameters_file, "w") as file:
    json.dump(
        parameters_to_save,
        file,
        indent=4
    )

print("\nFILES SAVED")
print("===========")
print(f"Window metrics:      {window_results_file}")
print(f"Monthly predictions: {monthly_results_file}")
print(f"Stability summary:   {stability_summary_file}")
print(f"Model parameters:    {parameters_file}")
print(f"RMSE plot:           {rmse_plot_file}")
print(f"Bias plot:           {bias_plot_file}")

print("\nSEND ME:")
print("1. WINDOW-BY-WINDOW SARIMA RESULTS")
print("2. SARIMA STABILITY SUMMARY")
print("3. BEST AND WORST WINDOWS")
print("4. RMSE plot")
print("5. Bias plot")
```

```python
# ============================================================
# LEVEL 7: ALL MODELS ACROSS ALL NINE 12-MONTH WINDOWS
#
# Models:
#   1. SARIMA(1,1,1)(1,1,1,12)
#   2. Prophet additive
#   3. Additive damped ETS
#   4. Seasonal naive
#
# Historical forecast windows:
#   July 2017–June 2018
#   ...
#   July 2025–June 2026
#
# All source rows are retained when computing monthly means,
# consistent with Joy and Khadijah's original processing.
# ============================================================


# ------------------------------------------------------------
# 0. Install Prophet
# ------------------------------------------------------------

!pip install -q prophet


# ------------------------------------------------------------
# 1. Imports
# ------------------------------------------------------------

from google.colab import drive
drive.mount("/content/drive")

from pathlib import Path
import calendar
import json
import logging
import time
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display
from prophet import Prophet
from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.tsa.holtwinters import ExponentialSmoothing

# Reduce Prophet/CmdStan messages.
logging.getLogger("cmdstanpy").setLevel(logging.WARNING)
logging.getLogger("prophet").setLevel(logging.WARNING)


# ------------------------------------------------------------
# 2. Project settings
# ------------------------------------------------------------

PROJECT = Path("/content/drive/MyDrive/Stanford_Weather_12M")
DATA_FILE = PROJECT / "data" / "OpenWeather19802026_StanfordOnly.csv"

OUTPUT_DIR = (
    PROJECT
    / "foundational_analysis"
    / "level_7_all_models_all_windows"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

CITY = "Stanford University"
FORECAST_HORIZON = 12

FORECAST_ORIGINS = pd.date_range(
    start="2017-07-01",
    end="2025-07-01",
    freq="12MS"
)

MODEL_PARAMETERS = {
    "sarima_111x111_12": {
        "order": (1, 1, 1),
        "seasonal_order": (1, 1, 1, 12),
        "enforce_stationarity": False,
        "enforce_invertibility": False
    },

    "prophet_additive": {
        "growth": "linear",
        "seasonality_mode": "additive",
        "yearly_seasonality": True,
        "weekly_seasonality": False,
        "daily_seasonality": False
    },

    "ets_add_add_damped12": {
        "trend": "add",
        "damped_trend": True,
        "seasonal": "add",
        "seasonal_periods": 12,
        "initialization_method": "estimated"
    },

    "seasonal_naive_12": {
        "seasonal_lag": 12
    }
}


# ============================================================
# PART A: LOAD AND VERIFY THE SOURCE DATA
# ============================================================


# ------------------------------------------------------------
# 3. Load Stanford hourly temperature data
# ------------------------------------------------------------

df = pd.read_csv(
    DATA_FILE,
    usecols=[
        "dt",
        "dt_iso",
        "timezone",
        "city_name",
        "temp"
    ]
)

df = df.loc[
    df["city_name"] == CITY
].copy()

df["temp"] = pd.to_numeric(
    df["temp"],
    errors="coerce"
)

df["timezone"] = pd.to_numeric(
    df["timezone"],
    errors="coerce"
)


# ------------------------------------------------------------
# 4. Convert timestamps to Stanford local time
# ------------------------------------------------------------

df["dt_utc"] = pd.to_datetime(
    df["dt"],
    unit="s",
    utc=True,
    errors="coerce"
)

df["timezone_offset"] = pd.to_timedelta(
    df["timezone"],
    unit="s"
)

df["dt_local"] = (
    df["dt_utc"] + df["timezone_offset"]
).dt.tz_localize(None)

df = df.dropna(
    subset=["dt_local", "temp"]
)

df["month"] = (
    df["dt_local"]
    .dt.to_period("M")
    .dt.to_timestamp()
)


# ------------------------------------------------------------
# 5. Confirm the complete source-data date range
# ------------------------------------------------------------

raw_month_coverage = (
    df.groupby("month")
      .agg(
          source_rows=("dt_local", "size"),
          unique_hourly_timestamps=("dt_local", "nunique"),
          valid_temperature_rows=("temp", "count"),
          first_local_timestamp=("dt_local", "min"),
          last_local_timestamp=("dt_local", "max")
      )
      .sort_index()
)

raw_month_coverage["days_in_month"] = (
    raw_month_coverage.index.days_in_month
)

raw_month_coverage["expected_hourly_timestamps"] = (
    raw_month_coverage["days_in_month"] * 24
)

raw_month_coverage["missing_unique_hours"] = (
    raw_month_coverage["expected_hourly_timestamps"]
    - raw_month_coverage["unique_hourly_timestamps"]
)

raw_month_coverage["extra_rows_from_duplicates"] = (
    raw_month_coverage["source_rows"]
    - raw_month_coverage["unique_hourly_timestamps"]
)

june_2026 = raw_month_coverage.loc[
    [pd.Timestamp("2026-06-01")]
].copy()

print("JUNE 2026 SOURCE-DATA CONFIRMATION")
print("==================================")
display(june_2026)

june_exists = (
    pd.Timestamp("2026-06-01")
    in raw_month_coverage.index
)

june_has_all_hours = (
    june_exists
    and june_2026.iloc[0]["unique_hourly_timestamps"]
    == june_2026.iloc[0]["expected_hourly_timestamps"]
)

print(f"June 2026 exists:                    {june_exists}")
print(f"June 2026 has all expected hours:   {june_has_all_hours}")

if june_exists:
    print(
        "June 2026 unique hourly timestamps: "
        f"{int(june_2026.iloc[0]['unique_hourly_timestamps'])}"
    )
    print(
        "June 2026 expected hourly timestamps: "
        f"{int(june_2026.iloc[0]['expected_hourly_timestamps'])}"
    )
    print(
        "June 2026 source rows: "
        f"{int(june_2026.iloc[0]['source_rows'])}"
    )
    print(
        "June 2026 first timestamp: "
        f"{june_2026.iloc[0]['first_local_timestamp']}"
    )
    print(
        "June 2026 last timestamp: "
        f"{june_2026.iloc[0]['last_local_timestamp']}"
    )

if not june_exists:
    raise ValueError(
        "June 2026 is missing from the source data."
    )

if not june_has_all_hours:
    raise ValueError(
        "June 2026 exists but does not contain all 720 "
        "unique hourly timestamps."
    )


# ------------------------------------------------------------
# 6. Display any observations after June 2026
#
# This tells us whether July 2026 is present but incomplete.
# It remains excluded from modeling.
# ------------------------------------------------------------

months_after_june_2026 = raw_month_coverage.loc[
    raw_month_coverage.index
    > pd.Timestamp("2026-06-01")
].copy()

print("\nSOURCE MONTHS AFTER JUNE 2026")
print("=============================")

if months_after_june_2026.empty:
    print("No source observations occur after June 2026.")
else:
    display(months_after_june_2026)

    print(
        "These later observations are not used. The official "
        "modeling endpoint remains June 2026."
    )


# ------------------------------------------------------------
# 7. Create the official monthly temperature series
#
# All rows are averaged. Duplicated timestamps are retained.
# ------------------------------------------------------------

series = (
    df.groupby("month")["temp"]
      .mean()
      .sort_index()
      .rename("mean_temp_F")
      .loc["1980-01-01":"2026-06-01"]
      .asfreq("MS")
)

print("\nOFFICIAL MONTHLY SERIES")
print("=======================")
print(f"First month: {series.index.min():%Y-%m}")
print(f"Last month:  {series.index.max():%Y-%m}")
print(f"Months:      {len(series)}")
print(f"Missing:     {series.isna().sum()}")

assert series.index.max() == pd.Timestamp("2026-06-01")
assert len(series) == 558
assert series.isna().sum() == 0

print(
    "CONFIRMED: The modeling data contain 558 consecutive "
    "months from January 1980 through June 2026."
)


# ============================================================
# PART B: DEFINE THE FOUR CANDIDATE MODELS
# ============================================================


# ------------------------------------------------------------
# 8. SARIMA
# ------------------------------------------------------------

def fit_sarima(training_series, forecast_dates):
    """Fit SARIMA and return its 12-month forecast."""

    parameters = MODEL_PARAMETERS[
        "sarima_111x111_12"
    ]

    with warnings.catch_warnings(record=True) as caught:
        warnings.simplefilter("always")

        model = SARIMAX(
            training_series,
            **parameters
        )

        fitted = model.fit(disp=False)

    prediction = (
        fitted
        .get_forecast(steps=len(forecast_dates))
        .predicted_mean
    )

    prediction.index = forecast_dates

    information = {
        "converged": bool(
            fitted.mle_retvals.get("converged", False)
        ),
        "warning_count": len(caught),
        "warning_messages": " | ".join(
            str(item.message) for item in caught
        ),
        "AIC": fitted.aic,
        "BIC": fitted.bic
    }

    return prediction.astype(float), information


# ------------------------------------------------------------
# 9. Prophet
# ------------------------------------------------------------

def fit_prophet(training_series, forecast_dates):
    """Fit additive Prophet with yearly seasonality."""

    parameters = MODEL_PARAMETERS[
        "prophet_additive"
    ]

    prophet_training = pd.DataFrame({
        "ds": training_series.index,
        "y": training_series.values
    })

    model = Prophet(**parameters)
    model.fit(prophet_training)

    future = pd.DataFrame({
        "ds": forecast_dates
    })

    prophet_output = model.predict(future)

    prediction = pd.Series(
        prophet_output["yhat"].to_numpy(),
        index=forecast_dates
    )

    information = {
        "converged": True,
        "warning_count": 0,
        "warning_messages": "",
        "AIC": np.nan,
        "BIC": np.nan
    }

    return prediction.astype(float), information


# ------------------------------------------------------------
# 10. ETS
# ------------------------------------------------------------

def fit_ets(training_series, forecast_dates):
    """Fit additive ETS with a damped trend."""

    parameters = MODEL_PARAMETERS[
        "ets_add_add_damped12"
    ]

    with warnings.catch_warnings(record=True) as caught:
        warnings.simplefilter("always")

        model = ExponentialSmoothing(
            training_series,
            **parameters
        )

        fitted = model.fit(optimized=True)

    prediction = fitted.forecast(
        len(forecast_dates)
    )

    prediction.index = forecast_dates

    information = {
        "converged": True,
        "warning_count": len(caught),
        "warning_messages": " | ".join(
            str(item.message) for item in caught
        ),
        "AIC": getattr(fitted, "aic", np.nan),
        "BIC": getattr(fitted, "bic", np.nan)
    }

    return prediction.astype(float), information


# ------------------------------------------------------------
# 11. Seasonal naive
# ------------------------------------------------------------

def fit_seasonal_naive(
    training_series,
    forecast_dates
):
    """
    Predict each month using the observed temperature from
    the same calendar month one year earlier.
    """

    seasonal_lag = MODEL_PARAMETERS[
        "seasonal_naive_12"
    ]["seasonal_lag"]

    predictions = []

    for forecast_date in forecast_dates:
        comparison_date = (
            forecast_date
            - pd.DateOffset(months=seasonal_lag)
        )

        if comparison_date not in training_series.index:
            raise ValueError(
                f"Required comparison month "
                f"{comparison_date:%Y-%m} is unavailable."
            )

        predictions.append(
            training_series.loc[comparison_date]
        )

    prediction = pd.Series(
        predictions,
        index=forecast_dates,
        dtype=float
    )

    information = {
        "converged": True,
        "warning_count": 0,
        "warning_messages": "",
        "AIC": np.nan,
        "BIC": np.nan
    }

    return prediction, information


# ------------------------------------------------------------
# 12. Store the model functions together
# ------------------------------------------------------------

MODELS = {
    "sarima_111x111_12": fit_sarima,
    "prophet_additive": fit_prophet,
    "ets_add_add_damped12": fit_ets,
    "seasonal_naive_12": fit_seasonal_naive
}


# ============================================================
# PART C: RUN ALL MODELS ACROSS ALL WINDOWS
# ============================================================


# ------------------------------------------------------------
# 13. Backtest loop
# ------------------------------------------------------------

window_results = []
monthly_results = []

print("\nALL-MODEL ROLLING BACKTEST")
print("==========================")

for origin in FORECAST_ORIGINS:

    test_months = pd.date_range(
        start=origin,
        periods=FORECAST_HORIZON,
        freq="MS"
    )

    test_end = test_months[-1]

    train_end = (
        origin - pd.offsets.MonthBegin(1)
    )

    train = series.loc[:train_end].copy()
    actual = series.reindex(test_months)

    print(
        f"\nForecast window: "
        f"{origin:%Y-%m} through {test_end:%Y-%m}"
    )
    print(f"Training through: {train_end:%Y-%m}")

    if train.isna().any():
        raise ValueError(
            f"Training data contain missing observations "
            f"for origin {origin:%Y-%m}."
        )

    if actual.isna().any():
        raise ValueError(
            f"Test data contain missing observations "
            f"for origin {origin:%Y-%m}."
        )

    for model_name, model_function in MODELS.items():

        start_time = time.perf_counter()

        try:
            prediction, model_info = model_function(
                train,
                test_months
            )

            runtime_seconds = (
                time.perf_counter() - start_time
            )

            prediction = prediction.reindex(test_months)

            if prediction.isna().any():
                raise ValueError(
                    "Model returned missing forecasts."
                )

            error = prediction - actual
            absolute_error = error.abs()

            rmse = np.sqrt(
                np.mean(error ** 2)
            )

            mae = np.mean(
                absolute_error
            )

            bias = np.mean(error)

            seasonal_shape_correlation = (
                actual.corr(prediction)
            )

            status = (
                "completed"
                if model_info["converged"]
                else "completed_not_converged"
            )

            window_results.append({
                "test_start": origin,
                "test_end": test_end,
                "train_end": train_end,
                "training_months": len(train),
                "model": model_name,
                "status": status,
                "converged": model_info["converged"],
                "warning_count":
                    model_info["warning_count"],
                "warning_messages":
                    model_info["warning_messages"],
                "RMSE_F": rmse,
                "MAE_F": mae,
                "bias_F": bias,
                "seasonal_shape_correlation":
                    seasonal_shape_correlation,
                "runtime_seconds": runtime_seconds,
                "AIC": model_info["AIC"],
                "BIC": model_info["BIC"]
            })

            for horizon, month in enumerate(
                test_months,
                start=1
            ):
                monthly_results.append({
                    "test_start": origin,
                    "test_end": test_end,
                    "train_end": train_end,
                    "model": model_name,
                    "forecast_horizon": horizon,
                    "month": month,
                    "actual_F": actual.loc[month],
                    "forecast_F": prediction.loc[month],
                    "error_F": error.loc[month],
                    "absolute_error_F":
                        absolute_error.loc[month],
                    "squared_error_F":
                        error.loc[month] ** 2,
                    "converged":
                        model_info["converged"]
                })

            print(
                f"  {model_name:<25} "
                f"RMSE={rmse:.3f}, "
                f"MAE={mae:.3f}, "
                f"Bias={bias:+.3f}, "
                f"Converged={model_info['converged']}"
            )

        except Exception as error_message:

            runtime_seconds = (
                time.perf_counter() - start_time
            )

            print(
                f"  {model_name:<25} "
                f"FAILED: {error_message}"
            )

            window_results.append({
                "test_start": origin,
                "test_end": test_end,
                "train_end": train_end,
                "training_months": len(train),
                "model": model_name,
                "status": "failed",
                "converged": False,
                "warning_count": np.nan,
                "warning_messages": str(error_message),
                "RMSE_F": np.nan,
                "MAE_F": np.nan,
                "bias_F": np.nan,
                "seasonal_shape_correlation": np.nan,
                "runtime_seconds": runtime_seconds,
                "AIC": np.nan,
                "BIC": np.nan
            })


# ------------------------------------------------------------
# 14. Create the detailed result tables
# ------------------------------------------------------------

window_results = pd.DataFrame(
    window_results
)

monthly_results = pd.DataFrame(
    monthly_results
)

print("\nWINDOW-BY-WINDOW RESULTS")
print("========================")

display(
    window_results[
        [
            "test_start",
            "test_end",
            "model",
            "status",
            "converged",
            "RMSE_F",
            "MAE_F",
            "bias_F",
            "seasonal_shape_correlation",
            "warning_count",
            "runtime_seconds"
        ]
    ]
    .sort_values(["test_start", "RMSE_F"])
    .round(3)
)


# ============================================================
# PART D: MEASURE STABILITY ACROSS WINDOWS
# ============================================================


# ------------------------------------------------------------
# 15. Create one stability-summary row per model
# ------------------------------------------------------------

summary_rows = []

for model_name in MODELS:

    model_windows = window_results.loc[
        (window_results["model"] == model_name)
        & window_results["RMSE_F"].notna()
    ].copy()

    model_months = monthly_results.loc[
        monthly_results["model"] == model_name
    ].copy()

    if model_windows.empty:
        summary_rows.append({
            "model": model_name,
            "windows_attempted": len(FORECAST_ORIGINS),
            "windows_completed": 0
        })
        continue

    pooled_errors = model_months["error_F"]

    summary_rows.append({
        "model": model_name,
        "windows_attempted": len(FORECAST_ORIGINS),
        "windows_completed": len(model_windows),
        "windows_converged":
            int(model_windows["converged"].sum()),
        "months_evaluated": len(model_months),

        "mean_RMSE_F":
            model_windows["RMSE_F"].mean(),

        "median_RMSE_F":
            model_windows["RMSE_F"].median(),

        "RMSE_std_F":
            model_windows["RMSE_F"].std(),

        "best_window_RMSE_F":
            model_windows["RMSE_F"].min(),

        "worst_window_RMSE_F":
            model_windows["RMSE_F"].max(),

        "mean_MAE_F":
            model_windows["MAE_F"].mean(),

        "mean_bias_F":
            model_windows["bias_F"].mean(),

        "worst_absolute_bias_F":
            model_windows["bias_F"].abs().max(),

        "pooled_RMSE_F":
            np.sqrt(np.mean(pooled_errors ** 2)),

        "pooled_MAE_F":
            np.mean(np.abs(pooled_errors)),

        "pooled_bias_F":
            np.mean(pooled_errors),

        "mean_seasonal_shape_correlation":
            model_windows[
                "seasonal_shape_correlation"
            ].mean(),

        "total_warning_count":
            model_windows["warning_count"].sum(),

        "total_runtime_seconds":
            model_windows["runtime_seconds"].sum()
    })

stability_summary = pd.DataFrame(
    summary_rows
)

# A simple diagnostic combining typical error and variability.
# Lower is better, but the individual columns remain important.
stability_summary["robustness_indicator"] = (
    stability_summary["mean_RMSE_F"]
    + stability_summary["RMSE_std_F"]
)

stability_summary = (
    stability_summary
    .sort_values(
        ["pooled_RMSE_F", "RMSE_std_F"]
    )
    .reset_index(drop=True)
)

print("\nMODEL STABILITY SUMMARY")
print("=======================")
display(stability_summary.round(3))


# ------------------------------------------------------------
# 16. Count which model won each historical window
# ------------------------------------------------------------

valid_window_results = window_results.dropna(
    subset=["RMSE_F"]
).copy()

window_winners = (
    valid_window_results
    .sort_values(["test_start", "RMSE_F"])
    .groupby("test_start", as_index=False)
    .first()
)

window_winners = window_winners[
    [
        "test_start",
        "test_end",
        "model",
        "RMSE_F",
        "MAE_F",
        "bias_F"
    ]
].rename(
    columns={"model": "winning_model"}
)

win_counts = (
    window_winners["winning_model"]
    .value_counts()
    .rename_axis("model")
    .reset_index(name="window_wins")
)

print("\nWINNER IN EACH HISTORICAL WINDOW")
print("================================")
display(window_winners.round(3))

print("\nNUMBER OF WINDOW WINS")
print("=====================")
display(win_counts)


# ============================================================
# PART E: VISUALIZE THE STABILITY RESULTS
# ============================================================


# ------------------------------------------------------------
# 17. Plot RMSE across forecast windows
# ------------------------------------------------------------

plot_results = valid_window_results.copy()

plot_results["window_label"] = (
    plot_results["test_start"].dt.strftime("%Y-%m")
)

fig, ax = plt.subplots(figsize=(14, 6))

for model_name in MODELS:

    model_plot = plot_results.loc[
        plot_results["model"] == model_name
    ].sort_values("test_start")

    ax.plot(
        model_plot["window_label"],
        model_plot["RMSE_F"],
        marker="o",
        linewidth=2,
        label=model_name
    )

ax.set_title(
    "Model Stability Across Nine 12-Month Forecast Windows"
)
ax.set_xlabel("Beginning of forecast window")
ax.set_ylabel("RMSE (°F)")
ax.grid(alpha=0.25)
ax.legend()

plt.xticks(rotation=35, ha="right")
plt.tight_layout()

rmse_plot_file = (
    OUTPUT_DIR
    / "all_models_rmse_across_windows.png"
)

plt.savefig(
    rmse_plot_file,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 18. Plot mean RMSE with variability
# ------------------------------------------------------------

summary_plot = stability_summary.sort_values(
    "mean_RMSE_F"
)

fig, ax = plt.subplots(figsize=(11, 6))

ax.bar(
    summary_plot["model"],
    summary_plot["mean_RMSE_F"],
    yerr=summary_plot["RMSE_std_F"],
    capsize=6,
    color="steelblue",
    alpha=0.85
)

ax.set_title(
    "Average 12-Month Forecast Error and Window-to-Window "
    "Variability"
)
ax.set_xlabel("Model")
ax.set_ylabel("Mean RMSE ± one standard deviation (°F)")
ax.grid(axis="y", alpha=0.25)

plt.xticks(rotation=25, ha="right")
plt.tight_layout()

stability_plot_file = (
    OUTPUT_DIR
    / "mean_rmse_with_variability.png"
)

plt.savefig(
    stability_plot_file,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 19. Plot bias across windows
# ------------------------------------------------------------

fig, ax = plt.subplots(figsize=(14, 6))

for model_name in MODELS:

    model_plot = plot_results.loc[
        plot_results["model"] == model_name
    ].sort_values("test_start")

    ax.plot(
        model_plot["window_label"],
        model_plot["bias_F"],
        marker="o",
        linewidth=2,
        label=model_name
    )

ax.axhline(
    0,
    color="black",
    linewidth=1
)

ax.set_title(
    "Forecast Bias Across Historical Windows\n"
    "Positive = Too Warm; Negative = Too Cold"
)
ax.set_xlabel("Beginning of forecast window")
ax.set_ylabel("Mean forecast error (°F)")
ax.grid(alpha=0.25)
ax.legend()

plt.xticks(rotation=35, ha="right")
plt.tight_layout()

bias_plot_file = (
    OUTPUT_DIR
    / "all_models_bias_across_windows.png"
)

plt.savefig(
    bias_plot_file,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# PART F: SAVE EVERYTHING
# ============================================================


# ------------------------------------------------------------
# 20. Save tables
# ------------------------------------------------------------

coverage_file = (
    OUTPUT_DIR
    / "raw_month_coverage.csv"
)

monthly_series_file = (
    OUTPUT_DIR
    / "monthly_data_used.csv"
)

window_results_file = (
    OUTPUT_DIR
    / "all_model_window_metrics.csv"
)

monthly_results_file = (
    OUTPUT_DIR
    / "all_model_monthly_predictions.csv"
)

stability_file = (
    OUTPUT_DIR
    / "model_stability_summary.csv"
)

winners_file = (
    OUTPUT_DIR
    / "window_winners.csv"
)

win_counts_file = (
    OUTPUT_DIR
    / "window_win_counts.csv"
)

parameters_file = (
    OUTPUT_DIR
    / "model_parameters.json"
)

raw_month_coverage.reset_index().to_csv(
    coverage_file,
    index=False
)

series.rename("mean_temp_F").to_csv(
    monthly_series_file,
    index_label="month"
)

window_results.to_csv(
    window_results_file,
    index=False
)

monthly_results.to_csv(
    monthly_results_file,
    index=False
)

stability_summary.to_csv(
    stability_file,
    index=False
)

window_winners.to_csv(
    winners_file,
    index=False
)

win_counts.to_csv(
    win_counts_file,
    index=False
)


# JSON cannot store tuples, so convert tuples to lists.
parameters_for_json = {
    model_name: {
        key: list(value) if isinstance(value, tuple) else value
        for key, value in parameters.items()
    }
    for model_name, parameters in MODEL_PARAMETERS.items()
}

with open(parameters_file, "w") as file:
    json.dump(
        {
            "models": parameters_for_json,
            "city": CITY,
            "forecast_horizon_months":
                FORECAST_HORIZON,
            "forecast_origins": [
                origin.strftime("%Y-%m-%d")
                for origin in FORECAST_ORIGINS
            ],
            "data_start": "1980-01-01",
            "data_end": "2026-06-01",
            "aggregation":
                "Mean of all source rows per local month",
            "duplicate_policy":
                "Repeated timestamps retained to match "
                "Joy/Khadijah",
            "error_definition":
                "forecast minus actual"
        },
        file,
        indent=4
    )

print("\nFILES SAVED")
print("===========")
print(f"Coverage audit:       {coverage_file}")
print(f"Monthly data used:    {monthly_series_file}")
print(f"Window metrics:       {window_results_file}")
print(f"Monthly predictions:  {monthly_results_file}")
print(f"Stability summary:    {stability_file}")
print(f"Window winners:       {winners_file}")
print(f"Window win counts:    {win_counts_file}")
print(f"Model parameters:     {parameters_file}")
print(f"RMSE plot:            {rmse_plot_file}")
print(f"Variability plot:     {stability_plot_file}")
print(f"Bias plot:            {bias_plot_file}")

print("\nSEND ME:")
print("1. JUNE 2026 SOURCE-DATA CONFIRMATION")
print("2. MODEL STABILITY SUMMARY")
print("3. WINNER IN EACH HISTORICAL WINDOW")
print("4. NUMBER OF WINDOW WINS")
print("5. The RMSE and variability plots")
```

```python
# ============================================================
# IMPROVEMENT STAGE 1–2
#
# 1. Lock the existing SARIMA benchmark.
# 2. Run nested tuning of:
#       - SARIMA parameter specification
#       - amount of historical training data
#
# Outer test windows:
#   July 2017–June 2018 through July 2025–June 2026
#
# Within every outer window:
#   - Use three earlier 12-month windows for parameter selection.
#   - Select parameters using inner validation only.
#   - Refit the selected model on the outer training data.
#   - Forecast the untouched outer 12-month period.
#
# This preserves the Joy/Khadijah all-row aggregation.
# ============================================================


# ------------------------------------------------------------
# 1. Imports and Drive
# ------------------------------------------------------------

from google.colab import drive
drive.mount("/content/drive")

from pathlib import Path
import json
import time
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display
from statsmodels.tsa.statespace.sarimax import SARIMAX


# ------------------------------------------------------------
# 2. Project settings
# ------------------------------------------------------------

PROJECT = Path("/content/drive/MyDrive/Stanford_Weather_12M")

# Use the exact monthly series created in Level 7.
DATA_FILE = (
    PROJECT
    / "foundational_analysis"
    / "level_7_all_models_all_windows"
    / "monthly_data_used.csv"
)

OUTPUT_DIR = (
    PROJECT
    / "model_improvement"
    / "nested_sarima_tuning"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

FORECAST_HORIZON = 12
INNER_FOLDS = 3

OUTER_ORIGINS = pd.date_range(
    start="2017-07-01",
    end="2025-07-01",
    freq="12MS"
)


# ============================================================
# PART A: LOCK THE BENCHMARK BEFORE TUNING
# ============================================================


# ------------------------------------------------------------
# 3. Benchmark specification
#
# Do not change this after seeing challenger results.
# ------------------------------------------------------------

BENCHMARK = {
    "model_name": "locked_sarima_111x111_12",
    "order": (1, 1, 1),
    "seasonal_order": (1, 1, 1, 12),
    "training_history": "all",
    "forecast_horizon": 12,
    "selection_status": "locked_before_nested_tuning"
}

# Previously established Level 7 results.
LOCKED_BENCHMARK_RESULTS = {
    "mean_RMSE_F": 1.953,
    "pooled_RMSE_F": 2.016,
    "RMSE_std_F": 0.528,
    "worst_window_RMSE_F": 2.789,
    "pooled_MAE_F": 1.564,
    "pooled_bias_F": -0.001,
    "windows_completed": 9
}

benchmark_file = OUTPUT_DIR / "locked_benchmark.json"

with open(benchmark_file, "w") as file:
    json.dump(
        {
            "model": {
                "model_name": BENCHMARK["model_name"],
                "order": list(BENCHMARK["order"]),
                "seasonal_order": list(
                    BENCHMARK["seasonal_order"]
                ),
                "training_history":
                    BENCHMARK["training_history"],
                "forecast_horizon":
                    BENCHMARK["forecast_horizon"],
                "selection_status":
                    BENCHMARK["selection_status"]
            },
            "previous_results":
                LOCKED_BENCHMARK_RESULTS
        },
        file,
        indent=4
    )

print("LOCKED BENCHMARK")
print("================")
print("Model: SARIMA(1,1,1)(1,1,1,12)")
print("Training history: all available data")
print("Previous pooled RMSE: 2.016 °F")
print("Previous mean RMSE:   1.953 °F")
print(f"Saved: {benchmark_file}")


# ============================================================
# PART B: LOAD AND VERIFY THE MONTHLY SERIES
# ============================================================


# ------------------------------------------------------------
# 4. Load monthly temperature
# ------------------------------------------------------------

monthly_data = pd.read_csv(
    DATA_FILE,
    parse_dates=["month"]
)

monthly_data = (
    monthly_data
    .set_index("month")
    .sort_index()
)

series = (
    monthly_data["mean_temp_F"]
    .astype(float)
    .asfreq("MS")
)

print("\nMONTHLY DATA")
print("============")
print(f"Source:      {DATA_FILE}")
print(f"First month: {series.index.min():%Y-%m}")
print(f"Last month:  {series.index.max():%Y-%m}")
print(f"Months:      {len(series)}")
print(f"Missing:     {series.isna().sum()}")

assert series.index.min() == pd.Timestamp("1980-01-01")
assert series.index.max() == pd.Timestamp("2026-06-01")
assert len(series) == 558
assert series.isna().sum() == 0


# ============================================================
# PART C: DEFINE THE CANDIDATE SEARCH SPACE
# ============================================================


# ------------------------------------------------------------
# 5. Small, interpretable SARIMA candidate set
#
# We vary nearby specifications around the inherited model.
# Seasonal period remains 12 because data are monthly and the
# earlier analysis established annual seasonality.
# ------------------------------------------------------------

SARIMA_CANDIDATES = [
    {
        "candidate": "sarima_111_111",
        "order": (1, 1, 1),
        "seasonal_order": (1, 1, 1, 12)
    },
    {
        "candidate": "sarima_011_111",
        "order": (0, 1, 1),
        "seasonal_order": (1, 1, 1, 12)
    },
    {
        "candidate": "sarima_110_111",
        "order": (1, 1, 0),
        "seasonal_order": (1, 1, 1, 12)
    },
    {
        "candidate": "sarima_211_111",
        "order": (2, 1, 1),
        "seasonal_order": (1, 1, 1, 12)
    },
    {
        "candidate": "sarima_112_111",
        "order": (1, 1, 2),
        "seasonal_order": (1, 1, 1, 12)
    },
    {
        "candidate": "sarima_111_011",
        "order": (1, 1, 1),
        "seasonal_order": (0, 1, 1, 12)
    },
    {
        "candidate": "sarima_111_110",
        "order": (1, 1, 1),
        "seasonal_order": (1, 1, 0, 12)
    },
    {
        "candidate": "sarima_011_011",
        "order": (0, 1, 1),
        "seasonal_order": (0, 1, 1, 12)
    },
    {
        "candidate": "sarima_110_110",
        "order": (1, 1, 0),
        "seasonal_order": (1, 1, 0, 12)
    }
]

# Test whether recent history is better than the complete
# historical record.
TRAINING_HISTORIES = [
    "all",
    30,
    20,
    15
]

print("\nSEARCH SPACE")
print("============")
print(f"SARIMA structures: {len(SARIMA_CANDIDATES)}")
print(f"Training histories: {TRAINING_HISTORIES}")
print(f"Inner folds per outer window: {INNER_FOLDS}")

total_inner_fits = (
    len(OUTER_ORIGINS)
    * len(SARIMA_CANDIDATES)
    * len(TRAINING_HISTORIES)
    * INNER_FOLDS
)

print(f"Maximum inner SARIMA fits: {total_inner_fits}")
print(
    "Additional outer fits: "
    f"{len(OUTER_ORIGINS) * 2}"
)


# ============================================================
# PART D: HELPER FUNCTIONS
# ============================================================


# ------------------------------------------------------------
# 6. Select the requested history length
# ------------------------------------------------------------

def select_training_history(training_series, history):
    """
    Return either all available training data or the requested
    number of most recent years.
    """

    if history == "all":
        return training_series.copy()

    months_requested = int(history) * 12

    return training_series.iloc[
        -months_requested:
    ].copy()


# ------------------------------------------------------------
# 7. Fit and forecast one SARIMA specification
# ------------------------------------------------------------

def sarima_forecast(
    training_series,
    forecast_dates,
    order,
    seasonal_order
):
    """
    Fit one SARIMA model and return its forecast and diagnostic
    information.
    """

    start_time = time.perf_counter()

    with warnings.catch_warnings(record=True) as caught:
        warnings.simplefilter("always")

        model = SARIMAX(
            training_series,
            order=order,
            seasonal_order=seasonal_order,
            enforce_stationarity=False,
            enforce_invertibility=False
        )

        fitted = model.fit(
            disp=False,
            maxiter=300
        )

    runtime = time.perf_counter() - start_time

    converged = bool(
        fitted.mle_retvals.get("converged", False)
    )

    prediction = (
        fitted
        .get_forecast(steps=len(forecast_dates))
        .predicted_mean
    )

    prediction.index = forecast_dates

    information = {
        "converged": converged,
        "warning_count": len(caught),
        "warning_messages": " | ".join(
            str(item.message) for item in caught
        ),
        "AIC": fitted.aic,
        "BIC": fitted.bic,
        "runtime_seconds": runtime
    }

    return prediction.astype(float), information


# ------------------------------------------------------------
# 8. Error metrics
# ------------------------------------------------------------

def calculate_metrics(actual, prediction):
    """Calculate RMSE, MAE and forecast-minus-actual bias."""

    error = prediction - actual

    return {
        "RMSE_F": np.sqrt(np.mean(error ** 2)),
        "MAE_F": np.mean(np.abs(error)),
        "bias_F": np.mean(error)
    }


# ------------------------------------------------------------
# 9. Build three inner validation origins
#
# Example:
# Outer origin July 2023
# Inner validation origins:
#   July 2020, July 2021 and July 2022
#
# All inner validation outcomes occur before July 2023.
# ------------------------------------------------------------

def get_inner_origins(outer_origin):
    return [
        outer_origin - pd.DateOffset(years=years_back)
        for years_back in range(INNER_FOLDS, 0, -1)
    ]


# ============================================================
# PART E: NESTED ROLLING BACKTEST
# ============================================================


# ------------------------------------------------------------
# 10. Run parameter selection inside every outer window
# ------------------------------------------------------------

inner_results = []
outer_results = []
outer_monthly_predictions = []
selected_models = []

print("\nNESTED SARIMA TUNING")
print("====================")

for outer_number, outer_origin in enumerate(
    OUTER_ORIGINS,
    start=1
):

    outer_test_months = pd.date_range(
        start=outer_origin,
        periods=FORECAST_HORIZON,
        freq="MS"
    )

    outer_test_end = outer_test_months[-1]

    outer_train_end = (
        outer_origin - pd.offsets.MonthBegin(1)
    )

    outer_train_all = series.loc[
        :outer_train_end
    ].copy()

    outer_actual = series.reindex(
        outer_test_months
    )

    print(
        f"\nOUTER WINDOW {outer_number}/"
        f"{len(OUTER_ORIGINS)}"
    )
    print(
        f"Test: {outer_origin:%Y-%m} through "
        f"{outer_test_end:%Y-%m}"
    )
    print(
        f"Outer training ends: "
        f"{outer_train_end:%Y-%m}"
    )

    # --------------------------------------------------------
    # INNER MODEL SELECTION
    # --------------------------------------------------------

    inner_origins = get_inner_origins(
        outer_origin
    )

    configuration_summaries = []

    for candidate in SARIMA_CANDIDATES:

        for history in TRAINING_HISTORIES:

            fold_metrics = []
            fold_failed = False

            for inner_origin in inner_origins:

                inner_test_months = pd.date_range(
                    start=inner_origin,
                    periods=FORECAST_HORIZON,
                    freq="MS"
                )

                inner_train_end = (
                    inner_origin
                    - pd.offsets.MonthBegin(1)
                )

                inner_train_all = series.loc[
                    :inner_train_end
                ].copy()

                inner_train = select_training_history(
                    inner_train_all,
                    history
                )

                inner_actual = series.reindex(
                    inner_test_months
                )

                result_row = {
                    "outer_test_start": outer_origin,
                    "inner_test_start": inner_origin,
                    "inner_test_end":
                        inner_test_months[-1],
                    "candidate":
                        candidate["candidate"],
                    "order": str(
                        candidate["order"]
                    ),
                    "seasonal_order": str(
                        candidate["seasonal_order"]
                    ),
                    "history_years": history,
                    "training_months":
                        len(inner_train)
                }

                try:
                    prediction, model_info = (
                        sarima_forecast(
                            inner_train,
                            inner_test_months,
                            candidate["order"],
                            candidate["seasonal_order"]
                        )
                    )

                    metrics = calculate_metrics(
                        inner_actual,
                        prediction
                    )

                    result_row.update({
                        "status": (
                            "completed"
                            if model_info["converged"]
                            else "not_converged"
                        ),
                        "converged":
                            model_info["converged"],
                        "RMSE_F":
                            metrics["RMSE_F"],
                        "MAE_F":
                            metrics["MAE_F"],
                        "bias_F":
                            metrics["bias_F"],
                        "warning_count":
                            model_info["warning_count"],
                        "runtime_seconds":
                            model_info["runtime_seconds"]
                    })

                    inner_results.append(
                        result_row
                    )

                    # Nonconverged models are not eligible.
                    if not model_info["converged"]:
                        fold_failed = True
                    else:
                        fold_metrics.append(metrics)

                except Exception as error_message:

                    fold_failed = True

                    result_row.update({
                        "status": "failed",
                        "converged": False,
                        "RMSE_F": np.nan,
                        "MAE_F": np.nan,
                        "bias_F": np.nan,
                        "warning_count": np.nan,
                        "runtime_seconds": np.nan,
                        "error_message":
                            str(error_message)
                    })

                    inner_results.append(
                        result_row
                    )

            # A configuration is eligible only if all inner
            # folds completed and converged.
            if (
                fold_failed
                or len(fold_metrics) != INNER_FOLDS
            ):
                configuration_summaries.append({
                    "candidate":
                        candidate["candidate"],
                    "order":
                        candidate["order"],
                    "seasonal_order":
                        candidate["seasonal_order"],
                    "history_years": history,
                    "inner_folds_completed":
                        len(fold_metrics),
                    "mean_inner_RMSE_F": np.inf,
                    "inner_RMSE_std_F": np.inf,
                    "mean_inner_MAE_F": np.inf,
                    "mean_inner_bias_F": np.nan,
                    "selection_score": np.inf
                })

                continue

            fold_rmse = np.array([
                item["RMSE_F"]
                for item in fold_metrics
            ])

            fold_mae = np.array([
                item["MAE_F"]
                for item in fold_metrics
            ])

            fold_bias = np.array([
                item["bias_F"]
                for item in fold_metrics
            ])

            mean_inner_rmse = fold_rmse.mean()
            inner_rmse_std = fold_rmse.std(ddof=1)

            # Lower is better.
            #
            # Mean RMSE is the main objective.
            # Adding half the RMSE standard deviation favors
            # configurations that are also stable across folds.
            selection_score = (
                mean_inner_rmse
                + 0.5 * inner_rmse_std
            )

            configuration_summaries.append({
                "candidate":
                    candidate["candidate"],
                "order":
                    candidate["order"],
                "seasonal_order":
                    candidate["seasonal_order"],
                "history_years": history,
                "inner_folds_completed":
                    INNER_FOLDS,
                "mean_inner_RMSE_F":
                    mean_inner_rmse,
                "inner_RMSE_std_F":
                    inner_rmse_std,
                "mean_inner_MAE_F":
                    fold_mae.mean(),
                "mean_inner_bias_F":
                    fold_bias.mean(),
                "selection_score":
                    selection_score
            })

    configuration_summaries = pd.DataFrame(
        configuration_summaries
    )

    eligible_configurations = (
        configuration_summaries
        .replace([np.inf, -np.inf], np.nan)
        .dropna(subset=["selection_score"])
        .sort_values(
            [
                "selection_score",
                "mean_inner_RMSE_F",
                "inner_RMSE_std_F"
            ]
        )
    )

    if eligible_configurations.empty:
        raise RuntimeError(
            f"No SARIMA candidate completed all inner folds "
            f"for outer origin {outer_origin:%Y-%m}."
        )

    selected = eligible_configurations.iloc[0]

    selected_order = tuple(
        selected["order"]
    )

    selected_seasonal_order = tuple(
        selected["seasonal_order"]
    )

    selected_history = (
        selected["history_years"]
    )

    selected_models.append({
        "outer_test_start": outer_origin,
        "outer_test_end": outer_test_end,
        "selected_candidate":
            selected["candidate"],
        "selected_order":
            str(selected_order),
        "selected_seasonal_order":
            str(selected_seasonal_order),
        "selected_history_years":
            selected_history,
        "inner_mean_RMSE_F":
            selected["mean_inner_RMSE_F"],
        "inner_RMSE_std_F":
            selected["inner_RMSE_std_F"],
        "inner_selection_score":
            selected["selection_score"]
    })

    print(
        "Selected: "
        f"{selected['candidate']}, "
        f"history={selected_history}, "
        f"inner mean RMSE="
        f"{selected['mean_inner_RMSE_F']:.3f}, "
        f"score={selected['selection_score']:.3f}"
    )

    # --------------------------------------------------------
    # OUTER FORECAST: LOCKED BENCHMARK
    # --------------------------------------------------------

    benchmark_prediction, benchmark_info = (
        sarima_forecast(
            outer_train_all,
            outer_test_months,
            BENCHMARK["order"],
            BENCHMARK["seasonal_order"]
        )
    )

    benchmark_metrics = calculate_metrics(
        outer_actual,
        benchmark_prediction
    )

    outer_results.append({
        "test_start": outer_origin,
        "test_end": outer_test_end,
        "model": "locked_benchmark",
        "candidate":
            BENCHMARK["model_name"],
        "order":
            str(BENCHMARK["order"]),
        "seasonal_order":
            str(BENCHMARK["seasonal_order"]),
        "history_years": "all",
        "training_months":
            len(outer_train_all),
        "converged":
            benchmark_info["converged"],
        "RMSE_F":
            benchmark_metrics["RMSE_F"],
        "MAE_F":
            benchmark_metrics["MAE_F"],
        "bias_F":
            benchmark_metrics["bias_F"],
        "runtime_seconds":
            benchmark_info["runtime_seconds"]
    })

    # --------------------------------------------------------
    # OUTER FORECAST: NESTED-TUNED CHALLENGER
    # --------------------------------------------------------

    tuned_outer_train = select_training_history(
        outer_train_all,
        selected_history
    )

    tuned_prediction, tuned_info = (
        sarima_forecast(
            tuned_outer_train,
            outer_test_months,
            selected_order,
            selected_seasonal_order
        )
    )

    tuned_metrics = calculate_metrics(
        outer_actual,
        tuned_prediction
    )

    outer_results.append({
        "test_start": outer_origin,
        "test_end": outer_test_end,
        "model": "nested_tuned_sarima",
        "candidate":
            selected["candidate"],
        "order":
            str(selected_order),
        "seasonal_order":
            str(selected_seasonal_order),
        "history_years":
            selected_history,
        "training_months":
            len(tuned_outer_train),
        "converged":
            tuned_info["converged"],
        "RMSE_F":
            tuned_metrics["RMSE_F"],
        "MAE_F":
            tuned_metrics["MAE_F"],
        "bias_F":
            tuned_metrics["bias_F"],
        "runtime_seconds":
            tuned_info["runtime_seconds"]
    })

    print(
        f"Outer benchmark: RMSE="
        f"{benchmark_metrics['RMSE_F']:.3f}"
    )

    print(
        f"Outer tuned:     RMSE="
        f"{tuned_metrics['RMSE_F']:.3f}"
    )

    # Save the monthly forecasts from both models.
    for horizon, month in enumerate(
        outer_test_months,
        start=1
    ):
        outer_monthly_predictions.append({
            "test_start": outer_origin,
            "month": month,
            "forecast_horizon": horizon,
            "actual_F": outer_actual.loc[month],
            "benchmark_forecast_F":
                benchmark_prediction.loc[month],
            "benchmark_error_F":
                benchmark_prediction.loc[month]
                - outer_actual.loc[month],
            "tuned_forecast_F":
                tuned_prediction.loc[month],
            "tuned_error_F":
                tuned_prediction.loc[month]
                - outer_actual.loc[month]
        })

    # Save checkpoints after every outer window.
    pd.DataFrame(inner_results).to_csv(
        OUTPUT_DIR / "inner_validation_results_checkpoint.csv",
        index=False
    )

    pd.DataFrame(outer_results).to_csv(
        OUTPUT_DIR / "outer_results_checkpoint.csv",
        index=False
    )

    pd.DataFrame(selected_models).to_csv(
        OUTPUT_DIR / "selected_models_checkpoint.csv",
        index=False
    )


# ============================================================
# PART F: COMPARE TUNED SARIMA WITH THE LOCKED BENCHMARK
# ============================================================


# ------------------------------------------------------------
# 11. Create result tables
# ------------------------------------------------------------

inner_results = pd.DataFrame(
    inner_results
)

outer_results = pd.DataFrame(
    outer_results
)

selected_models = pd.DataFrame(
    selected_models
)

outer_monthly_predictions = pd.DataFrame(
    outer_monthly_predictions
)

print("\nSELECTED MODEL IN EACH OUTER WINDOW")
print("===================================")
display(selected_models.round(3))

print("\nOUTER-WINDOW RESULTS")
print("====================")
display(
    outer_results[
        [
            "test_start",
            "test_end",
            "model",
            "candidate",
            "order",
            "seasonal_order",
            "history_years",
            "training_months",
            "converged",
            "RMSE_F",
            "MAE_F",
            "bias_F"
        ]
    ].round(3)
)


# ------------------------------------------------------------
# 12. Summarize each outer model
# ------------------------------------------------------------

summary_rows = []

for model_name in [
    "locked_benchmark",
    "nested_tuned_sarima"
]:
    model_windows = outer_results.loc[
        outer_results["model"] == model_name
    ].copy()

    if model_name == "locked_benchmark":
        error = (
            outer_monthly_predictions[
                "benchmark_error_F"
            ]
        )
    else:
        error = (
            outer_monthly_predictions[
                "tuned_error_F"
            ]
        )

    summary_rows.append({
        "model": model_name,
        "windows_completed":
            len(model_windows),
        "windows_converged":
            int(model_windows["converged"].sum()),
        "mean_RMSE_F":
            model_windows["RMSE_F"].mean(),
        "median_RMSE_F":
            model_windows["RMSE_F"].median(),
        "RMSE_std_F":
            model_windows["RMSE_F"].std(),
        "best_window_RMSE_F":
            model_windows["RMSE_F"].min(),
        "worst_window_RMSE_F":
            model_windows["RMSE_F"].max(),
        "mean_MAE_F":
            model_windows["MAE_F"].mean(),
        "pooled_RMSE_F":
            np.sqrt(np.mean(error ** 2)),
        "pooled_MAE_F":
            np.mean(np.abs(error)),
        "pooled_bias_F":
            np.mean(error)
    })

comparison_summary = pd.DataFrame(
    summary_rows
)

print("\nBENCHMARK VS NESTED-TUNED SARIMA")
print("================================")
display(comparison_summary.round(3))


# ------------------------------------------------------------
# 13. Matched window comparison
# ------------------------------------------------------------

matched_windows = (
    outer_results
    .pivot(
        index=["test_start", "test_end"],
        columns="model",
        values="RMSE_F"
    )
    .reset_index()
)

matched_windows["RMSE_improvement_F"] = (
    matched_windows["locked_benchmark"]
    - matched_windows["nested_tuned_sarima"]
)

matched_windows["RMSE_improvement_percent"] = (
    100
    * matched_windows["RMSE_improvement_F"]
    / matched_windows["locked_benchmark"]
)

matched_windows["winner"] = np.where(
    matched_windows["nested_tuned_sarima"]
    < matched_windows["locked_benchmark"],
    "nested_tuned_sarima",
    np.where(
        matched_windows["nested_tuned_sarima"]
        > matched_windows["locked_benchmark"],
        "locked_benchmark",
        "tie"
    )
)

print("\nMATCHED WINDOW COMPARISON")
print("=========================")
display(matched_windows.round(3))


# ------------------------------------------------------------
# 14. Automatic acceptance checks
# ------------------------------------------------------------

benchmark_summary = comparison_summary.loc[
    comparison_summary["model"]
    == "locked_benchmark"
].iloc[0]

tuned_summary = comparison_summary.loc[
    comparison_summary["model"]
    == "nested_tuned_sarima"
].iloc[0]

acceptance_checks = pd.DataFrame({
    "requirement": [
        "Lower pooled RMSE",
        "Lower mean RMSE",
        "No higher RMSE variability",
        "No worse worst-window RMSE",
        "Absolute pooled bias below 0.25°F",
        "All nine outer windows converged"
    ],

    "benchmark": [
        benchmark_summary["pooled_RMSE_F"],
        benchmark_summary["mean_RMSE_F"],
        benchmark_summary["RMSE_std_F"],
        benchmark_summary["worst_window_RMSE_F"],
        abs(benchmark_summary["pooled_bias_F"]),
        benchmark_summary["windows_converged"]
    ],

    "challenger": [
        tuned_summary["pooled_RMSE_F"],
        tuned_summary["mean_RMSE_F"],
        tuned_summary["RMSE_std_F"],
        tuned_summary["worst_window_RMSE_F"],
        abs(tuned_summary["pooled_bias_F"]),
        tuned_summary["windows_converged"]
    ],

    "passed": [
        tuned_summary["pooled_RMSE_F"]
        < benchmark_summary["pooled_RMSE_F"],

        tuned_summary["mean_RMSE_F"]
        < benchmark_summary["mean_RMSE_F"],

        tuned_summary["RMSE_std_F"]
        <= benchmark_summary["RMSE_std_F"],

        tuned_summary["worst_window_RMSE_F"]
        <= benchmark_summary["worst_window_RMSE_F"],

        abs(tuned_summary["pooled_bias_F"]) < 0.25,

        tuned_summary["windows_converged"] == 9
    ]
})

print("\nACCEPTANCE CHECKS")
print("=================")
display(acceptance_checks.round(3))

all_checks_passed = bool(
    acceptance_checks["passed"].all()
)

pooled_improvement_percent = (
    100
    * (
        benchmark_summary["pooled_RMSE_F"]
        - tuned_summary["pooled_RMSE_F"]
    )
    / benchmark_summary["pooled_RMSE_F"]
)

print("\nPRELIMINARY DECISION")
print("====================")
print(
    f"Pooled RMSE improvement: "
    f"{pooled_improvement_percent:+.2f}%"
)

if all_checks_passed:
    print(
        "The nested-tuned SARIMA passed all stability checks."
    )

    if pooled_improvement_percent >= 3:
        print(
            "It also achieved at least a 3% pooled-RMSE "
            "improvement and is a meaningful challenger."
        )
    else:
        print(
            "However, its RMSE improvement is below 3%. "
            "Treat it as a marginal rather than clearly "
            "meaningful improvement."
        )
else:
    print(
        "The nested-tuned SARIMA did not pass every stability "
        "check. Retain the locked SARIMA benchmark for now."
    )


# ============================================================
# PART G: PLOTS
# ============================================================


# ------------------------------------------------------------
# 15. RMSE by outer window
# ------------------------------------------------------------

plot_table = (
    outer_results
    .pivot(
        index="test_start",
        columns="model",
        values="RMSE_F"
    )
    .sort_index()
)

fig, ax = plt.subplots(figsize=(14, 6))

ax.plot(
    plot_table.index,
    plot_table["locked_benchmark"],
    marker="o",
    linewidth=2.5,
    label="Locked SARIMA benchmark"
)

ax.plot(
    plot_table.index,
    plot_table["nested_tuned_sarima"],
    marker="o",
    linewidth=2.5,
    label="Nested-tuned SARIMA"
)

ax.set_title(
    "Locked Benchmark vs Nested-Tuned SARIMA\n"
    "Nine Outer 12-Month Forecast Windows"
)
ax.set_xlabel("Forecast-window start")
ax.set_ylabel("RMSE (°F)")
ax.grid(alpha=0.25)
ax.legend()

plt.tight_layout()

window_plot_file = (
    OUTPUT_DIR
    / "benchmark_vs_tuned_rmse_by_window.png"
)

plt.savefig(
    window_plot_file,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 16. Overall stability comparison
# ------------------------------------------------------------

fig, ax = plt.subplots(figsize=(9, 6))

ax.bar(
    comparison_summary["model"],
    comparison_summary["mean_RMSE_F"],
    yerr=comparison_summary["RMSE_std_F"],
    capsize=7,
    color=["gray", "darkred"],
    alpha=0.85
)

ax.set_title(
    "Mean RMSE and Window-to-Window Variability"
)
ax.set_ylabel(
    "Mean RMSE ± one standard deviation (°F)"
)
ax.grid(axis="y", alpha=0.25)

plt.xticks(rotation=15)
plt.tight_layout()

summary_plot_file = (
    OUTPUT_DIR
    / "benchmark_vs_tuned_stability.png"
)

plt.savefig(
    summary_plot_file,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# PART H: SAVE FINAL OUTPUTS
# ============================================================

inner_results_file = (
    OUTPUT_DIR / "inner_validation_results.csv"
)

selected_models_file = (
    OUTPUT_DIR / "selected_model_by_outer_window.csv"
)

outer_results_file = (
    OUTPUT_DIR / "outer_window_results.csv"
)

monthly_predictions_file = (
    OUTPUT_DIR / "outer_monthly_predictions.csv"
)

comparison_summary_file = (
    OUTPUT_DIR / "benchmark_vs_tuned_summary.csv"
)

matched_windows_file = (
    OUTPUT_DIR / "matched_window_comparison.csv"
)

acceptance_file = (
    OUTPUT_DIR / "acceptance_checks.csv"
)

inner_results.to_csv(
    inner_results_file,
    index=False
)

selected_models.to_csv(
    selected_models_file,
    index=False
)

outer_results.to_csv(
    outer_results_file,
    index=False
)

outer_monthly_predictions.to_csv(
    monthly_predictions_file,
    index=False
)

comparison_summary.to_csv(
    comparison_summary_file,
    index=False
)

matched_windows.to_csv(
    matched_windows_file,
    index=False
)

acceptance_checks.to_csv(
    acceptance_file,
    index=False
)

print("\nFILES SAVED")
print("===========")
print(f"Locked benchmark:     {benchmark_file}")
print(f"Inner validation:     {inner_results_file}")
print(f"Selected models:      {selected_models_file}")
print(f"Outer results:        {outer_results_file}")
print(f"Monthly predictions:  {monthly_predictions_file}")
print(f"Summary comparison:   {comparison_summary_file}")
print(f"Matched windows:      {matched_windows_file}")
print(f"Acceptance checks:    {acceptance_file}")
print(f"Window plot:          {window_plot_file}")
print(f"Stability plot:       {summary_plot_file}")

print("\nSEND ME:")
print("1. SELECTED MODEL IN EACH OUTER WINDOW")
print("2. BENCHMARK VS NESTED-TUNED SARIMA")
print("3. MATCHED WINDOW COMPARISON")
print("4. ACCEPTANCE CHECKS")
print("5. PRELIMINARY DECISION")
```

```python
# ============================================================
# ENSEMBLE CHALLENGER TEST — 12-MONTH FORECASTING
#
# Goal:
# Determine whether combining stable temperature-only models
# improves upon the locked SARIMA benchmark.
#
# This code uses the genuine out-of-sample forecasts created
# in Level 7. It does not refit the existing models.
#
# All ensemble specifications and weights are fixed in advance.
# ============================================================


# ------------------------------------------------------------
# 1. Imports and Drive
# ------------------------------------------------------------

from google.colab import drive
drive.mount("/content/drive")

from pathlib import Path
import json

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display


# ------------------------------------------------------------
# 2. File locations
# ------------------------------------------------------------

PROJECT = Path("/content/drive/MyDrive/Stanford_Weather_12M")

LEVEL_7_DIR = (
    PROJECT
    / "foundational_analysis"
    / "level_7_all_models_all_windows"
)

PREDICTIONS_FILE = (
    LEVEL_7_DIR
    / "all_model_monthly_predictions.csv"
)

MONTHLY_DATA_FILE = (
    LEVEL_7_DIR
    / "monthly_data_used.csv"
)

OUTPUT_DIR = (
    PROJECT
    / "model_improvement"
    / "12m_ensemble_challenger"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

LOCKED_BENCHMARK = "sarima_111x111_12"


# ============================================================
# PART A: LOAD THE EXISTING OUT-OF-SAMPLE PREDICTIONS
# ============================================================


# ------------------------------------------------------------
# 3. Load Level 7 predictions
# ------------------------------------------------------------

predictions_long = pd.read_csv(
    PREDICTIONS_FILE,
    parse_dates=[
        "test_start",
        "test_end",
        "train_end",
        "month"
    ]
)

required_models = [
    "sarima_111x111_12",
    "ets_add_add_damped12",
    "prophet_additive",
    "seasonal_naive_12"
]

available_models = sorted(
    predictions_long["model"].unique()
)

print("AVAILABLE OUT-OF-SAMPLE MODELS")
print("==============================")
print(available_models)

missing_models = sorted(
    set(required_models) - set(available_models)
)

if missing_models:
    raise ValueError(
        f"Required models are missing: {missing_models}"
    )


# ------------------------------------------------------------
# 4. Convert from long format to one forecast column per model
# ------------------------------------------------------------

index_columns = [
    "test_start",
    "test_end",
    "train_end",
    "forecast_horizon",
    "month",
    "actual_F"
]

forecast_table = (
    predictions_long
    .pivot_table(
        index=index_columns,
        columns="model",
        values="forecast_F",
        aggfunc="first"
    )
    .reset_index()
)

forecast_table.columns.name = None

forecast_table = forecast_table.sort_values(
    ["test_start", "forecast_horizon"]
).reset_index(drop=True)

print("\nFORECAST DATA CHECK")
print("===================")
print(
    f"First window: "
    f"{forecast_table['test_start'].min():%Y-%m}"
)
print(
    f"Last window:  "
    f"{forecast_table['test_start'].max():%Y-%m}"
)
print(
    f"Historical windows: "
    f"{forecast_table['test_start'].nunique()}"
)
print(f"Monthly predictions: {len(forecast_table)}")
print(
    "Missing model predictions: "
    f"{forecast_table[required_models].isna().sum().sum()}"
)

assert forecast_table["test_start"].nunique() == 9
assert len(forecast_table) == 108
assert forecast_table[required_models].isna().sum().sum() == 0


# ============================================================
# PART B: CREATE A 15-YEAR CLIMATOLOGY FORECAST
# ============================================================


# ------------------------------------------------------------
# 5. Load the monthly temperature series
# ------------------------------------------------------------

monthly_data = pd.read_csv(
    MONTHLY_DATA_FILE,
    parse_dates=["month"]
)

series = (
    monthly_data
    .set_index("month")["mean_temp_F"]
    .astype(float)
    .sort_index()
    .asfreq("MS")
)

assert series.index.min() == pd.Timestamp("1980-01-01")
assert series.index.max() == pd.Timestamp("2026-06-01")
assert series.isna().sum() == 0


# ------------------------------------------------------------
# 6. Forecast each month using the average of the preceding
#    15 observations for that calendar month
#
# Example:
# To forecast July 2023, average the July temperatures in the
# training data from July 2008 through July 2022.
# ------------------------------------------------------------

def climatology_15y_forecast(
    training_series,
    forecast_dates
):
    predictions = []

    for forecast_date in forecast_dates:

        same_calendar_month = training_series.loc[
            training_series.index.month
            == forecast_date.month
        ]

        recent_same_month = same_calendar_month.iloc[
            -15:
        ]

        if len(recent_same_month) < 15:
            raise ValueError(
                f"Fewer than 15 historical values are "
                f"available for month {forecast_date.month}."
            )

        predictions.append(
            recent_same_month.mean()
        )

    return pd.Series(
        predictions,
        index=forecast_dates,
        dtype=float
    )


forecast_table["climatology_15y"] = np.nan

for test_start, window_data in forecast_table.groupby(
    "test_start"
):

    train_end = (
        test_start - pd.offsets.MonthBegin(1)
    )

    training_series = series.loc[
        :train_end
    ]

    forecast_dates = pd.DatetimeIndex(
        window_data["month"]
    )

    climatology_forecast = (
        climatology_15y_forecast(
            training_series,
            forecast_dates
        )
    )

    forecast_table.loc[
        window_data.index,
        "climatology_15y"
    ] = climatology_forecast.to_numpy()


# ============================================================
# PART C: DEFINE THE ENSEMBLES
# ============================================================


# ------------------------------------------------------------
# 7. Short aliases for readability
# ------------------------------------------------------------

sarima = forecast_table["sarima_111x111_12"]
ets = forecast_table["ets_add_add_damped12"]
prophet = forecast_table["prophet_additive"]
climatology = forecast_table["climatology_15y"]


# ------------------------------------------------------------
# 8. Predefined fixed-weight ensembles
#
# The weights do not change between historical windows.
# ------------------------------------------------------------

# Equal combination of SARIMA and ETS.
forecast_table["ensemble_sarima_ets_50_50"] = (
    0.50 * sarima
    + 0.50 * ets
)

# Equal combination of SARIMA and Prophet.
forecast_table["ensemble_sarima_prophet_50_50"] = (
    0.50 * sarima
    + 0.50 * prophet
)

# Equal combination of SARIMA and recent climatology.
forecast_table["ensemble_sarima_climatology_50_50"] = (
    0.50 * sarima
    + 0.50 * climatology
)

# SARIMA remains dominant; ETS provides adjustment.
forecast_table["ensemble_sarima_ets_75_25"] = (
    0.75 * sarima
    + 0.25 * ets
)

# SARIMA remains dominant; climatology provides stabilization.
forecast_table["ensemble_sarima_climatology_75_25"] = (
    0.75 * sarima
    + 0.25 * climatology
)

# Equal average of the three advanced models.
forecast_table["ensemble_three_model_equal"] = (
    sarima + ets + prophet
) / 3

# SARIMA-dominant three-model ensemble.
forecast_table["ensemble_sarima_dominant"] = (
    0.60 * sarima
    + 0.20 * ets
    + 0.20 * prophet
)

# Median ensemble: middle forecast among SARIMA, ETS and Prophet.
forecast_table["ensemble_three_model_median"] = (
    forecast_table[
        [
            "sarima_111x111_12",
            "ets_add_add_damped12",
            "prophet_additive"
        ]
    ]
    .median(axis=1)
)


# ------------------------------------------------------------
# 9. Models to evaluate
# ------------------------------------------------------------

BASE_MODELS = [
    "sarima_111x111_12",
    "ets_add_add_damped12",
    "prophet_additive",
    "seasonal_naive_12",
    "climatology_15y"
]

ENSEMBLE_MODELS = [
    "ensemble_sarima_ets_50_50",
    "ensemble_sarima_prophet_50_50",
    "ensemble_sarima_climatology_50_50",
    "ensemble_sarima_ets_75_25",
    "ensemble_sarima_climatology_75_25",
    "ensemble_three_model_equal",
    "ensemble_sarima_dominant",
    "ensemble_three_model_median"
]

ALL_MODELS = BASE_MODELS + ENSEMBLE_MODELS


# ============================================================
# PART D: EVALUATE EVERY MODEL IN EVERY WINDOW
# ============================================================


# ------------------------------------------------------------
# 10. Calculate window-level metrics
# ------------------------------------------------------------

window_results = []

for test_start, window_data in forecast_table.groupby(
    "test_start"
):

    test_end = window_data["test_end"].iloc[0]
    actual = window_data["actual_F"]

    for model_name in ALL_MODELS:

        prediction = window_data[model_name]
        error = prediction - actual

        window_results.append({
            "test_start": test_start,
            "test_end": test_end,
            "model": model_name,
            "RMSE_F": np.sqrt(
                np.mean(error ** 2)
            ),
            "MAE_F": np.mean(
                np.abs(error)
            ),
            "bias_F": np.mean(error),
            "maximum_absolute_error_F":
                np.abs(error).max()
        })

window_results = pd.DataFrame(
    window_results
)


# ------------------------------------------------------------
# 11. Calculate stability summary
# ------------------------------------------------------------

summary_rows = []

for model_name in ALL_MODELS:

    model_windows = window_results.loc[
        window_results["model"] == model_name
    ].copy()

    pooled_error = (
        forecast_table[model_name]
        - forecast_table["actual_F"]
    )

    summary_rows.append({
        "model": model_name,
        "model_type": (
            "ensemble"
            if model_name in ENSEMBLE_MODELS
            else "individual"
        ),
        "windows_completed":
            len(model_windows),
        "months_evaluated":
            len(pooled_error),
        "mean_RMSE_F":
            model_windows["RMSE_F"].mean(),
        "median_RMSE_F":
            model_windows["RMSE_F"].median(),
        "RMSE_std_F":
            model_windows["RMSE_F"].std(),
        "best_window_RMSE_F":
            model_windows["RMSE_F"].min(),
        "worst_window_RMSE_F":
            model_windows["RMSE_F"].max(),
        "mean_MAE_F":
            model_windows["MAE_F"].mean(),
        "worst_absolute_bias_F":
            model_windows["bias_F"].abs().max(),
        "pooled_RMSE_F":
            np.sqrt(np.mean(pooled_error ** 2)),
        "pooled_MAE_F":
            np.mean(np.abs(pooled_error)),
        "pooled_bias_F":
            np.mean(pooled_error)
    })

stability_summary = pd.DataFrame(
    summary_rows
)

stability_summary["robustness_indicator"] = (
    stability_summary["mean_RMSE_F"]
    + stability_summary["RMSE_std_F"]
)

stability_summary = (
    stability_summary
    .sort_values(
        [
            "pooled_RMSE_F",
            "RMSE_std_F"
        ]
    )
    .reset_index(drop=True)
)

print("FULL ENSEMBLE STABILITY SUMMARY")
print("===============================")
display(stability_summary.round(3))


# ============================================================
# PART E: FAIR COMPARISON AGAINST LOCKED SARIMA
# ============================================================


# ------------------------------------------------------------
# 12. Compare every ensemble with SARIMA on matched windows
# ------------------------------------------------------------

benchmark_windows = (
    window_results.loc[
        window_results["model"]
        == LOCKED_BENCHMARK,
        [
            "test_start",
            "RMSE_F",
            "MAE_F",
            "bias_F"
        ]
    ]
    .rename(
        columns={
            "RMSE_F": "benchmark_RMSE_F",
            "MAE_F": "benchmark_MAE_F",
            "bias_F": "benchmark_bias_F"
        }
    )
)

matched_rows = []

for ensemble_name in ENSEMBLE_MODELS:

    ensemble_windows = (
        window_results.loc[
            window_results["model"]
            == ensemble_name,
            [
                "test_start",
                "RMSE_F",
                "MAE_F",
                "bias_F"
            ]
        ]
        .rename(
            columns={
                "RMSE_F": "challenger_RMSE_F",
                "MAE_F": "challenger_MAE_F",
                "bias_F": "challenger_bias_F"
            }
        )
    )

    matched = benchmark_windows.merge(
        ensemble_windows,
        on="test_start",
        how="inner"
    )

    benchmark_error = (
        forecast_table[LOCKED_BENCHMARK]
        - forecast_table["actual_F"]
    )

    challenger_error = (
        forecast_table[ensemble_name]
        - forecast_table["actual_F"]
    )

    benchmark_pooled_rmse = np.sqrt(
        np.mean(benchmark_error ** 2)
    )

    challenger_pooled_rmse = np.sqrt(
        np.mean(challenger_error ** 2)
    )

    improvement_F = (
        benchmark_pooled_rmse
        - challenger_pooled_rmse
    )

    improvement_percent = (
        100
        * improvement_F
        / benchmark_pooled_rmse
    )

    matched_rows.append({
        "challenger": ensemble_name,
        "matched_windows": len(matched),
        "matched_months":
            len(challenger_error),
        "benchmark_pooled_RMSE_F":
            benchmark_pooled_rmse,
        "challenger_pooled_RMSE_F":
            challenger_pooled_rmse,
        "RMSE_improvement_F":
            improvement_F,
        "RMSE_improvement_percent":
            improvement_percent,
        "benchmark_mean_window_RMSE_F":
            matched["benchmark_RMSE_F"].mean(),
        "challenger_mean_window_RMSE_F":
            matched["challenger_RMSE_F"].mean(),
        "benchmark_RMSE_std_F":
            matched["benchmark_RMSE_F"].std(),
        "challenger_RMSE_std_F":
            matched["challenger_RMSE_F"].std(),
        "benchmark_worst_window_RMSE_F":
            matched["benchmark_RMSE_F"].max(),
        "challenger_worst_window_RMSE_F":
            matched["challenger_RMSE_F"].max(),
        "benchmark_pooled_bias_F":
            benchmark_error.mean(),
        "challenger_pooled_bias_F":
            challenger_error.mean(),
        "challenger_window_wins":
            int(
                (
                    matched["challenger_RMSE_F"]
                    < matched["benchmark_RMSE_F"]
                ).sum()
            ),
        "benchmark_window_wins":
            int(
                (
                    matched["benchmark_RMSE_F"]
                    < matched["challenger_RMSE_F"]
                ).sum()
            )
    })

matched_summary = pd.DataFrame(
    matched_rows
)

# Stability checks for each ensemble.
matched_summary["lower_pooled_RMSE"] = (
    matched_summary["challenger_pooled_RMSE_F"]
    < matched_summary["benchmark_pooled_RMSE_F"]
)

matched_summary["lower_mean_RMSE"] = (
    matched_summary["challenger_mean_window_RMSE_F"]
    < matched_summary["benchmark_mean_window_RMSE_F"]
)

matched_summary["no_higher_variability"] = (
    matched_summary["challenger_RMSE_std_F"]
    <= matched_summary["benchmark_RMSE_std_F"]
)

matched_summary["no_worse_worst_window"] = (
    matched_summary["challenger_worst_window_RMSE_F"]
    <= matched_summary["benchmark_worst_window_RMSE_F"]
)

matched_summary["acceptable_bias"] = (
    matched_summary[
        "challenger_pooled_bias_F"
    ].abs() < 0.25
)

matched_summary["passes_all_stability_checks"] = (
    matched_summary[
        [
            "lower_pooled_RMSE",
            "lower_mean_RMSE",
            "no_higher_variability",
            "no_worse_worst_window",
            "acceptable_bias"
        ]
    ]
    .all(axis=1)
)

matched_summary = (
    matched_summary
    .sort_values(
        "challenger_pooled_RMSE_F"
    )
    .reset_index(drop=True)
)

print("\nENSEMBLES VS LOCKED SARIMA")
print("==========================")
display(matched_summary.round(3))


# ------------------------------------------------------------
# 13. Winner in each historical window
# ------------------------------------------------------------

ensemble_and_benchmark = (
    window_results.loc[
        window_results["model"].isin(
            [LOCKED_BENCHMARK] + ENSEMBLE_MODELS
        )
    ]
)

window_winners = (
    ensemble_and_benchmark
    .sort_values(["test_start", "RMSE_F"])
    .groupby("test_start", as_index=False)
    .first()
)

window_winners = window_winners[
    [
        "test_start",
        "test_end",
        "model",
        "RMSE_F",
        "MAE_F",
        "bias_F"
    ]
].rename(
    columns={"model": "winning_model"}
)

print("\nWINNER IN EACH WINDOW")
print("=====================")
display(window_winners.round(3))


# ============================================================
# PART F: PRELIMINARY DECISION
# ============================================================


# ------------------------------------------------------------
# 14. Find the leading ensemble
# ------------------------------------------------------------

passing_ensembles = matched_summary.loc[
    matched_summary[
        "passes_all_stability_checks"
    ]
].copy()

print("\nPRELIMINARY ENSEMBLE DECISION")
print("=============================")

if passing_ensembles.empty:

    print(
        "No ensemble passed every stability requirement."
    )
    print(
        "Retain the locked SARIMA benchmark."
    )

else:
    leading_ensemble = passing_ensembles.iloc[0]

    print(
        f"Leading ensemble: "
        f"{leading_ensemble['challenger']}"
    )

    print(
        f"Pooled RMSE: "
        f"{leading_ensemble['challenger_pooled_RMSE_F']:.3f} °F"
    )

    print(
        f"Improvement over SARIMA: "
        f"{leading_ensemble['RMSE_improvement_percent']:.2f}%"
    )

    if (
        leading_ensemble[
            "RMSE_improvement_percent"
        ] >= 3
    ):
        print(
            "The ensemble achieved at least a 3% RMSE "
            "improvement while passing the stability checks."
        )
    else:
        print(
            "The ensemble passed the stability checks, but "
            "the RMSE improvement is below 3% and should be "
            "treated as marginal."
        )


# ============================================================
# PART G: VISUALIZATIONS
# ============================================================


# ------------------------------------------------------------
# 15. Plot RMSE across windows for SARIMA and leading ensembles
# ------------------------------------------------------------

top_ensemble_names = (
    matched_summary
    .head(3)["challenger"]
    .tolist()
)

models_to_plot = (
    [LOCKED_BENCHMARK]
    + top_ensemble_names
)

plot_data = window_results.loc[
    window_results["model"].isin(models_to_plot)
].copy()

plot_data["window_label"] = (
    plot_data["test_start"].dt.strftime("%Y-%m")
)

fig, ax = plt.subplots(figsize=(14, 6))

for model_name in models_to_plot:

    model_data = plot_data.loc[
        plot_data["model"] == model_name
    ].sort_values("test_start")

    ax.plot(
        model_data["window_label"],
        model_data["RMSE_F"],
        marker="o",
        linewidth=2.2,
        label=model_name
    )

ax.set_title(
    "Top Ensemble Challengers vs Locked SARIMA\n"
    "Nine Historical 12-Month Forecast Windows"
)
ax.set_xlabel("Forecast-window start")
ax.set_ylabel("RMSE (°F)")
ax.grid(alpha=0.25)
ax.legend()

plt.xticks(rotation=35, ha="right")
plt.tight_layout()

window_plot_file = (
    OUTPUT_DIR
    / "top_ensembles_rmse_by_window.png"
)

plt.savefig(
    window_plot_file,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 16. Plot pooled RMSE for all ensembles
# ------------------------------------------------------------

bar_data = matched_summary.sort_values(
    "challenger_pooled_RMSE_F"
)

fig, ax = plt.subplots(figsize=(13, 6))

ax.barh(
    bar_data["challenger"],
    bar_data["challenger_pooled_RMSE_F"],
    color="steelblue"
)

benchmark_rmse = (
    bar_data["benchmark_pooled_RMSE_F"].iloc[0]
)

ax.axvline(
    benchmark_rmse,
    color="darkred",
    linestyle="--",
    linewidth=2,
    label=f"Locked SARIMA = {benchmark_rmse:.3f}°F"
)

ax.set_title(
    "Pooled RMSE of Ensemble Challengers"
)
ax.set_xlabel("Pooled RMSE (°F)")
ax.set_ylabel("Ensemble")
ax.grid(axis="x", alpha=0.25)
ax.legend()

plt.tight_layout()

pooled_plot_file = (
    OUTPUT_DIR
    / "ensemble_pooled_rmse_comparison.png"
)

plt.savefig(
    pooled_plot_file,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# PART H: SAVE OUTPUTS
# ============================================================


# ------------------------------------------------------------
# 17. Save ensemble definitions
# ------------------------------------------------------------

ensemble_definitions = {
    "ensemble_sarima_ets_50_50": {
        "sarima_111x111_12": 0.50,
        "ets_add_add_damped12": 0.50
    },
    "ensemble_sarima_prophet_50_50": {
        "sarima_111x111_12": 0.50,
        "prophet_additive": 0.50
    },
    "ensemble_sarima_climatology_50_50": {
        "sarima_111x111_12": 0.50,
        "climatology_15y": 0.50
    },
    "ensemble_sarima_ets_75_25": {
        "sarima_111x111_12": 0.75,
        "ets_add_add_damped12": 0.25
    },
    "ensemble_sarima_climatology_75_25": {
        "sarima_111x111_12": 0.75,
        "climatology_15y": 0.25
    },
    "ensemble_three_model_equal": {
        "sarima_111x111_12": 1 / 3,
        "ets_add_add_damped12": 1 / 3,
        "prophet_additive": 1 / 3
    },
    "ensemble_sarima_dominant": {
        "sarima_111x111_12": 0.60,
        "ets_add_add_damped12": 0.20,
        "prophet_additive": 0.20
    },
    "ensemble_three_model_median": {
        "method": "row-wise median",
        "models": [
            "sarima_111x111_12",
            "ets_add_add_damped12",
            "prophet_additive"
        ]
    }
}

definitions_file = (
    OUTPUT_DIR
    / "ensemble_definitions.json"
)

with open(definitions_file, "w") as file:
    json.dump(
        ensemble_definitions,
        file,
        indent=4
    )


# ------------------------------------------------------------
# 18. Save tables
# ------------------------------------------------------------

forecast_file = (
    OUTPUT_DIR
    / "ensemble_monthly_predictions.csv"
)

window_results_file = (
    OUTPUT_DIR
    / "ensemble_window_metrics.csv"
)

stability_file = (
    OUTPUT_DIR
    / "full_ensemble_stability_summary.csv"
)

matched_file = (
    OUTPUT_DIR
    / "ensembles_vs_locked_sarima.csv"
)

winners_file = (
    OUTPUT_DIR
    / "ensemble_window_winners.csv"
)

forecast_table.to_csv(
    forecast_file,
    index=False
)

window_results.to_csv(
    window_results_file,
    index=False
)

stability_summary.to_csv(
    stability_file,
    index=False
)

matched_summary.to_csv(
    matched_file,
    index=False
)

window_winners.to_csv(
    winners_file,
    index=False
)

print("\nFILES SAVED")
print("===========")
print(f"Ensemble definitions: {definitions_file}")
print(f"Monthly predictions:  {forecast_file}")
print(f"Window metrics:       {window_results_file}")
print(f"Stability summary:    {stability_file}")
print(f"Matched comparison:   {matched_file}")
print(f"Window winners:       {winners_file}")
print(f"Window RMSE plot:      {window_plot_file}")
print(f"Pooled RMSE plot:      {pooled_plot_file}")

print("\nSEND ME:")
print("1. FULL ENSEMBLE STABILITY SUMMARY")
print("2. ENSEMBLES VS LOCKED SARIMA")
print("3. WINNER IN EACH WINDOW")
print("4. PRELIMINARY ENSEMBLE DECISION")
print("5. Both plots")
```

```python
# ============================================================
# FINAL INDEPENDENT VALIDATION:
# LOCKED SARIMA VS FIXED 75/25 ENSEMBLE
#
# Validation windows:
# July 2008–June 2009 through July 2016–June 2017
#
# IMPORTANT:
# These earlier windows were not used to select the 75/25 weight.
# ============================================================

from google.colab import drive
drive.mount("/content/drive")

import json
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display
from sklearn.metrics import mean_absolute_error, mean_squared_error
from statsmodels.tsa.statespace.sarimax import SARIMAX


# ============================================================
# 1. PATHS AND LOCKED SETTINGS
# ============================================================

PROJECT = Path("/content/drive/MyDrive/Stanford_Weather_12M")

DATA_FILE = (
    PROJECT
    / "foundational_analysis"
    / "level_7_all_models_all_windows"
    / "monthly_data_used.csv"
)

# Fallback in case the foundational-analysis copy is unavailable
if not DATA_FILE.exists():
    DATA_FILE = (
        PROJECT
        / "fair_12m_stability_comparison"
        / "monthly_data_used.csv"
    )

OUTPUT_DIR = (
    PROJECT
    / "model_improvement"
    / "independent_ensemble_validation"
)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

SARIMA_ORDER = (1, 1, 1)
SARIMA_SEASONAL_ORDER = (1, 1, 1, 12)

CLIMATOLOGY_YEARS = 15

SARIMA_WEIGHT = 0.75
CLIMATOLOGY_WEIGHT = 0.25

FORECAST_MONTHS = 12

# Nine earlier windows that were not used to select the ensemble
FORECAST_ORIGINS = pd.date_range(
    start="2008-07-01",
    end="2016-07-01",
    freq="12MS"
)


# ============================================================
# 2. LOAD THE MONTHLY TEMPERATURE SERIES
# ============================================================

monthly_data = pd.read_csv(DATA_FILE)

print("AVAILABLE COLUMNS")
print("=================")
print(monthly_data.columns.tolist())


# Find the month column
if "month" in monthly_data.columns:
    month_column = "month"
elif "date" in monthly_data.columns:
    month_column = "date"
else:
    raise ValueError(
        "The monthly file must contain a 'month' or 'date' column."
    )


# Find the temperature column
temperature_candidates = [
    "mean_temp_F",
    "temp",
    "temperature",
    "monthly_mean_temp_F"
]

temperature_column = next(
    (
        column
        for column in temperature_candidates
        if column in monthly_data.columns
    ),
    None
)

if temperature_column is None:
    raise ValueError(
        "Could not find the monthly temperature column. "
        f"Columns found: {monthly_data.columns.tolist()}"
    )


monthly_data[month_column] = pd.to_datetime(
    monthly_data[month_column]
).dt.to_period("M").dt.to_timestamp()

monthly_data[temperature_column] = pd.to_numeric(
    monthly_data[temperature_column],
    errors="coerce"
)

series = (
    monthly_data
    .dropna(subset=[month_column, temperature_column])
    .drop_duplicates(subset=[month_column], keep="last")
    .set_index(month_column)[temperature_column]
    .sort_index()
    .astype(float)
    .asfreq("MS")
)

series.name = "actual_F"


print("\nMONTHLY DATA")
print("============")
print(f"Source:       {DATA_FILE}")
print(f"First month:  {series.index.min():%Y-%m}")
print(f"Last month:   {series.index.max():%Y-%m}")
print(f"Months:       {len(series):,}")
print(f"Missing:      {series.isna().sum():,}")


if series.isna().any():
    missing_months = series[series.isna()].index.strftime("%Y-%m").tolist()
    raise ValueError(
        "The series contains missing months: "
        + ", ".join(missing_months)
    )


# Confirm that all validation months are available
final_required_month = (
    FORECAST_ORIGINS[-1]
    + pd.offsets.MonthBegin(FORECAST_MONTHS - 1)
)

if final_required_month > series.index.max():
    raise ValueError(
        f"Validation requires data through {final_required_month:%Y-%m}, "
        f"but the series ends in {series.index.max():%Y-%m}."
    )


# ============================================================
# 3. FORECASTING FUNCTIONS
# ============================================================

def fit_locked_sarima(train_series, forecast_index):
    """
    Fit the previously selected SARIMA structure using all data
    available before the forecast window.
    """

    with warnings.catch_warnings():
        warnings.simplefilter("ignore")

        model = SARIMAX(
            train_series,
            order=SARIMA_ORDER,
            seasonal_order=SARIMA_SEASONAL_ORDER,
            enforce_stationarity=False,
            enforce_invertibility=False
        )

        fitted_model = model.fit(
            disp=False,
            maxiter=500
        )

    forecast = fitted_model.get_forecast(
        steps=len(forecast_index)
    ).predicted_mean

    forecast.index = forecast_index

    converged = bool(
        fitted_model.mle_retvals.get("converged", False)
    )

    return forecast.astype(float), converged


def forecast_15_year_climatology(train_series, forecast_index):
    """
    For each forecast month, average that same calendar month
    from the preceding 15 years.

    Only information available before the forecast window is used.
    """

    forecasts = []

    for forecast_month in forecast_index:
        same_calendar_month = train_series[
            train_series.index.month == forecast_month.month
        ]

        recent_values = same_calendar_month.tail(
            CLIMATOLOGY_YEARS
        )

        if len(recent_values) < CLIMATOLOGY_YEARS:
            raise ValueError(
                f"Only {len(recent_values)} historical values were "
                f"available for calendar month {forecast_month.month}."
            )

        forecasts.append(recent_values.mean())

    return pd.Series(
        forecasts,
        index=forecast_index,
        dtype=float
    )


def calculate_metrics(actual, forecast):
    """Calculate forecast error statistics."""

    valid = actual.notna() & forecast.notna()

    actual_valid = actual.loc[valid].astype(float)
    forecast_valid = forecast.loc[valid].astype(float)

    error = forecast_valid - actual_valid

    return {
        "months_evaluated": len(error),
        "RMSE_F": np.sqrt(np.mean(error ** 2)),
        "MAE_F": np.mean(np.abs(error)),
        "bias_F": np.mean(error),
        "minimum_error_F": error.min(),
        "maximum_error_F": error.max()
    }


# ============================================================
# 4. RUN THE NINE INDEPENDENT WINDOWS
# ============================================================

monthly_predictions = []
window_metrics = []

print("\nINDEPENDENT VALIDATION WINDOWS")
print("==============================")

for window_number, test_start in enumerate(
    FORECAST_ORIGINS,
    start=1
):
    test_months = pd.date_range(
        start=test_start,
        periods=FORECAST_MONTHS,
        freq="MS"
    )

    train_end = test_start - pd.offsets.MonthBegin(1)

    train = series.loc[:train_end].copy()
    actual = series.reindex(test_months)

    if actual.isna().any():
        raise ValueError(
            f"Missing actual temperature in window beginning "
            f"{test_start:%Y-%m}."
        )

    sarima_forecast, sarima_converged = fit_locked_sarima(
        train,
        test_months
    )

    climatology_forecast = forecast_15_year_climatology(
        train,
        test_months
    )

    ensemble_forecast = (
        SARIMA_WEIGHT * sarima_forecast
        + CLIMATOLOGY_WEIGHT * climatology_forecast
    )

    forecasts = {
        "sarima_111x111_12": sarima_forecast,
        "ensemble_sarima_climatology_75_25": ensemble_forecast
    }

    for model_name, forecast in forecasts.items():
        metrics = calculate_metrics(
            actual,
            forecast
        )

        window_metrics.append({
            "window_number": window_number,
            "test_start": test_months.min(),
            "test_end": test_months.max(),
            "training_end": train_end,
            "training_months": len(train),
            "model": model_name,
            "converged": (
                sarima_converged
                if model_name == "sarima_111x111_12"
                else sarima_converged
            ),
            **metrics
        })

    for horizon, month in enumerate(test_months, start=1):
        monthly_predictions.append({
            "window_number": window_number,
            "test_start": test_months.min(),
            "test_end": test_months.max(),
            "month": month,
            "forecast_horizon": horizon,
            "actual_F": actual.loc[month],
            "sarima_forecast_F": sarima_forecast.loc[month],
            "climatology_15y_forecast_F":
                climatology_forecast.loc[month],
            "ensemble_forecast_F": ensemble_forecast.loc[month],
            "sarima_error_F":
                sarima_forecast.loc[month] - actual.loc[month],
            "ensemble_error_F":
                ensemble_forecast.loc[month] - actual.loc[month]
        })

    sarima_metrics = calculate_metrics(
        actual,
        sarima_forecast
    )

    ensemble_metrics = calculate_metrics(
        actual,
        ensemble_forecast
    )

    print(
        f"\nWindow {window_number}: "
        f"{test_months.min():%Y-%m} through "
        f"{test_months.max():%Y-%m}"
    )
    print(
        f"  SARIMA:       "
        f"RMSE={sarima_metrics['RMSE_F']:.3f}, "
        f"MAE={sarima_metrics['MAE_F']:.3f}, "
        f"Bias={sarima_metrics['bias_F']:+.3f}, "
        f"Converged={sarima_converged}"
    )
    print(
        f"  75/25 ensemble: "
        f"RMSE={ensemble_metrics['RMSE_F']:.3f}, "
        f"MAE={ensemble_metrics['MAE_F']:.3f}, "
        f"Bias={ensemble_metrics['bias_F']:+.3f}"
    )


window_metrics = pd.DataFrame(window_metrics)
monthly_predictions = pd.DataFrame(monthly_predictions)


# ============================================================
# 5. SUMMARIZE STABILITY ACROSS ALL NINE WINDOWS
# ============================================================

def create_stability_summary(
    metrics_table,
    predictions_table
):
    rows = []

    model_settings = {
        "sarima_111x111_12": {
            "forecast_column": "sarima_forecast_F",
            "error_column": "sarima_error_F"
        },
        "ensemble_sarima_climatology_75_25": {
            "forecast_column": "ensemble_forecast_F",
            "error_column": "ensemble_error_F"
        }
    }

    for model_name, columns in model_settings.items():
        model_windows = metrics_table[
            metrics_table["model"] == model_name
        ].copy()

        pooled_error = predictions_table[
            columns["error_column"]
        ].dropna()

        rows.append({
            "model": model_name,
            "windows_completed": len(model_windows),
            "windows_converged":
                int(model_windows["converged"].sum()),
            "months_evaluated": len(pooled_error),
            "mean_RMSE_F": model_windows["RMSE_F"].mean(),
            "median_RMSE_F": model_windows["RMSE_F"].median(),
            "RMSE_std_F": model_windows["RMSE_F"].std(ddof=0),
            "best_window_RMSE_F": model_windows["RMSE_F"].min(),
            "worst_window_RMSE_F": model_windows["RMSE_F"].max(),
            "mean_MAE_F": model_windows["MAE_F"].mean(),
            "worst_absolute_bias_F":
                model_windows["bias_F"].abs().max(),
            "pooled_RMSE_F":
                np.sqrt(np.mean(pooled_error ** 2)),
            "pooled_MAE_F":
                np.mean(np.abs(pooled_error)),
            "pooled_bias_F":
                np.mean(pooled_error)
        })

    summary = pd.DataFrame(rows)

    summary["robustness_indicator"] = (
        summary["mean_RMSE_F"]
        + summary["RMSE_std_F"]
    )

    return summary.sort_values(
        "pooled_RMSE_F"
    ).reset_index(drop=True)


stability_summary = create_stability_summary(
    window_metrics,
    monthly_predictions
)


# ============================================================
# 6. CREATE A FAIR WINDOW-BY-WINDOW COMPARISON
# ============================================================

rmse_comparison = (
    window_metrics
    .pivot(
        index=["test_start", "test_end"],
        columns="model",
        values="RMSE_F"
    )
    .reset_index()
)

rmse_comparison.columns.name = None

sarima_name = "sarima_111x111_12"
ensemble_name = "ensemble_sarima_climatology_75_25"

rmse_comparison["RMSE_improvement_F"] = (
    rmse_comparison[sarima_name]
    - rmse_comparison[ensemble_name]
)

rmse_comparison["RMSE_improvement_percent"] = (
    100
    * rmse_comparison["RMSE_improvement_F"]
    / rmse_comparison[sarima_name]
)

rmse_comparison["winner"] = np.where(
    rmse_comparison[ensemble_name]
    < rmse_comparison[sarima_name],
    ensemble_name,
    sarima_name
)


# ============================================================
# 7. ACCEPTANCE CHECKS
# ============================================================

summary_by_model = stability_summary.set_index("model")

sarima_summary = summary_by_model.loc[sarima_name]
ensemble_summary = summary_by_model.loc[ensemble_name]

pooled_improvement_F = (
    sarima_summary["pooled_RMSE_F"]
    - ensemble_summary["pooled_RMSE_F"]
)

pooled_improvement_percent = (
    100
    * pooled_improvement_F
    / sarima_summary["pooled_RMSE_F"]
)

acceptance_checks = pd.DataFrame([
    {
        "requirement": "Lower pooled RMSE",
        "SARIMA": sarima_summary["pooled_RMSE_F"],
        "ensemble": ensemble_summary["pooled_RMSE_F"],
        "passed":
            ensemble_summary["pooled_RMSE_F"]
            < sarima_summary["pooled_RMSE_F"]
    },
    {
        "requirement": "Lower mean window RMSE",
        "SARIMA": sarima_summary["mean_RMSE_F"],
        "ensemble": ensemble_summary["mean_RMSE_F"],
        "passed":
            ensemble_summary["mean_RMSE_F"]
            < sarima_summary["mean_RMSE_F"]
    },
    {
        "requirement": "No higher RMSE variability",
        "SARIMA": sarima_summary["RMSE_std_F"],
        "ensemble": ensemble_summary["RMSE_std_F"],
        "passed":
            ensemble_summary["RMSE_std_F"]
            <= sarima_summary["RMSE_std_F"]
    },
    {
        "requirement": "No worse worst-window RMSE",
        "SARIMA": sarima_summary["worst_window_RMSE_F"],
        "ensemble": ensemble_summary["worst_window_RMSE_F"],
        "passed":
            ensemble_summary["worst_window_RMSE_F"]
            <= sarima_summary["worst_window_RMSE_F"]
    },
    {
        "requirement": "Absolute pooled bias below 0.25°F",
        "SARIMA": abs(sarima_summary["pooled_bias_F"]),
        "ensemble": abs(ensemble_summary["pooled_bias_F"]),
        "passed":
            abs(ensemble_summary["pooled_bias_F"]) < 0.25
    },
    {
        "requirement": "SARIMA converged in all nine windows",
        "SARIMA": sarima_summary["windows_converged"],
        "ensemble": ensemble_summary["windows_converged"],
        "passed":
            ensemble_summary["windows_converged"] == 9
    }
])


ensemble_wins = int(
    (
        rmse_comparison["winner"] == ensemble_name
    ).sum()
)

sarima_wins = int(
    (
        rmse_comparison["winner"] == sarima_name
    ).sum()
)

passes_all_checks = bool(
    acceptance_checks["passed"].all()
)


# ============================================================
# 8. DISPLAY RESULTS
# ============================================================

print("\n\nINDEPENDENT STABILITY SUMMARY")
print("=============================")

display(
    stability_summary.round(3)
)


print("\nWINDOW-BY-WINDOW COMPARISON")
print("===========================")

display(
    rmse_comparison.round(3)
)


print("\nACCEPTANCE CHECKS")
print("=================")

display(
    acceptance_checks.round(3)
)


print("\nFINAL INDEPENDENT VALIDATION DECISION")
print("=====================================")

print(f"Ensemble window wins: {ensemble_wins}")
print(f"SARIMA window wins:   {sarima_wins}")

print(
    f"Pooled RMSE improvement: "
    f"{pooled_improvement_F:.3f} °F "
    f"({pooled_improvement_percent:.2f}%)"
)

if passes_all_checks and pooled_improvement_F > 0:
    print(
        "\nThe fixed 75/25 ensemble passed every stability check "
        "on the earlier validation period."
    )
    print(
        "This provides independent support for using the ensemble "
        "instead of SARIMA alone."
    )

elif pooled_improvement_F > 0:
    failed_requirements = acceptance_checks.loc[
        ~acceptance_checks["passed"],
        "requirement"
    ].tolist()

    print(
        "\nThe ensemble lowered pooled RMSE, but it did not pass "
        "every stability requirement."
    )
    print("Failed checks:")
    for requirement in failed_requirements:
        print(f"  - {requirement}")

    print(
        "Treat the ensemble improvement as inconclusive and retain "
        "SARIMA as the safer production model."
    )

else:
    print(
        "\nThe ensemble did not lower pooled RMSE on the untouched "
        "earlier windows."
    )
    print(
        "Retain SARIMA alone as the stable 12-month production model."
    )


# ============================================================
# 9. PLOTS
# ============================================================

plot_table = rmse_comparison.copy()

plot_table["window_label"] = (
    plot_table["test_start"].dt.strftime("%Y-%m")
    + "\nto\n"
    + plot_table["test_end"].dt.strftime("%Y-%m")
)

x_positions = np.arange(len(plot_table))

plt.figure(figsize=(14, 6))

plt.plot(
    x_positions,
    plot_table[sarima_name],
    marker="o",
    linewidth=2,
    label="SARIMA"
)

plt.plot(
    x_positions,
    plot_table[ensemble_name],
    marker="o",
    linewidth=2,
    label="75% SARIMA + 25% climatology"
)

plt.xticks(
    x_positions,
    plot_table["window_label"],
    rotation=45,
    ha="right"
)

plt.ylabel("RMSE (°F)")
plt.xlabel("Independent 12-month validation window")
plt.title(
    "Independent Validation: SARIMA vs Fixed 75/25 Ensemble"
)
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()

window_plot_file = (
    OUTPUT_DIR
    / "independent_rmse_by_window.png"
)

plt.savefig(
    window_plot_file,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


comparison_for_plot = stability_summary.set_index("model")

plt.figure(figsize=(9, 5))

bars = plt.bar(
    ["SARIMA", "75/25 ensemble"],
    [
        comparison_for_plot.loc[
            sarima_name,
            "pooled_RMSE_F"
        ],
        comparison_for_plot.loc[
            ensemble_name,
            "pooled_RMSE_F"
        ]
    ],
    color=["steelblue", "darkorange"]
)

plt.ylabel("Pooled RMSE (°F)")
plt.title(
    "Independent Validation: Overall Forecast Error"
)
plt.grid(axis="y", alpha=0.3)

for bar in bars:
    value = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value,
        f"{value:.3f}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()

pooled_plot_file = (
    OUTPUT_DIR
    / "independent_pooled_rmse.png"
)

plt.savefig(
    pooled_plot_file,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 10. SAVE RESULTS AND REPRODUCIBILITY INFORMATION
# ============================================================

window_metrics.to_csv(
    OUTPUT_DIR / "independent_window_metrics.csv",
    index=False
)

monthly_predictions.to_csv(
    OUTPUT_DIR / "independent_monthly_predictions.csv",
    index=False
)

stability_summary.to_csv(
    OUTPUT_DIR / "independent_stability_summary.csv",
    index=False
)

rmse_comparison.to_csv(
    OUTPUT_DIR / "independent_window_comparison.csv",
    index=False
)

acceptance_checks.to_csv(
    OUTPUT_DIR / "independent_acceptance_checks.csv",
    index=False
)

metadata = {
    "purpose":
        "Independent validation of the previously selected "
        "12-month ensemble",
    "data_file": str(DATA_FILE),
    "data_first_month": str(series.index.min().date()),
    "data_last_month": str(series.index.max().date()),
    "validation_first_window":
        str(FORECAST_ORIGINS.min().date()),
    "validation_last_window":
        str(FORECAST_ORIGINS.max().date()),
    "validation_windows": len(FORECAST_ORIGINS),
    "forecast_horizon_months": FORECAST_MONTHS,
    "sarima_order": list(SARIMA_ORDER),
    "sarima_seasonal_order":
        list(SARIMA_SEASONAL_ORDER),
    "sarima_training_history":
        "All information available before each forecast origin",
    "climatology_years": CLIMATOLOGY_YEARS,
    "sarima_weight": SARIMA_WEIGHT,
    "climatology_weight": CLIMATOLOGY_WEIGHT,
    "weights_tuned_on_this_validation_set": False,
    "ensemble_pooled_rmse_improvement_F":
        float(pooled_improvement_F),
    "ensemble_pooled_rmse_improvement_percent":
        float(pooled_improvement_percent),
    "ensemble_window_wins": ensemble_wins,
    "sarima_window_wins": sarima_wins,
    "passed_all_checks": passes_all_checks
}

with open(
    OUTPUT_DIR / "validation_metadata.json",
    "w"
) as metadata_file:
    json.dump(
        metadata,
        metadata_file,
        indent=4
    )


print("\nFILES SAVED")
print("===========")
print(f"Directory:           {OUTPUT_DIR}")
print(
    "Stability summary:  "
    f"{OUTPUT_DIR / 'independent_stability_summary.csv'}"
)
print(
    "Window comparison:  "
    f"{OUTPUT_DIR / 'independent_window_comparison.csv'}"
)
print(
    "Acceptance checks:  "
    f"{OUTPUT_DIR / 'independent_acceptance_checks.csv'}"
)
print(
    "Monthly forecasts:  "
    f"{OUTPUT_DIR / 'independent_monthly_predictions.csv'}"
)
print(f"Window plot:         {window_plot_file}")
print(f"Pooled RMSE plot:    {pooled_plot_file}")

print("\nSEND ME:")
print("1. INDEPENDENT STABILITY SUMMARY")
print("2. WINDOW-BY-WINDOW COMPARISON")
print("3. ACCEPTANCE CHECKS")
print("4. FINAL INDEPENDENT VALIDATION DECISION")
```
