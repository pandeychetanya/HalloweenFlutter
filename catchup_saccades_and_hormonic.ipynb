```python
# ============================================================
# PD vs CONTROL: STATIC-TASK "CATCH-UP EQUIVALENT" SACCADES
# ============================================================
#
# INPUT FILES:
#   /content/drive/MyDrive/steer_search_PD/all_remodnav/
#       all_remodnav_events_p001.csv, ...
#       all_remodnav_events_s178.csv, ...
#
# TERMINOLOGY:
# A literal catch-up saccade is defined during smooth pursuit of a moving
# target. In a static-image task, the closest gaze-only equivalent is a
# short-latency, small corrective / multiple-step saccade that follows a
# larger saccade and continues in approximately the same direction.
#
# THIS CELL ONLY:
#   1. loads all all_remodnav_events_<subject>.csv files;
#   2. constructs consecutive saccade pairs within each trial;
#   3. flags candidate static-task catch-up equivalents;
#   4. compares their frequency between PD and Control participants;
#   5. saves row-level candidates, participant summaries, statistics, and plot.
#
# It does NOT analyze pursuit context, fixation context, PSO context, or all
# REMoDNaV event classes beyond using SACC/ISAC rows as saccadic movements.
# ============================================================

from pathlib import Path
import re
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.stats import ttest_ind, mannwhitneyu
from statsmodels.discrete.discrete_model import NegativeBinomial
from IPython.display import display

# ============================================================
# COLAB SETUP: MOUNT DRIVE, IMPORT PACKAGES, AND SET PATHS
# ============================================================
# ============================================================
# FLUSH, UNMOUNT, AND REMOUNT GOOGLE DRIVE
# ============================================================

from google.colab import drive
from pathlib import Path
import shutil
import os

MOUNT_PATH = Path("/content/drive")

# Flush pending writes and unmount Drive
try:
    drive.flush_and_unmount()
    print("Google Drive flushed and unmounted.")
except Exception as e:
    print(f"Unmount note: {e}")

# Remove the old mount-point directory and its temporary contents
if MOUNT_PATH.exists():
    shutil.rmtree(MOUNT_PATH)

# Recreate an empty mount point
MOUNT_PATH.mkdir(parents=True, exist_ok=True)

# Mount Drive again
drive.mount(str(MOUNT_PATH), force_remount=True)

print("\nDrive remounted successfully.")
print("MyDrive exists:", Path("/content/drive/MyDrive").exists())

# ============================================================
# 0. CHANGEABLE DEFINITION PARAMETERS
# ============================================================

ROOT = Path("/content/drive/MyDrive/steer_search_PD/all_remodnav")
FILE_GLOB = "all_remodnav_events_*.csv"
OUT_DIR = ROOT / "static_catchup_equivalent_pd_control"
OUT_DIR.mkdir(parents=True, exist_ok=True)

# REMoDNaV labels treated as saccades when making consecutive pairs.
SACCADE_LABELS = ("SACC", "ISAC")

# Allowed label for the SECOND movement in the pair.
# Recommended default: allow both because a true small corrective movement
# may be classified as either SACC or ISAC depending on segmentation.
CURRENT_ALLOWED_LABELS = ("SACC", "ISAC")

# To run an ISAC-only sensitivity version, use:
# CURRENT_ALLOWED_LABELS = ("ISAC",)

# Candidate second-saccade amplitude window.
MIN_CURRENT_AMP_DEG = 0.20
MAX_CURRENT_AMP_DEG = 2.00

# The first movement should be a reasonably large primary saccade.
MIN_PREVIOUS_AMP_DEG = 2.00

# Time from previous saccade offset to current saccade onset.
MIN_ISI_MS = 0.0
MAX_ISI_MS = 250.0

# Catch-up-like means the second movement continues in approximately the
# SAME direction as the first movement. 45 degrees is a common, interpretable
# starting tolerance; change it here for sensitivity analyses.
MAX_DIRECTION_DIFFERENCE_DEG = 45.0

# Spatial continuity check: current start should be close to previous endpoint.
# Set to None to disable this condition.
MAX_START_TO_PREVIOUS_END_GAP_DEG = 1.50

# Optional size-ratio restriction. Example: 0.50 means the second movement
# must be no more than half the amplitude of the first.
# Set to None to disable.
MAX_CURRENT_TO_PREVIOUS_AMP_RATIO = None

# Keep both movements fully inside IMAGE_ONSET -> RESPONSE.
REQUIRE_FULLY_INSIDE_RESPONSE_WINDOW = True
RESPONSE_TOLERANCE_MS = 2.0

# Participant-level primary metric used for PD-vs-Control testing.
# Other metrics are still saved in the summary table.
PRIMARY_METRIC = "candidates_per_100_saccades"

RANDOM_SEED = None


# ============================================================
# 1. HELPERS
# ============================================================

def read_csv_flexible(path):
    """Read comma- or tab-separated CSV exports."""
    df = pd.read_csv(path, low_memory=False)
    if df.shape[1] == 1:
        df = pd.read_csv(path, sep="\t", low_memory=False)
    df.columns = [str(c).strip() for c in df.columns]
    return df


def infer_subject_from_filename(path):
    match = re.search(r"all_remodnav_events_([ps]\d+)", path.stem, re.I)
    return match.group(1).lower() if match else None


def numeric(series):
    return pd.to_numeric(series, errors="coerce")


def safe_rate(numerator, denominator, multiplier=1.0):
    numerator = numeric(numerator)
    denominator = numeric(denominator)
    result = multiplier * numerator / denominator
    return result.where(denominator > 0)


def angular_difference_deg(angle1, angle2):
    """Smallest absolute angular difference, from 0 to 180 degrees."""
    return np.abs((angle1 - angle2 + 180.0) % 360.0 - 180.0)


def hedges_g(pd_values, control_values):
    """Positive value means the participant-level metric is higher in PD."""
    x = np.asarray(pd_values, dtype=float)
    y = np.asarray(control_values, dtype=float)
    x = x[np.isfinite(x)]
    y = y[np.isfinite(y)]

    nx, ny = len(x), len(y)
    if nx < 2 or ny < 2:
        return np.nan

    pooled_variance = (
        (nx - 1) * np.var(x, ddof=1)
        + (ny - 1) * np.var(y, ddof=1)
    ) / (nx + ny - 2)

    if not np.isfinite(pooled_variance) or pooled_variance <= 0:
        return np.nan

    cohen_d = (np.mean(x) - np.mean(y)) / np.sqrt(pooled_variance)
    correction = 1.0 - 3.0 / (4.0 * (nx + ny) - 9.0)
    return float(correction * cohen_d)


def participant_level_test(subject_summary, metric):
    pd_values = numeric(
        subject_summary.loc[subject_summary["group"] == "PD", metric]
    ).dropna().to_numpy()
    control_values = numeric(
        subject_summary.loc[subject_summary["group"] == "Control", metric]
    ).dropna().to_numpy()

    result = {
        "metric": metric,
        "n_PD": len(pd_values),
        "n_Control": len(control_values),
        "PD_mean": np.mean(pd_values) if len(pd_values) else np.nan,
        "Control_mean": np.mean(control_values) if len(control_values) else np.nan,
        "PD_median": np.median(pd_values) if len(pd_values) else np.nan,
        "Control_median": np.median(control_values) if len(control_values) else np.nan,
        "mean_difference_PD_minus_Control": (
            np.mean(pd_values) - np.mean(control_values)
            if len(pd_values) and len(control_values)
            else np.nan
        ),
        "hedges_g_PD_minus_Control": hedges_g(pd_values, control_values),
    }

    if len(pd_values) >= 2 and len(control_values) >= 2:
        welch = ttest_ind(pd_values, control_values, equal_var=False)
        mann_whitney = mannwhitneyu(
            pd_values, control_values, alternative="two-sided"
        )
        result.update({
            "welch_t": float(welch.statistic),
            "welch_p": float(welch.pvalue),
            "mannwhitney_U": float(mann_whitney.statistic),
            "mannwhitney_p": float(mann_whitney.pvalue),
        })
    else:
        result.update({
            "welch_t": np.nan,
            "welch_p": np.nan,
            "mannwhitney_U": np.nan,
            "mannwhitney_p": np.nan,
        })

    return pd.DataFrame([result])


def fit_negative_binomial_rate(subject_summary):
    """
    Participant-level count model:
        candidate_count ~ PD + offset(log(total saccades))

    IRR > 1 means a higher candidate rate in PD.
    """
    d = subject_summary[
        ["group", "candidate_count", "total_saccades"]
    ].copy()
    d["candidate_count"] = numeric(d["candidate_count"])
    d["total_saccades"] = numeric(d["total_saccades"])
    d = d.dropna()
    d = d[d["total_saccades"] > 0]

    if len(d) < 4 or d["group"].nunique() < 2:
        return pd.DataFrame([{
            "model": "Negative binomial with log(total_saccades) offset",
            "IRR_PD_vs_Control": np.nan,
            "CI_low": np.nan,
            "CI_high": np.nan,
            "p": np.nan,
            "converged": False,
            "error": "Insufficient data",
        }])

    X = pd.DataFrame({
        "const": np.ones(len(d), dtype=float),
        "PD": (d["group"] == "PD").astype(float).to_numpy(),
    })
    y = d["candidate_count"].to_numpy(dtype=float)
    offset = np.log(d["total_saccades"].to_numpy(dtype=float))

    try:
        fit = NegativeBinomial(
            y,
            X,
            offset=offset,
            loglike_method="nb2",
        ).fit(disp=False, maxiter=1000, cov_type="HC3")

        beta = float(fit.params["PD"])
        ci = fit.conf_int().loc["PD"]

        return pd.DataFrame([{
            "model": "Negative binomial with log(total_saccades) offset",
            "IRR_PD_vs_Control": float(np.exp(beta)),
            "CI_low": float(np.exp(ci.iloc[0])),
            "CI_high": float(np.exp(ci.iloc[1])),
            "p": float(fit.pvalues["PD"]),
            "converged": bool(fit.mle_retvals.get("converged", True)),
            "error": "",
        }])
    except Exception as exc:
        return pd.DataFrame([{
            "model": "Negative binomial with log(total_saccades) offset",
            "IRR_PD_vs_Control": np.nan,
            "CI_low": np.nan,
            "CI_high": np.nan,
            "p": np.nan,
            "converged": False,
            "error": repr(exc),
        }])


# ============================================================
# 2. LOAD ALL SUBJECT FILES
# ============================================================

files = sorted(ROOT.glob(FILE_GLOB))
if not files:
    raise FileNotFoundError(f"No files matched: {ROOT / FILE_GLOB}")

frames = []
for path in files:
    df = read_csv_flexible(path)
    file_subject = infer_subject_from_filename(path)

    if "subject" not in df.columns:
        if file_subject is None:
            raise ValueError(f"Cannot infer subject from {path.name}")
        df["subject"] = file_subject
    else:
        df["subject"] = df["subject"].astype(str).str.strip().str.lower()
        missing_subject = df["subject"].isin(["", "nan", "none"])
        if file_subject is not None:
            df.loc[missing_subject, "subject"] = file_subject

    df["source_file"] = path.name
    frames.append(df)

all_events = pd.concat(frames, ignore_index=True)
print(f"Loaded {len(files)} files with {len(all_events):,} total events.")


# ============================================================
# 3. STANDARDIZE REQUIRED VARIABLES
# ============================================================

required = [
    "subject", "block_id", "trial", "label",
    "start_abs_ms", "end_abs_ms",
]
missing = [column for column in required if column not in all_events.columns]
if missing:
    raise ValueError(f"Missing required columns: {missing}")

# Amplitude fallback.
if "amplitude_deg" not in all_events.columns:
    if "amp" in all_events.columns:
        all_events["amplitude_deg"] = all_events["amp"]
    else:
        raise ValueError("Neither amplitude_deg nor amp is available.")

# Displacement fallback.
if "dx_deg" not in all_events.columns:
    if {"start_x", "end_x"}.issubset(all_events.columns):
        all_events["dx_deg"] = numeric(all_events["end_x"]) - numeric(all_events["start_x"])
    else:
        all_events["dx_deg"] = np.nan

if "dy_deg" not in all_events.columns:
    if {"start_y", "end_y"}.issubset(all_events.columns):
        all_events["dy_deg"] = numeric(all_events["end_y"]) - numeric(all_events["start_y"])
    else:
        all_events["dy_deg"] = np.nan

numeric_columns = [
    "trial", "event_index_block", "event_index_trial",
    "start_abs_ms", "end_abs_ms", "onset_ms_from_image",
    "offset_ms_from_image", "rt_ms", "image_onset_ms", "response_ms",
    "duration_ms", "amplitude_deg", "direction_deg",
    "start_x", "start_y", "end_x", "end_y", "dx_deg", "dy_deg",
    "peak_velocity_deg_s",
]
for column in numeric_columns:
    if column in all_events.columns:
        all_events[column] = numeric(all_events[column])

all_events["subject"] = all_events["subject"].astype(str).str.strip().str.lower()
all_events["block_id"] = all_events["block_id"].astype(str).str.strip()
all_events["label"] = all_events["label"].astype(str).str.strip().str.upper()

# Group is determined from subject ID exactly as requested.
all_events["group"] = pd.NA
all_events.loc[all_events["subject"].str.startswith("p"), "group"] = "PD"
all_events.loc[all_events["subject"].str.startswith("s"), "group"] = "Control"
all_events = all_events[all_events["group"].notna()].copy()

# Keep only events assigned to a numerical trial.
all_events = all_events[all_events["trial"].notna()].copy()
all_events["trial"] = all_events["trial"].astype(int)

# Ensure valid event timing.
all_events = all_events[
    all_events["start_abs_ms"].notna()
    & all_events["end_abs_ms"].notna()
    & (all_events["end_abs_ms"] >= all_events["start_abs_ms"])
].copy()

# Compute direction from dx/dy where direction_deg is absent.
computed_direction = (
    np.degrees(np.arctan2(all_events["dy_deg"], all_events["dx_deg"]))
    + 360.0
) % 360.0
all_events["direction_deg"] = all_events["direction_deg"].fillna(computed_direction)

# Use onset/offset relative to image if present; otherwise reconstruct from
# absolute image onset when possible.
if "onset_ms_from_image" not in all_events.columns:
    all_events["onset_ms_from_image"] = np.nan
if "offset_ms_from_image" not in all_events.columns:
    all_events["offset_ms_from_image"] = np.nan

if "image_onset_ms" in all_events.columns:
    all_events["onset_ms_from_image"] = all_events["onset_ms_from_image"].fillna(
        all_events["start_abs_ms"] - all_events["image_onset_ms"]
    )
    all_events["offset_ms_from_image"] = all_events["offset_ms_from_image"].fillna(
        all_events["end_abs_ms"] - all_events["image_onset_ms"]
    )

# Trial response duration fallback.
if "rt_ms" not in all_events.columns:
    all_events["rt_ms"] = np.nan
if {"response_ms", "image_onset_ms"}.issubset(all_events.columns):
    all_events["rt_ms"] = all_events["rt_ms"].fillna(
        all_events["response_ms"] - all_events["image_onset_ms"]
    )

print("PD subjects:", sorted(all_events.loc[all_events.group == "PD", "subject"].unique()))
print("Control subjects:", sorted(all_events.loc[all_events.group == "Control", "subject"].unique()))


# ============================================================
# 4. KEEP ONLY SACCADES AND CONSTRUCT CONSECUTIVE PAIRS
# ============================================================

saccades = all_events[all_events["label"].isin(SACCADE_LABELS)].copy()

sort_columns = ["subject", "block_id", "trial", "start_abs_ms", "end_abs_ms"]
if "event_index_trial" in saccades.columns:
    sort_columns.append("event_index_trial")
saccades = saccades.sort_values(sort_columns).reset_index(drop=True)

trial_keys = ["subject", "block_id", "trial"]
grouped = saccades.groupby(trial_keys, sort=False, observed=True)

# Shift all previous-saccade information within trial.
previous_columns = [
    "label", "start_abs_ms", "end_abs_ms",
    "onset_ms_from_image", "offset_ms_from_image", "rt_ms",
    "amplitude_deg", "direction_deg",
    "start_x", "start_y", "end_x", "end_y",
    "peak_velocity_deg_s",
]
for column in previous_columns:
    if column in saccades.columns:
        saccades[f"previous_{column}"] = grouped[column].shift(1)

saccades["has_previous_saccade"] = saccades["previous_end_abs_ms"].notna()
saccades["isi_ms"] = saccades["start_abs_ms"] - saccades["previous_end_abs_ms"]
saccades["direction_difference_deg"] = angular_difference_deg(
    saccades["direction_deg"], saccades["previous_direction_deg"]
)
saccades["current_to_previous_amp_ratio"] = (
    saccades["amplitude_deg"] / saccades["previous_amplitude_deg"]
)

if {"start_x", "start_y", "previous_end_x", "previous_end_y"}.issubset(saccades.columns):
    saccades["start_to_previous_end_gap_deg"] = np.sqrt(
        (saccades["start_x"] - saccades["previous_end_x"]) ** 2
        + (saccades["start_y"] - saccades["previous_end_y"]) ** 2
    )
else:
    saccades["start_to_previous_end_gap_deg"] = np.nan


# ============================================================
# 5. DEFINE STATIC-TASK CATCH-UP-EQUIVALENT CANDIDATES
# ============================================================

condition_previous_exists = saccades["has_previous_saccade"]
condition_current_label = saccades["label"].isin(CURRENT_ALLOWED_LABELS)
condition_current_amplitude = saccades["amplitude_deg"].between(
    MIN_CURRENT_AMP_DEG, MAX_CURRENT_AMP_DEG, inclusive="both"
)
condition_previous_amplitude = (
    saccades["previous_amplitude_deg"] >= MIN_PREVIOUS_AMP_DEG
)
condition_short_isi = saccades["isi_ms"].between(
    MIN_ISI_MS, MAX_ISI_MS, inclusive="both"
)
condition_same_direction = (
    saccades["direction_difference_deg"] <= MAX_DIRECTION_DIFFERENCE_DEG
)

candidate = (
    condition_previous_exists
    & condition_current_label
    & condition_current_amplitude
    & condition_previous_amplitude
    & condition_short_isi
    & condition_same_direction
)

if MAX_START_TO_PREVIOUS_END_GAP_DEG is not None:
    candidate &= (
        saccades["start_to_previous_end_gap_deg"]
        <= MAX_START_TO_PREVIOUS_END_GAP_DEG
    )

if MAX_CURRENT_TO_PREVIOUS_AMP_RATIO is not None:
    candidate &= (
        saccades["current_to_previous_amp_ratio"]
        <= MAX_CURRENT_TO_PREVIOUS_AMP_RATIO
    )

if REQUIRE_FULLY_INSIDE_RESPONSE_WINDOW:
    current_inside = (
        saccades["onset_ms_from_image"].notna()
        & saccades["offset_ms_from_image"].notna()
        & saccades["rt_ms"].notna()
        & (saccades["onset_ms_from_image"] >= -RESPONSE_TOLERANCE_MS)
        & (
            saccades["offset_ms_from_image"]
            <= saccades["rt_ms"] + RESPONSE_TOLERANCE_MS
        )
    )
    previous_inside = (
        saccades["previous_onset_ms_from_image"].notna()
        & saccades["previous_offset_ms_from_image"].notna()
        & saccades["rt_ms"].notna()
        & (saccades["previous_onset_ms_from_image"] >= -RESPONSE_TOLERANCE_MS)
        & (
            saccades["previous_offset_ms_from_image"]
            <= saccades["rt_ms"] + RESPONSE_TOLERANCE_MS
        )
    )
    candidate &= current_inside & previous_inside

saccades["is_static_catchup_equivalent"] = candidate.fillna(False)

candidates = saccades[saccades["is_static_catchup_equivalent"]].copy()

print("\nDefinition used")
print(f"  Current labels: {CURRENT_ALLOWED_LABELS}")
print(f"  Current amplitude: {MIN_CURRENT_AMP_DEG}-{MAX_CURRENT_AMP_DEG} deg")
print(f"  Previous amplitude: >= {MIN_PREVIOUS_AMP_DEG} deg")
print(f"  ISI: {MIN_ISI_MS}-{MAX_ISI_MS} ms")
print(f"  Direction difference: <= {MAX_DIRECTION_DIFFERENCE_DEG} deg")
print(f"  Start-to-previous-end gap: <= {MAX_START_TO_PREVIOUS_END_GAP_DEG} deg")
print(f"\nDetected {len(candidates):,} candidate events.")


# ============================================================
# 6. PARTICIPANT-LEVEL PD-vs-CONTROL SUMMARY
# ============================================================

# Count all trial-assigned trials, including trials with no saccades.
trial_table = (
    all_events[["group", "subject", "block_id", "trial"]]
    .drop_duplicates()
    .copy()
)
trial_table["trial_uid"] = (
    trial_table["subject"].astype(str)
    + "|" + trial_table["block_id"].astype(str)
    + "|" + trial_table["trial"].astype(str)
)

saccades["trial_uid"] = (
    saccades["subject"].astype(str)
    + "|" + saccades["block_id"].astype(str)
    + "|" + saccades["trial"].astype(str)
)
candidates["trial_uid"] = (
    candidates["subject"].astype(str)
    + "|" + candidates["block_id"].astype(str)
    + "|" + candidates["trial"].astype(str)
)

subject_trials = (
    trial_table.groupby(["group", "subject"], as_index=False, observed=True)
    .agg(total_trials=("trial_uid", "nunique"))
)

subject_saccades = (
    saccades.groupby(["group", "subject"], as_index=False, observed=True)
    .agg(
        total_saccades=("label", "size"),
        eligible_pairs=("has_previous_saccade", "sum"),
    )
)

subject_candidates = (
    candidates.groupby(["group", "subject"], as_index=False, observed=True)
    .agg(
        candidate_count=("is_static_catchup_equivalent", "size"),
        trials_with_candidate=("trial_uid", "nunique"),
        median_candidate_amp_deg=("amplitude_deg", "median"),
        median_candidate_isi_ms=("isi_ms", "median"),
        median_direction_difference_deg=("direction_difference_deg", "median"),
    )
)

subject_summary = (
    subject_trials
    .merge(subject_saccades, on=["group", "subject"], how="left")
    .merge(subject_candidates, on=["group", "subject"], how="left")
)

for column in [
    "total_saccades", "eligible_pairs", "candidate_count", "trials_with_candidate"
]:
    subject_summary[column] = subject_summary[column].fillna(0)

subject_summary["candidates_per_100_saccades"] = safe_rate(
    subject_summary["candidate_count"], subject_summary["total_saccades"], 100
)
subject_summary["candidates_per_100_pairs"] = safe_rate(
    subject_summary["candidate_count"], subject_summary["eligible_pairs"], 100
)
subject_summary["candidates_per_100_trials"] = safe_rate(
    subject_summary["candidate_count"], subject_summary["total_trials"], 100
)
subject_summary["percent_trials_with_candidate"] = safe_rate(
    subject_summary["trials_with_candidate"], subject_summary["total_trials"], 100
)

subject_summary = subject_summary.sort_values(["group", "subject"]).reset_index(drop=True)

summary_metrics = [
    "candidate_count",
    "candidates_per_100_saccades",
    "candidates_per_100_pairs",
    "candidates_per_100_trials",
    "percent_trials_with_candidate",
]

group_summary_rows = []
for group_name, d in subject_summary.groupby("group", observed=True):
    row = {
        "group": group_name,
        "n_subjects": d["subject"].nunique(),
        "total_candidates": d["candidate_count"].sum(),
        "total_saccades": d["total_saccades"].sum(),
        "total_trials": d["total_trials"].sum(),
    }
    for metric in summary_metrics:
        row[f"mean_{metric}"] = numeric(d[metric]).mean()
        row[f"sd_{metric}"] = numeric(d[metric]).std(ddof=1)
        row[f"median_{metric}"] = numeric(d[metric]).median()
    group_summary_rows.append(row)

group_summary = pd.DataFrame(group_summary_rows)

all_tests = pd.concat(
    [participant_level_test(subject_summary, metric) for metric in summary_metrics],
    ignore_index=True,
)

nb_model = fit_negative_binomial_rate(subject_summary)

print("\nParticipant summary")
display(subject_summary)

print("\nGroup summary")
display(group_summary)

print("\nParticipant-level statistical comparisons")
display(all_tests)

print("\nNegative-binomial rate model")
display(nb_model)


# ============================================================
# 7. PRIMARY PLOT
# ============================================================

plot_data = subject_summary[["group", "subject", PRIMARY_METRIC]].dropna().copy()
group_order = [group for group in ["Control", "PD"] if group in plot_data["group"].unique()]

fig, ax = plt.subplots(figsize=(7.5, 5.5))
rng = np.random.default_rng(RANDOM_SEED)

for x_position, group_name in enumerate(group_order):
    values = numeric(
        plot_data.loc[plot_data["group"] == group_name, PRIMARY_METRIC]
    ).dropna().to_numpy()

    jitter = rng.uniform(-0.09, 0.09, size=len(values))
    ax.scatter(
        np.full(len(values), x_position) + jitter,
        values,
        s=50,
        alpha=0.8,
        edgecolors="black",
        linewidths=0.5,
        zorder=3,
    )

    if len(values):
        mean = np.mean(values)
        sem = np.std(values, ddof=1) / np.sqrt(len(values)) if len(values) > 1 else 0.0
        ax.errorbar(
            x_position,
            mean,
            yerr=sem,
            fmt="o",
            markersize=9,
            capsize=5,
            linewidth=2,
            zorder=4,
        )

ax.set_xticks(range(len(group_order)))
ax.set_xticklabels(group_order)
ax.set_ylabel(PRIMARY_METRIC.replace("_", " "))
ax.set_xlabel("")
ax.set_title("Static-task catch-up-equivalent saccades")
ax.grid(axis="y", alpha=0.25)
fig.tight_layout()

plot_path = OUT_DIR / f"PD_vs_Control_{PRIMARY_METRIC}.png"
fig.savefig(plot_path, dpi=300, bbox_inches="tight")
plt.show()


# ============================================================
# 8. SAVE OUTPUTS
# ============================================================

saccades.to_csv(
    OUT_DIR / "all_saccades_with_static_catchup_flag.csv", index=False
)
candidates.to_csv(
    OUT_DIR / "static_catchup_equivalent_candidates.csv", index=False
)
subject_summary.to_csv(
    OUT_DIR / "subject_summary.csv", index=False
)
group_summary.to_csv(
    OUT_DIR / "group_summary.csv", index=False
)
all_tests.to_csv(
    OUT_DIR / "participant_level_tests.csv", index=False
)
nb_model.to_csv(
    OUT_DIR / "negative_binomial_rate_model.csv", index=False
)

# Save the exact definition so results remain reproducible.
definition_table = pd.DataFrame([{
    "SACCADE_LABELS": ",".join(SACCADE_LABELS),
    "CURRENT_ALLOWED_LABELS": ",".join(CURRENT_ALLOWED_LABELS),
    "MIN_CURRENT_AMP_DEG": MIN_CURRENT_AMP_DEG,
    "MAX_CURRENT_AMP_DEG": MAX_CURRENT_AMP_DEG,
    "MIN_PREVIOUS_AMP_DEG": MIN_PREVIOUS_AMP_DEG,
    "MIN_ISI_MS": MIN_ISI_MS,
    "MAX_ISI_MS": MAX_ISI_MS,
    "MAX_DIRECTION_DIFFERENCE_DEG": MAX_DIRECTION_DIFFERENCE_DEG,
    "MAX_START_TO_PREVIOUS_END_GAP_DEG": MAX_START_TO_PREVIOUS_END_GAP_DEG,
    "MAX_CURRENT_TO_PREVIOUS_AMP_RATIO": MAX_CURRENT_TO_PREVIOUS_AMP_RATIO,
    "REQUIRE_FULLY_INSIDE_RESPONSE_WINDOW": REQUIRE_FULLY_INSIDE_RESPONSE_WINDOW,
    "PRIMARY_METRIC": PRIMARY_METRIC,
}])
definition_table.to_csv(OUT_DIR / "definition_used.csv", index=False)

print("\nSaved outputs to:")
print(OUT_DIR)
print("\nPrimary interpretation:")
print("  - Welch/Mann-Whitney tests compare participant-level rates.")
print("  - Hedges' g > 0 means PD participants have a higher rate.")
print("  - Negative-binomial IRR > 1 means a higher candidate rate in PD,")
print("    after accounting for each participant's total number of saccades.")
```

```python
# ============================================================
# DIAGNOSE >500-MS REMODNAV FIXATIONS
#
# Uses:
#   /content/drive/MyDrive/steer_search_PD/all_remodnav/
#       all_remodnav_events_p001.csv
#       all_remodnav_events_p002.csv
#       ...
#       all_remodnav_events_s178.csv
#       ...
#
# This cell:
#   1. Keeps FIXA events fully inside IMAGE_ONSET -> RESPONSE
#   2. Finds fixations >= 500 ms
#   3. Determines whether they are first/final fixations
#   4. Checks proximity to response
#   5. Checks block quality and missing-data gaps
#   6. Flags unusually high-motion FIXA events
#   7. Compares long-fixation proportions between PD/control
#   8. Repeats the comparison across multiple duration cutoffs
#   9. Saves all diagnostic tables and plots
#
# It does NOT alter any original CSV.
# ============================================================

from pathlib import Path
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.stats import ttest_ind, mannwhitneyu
from IPython.display import display


# ============================================================
# 0. CHANGEABLE SETTINGS
# ============================================================

ROOT = Path(
    "/content/drive/MyDrive/steer_search_PD/all_remodnav"
)

FILE_PATTERN = "all_remodnav_events_*.csv"

OUT_DIR = ROOT / "long_fixation_diagnostics"
OUT_DIR.mkdir(parents=True, exist_ok=True)


# Main definition of a long fixation
LONG_FIXATION_MS = 500.0

# An additional very-long category
VERY_LONG_FIXATION_MS = 1000.0

# A fixation ending this close to the response is considered
# potentially response-related
NEAR_RESPONSE_MS = 100.0

# Same idea for the beginning of the image
NEAR_IMAGE_ONSET_MS = 100.0

# Flag blocks below this valid-data fraction
LOW_VALID_BLOCK_THRESHOLD = 0.85

# Gap between consecutive classified events that may indicate
# missing data / NaN periods / blink-related interruption
LARGE_EVENT_GAP_MS = 20.0

# Ordinary fixations used to estimate unusually high motion
MOTION_REFERENCE_MIN_MS = 100.0
MOTION_REFERENCE_MAX_MS = 500.0

# A long FIXA above this percentile of ordinary FIXA motion
# will be flagged as a possible merged drift/movement event
MOTION_PERCENTILE = 95.0

# Duration cutoffs used for sensitivity analysis
DURATION_CUTOFFS_MS = [
    300, 400, 500, 600, 750, 1000
]

# Number of longest rows to save separately
N_TOP_LONGEST = 300


# ============================================================
# 1. HELPER FUNCTIONS
# ============================================================

def hedges_g(group_a, group_b):
    """
    Hedges' g for group_a minus group_b.
    In this analysis:
        group_a = PD
        group_b = Control
    """
    a = np.asarray(group_a, dtype=float)
    b = np.asarray(group_b, dtype=float)

    a = a[np.isfinite(a)]
    b = b[np.isfinite(b)]

    n_a = len(a)
    n_b = len(b)

    if n_a < 2 or n_b < 2:
        return np.nan

    var_a = np.var(a, ddof=1)
    var_b = np.var(b, ddof=1)

    pooled_df = n_a + n_b - 2

    if pooled_df <= 0:
        return np.nan

    pooled_sd = np.sqrt(
        ((n_a - 1) * var_a + (n_b - 1) * var_b)
        / pooled_df
    )

    if not np.isfinite(pooled_sd) or pooled_sd == 0:
        return np.nan

    d = (np.mean(a) - np.mean(b)) / pooled_sd

    # Small-sample correction
    correction = 1.0 - 3.0 / (4.0 * (n_a + n_b) - 9.0)

    return correction * d


def safe_welch(a, b):
    a = np.asarray(a, dtype=float)
    b = np.asarray(b, dtype=float)

    a = a[np.isfinite(a)]
    b = b[np.isfinite(b)]

    if len(a) < 2 or len(b) < 2:
        return np.nan, np.nan

    result = ttest_ind(a, b, equal_var=False, nan_policy="omit")
    return float(result.statistic), float(result.pvalue)


def safe_mannwhitney(a, b):
    a = np.asarray(a, dtype=float)
    b = np.asarray(b, dtype=float)

    a = a[np.isfinite(a)]
    b = b[np.isfinite(b)]

    if len(a) < 1 or len(b) < 1:
        return np.nan, np.nan

    try:
        result = mannwhitneyu(
            a,
            b,
            alternative="two-sided"
        )
        return float(result.statistic), float(result.pvalue)
    except ValueError:
        return np.nan, np.nan


def save_current_figure(filename):
    path = OUT_DIR / filename
    plt.tight_layout()
    plt.savefig(path, dpi=300, bbox_inches="tight")
    plt.show()
    plt.close()
    print(f"Saved: {path}")


# ============================================================
# 2. FIND AND LOAD ALL EVENT FILES
# ============================================================

if not ROOT.exists():
    raise FileNotFoundError(
        f"Input folder does not exist:\n{ROOT}\n\n"
        "Make sure Google Drive is mounted and the path is correct."
    )

event_files = sorted(ROOT.glob(FILE_PATTERN))

if not event_files:
    raise FileNotFoundError(
        f"No files matching '{FILE_PATTERN}' were found in:\n{ROOT}"
    )

print(f"Found {len(event_files)} participant files.\n")

all_frames = []

for file_path in event_files:

    df = pd.read_csv(file_path, low_memory=False)

    # Recover subject ID from filename when needed
    filename_subject = (
        file_path.stem
        .replace("all_remodnav_events_", "")
        .lower()
    )

    if "subject" not in df.columns:
        df["subject"] = filename_subject
    else:
        df["subject"] = (
            df["subject"]
            .astype("string")
            .str.strip()
            .str.lower()
        )

        df["subject"] = df["subject"].fillna(filename_subject)

    df["source_file"] = file_path.name

    all_frames.append(df)

events = pd.concat(
    all_frames,
    ignore_index=True,
    sort=False
)

print(f"Total event rows loaded: {len(events):,}")


# ============================================================
# 3. STANDARDIZE SUBJECT, GROUP, LABEL, AND NUMERIC COLUMNS
# ============================================================

events["subject"] = (
    events["subject"]
    .astype("string")
    .str.strip()
    .str.lower()
)

# Use subject prefix as the authoritative group definition
events["group"] = np.where(
    events["subject"].str.startswith("p", na=False),
    "PD",
    np.where(
        events["subject"].str.startswith("s", na=False),
        "Control",
        "Unknown"
    )
)

events["label"] = (
    events["label"]
    .astype("string")
    .str.strip()
    .str.upper()
)

numeric_columns = [
    "block_number",
    "event_index_block",
    "trial",
    "event_index_trial",
    "rt_ms",
    "start_time",
    "end_time",
    "start_abs_ms",
    "end_abs_ms",
    "onset_ms_from_image",
    "offset_ms_from_image",
    "duration_ms",
    "amp",
    "amplitude_deg",
    "peak_vel",
    "peak_velocity_deg_s",
    "med_vel",
    "avg_vel",
    "start_x",
    "start_y",
    "end_x",
    "end_y",
    "image_onset_ms",
    "response_ms",
    "valid_block_fraction",
    "n_blink_windows_block",
    "n_blink_samples_block",
    "sampling_rate_hz",
]

for column in numeric_columns:
    if column in events.columns:
        events[column] = pd.to_numeric(
            events[column],
            errors="coerce"
        )


# Required columns
required_columns = [
    "subject",
    "group",
    "block_id",
    "label",
    "trial",
    "duration_ms",
]

missing_columns = [
    column
    for column in required_columns
    if column not in events.columns
]

if missing_columns:
    raise ValueError(
        "Required columns are missing:\n"
        + "\n".join(missing_columns)
    )


# ============================================================
# 4. SORT THE COMPLETE EVENT STREAM
# ============================================================

sort_columns = [
    column
    for column in [
        "subject",
        "block_id",
        "event_index_block",
        "start_abs_ms",
    ]
    if column in events.columns
]

events = (
    events
    .sort_values(sort_columns, kind="mergesort")
    .reset_index(drop=True)
)

block_keys = ["subject", "block_id"]

block_grouped = events.groupby(
    block_keys,
    sort=False,
    dropna=False
)


# Previous and next event labels
events["previous_event_label"] = (
    block_grouped["label"].shift(1)
)

events["next_event_label"] = (
    block_grouped["label"].shift(-1)
)


# Previous event end and next event start
if "end_abs_ms" in events.columns:
    events["previous_event_end_abs_ms"] = (
        block_grouped["end_abs_ms"].shift(1)
    )
else:
    events["previous_event_end_abs_ms"] = np.nan

if "start_abs_ms" in events.columns:
    events["next_event_start_abs_ms"] = (
        block_grouped["start_abs_ms"].shift(-1)
    )
else:
    events["next_event_start_abs_ms"] = np.nan


events["gap_before_event_ms"] = (
    events["start_abs_ms"]
    - events["previous_event_end_abs_ms"]
)

events["gap_after_event_ms"] = (
    events["next_event_start_abs_ms"]
    - events["end_abs_ms"]
)


# Small negative values can occur because of rounding
events.loc[
    events["gap_before_event_ms"].between(-2, 0),
    "gap_before_event_ms"
] = 0

events.loc[
    events["gap_after_event_ms"].between(-2, 0),
    "gap_after_event_ms"
] = 0


# ============================================================
# 5. IDENTIFY EVENTS STRICTLY INSIDE IMAGE ONSET -> RESPONSE
# ============================================================

strict_inside = pd.Series(
    False,
    index=events.index,
    dtype=bool
)

# Preferred relative-time method
relative_available = (
    events["onset_ms_from_image"].notna()
    & events["offset_ms_from_image"].notna()
    & events["rt_ms"].notna()
)

strict_inside.loc[relative_available] = (
    (events.loc[relative_available, "onset_ms_from_image"] >= 0)
    & (
        events.loc[relative_available, "offset_ms_from_image"]
        <= events.loc[relative_available, "rt_ms"]
    )
)


# Absolute-time fallback
absolute_available = (
    events["start_abs_ms"].notna()
    & events["end_abs_ms"].notna()
    & events["image_onset_ms"].notna()
    & events["response_ms"].notna()
)

absolute_fallback = (
    ~relative_available
    & absolute_available
)

strict_inside.loc[absolute_fallback] = (
    (
        events.loc[absolute_fallback, "start_abs_ms"]
        >= events.loc[absolute_fallback, "image_onset_ms"]
    )
    & (
        events.loc[absolute_fallback, "end_abs_ms"]
        <= events.loc[absolute_fallback, "response_ms"]
    )
)


# Must also have a trial assignment
strict_inside &= events["trial"].notna()

events["strictly_inside_trial"] = strict_inside


# ============================================================
# 6. KEEP STRICTLY INSIDE FIXATIONS
# ============================================================

fixations = events.loc[
    (events["label"] == "FIXA")
    & events["strictly_inside_trial"]
].copy()

fixations = fixations.loc[
    fixations["duration_ms"].notna()
    & (fixations["duration_ms"] >= 0)
].copy()

if fixations.empty:
    raise ValueError(
        "No strictly in-trial FIXA events were found.\n"
        "Check the trial timing columns and label values."
    )

print(f"Strictly in-trial FIXA events: {len(fixations):,}")


# ============================================================
# 7. DETERMINE FIXATION ORDER WITHIN EACH TRIAL
# ============================================================

trial_keys = [
    "subject",
    "group",
    "block_id",
    "trial",
]

fixations = (
    fixations
    .sort_values(
        [
            "subject",
            "block_id",
            "trial",
            "start_abs_ms",
            "event_index_block",
        ],
        kind="mergesort"
    )
    .reset_index(drop=True)
)

trial_grouped = fixations.groupby(
    trial_keys,
    sort=False,
    dropna=False
)

fixations["fixation_number_in_trial"] = (
    trial_grouped.cumcount() + 1
)

fixations["n_fixations_in_trial"] = (
    trial_grouped["duration_ms"].transform("size")
)

fixations["is_first_fixation"] = (
    fixations["fixation_number_in_trial"] == 1
)

fixations["is_final_fixation"] = (
    fixations["fixation_number_in_trial"]
    == fixations["n_fixations_in_trial"]
)


# ============================================================
# 8. CALCULATE DISTANCE FROM IMAGE ONSET AND RESPONSE
# ============================================================

fixations["time_from_image_onset_ms"] = (
    fixations["onset_ms_from_image"]
)

fixations["time_to_response_ms"] = (
    fixations["rt_ms"]
    - fixations["offset_ms_from_image"]
)


# Absolute-time fallback where necessary
missing_onset_distance = (
    fixations["time_from_image_onset_ms"].isna()
    & fixations["start_abs_ms"].notna()
    & fixations["image_onset_ms"].notna()
)

fixations.loc[
    missing_onset_distance,
    "time_from_image_onset_ms"
] = (
    fixations.loc[missing_onset_distance, "start_abs_ms"]
    - fixations.loc[missing_onset_distance, "image_onset_ms"]
)


missing_response_distance = (
    fixations["time_to_response_ms"].isna()
    & fixations["response_ms"].notna()
    & fixations["end_abs_ms"].notna()
)

fixations.loc[
    missing_response_distance,
    "time_to_response_ms"
] = (
    fixations.loc[missing_response_distance, "response_ms"]
    - fixations.loc[missing_response_distance, "end_abs_ms"]
)


fixations["near_image_onset"] = (
    fixations["time_from_image_onset_ms"]
    .between(0, NEAR_IMAGE_ONSET_MS, inclusive="both")
)

fixations["near_response"] = (
    fixations["time_to_response_ms"]
    .between(0, NEAR_RESPONSE_MS, inclusive="both")
)


# ============================================================
# 9. LONG-FIXATION FLAGS
# ============================================================

fixations["is_long_fixation"] = (
    fixations["duration_ms"] >= LONG_FIXATION_MS
)

fixations["is_very_long_fixation"] = (
    fixations["duration_ms"] >= VERY_LONG_FIXATION_MS
)


# ============================================================
# 10. MOTION WITHIN FIXA EVENTS
# ============================================================

# REMoDNaV's amplitude_deg for FIXA can be treated here as an
# event-level displacement measure, not as saccade amplitude.

fixations["fixation_displacement_deg"] = (
    fixations["amplitude_deg"]
)

fixations["duration_s"] = (
    fixations["duration_ms"] / 1000.0
)

fixations["displacement_rate_deg_s"] = (
    fixations["fixation_displacement_deg"]
    / fixations["duration_s"].replace(0, np.nan)
)


# Reference distribution from ordinary-duration fixations
motion_reference = fixations.loc[
    fixations["duration_ms"].between(
        MOTION_REFERENCE_MIN_MS,
        MOTION_REFERENCE_MAX_MS,
        inclusive="left"
    )
].copy()


def percentile_or_nan(series, percentile):
    values = pd.to_numeric(series, errors="coerce").dropna()

    if values.empty:
        return np.nan

    return float(np.nanpercentile(values, percentile))


displacement_rate_threshold = percentile_or_nan(
    motion_reference["displacement_rate_deg_s"],
    MOTION_PERCENTILE
)

average_velocity_threshold = percentile_or_nan(
    motion_reference["avg_vel"],
    MOTION_PERCENTILE
)

median_velocity_threshold = percentile_or_nan(
    motion_reference["med_vel"],
    MOTION_PERCENTILE
)


print("\nMotion thresholds derived from ordinary FIXA events:")
print(
    f"  Displacement rate {MOTION_PERCENTILE:.0f}th percentile: "
    f"{displacement_rate_threshold:.3f} deg/s"
)

print(
    f"  Average velocity {MOTION_PERCENTILE:.0f}th percentile: "
    f"{average_velocity_threshold:.3f} deg/s"
)

print(
    f"  Median velocity {MOTION_PERCENTILE:.0f}th percentile: "
    f"{median_velocity_threshold:.3f} deg/s"
)


fixations["high_displacement_rate"] = False
fixations["high_average_velocity"] = False
fixations["high_median_velocity"] = False

if np.isfinite(displacement_rate_threshold):
    fixations["high_displacement_rate"] = (
        fixations["displacement_rate_deg_s"]
        > displacement_rate_threshold
    )

if np.isfinite(average_velocity_threshold):
    fixations["high_average_velocity"] = (
        fixations["avg_vel"]
        > average_velocity_threshold
    )

if np.isfinite(median_velocity_threshold):
    fixations["high_median_velocity"] = (
        fixations["med_vel"]
        > median_velocity_threshold
    )


fixations["high_motion_FIXA"] = (
    fixations["high_displacement_rate"]
    | fixations["high_average_velocity"]
    | fixations["high_median_velocity"]
)


# ============================================================
# 11. DATA-QUALITY AND EVENT-GAP FLAGS
# ============================================================

fixations["low_validity_block"] = (
    fixations["valid_block_fraction"]
    < LOW_VALID_BLOCK_THRESHOLD
)

fixations["large_gap_before"] = (
    fixations["gap_before_event_ms"]
    > LARGE_EVENT_GAP_MS
)

fixations["large_gap_after"] = (
    fixations["gap_after_event_ms"]
    > LARGE_EVENT_GAP_MS
)

fixations["adjacent_to_large_event_gap"] = (
    fixations["large_gap_before"]
    | fixations["large_gap_after"]
)


# ============================================================
# 12. POSSIBLE EXPLANATIONS FOR EACH LONG FIXATION
# ============================================================

fixations["long_and_first"] = (
    fixations["is_long_fixation"]
    & fixations["is_first_fixation"]
)

fixations["long_and_final"] = (
    fixations["is_long_fixation"]
    & fixations["is_final_fixation"]
)

fixations["long_and_near_response"] = (
    fixations["is_long_fixation"]
    & fixations["near_response"]
)

fixations["long_and_near_onset"] = (
    fixations["is_long_fixation"]
    & fixations["near_image_onset"]
)

fixations["long_and_high_motion"] = (
    fixations["is_long_fixation"]
    & fixations["high_motion_FIXA"]
)

fixations["long_and_low_validity"] = (
    fixations["is_long_fixation"]
    & fixations["low_validity_block"]
)

fixations["long_and_adjacent_gap"] = (
    fixations["is_long_fixation"]
    & fixations["adjacent_to_large_event_gap"]
)


# A relatively clean, interior long fixation:
# not final, not near response, not high motion,
# not in a low-validity block, and not near a data gap
fixations["clean_interior_long_fixation"] = (
    fixations["is_long_fixation"]
    & ~fixations["is_final_fixation"]
    & ~fixations["near_response"]
    & ~fixations["high_motion_FIXA"]
    & ~fixations["low_validity_block"]
    & ~fixations["adjacent_to_large_event_gap"]
)


# Human-readable, non-exclusive explanation string
def make_reason_string(row):

    if not row["is_long_fixation"]:
        return ""

    reasons = []

    if row["is_first_fixation"]:
        reasons.append("first fixation")

    if row["is_final_fixation"]:
        reasons.append("final fixation")

    if row["near_image_onset"]:
        reasons.append("near image onset")

    if row["near_response"]:
        reasons.append("near response")

    if row["high_motion_FIXA"]:
        reasons.append("high motion within FIXA")

    if row["low_validity_block"]:
        reasons.append("low-validity block")

    if row["adjacent_to_large_event_gap"]:
        reasons.append("adjacent to event/data gap")

    if not reasons:
        reasons.append("clean interior long fixation")

    return "; ".join(reasons)


fixations["possible_explanations"] = fixations.apply(
    make_reason_string,
    axis=1
)


# ============================================================
# 13. PARTICIPANT-LEVEL SUMMARY
# ============================================================

participant_summary = (
    fixations
    .groupby(
        ["group", "subject"],
        as_index=False,
        dropna=False
    )
    .agg(
        n_fixations=("duration_ms", "size"),
        n_long_fixations=("is_long_fixation", "sum"),
        n_very_long_fixations=("is_very_long_fixation", "sum"),
        median_fixation_duration_ms=("duration_ms", "median"),
        mean_fixation_duration_ms=("duration_ms", "mean"),
        p90_fixation_duration_ms=(
            "duration_ms",
            lambda x: np.nanpercentile(x, 90)
        ),
        p95_fixation_duration_ms=(
            "duration_ms",
            lambda x: np.nanpercentile(x, 95)
        ),
        p99_fixation_duration_ms=(
            "duration_ms",
            lambda x: np.nanpercentile(x, 99)
        ),
        maximum_fixation_duration_ms=("duration_ms", "max"),
        n_long_first=("long_and_first", "sum"),
        n_long_final=("long_and_final", "sum"),
        n_long_near_response=("long_and_near_response", "sum"),
        n_long_near_onset=("long_and_near_onset", "sum"),
        n_long_high_motion=("long_and_high_motion", "sum"),
        n_long_low_validity=("long_and_low_validity", "sum"),
        n_long_adjacent_gap=("long_and_adjacent_gap", "sum"),
        n_clean_interior_long=(
            "clean_interior_long_fixation",
            "sum"
        ),
    )
)

participant_summary["fixation_long_proportion"] = (
    participant_summary["n_long_fixations"]
    / participant_summary["n_fixations"]
)

participant_summary["fixation_very_long_proportion"] = (
    participant_summary["n_very_long_fixations"]
    / participant_summary["n_fixations"]
)

participant_summary["long_fixations_per_100"] = (
    100.0
    * participant_summary["fixation_long_proportion"]
)


# Reason proportions within long fixations
reason_count_columns = [
    "n_long_first",
    "n_long_final",
    "n_long_near_response",
    "n_long_near_onset",
    "n_long_high_motion",
    "n_long_low_validity",
    "n_long_adjacent_gap",
    "n_clean_interior_long",
]

for column in reason_count_columns:

    proportion_column = column.replace("n_", "proportion_")

    participant_summary[proportion_column] = np.where(
        participant_summary["n_long_fixations"] > 0,
        participant_summary[column]
        / participant_summary["n_long_fixations"],
        np.nan
    )


participant_summary = participant_summary.sort_values(
    ["group", "fixation_long_proportion"],
    ascending=[True, False]
)

display(participant_summary)


# ============================================================
# 14. GROUP SUMMARY AND STATISTICAL COMPARISON
# ============================================================

group_summary = (
    participant_summary
    .groupby("group", as_index=False)
    .agg(
        n_participants=("subject", "nunique"),
        median_n_fixations=("n_fixations", "median"),
        median_n_long=("n_long_fixations", "median"),
        mean_long_proportion=(
            "fixation_long_proportion",
            "mean"
        ),
        median_long_proportion=(
            "fixation_long_proportion",
            "median"
        ),
        std_long_proportion=(
            "fixation_long_proportion",
            "std"
        ),
        minimum_long_proportion=(
            "fixation_long_proportion",
            "min"
        ),
        maximum_long_proportion=(
            "fixation_long_proportion",
            "max"
        ),
        median_fixation_duration_ms=(
            "median_fixation_duration_ms",
            "median"
        ),
        median_p95_duration_ms=(
            "p95_fixation_duration_ms",
            "median"
        ),
    )
)

print("\nGROUP SUMMARY")
display(group_summary)


pd_values = participant_summary.loc[
    participant_summary["group"] == "PD",
    "fixation_long_proportion"
].dropna()

control_values = participant_summary.loc[
    participant_summary["group"] == "Control",
    "fixation_long_proportion"
].dropna()


welch_t, welch_p = safe_welch(
    pd_values,
    control_values
)

mw_u, mw_p = safe_mannwhitney(
    pd_values,
    control_values
)

g_value = hedges_g(
    pd_values,
    control_values
)


print("\nPD VS CONTROL: PARTICIPANT-LEVEL LONG-FIXATION PROPORTION")
print(
    f"PD median:      {pd_values.median():.4f} "
    f"({100 * pd_values.median():.2f}%)"
)

print(
    f"Control median: {control_values.median():.4f} "
    f"({100 * control_values.median():.2f}%)"
)

print(f"Welch t = {welch_t:.4f}, p = {welch_p:.6g}")
print(f"Mann–Whitney U = {mw_u:.4f}, p = {mw_p:.6g}")
print(f"Hedges' g, PD minus Control = {g_value:.4f}")


# ============================================================
# 15. REASON SUMMARY AMONG LONG FIXATIONS
# ============================================================

long_fixations = fixations.loc[
    fixations["is_long_fixation"]
].copy()

reason_columns = {
    "First fixation": "is_first_fixation",
    "Final fixation": "is_final_fixation",
    "Near image onset": "near_image_onset",
    "Near response": "near_response",
    "High motion within FIXA": "high_motion_FIXA",
    "Low-validity block": "low_validity_block",
    "Adjacent event/data gap": "adjacent_to_large_event_gap",
    "Clean interior long fixation": "clean_interior_long_fixation",
}

reason_rows = []

for (group, subject), subject_long in long_fixations.groupby(
    ["group", "subject"],
    sort=False
):

    row = {
        "group": group,
        "subject": subject,
        "n_long_fixations": len(subject_long),
    }

    for readable_name, column in reason_columns.items():
        row[readable_name] = subject_long[column].mean()

    reason_rows.append(row)

reason_by_subject = pd.DataFrame(reason_rows)

reason_group_rows = []

for group, group_df in reason_by_subject.groupby("group"):

    for reason_name in reason_columns:

        values = group_df[reason_name].dropna()

        reason_group_rows.append({
            "group": group,
            "reason": reason_name,
            "n_participants_with_long_fixations": len(values),
            "mean_participant_proportion": values.mean(),
            "median_participant_proportion": values.median(),
        })

reason_group_summary = pd.DataFrame(reason_group_rows)

print("\nPOSSIBLE EXPLANATIONS AMONG >500-MS FIXATIONS")
print(
    "Categories overlap, so percentages are not expected to sum to 100%."
)

display(reason_group_summary)


# ============================================================
# 16. TRIAL-LEVEL SUMMARY
# ============================================================

trial_summary = (
    fixations
    .groupby(
        ["group", "subject", "block_id", "trial"],
        as_index=False,
        dropna=False
    )
    .agg(
        image=("image", "first")
            if "image" in fixations.columns
            else ("trial", "first"),
        stim_id=("stim_id", "first")
            if "stim_id" in fixations.columns
            else ("trial", "first"),
        rt_ms=("rt_ms", "first"),
        n_fixations=("duration_ms", "size"),
        n_long_fixations=("is_long_fixation", "sum"),
        longest_fixation_ms=("duration_ms", "max"),
        median_fixation_ms=("duration_ms", "median"),
        final_fixation_duration_ms=(
            "duration_ms",
            lambda x: x.iloc[-1]
        ),
    )
)

trial_summary["has_long_fixation"] = (
    trial_summary["n_long_fixations"] > 0
)

trial_summary["long_fixation_proportion"] = (
    trial_summary["n_long_fixations"]
    / trial_summary["n_fixations"]
)


# Participant-level RT comparison:
# median RT on trials with a long fixation versus without one
rt_comparison_rows = []

for (group, subject), subject_trials in trial_summary.groupby(
    ["group", "subject"]
):

    rt_with_long = subject_trials.loc[
        subject_trials["has_long_fixation"],
        "rt_ms"
    ].dropna()

    rt_without_long = subject_trials.loc[
        ~subject_trials["has_long_fixation"],
        "rt_ms"
    ].dropna()

    rt_comparison_rows.append({
        "group": group,
        "subject": subject,
        "n_trials_with_long_fixation": len(rt_with_long),
        "n_trials_without_long_fixation": len(rt_without_long),
        "median_rt_with_long_ms": (
            rt_with_long.median()
            if len(rt_with_long)
            else np.nan
        ),
        "median_rt_without_long_ms": (
            rt_without_long.median()
            if len(rt_without_long)
            else np.nan
        ),
    })

rt_comparison = pd.DataFrame(rt_comparison_rows)

rt_comparison["median_rt_difference_ms"] = (
    rt_comparison["median_rt_with_long_ms"]
    - rt_comparison["median_rt_without_long_ms"]
)

print("\nARE LONG FIXATIONS CONCENTRATED IN LONGER-RT TRIALS?")
display(
    rt_comparison.sort_values(
        "median_rt_difference_ms",
        ascending=False
    )
)


# ============================================================
# 17. DURATION-CUTOFF SENSITIVITY ANALYSIS
# ============================================================

sensitivity_rows = []

for cutoff_ms in DURATION_CUTOFFS_MS:

    temporary = fixations[
        ["group", "subject", "duration_ms"]
    ].copy()

    temporary["above_cutoff"] = (
        temporary["duration_ms"] >= cutoff_ms
    )

    cutoff_subject = (
        temporary
        .groupby(
            ["group", "subject"],
            as_index=False
        )
        .agg(
            n_fixations=("duration_ms", "size"),
            n_above_cutoff=("above_cutoff", "sum"),
        )
    )

    cutoff_subject["proportion"] = (
        cutoff_subject["n_above_cutoff"]
        / cutoff_subject["n_fixations"]
    )

    pd_cutoff = cutoff_subject.loc[
        cutoff_subject["group"] == "PD",
        "proportion"
    ]

    control_cutoff = cutoff_subject.loc[
        cutoff_subject["group"] == "Control",
        "proportion"
    ]

    cutoff_t, cutoff_p = safe_welch(
        pd_cutoff,
        control_cutoff
    )

    cutoff_u, cutoff_mw_p = safe_mannwhitney(
        pd_cutoff,
        control_cutoff
    )

    cutoff_g = hedges_g(
        pd_cutoff,
        control_cutoff
    )

    sensitivity_rows.append({
        "cutoff_ms": cutoff_ms,
        "pd_mean_proportion": pd_cutoff.mean(),
        "pd_median_proportion": pd_cutoff.median(),
        "control_mean_proportion": control_cutoff.mean(),
        "control_median_proportion": control_cutoff.median(),
        "welch_t": cutoff_t,
        "welch_p": cutoff_p,
        "mannwhitney_u": cutoff_u,
        "mannwhitney_p": cutoff_mw_p,
        "hedges_g_pd_minus_control": cutoff_g,
    })

cutoff_sensitivity = pd.DataFrame(sensitivity_rows)

print("\nDURATION-CUTOFF SENSITIVITY")
display(cutoff_sensitivity)


# ============================================================
# 18. SAVE DIAGNOSTIC TABLES
# ============================================================

fixation_output_path = (
    OUT_DIR / "all_in_trial_fixations_with_diagnostics.csv"
)

long_output_path = (
    OUT_DIR / "long_fixations_500ms_with_diagnostics.csv"
)

participant_output_path = (
    OUT_DIR / "participant_long_fixation_summary.csv"
)

group_output_path = (
    OUT_DIR / "group_long_fixation_summary.csv"
)

reason_subject_output_path = (
    OUT_DIR / "long_fixation_reasons_by_subject.csv"
)

reason_group_output_path = (
    OUT_DIR / "long_fixation_reasons_by_group.csv"
)

trial_output_path = (
    OUT_DIR / "trial_long_fixation_summary.csv"
)

rt_output_path = (
    OUT_DIR / "long_fixation_trial_rt_comparison.csv"
)

sensitivity_output_path = (
    OUT_DIR / "long_fixation_cutoff_sensitivity.csv"
)


fixations.to_csv(
    fixation_output_path,
    index=False
)

long_fixations.to_csv(
    long_output_path,
    index=False
)

participant_summary.to_csv(
    participant_output_path,
    index=False
)

group_summary.to_csv(
    group_output_path,
    index=False
)

reason_by_subject.to_csv(
    reason_subject_output_path,
    index=False
)

reason_group_summary.to_csv(
    reason_group_output_path,
    index=False
)

trial_summary.to_csv(
    trial_output_path,
    index=False
)

rt_comparison.to_csv(
    rt_output_path,
    index=False
)

cutoff_sensitivity.to_csv(
    sensitivity_output_path,
    index=False
)


# Save the longest events separately
preferred_top_columns = [
    "group",
    "subject",
    "block_id",
    "block_number",
    "trial",
    "image",
    "stim_id",
    "event_index_block",
    "event_index_trial",
    "duration_ms",
    "fixation_number_in_trial",
    "n_fixations_in_trial",
    "is_first_fixation",
    "is_final_fixation",
    "time_from_image_onset_ms",
    "time_to_response_ms",
    "near_image_onset",
    "near_response",
    "fixation_displacement_deg",
    "displacement_rate_deg_s",
    "avg_vel",
    "med_vel",
    "high_motion_FIXA",
    "valid_block_fraction",
    "low_validity_block",
    "previous_event_label",
    "next_event_label",
    "gap_before_event_ms",
    "gap_after_event_ms",
    "adjacent_to_large_event_gap",
    "possible_explanations",
    "start_abs_ms",
    "end_abs_ms",
    "onset_ms_from_image",
    "offset_ms_from_image",
    "start_x",
    "start_y",
    "end_x",
    "end_y",
    "behav_file",
    "edf_file",
    "source_file",
]

top_columns = [
    column
    for column in preferred_top_columns
    if column in long_fixations.columns
]

top_longest = (
    long_fixations
    .sort_values(
        "duration_ms",
        ascending=False
    )
    .head(N_TOP_LONGEST)
    [top_columns]
)

top_longest_path = (
    OUT_DIR / f"top_{N_TOP_LONGEST}_longest_fixations.csv"
)

top_longest.to_csv(
    top_longest_path,
    index=False
)


# ============================================================
# 19. FIGURE 1: FIXATION-DURATION DISTRIBUTION
# ============================================================

plot_maximum = min(
    2000,
    float(np.nanpercentile(
        fixations["duration_ms"],
        99.5
    ))
)

plot_maximum = max(plot_maximum, 600)

bins = np.linspace(
    0,
    plot_maximum,
    80
)

plt.figure(figsize=(9, 6))

for group in ["Control", "PD"]:

    values = fixations.loc[
        fixations["group"] == group,
        "duration_ms"
    ].dropna()

    plt.hist(
        values,
        bins=bins,
        density=True,
        histtype="step",
        linewidth=2,
        label=group
    )

plt.axvline(
    LONG_FIXATION_MS,
    linestyle="--",
    linewidth=1.5,
    label=f"{LONG_FIXATION_MS:.0f}-ms cutoff"
)

plt.xlabel("Fixation duration (ms)")
plt.ylabel("Density")
plt.title("Distribution of strictly in-trial FIXA durations")
plt.legend()

save_current_figure(
    "fixation_duration_distribution.png"
)


# ============================================================
# 20. FIGURE 2: PARTICIPANT LONG-FIXATION PROPORTIONS
# ============================================================

control_plot = participant_summary.loc[
    participant_summary["group"] == "Control",
    "fixation_long_proportion"
].dropna().to_numpy()

pd_plot = participant_summary.loc[
    participant_summary["group"] == "PD",
    "fixation_long_proportion"
].dropna().to_numpy()

plt.figure(figsize=(7, 6))

plt.boxplot(
    [control_plot, pd_plot],
    labels=["Control", "PD"],
    showfliers=False
)

rng = np.random.default_rng(12345)

for position, values in enumerate(
    [control_plot, pd_plot],
    start=1
):

    jitter = rng.normal(
        loc=0,
        scale=0.035,
        size=len(values)
    )

    plt.scatter(
        np.full(len(values), position) + jitter,
        values,
        alpha=0.8
    )

plt.ylabel(
    f"Proportion of FIXA events ≥{LONG_FIXATION_MS:.0f} ms"
)

plt.title(
    "Participant-level long-fixation proportion\n"
    f"Welch p={welch_p:.4g}, "
    f"Hedges' g={g_value:.3f}"
)

save_current_figure(
    "participant_long_fixation_proportion.png"
)


# ============================================================
# 21. FIGURE 3: POSSIBLE EXPLANATIONS
# ============================================================

reason_plot = (
    reason_group_summary
    .pivot(
        index="reason",
        columns="group",
        values="mean_participant_proportion"
    )
)

reason_plot = reason_plot.reindex(
    list(reason_columns.keys())
)

reason_plot_percent = 100 * reason_plot

ax = reason_plot_percent.plot(
    kind="bar",
    figsize=(11, 6)
)

ax.set_ylabel(
    "Mean participant percentage of long fixations"
)

ax.set_xlabel("Possible explanation")

ax.set_title(
    f"Characteristics of FIXA events ≥{LONG_FIXATION_MS:.0f} ms\n"
    "Categories are overlapping"
)

plt.xticks(
    rotation=35,
    ha="right"
)

save_current_figure(
    "long_fixation_possible_explanations.png"
)


# ============================================================
# 22. FIGURE 4: CUTOFF SENSITIVITY
# ============================================================

plt.figure(figsize=(8, 6))

plt.plot(
    cutoff_sensitivity["cutoff_ms"],
    100 * cutoff_sensitivity["control_median_proportion"],
    marker="o",
    label="Control"
)

plt.plot(
    cutoff_sensitivity["cutoff_ms"],
    100 * cutoff_sensitivity["pd_median_proportion"],
    marker="o",
    label="PD"
)

plt.xlabel("Long-fixation cutoff (ms)")
plt.ylabel(
    "Median participant percentage above cutoff"
)

plt.title(
    "Sensitivity to the definition of a long fixation"
)

plt.legend()

save_current_figure(
    "long_fixation_cutoff_sensitivity.png"
)


# ============================================================
# 23. FINAL CONSOLE SUMMARY
# ============================================================

n_total_fixations = len(fixations)
n_long_fixations = int(
    fixations["is_long_fixation"].sum()
)

n_very_long_fixations = int(
    fixations["is_very_long_fixation"].sum()
)

print("\n" + "=" * 70)
print("FINAL LONG-FIXATION DIAGNOSTIC SUMMARY")
print("=" * 70)

print(
    f"All strictly in-trial FIXA events: "
    f"{n_total_fixations:,}"
)

print(
    f"FIXA ≥{LONG_FIXATION_MS:.0f} ms: "
    f"{n_long_fixations:,} "
    f"({100 * n_long_fixations / n_total_fixations:.2f}%)"
)

print(
    f"FIXA ≥{VERY_LONG_FIXATION_MS:.0f} ms: "
    f"{n_very_long_fixations:,} "
    f"({100 * n_very_long_fixations / n_total_fixations:.2f}%)"
)


if n_long_fixations > 0:

    def event_percentage(column):
        return (
            100
            * long_fixations[column].mean()
        )

    print("\nCharacteristics of the long FIXA events:")

    print(
        f"  Final fixation in trial: "
        f"{event_percentage('is_final_fixation'):.2f}%"
    )

    print(
        f"  Within {NEAR_RESPONSE_MS:.0f} ms of response: "
        f"{event_percentage('near_response'):.2f}%"
    )

    print(
        f"  First fixation in trial: "
        f"{event_percentage('is_first_fixation'):.2f}%"
    )

    print(
        f"  High motion within FIXA: "
        f"{event_percentage('high_motion_FIXA'):.2f}%"
    )

    print(
        f"  Low-validity block: "
        f"{event_percentage('low_validity_block'):.2f}%"
    )

    print(
        f"  Adjacent to event/data gap: "
        f"{event_percentage('adjacent_to_large_event_gap'):.2f}%"
    )

    print(
        f"  Clean interior long fixation: "
        f"{event_percentage('clean_interior_long_fixation'):.2f}%"
    )


print("\nParticipants with the highest long-fixation proportions:")

display(
    participant_summary[
        [
            "group",
            "subject",
            "n_fixations",
            "n_long_fixations",
            "long_fixations_per_100",
            "maximum_fixation_duration_ms",
            "n_long_final",
            "n_long_near_response",
            "n_long_high_motion",
            "n_long_low_validity",
            "n_long_adjacent_gap",
            "n_clean_interior_long",
        ]
    ]
    .sort_values(
        "long_fixations_per_100",
        ascending=False
    )
    .head(15)
)


print("\nSaved output folder:")
print(OUT_DIR)

print("\nMost useful file for manual inspection:")
print(top_longest_path)

print(
    "\nInterpretation guide:\n"
    "• Mostly final/near-response events: likely target recognition or "
    "response-decision dwell.\n"
    "• Mostly high-motion FIXA events: inspect raw traces for merged drift, "
    "pursuit-like motion, or missed small saccades.\n"
    "• Mostly low-validity/gap-adjacent events: segmentation or missing-data "
    "effects are more plausible.\n"
    "• Mostly clean interior events: likely genuine long visual-processing "
    "periods rather than obvious classification artifacts."
)
```

```python
# ============================================================
# COMPLETE ≥500-MS FIXATION + PARTICIPANT AGE ANALYSIS
#
# Run this cell AFTER the original long-fixation diagnostic cell.
#
# Uses:
#   1. Existing `fixations` dataframe, if available
#
#      OR, as a fallback:
#
#   2. /content/drive/MyDrive/steer_search_PD/all_remodnav/
#      long_fixation_diagnostics/
#      all_in_trial_fixations_with_diagnostics.csv
#
# Participant information:
#   /content/drive/MyDrive/steer_search_PD/steer-search-PD/
#   2025_spring_participants_info (1).csv
#
# Main participant-level outcome:
#
#       number of FIXA events ≥500 ms
#       --------------------------------
#       total number of FIXA events
#
# Analyses:
#   1. Total and ≥500-ms fixation counts by subject
#   2. Percentage of fixations ≥500 ms by subject
#   3. PD versus Control comparison
#   4. Overall age correlation
#   5. Separate age correlations for PD and Control
#   6. Age × group participant-level regression
#   7. Optional age-adjusted PD versus Control regression
#   8. Figures and saved CSV tables
#
# Important:
#   p001 maps to participant-information SID S001j
#   p002 maps to S002j
#   ...
#   s179 maps directly to S179
#
# S179 is therefore classified as Control.
# ============================================================


# ============================================================
# 0. IMPORTS
# ============================================================

from pathlib import Path
import re
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.stats import (
    ttest_ind,
    mannwhitneyu,
    pearsonr,
    spearmanr,
)

import statsmodels.api as sm

from IPython.display import display


# ============================================================
# 1. CHANGEABLE SETTINGS
# ============================================================

LONG_FIXATION_CUTOFF_MS = 500.0

DRIVE_PROJECT_ROOT = Path(
    "/content/drive/MyDrive/steer_search_PD"
)

# Expected metadata-file locations
PARTICIPANT_INFO_CANDIDATES = [
    DRIVE_PROJECT_ROOT
    / "steer-search-PD"
    / "2025_spring_participants_info (1).csv",

    DRIVE_PROJECT_ROOT
    / "2025_spring_participants_info (1).csv",
]

# Fallback fixation file produced by the original diagnostic cell
DEFAULT_FIXATION_FILE = (
    DRIVE_PROJECT_ROOT
    / "all_remodnav"
    / "long_fixation_diagnostics"
    / "all_in_trial_fixations_with_diagnostics.csv"
)

# Output folder
if "OUT_DIR" in globals():
    BASE_OUTPUT_DIR = Path(OUT_DIR)
else:
    BASE_OUTPUT_DIR = (
        DRIVE_PROJECT_ROOT
        / "all_remodnav"
        / "long_fixation_diagnostics"
    )

AGE_OUTPUT_DIR = (
    BASE_OUTPUT_DIR
    / "500ms_age_analysis"
)

AGE_OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

print("Output folder:")
print(AGE_OUTPUT_DIR)


# ============================================================
# 2. HELPER FUNCTIONS
# ============================================================

def normalize_remodnav_subject(subject):
    """
    Normalize eye-tracking subject IDs.

    Examples:
        P001 -> p001
        p1   -> p001
        S179 -> s179
        s179 -> s179
    """

    if pd.isna(subject):
        return pd.NA

    value = str(subject).strip().lower()
    value = re.sub(r"\s+", "", value)

    pd_match = re.fullmatch(r"p(\d+)", value)

    if pd_match:
        subject_number = int(
            pd_match.group(1)
        )

        return f"p{subject_number:03d}"

    control_match = re.fullmatch(
        r"s(\d+)",
        value
    )

    if control_match:
        subject_number = int(
            control_match.group(1)
        )

        return f"s{subject_number:03d}"

    return value


def normalize_metadata_subject(subject):
    """
    Normalize participant-information SIDs.

    Examples:
        S001j -> s001j
        S012j -> s012j
        S179  -> s179
    """

    if pd.isna(subject):
        return pd.NA

    value = str(subject).strip().lower()
    value = re.sub(r"\s+", "", value)

    pd_match = re.fullmatch(
        r"s(\d+)j",
        value
    )

    if pd_match:
        subject_number = int(
            pd_match.group(1)
        )

        return f"s{subject_number:03d}j"

    control_match = re.fullmatch(
        r"s(\d+)",
        value
    )

    if control_match:
        subject_number = int(
            control_match.group(1)
        )

        return f"s{subject_number:03d}"

    return value


def make_metadata_match_id(remodnav_subject):
    """
    Convert the eye-tracking subject ID into the ID used
    in the participant-information CSV.

    Examples:
        p001 -> s001j
        p012 -> s012j
        s179 -> s179
    """

    subject = normalize_remodnav_subject(
        remodnav_subject
    )

    if pd.isna(subject):
        return pd.NA

    pd_match = re.fullmatch(
        r"p(\d{3})",
        str(subject)
    )

    if pd_match:
        return f"s{pd_match.group(1)}j"

    control_match = re.fullmatch(
        r"s(\d{3})",
        str(subject)
    )

    if control_match:
        return subject

    return subject


def derive_group_from_subject(subject):
    """
    Classify group from the REMoDNaV subject prefix.

        p### -> PD
        s### -> Control
    """

    subject = normalize_remodnav_subject(
        subject
    )

    if pd.isna(subject):
        return "Unknown"

    subject = str(subject)

    if subject.startswith("p"):
        return "PD"

    if subject.startswith("s"):
        return "Control"

    return "Unknown"


def safe_pearson(x, y):
    """
    Pearson correlation with missing-value handling.
    """

    temporary = pd.DataFrame({
        "x": pd.to_numeric(
            x,
            errors="coerce"
        ),
        "y": pd.to_numeric(
            y,
            errors="coerce"
        ),
    }).dropna()

    n = len(temporary)

    if (
        n < 3
        or temporary["x"].nunique() < 2
        or temporary["y"].nunique() < 2
    ):
        return np.nan, np.nan, n

    result = pearsonr(
        temporary["x"],
        temporary["y"]
    )

    return (
        float(result.statistic),
        float(result.pvalue),
        n,
    )


def safe_spearman(x, y):
    """
    Spearman correlation with missing-value handling.
    """

    temporary = pd.DataFrame({
        "x": pd.to_numeric(
            x,
            errors="coerce"
        ),
        "y": pd.to_numeric(
            y,
            errors="coerce"
        ),
    }).dropna()

    n = len(temporary)

    if (
        n < 3
        or temporary["x"].nunique() < 2
        or temporary["y"].nunique() < 2
    ):
        return np.nan, np.nan, n

    result = spearmanr(
        temporary["x"],
        temporary["y"]
    )

    return (
        float(result.statistic),
        float(result.pvalue),
        n,
    )


def hedges_g(group_a, group_b):
    """
    Hedges' g for group_a minus group_b.

    In this analysis:
        group_a = PD
        group_b = Control
    """

    group_a = pd.to_numeric(
        pd.Series(group_a),
        errors="coerce"
    ).dropna().to_numpy()

    group_b = pd.to_numeric(
        pd.Series(group_b),
        errors="coerce"
    ).dropna().to_numpy()

    n_a = len(group_a)
    n_b = len(group_b)

    if n_a < 2 or n_b < 2:
        return np.nan

    variance_a = np.var(
        group_a,
        ddof=1
    )

    variance_b = np.var(
        group_b,
        ddof=1
    )

    pooled_degrees_of_freedom = (
        n_a + n_b - 2
    )

    if pooled_degrees_of_freedom <= 0:
        return np.nan

    pooled_variance = (
        ((n_a - 1) * variance_a)
        + ((n_b - 1) * variance_b)
    ) / pooled_degrees_of_freedom

    if (
        not np.isfinite(pooled_variance)
        or pooled_variance <= 0
    ):
        return np.nan

    pooled_standard_deviation = np.sqrt(
        pooled_variance
    )

    cohens_d = (
        np.mean(group_a)
        - np.mean(group_b)
    ) / pooled_standard_deviation

    correction = (
        1.0
        - 3.0
        / (
            4.0 * (n_a + n_b)
            - 9.0
        )
    )

    return correction * cohens_d


def save_current_figure(filename):
    """
    Save and display a matplotlib figure.
    """

    output_path = (
        AGE_OUTPUT_DIR
        / filename
    )

    plt.tight_layout()

    plt.savefig(
        output_path,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()

    print(f"Saved: {output_path}")


# ============================================================
# 3. LOAD THE FIXATION DATA
# ============================================================

if (
    "fixations" in globals()
    and isinstance(fixations, pd.DataFrame)
    and not fixations.empty
):

    fixation_data = fixations.copy()

    print(
        "\nUsing the existing `fixations` dataframe "
        "from the previous cell."
    )

elif DEFAULT_FIXATION_FILE.exists():

    fixation_data = pd.read_csv(
        DEFAULT_FIXATION_FILE,
        low_memory=False
    )

    print(
        "\nThe `fixations` variable was not found."
    )

    print(
        "Loaded the saved fixation table:"
    )

    print(DEFAULT_FIXATION_FILE)

else:

    raise FileNotFoundError(
        "The `fixations` dataframe was not found, "
        "and the saved fixation CSV was not found.\n\n"
        "Run the original long-fixation diagnostic "
        "cell before this cell."
    )


required_fixation_columns = [
    "subject",
    "duration_ms",
]

missing_fixation_columns = [
    column
    for column in required_fixation_columns
    if column not in fixation_data.columns
]

if missing_fixation_columns:

    raise ValueError(
        "Required fixation columns are missing:\n"
        + "\n".join(
            missing_fixation_columns
        )
    )


# Normalize subject IDs
fixation_data["subject"] = (
    fixation_data["subject"]
    .apply(
        normalize_remodnav_subject
    )
)


# Ensure duration is numeric
fixation_data["duration_ms"] = pd.to_numeric(
    fixation_data["duration_ms"],
    errors="coerce"
)


# Remove rows without a valid duration
fixation_data = fixation_data.loc[
    fixation_data["duration_ms"].notna()
    & (
        fixation_data["duration_ms"] >= 0
    )
].copy()


# Define group based on subject prefix.
# This guarantees that s179 is Control.
fixation_data["group"] = (
    fixation_data["subject"]
    .apply(
        derive_group_from_subject
    )
)


# Flag fixations at or above the 500-ms threshold
fixation_data["is_fixation_ge_500ms"] = (
    fixation_data["duration_ms"]
    >= LONG_FIXATION_CUTOFF_MS
)


print(
    f"\nStrictly in-trial fixation rows used: "
    f"{len(fixation_data):,}"
)

print(
    f"Fixations ≥{LONG_FIXATION_CUTOFF_MS:.0f} ms: "
    f"{int(fixation_data['is_fixation_ge_500ms'].sum()):,}"
)


# ============================================================
# 4. CREATE PARTICIPANT-LEVEL FIXATION SUMMARY
# ============================================================

participant_500ms_summary = (
    fixation_data
    .groupby(
        [
            "group",
            "subject",
        ],
        as_index=False,
        dropna=False
    )
    .agg(
        n_total_fixations=(
            "duration_ms",
            "size"
        ),

        n_fixations_ge_500ms=(
            "is_fixation_ge_500ms",
            "sum"
        ),

        mean_fixation_duration_ms=(
            "duration_ms",
            "mean"
        ),

        median_fixation_duration_ms=(
            "duration_ms",
            "median"
        ),

        p90_fixation_duration_ms=(
            "duration_ms",
            lambda values: np.nanpercentile(
                values,
                90
            )
        ),

        p95_fixation_duration_ms=(
            "duration_ms",
            lambda values: np.nanpercentile(
                values,
                95
            )
        ),

        maximum_fixation_duration_ms=(
            "duration_ms",
            "max"
        ),
    )
)


participant_500ms_summary[
    "proportion_fixations_ge_500ms"
] = (
    participant_500ms_summary[
        "n_fixations_ge_500ms"
    ]
    / participant_500ms_summary[
        "n_total_fixations"
    ]
)


participant_500ms_summary[
    "percentage_fixations_ge_500ms"
] = (
    100.0
    * participant_500ms_summary[
        "proportion_fixations_ge_500ms"
    ]
)


participant_500ms_summary[
    "fixations_ge_500ms_per_100"
] = (
    participant_500ms_summary[
        "percentage_fixations_ge_500ms"
    ]
)


participant_500ms_summary[
    "metadata_match_id"
] = (
    participant_500ms_summary[
        "subject"
    ]
    .apply(
        make_metadata_match_id
    )
)


# ============================================================
# 5. LOCATE PARTICIPANT-INFORMATION FILE
# ============================================================

participant_info_path = None

for candidate_path in PARTICIPANT_INFO_CANDIDATES:

    if candidate_path.exists():

        participant_info_path = (
            candidate_path
        )

        break


# Recursive fallback if the expected path was not found
if participant_info_path is None:

    possible_matches = sorted(
        DRIVE_PROJECT_ROOT.rglob(
            "2025_spring_participants_info (1).csv"
        )
    )

    if possible_matches:

        participant_info_path = (
            possible_matches[0]
        )


if participant_info_path is None:

    raise FileNotFoundError(
        "Could not locate:\n"
        "2025_spring_participants_info (1).csv\n\n"
        "The code searched the steer_search_PD "
        "folder and all of its subfolders."
    )


print(
    "\nParticipant-information file:"
)

print(participant_info_path)


# ============================================================
# 6. LOAD PARTICIPANT-INFORMATION FILE
#
# FIX:
# The file is explicitly loaded as comma-separated.
# ============================================================

participant_info = pd.read_csv(
    participant_info_path,
    sep=",",
    encoding="utf-8-sig",
    low_memory=False
)


# Clean the column names
participant_info.columns = (
    participant_info.columns
    .astype(str)
    .str.replace(
        "\ufeff",
        "",
        regex=False
    )
    .str.strip()
)


print(
    "\nParticipant-information columns:"
)

print(
    participant_info.columns.tolist()
)


print(
    "\nFirst five participant-information rows:"
)

display(
    participant_info.head()
)


# ============================================================
# 7. IDENTIFY AND CLEAN METADATA COLUMNS
# ============================================================

column_lookup = {
    column.lower().strip(): column
    for column in participant_info.columns
}


if "sid" not in column_lookup:

    raise ValueError(
        "The participant-information file does not "
        "contain an SID column.\n\n"
        f"Columns found:\n"
        f"{participant_info.columns.tolist()}"
    )


if "age" not in column_lookup:

    raise ValueError(
        "The participant-information file does not "
        "contain an Age column.\n\n"
        f"Columns found:\n"
        f"{participant_info.columns.tolist()}"
    )


sid_column = column_lookup["sid"]
age_column = column_lookup["age"]

condition_column = column_lookup.get(
    "condition"
)

years_condition_column = column_lookup.get(
    "years of condition"
)

moca_column = column_lookup.get(
    "moca score"
)

gender_column = column_lookup.get(
    "gender"
)

dob_column = column_lookup.get(
    "dob"
)

experiment_date_column = column_lookup.get(
    "date of experiment"
)


participant_info["metadata_subject"] = (
    participant_info[sid_column]
    .apply(
        normalize_metadata_subject
    )
)


participant_info["age"] = pd.to_numeric(
    participant_info[age_column],
    errors="coerce"
)


metadata_columns = [
    "metadata_subject",
    "age",
]


optional_metadata_mapping = {
    "metadata_condition": condition_column,
    "years_of_condition": years_condition_column,
    "moca_score": moca_column,
    "gender": gender_column,
    "dob": dob_column,
    "date_of_experiment": experiment_date_column,
}


for new_column, original_column in (
    optional_metadata_mapping.items()
):

    if original_column is not None:

        participant_info[new_column] = (
            participant_info[
                original_column
            ]
        )

        metadata_columns.append(
            new_column
        )


participant_metadata = (
    participant_info[
        metadata_columns
    ]
    .drop_duplicates(
        subset=[
            "metadata_subject"
        ],
        keep="first"
    )
    .copy()
)


# ============================================================
# 8. MERGE FIXATION SUMMARY WITH AGE INFORMATION
# ============================================================

analysis_500ms = (
    participant_500ms_summary
    .merge(
        participant_metadata,
        left_on="metadata_match_id",
        right_on="metadata_subject",
        how="left",
        validate="many_to_one"
    )
)


analysis_500ms = (
    analysis_500ms
    .sort_values(
        [
            "group",
            "subject",
        ]
    )
    .reset_index(
        drop=True
    )
)


print(
    "\nSUBJECT-LEVEL ≥500-MS FIXATION SUMMARY"
)


summary_display_columns = [
    "group",
    "subject",
    "metadata_subject",
    "metadata_condition",
    "age",
    "gender",
    "n_total_fixations",
    "n_fixations_ge_500ms",
    "percentage_fixations_ge_500ms",
    "median_fixation_duration_ms",
    "p95_fixation_duration_ms",
    "maximum_fixation_duration_ms",
]


summary_display_columns = [
    column
    for column in summary_display_columns
    if column in analysis_500ms.columns
]


display(
    analysis_500ms[
        summary_display_columns
    ]
    .sort_values(
        "percentage_fixations_ge_500ms",
        ascending=False
    )
)


# ============================================================
# 9. CHECK WHETHER AGES MATCHED CORRECTLY
# ============================================================

missing_age = analysis_500ms.loc[
    analysis_500ms["age"].isna(),
    [
        "group",
        "subject",
        "metadata_match_id",
    ]
]


if missing_age.empty:

    print(
        "\nAge successfully matched for every "
        "eye-tracking participant."
    )

else:

    print(
        "\nWARNING: Age could not be matched "
        "for the following subjects:"
    )

    display(
        missing_age
    )


# Check that S179 was classified correctly
s179_check = analysis_500ms.loc[
    analysis_500ms["subject"] == "s179",
    [
        "subject",
        "group",
        "metadata_subject",
        "metadata_condition",
        "age",
    ]
]


print(
    "\nS179 classification check:"
)

if s179_check.empty:

    print(
        "s179 was not present in the fixation dataset."
    )

else:

    display(
        s179_check
    )


# Show metadata conditions that differ from the
# eye-tracking PD/Control grouping
if "metadata_condition" in analysis_500ms.columns:

    condition_check = analysis_500ms[
        [
            "subject",
            "group",
            "metadata_condition",
            "age",
        ]
    ].copy()

    condition_check[
        "metadata_condition_clean"
    ] = (
        condition_check[
            "metadata_condition"
        ]
        .astype("string")
        .str.strip()
        .str.lower()
    )

    unusual_conditions = (
        condition_check.loc[
            condition_check[
                "metadata_condition_clean"
            ].notna()
            & ~condition_check[
                "metadata_condition_clean"
            ].isin(
                [
                    "control",
                    "parkinson's",
                    "parkinsons",
                    "parkinson",
                ]
            ),
            [
                "subject",
                "group",
                "metadata_condition",
                "age",
            ]
        ]
    )

    if not unusual_conditions.empty:

        print(
            "\nMetadata conditions requiring attention:"
        )

        display(
            unusual_conditions
        )

        print(
            "The statistical group remains based on "
            "the REMoDNaV subject prefix:"
            "\n  p### = PD"
            "\n  s### = Control"
        )


# ============================================================
# 10. GROUP SUMMARY
# ============================================================

group_summary_500ms = (
    analysis_500ms
    .groupby(
        "group",
        as_index=False
    )
    .agg(
        n_participants=(
            "subject",
            "nunique"
        ),

        total_fixations=(
            "n_total_fixations",
            "sum"
        ),

        total_fixations_ge_500ms=(
            "n_fixations_ge_500ms",
            "sum"
        ),

        mean_subject_percentage_ge_500ms=(
            "percentage_fixations_ge_500ms",
            "mean"
        ),

        median_subject_percentage_ge_500ms=(
            "percentage_fixations_ge_500ms",
            "median"
        ),

        standard_deviation_subject_percentage=(
            "percentage_fixations_ge_500ms",
            "std"
        ),

        minimum_subject_percentage=(
            "percentage_fixations_ge_500ms",
            "min"
        ),

        maximum_subject_percentage=(
            "percentage_fixations_ge_500ms",
            "max"
        ),

        mean_age=(
            "age",
            "mean"
        ),

        median_age=(
            "age",
            "median"
        ),
    )
)


group_summary_500ms[
    "pooled_event_percentage_ge_500ms"
] = (
    100.0
    * group_summary_500ms[
        "total_fixations_ge_500ms"
    ]
    / group_summary_500ms[
        "total_fixations"
    ]
)


print(
    "\nGROUP SUMMARY"
)

display(
    group_summary_500ms
)


# ============================================================
# 11. PD VERSUS CONTROL COMPARISON
#
# The statistical unit is the participant.
# Each participant contributes one proportion.
# ============================================================

pd_proportions = analysis_500ms.loc[
    analysis_500ms["group"] == "PD",
    "proportion_fixations_ge_500ms"
].dropna()


control_proportions = analysis_500ms.loc[
    analysis_500ms["group"] == "Control",
    "proportion_fixations_ge_500ms"
].dropna()


if (
    len(pd_proportions) >= 2
    and len(control_proportions) >= 2
):

    welch_result = ttest_ind(
        pd_proportions,
        control_proportions,
        equal_var=False,
        nan_policy="omit"
    )

    mannwhitney_result = mannwhitneyu(
        pd_proportions,
        control_proportions,
        alternative="two-sided"
    )

    hedges_g_value = hedges_g(
        pd_proportions,
        control_proportions
    )

else:

    welch_result = None
    mannwhitney_result = None
    hedges_g_value = np.nan


group_comparison_500ms = pd.DataFrame([
    {
        "outcome": (
            "Participant proportion of "
            "fixations >=500 ms"
        ),

        "n_pd": len(
            pd_proportions
        ),

        "n_control": len(
            control_proportions
        ),

        "pd_mean_proportion": (
            pd_proportions.mean()
        ),

        "pd_median_proportion": (
            pd_proportions.median()
        ),

        "control_mean_proportion": (
            control_proportions.mean()
        ),

        "control_median_proportion": (
            control_proportions.median()
        ),

        "pd_mean_percentage": (
            100.0
            * pd_proportions.mean()
        ),

        "control_mean_percentage": (
            100.0
            * control_proportions.mean()
        ),

        "mean_difference_percentage_points_pd_minus_control": (
            100.0
            * (
                pd_proportions.mean()
                - control_proportions.mean()
            )
        ),

        "welch_t": (
            float(
                welch_result.statistic
            )
            if welch_result is not None
            else np.nan
        ),

        "welch_p": (
            float(
                welch_result.pvalue
            )
            if welch_result is not None
            else np.nan
        ),

        "mannwhitney_u": (
            float(
                mannwhitney_result.statistic
            )
            if mannwhitney_result is not None
            else np.nan
        ),

        "mannwhitney_p": (
            float(
                mannwhitney_result.pvalue
            )
            if mannwhitney_result is not None
            else np.nan
        ),

        "hedges_g_pd_minus_control": (
            hedges_g_value
        ),
    }
])


print(
    "\nPD VERSUS CONTROL: "
    "PARTICIPANT-LEVEL ≥500-MS FIXATION PROPORTION"
)

display(
    group_comparison_500ms
)


# ============================================================
# 12. AGE CORRELATIONS
#
# Correlations are computed using participant-level percentages.
# ============================================================

age_analysis_data = analysis_500ms.loc[
    analysis_500ms["age"].notna()
].copy()


age_correlation_rows = []


for group_name in [
    "All",
    "Control",
    "PD",
]:

    if group_name == "All":

        temporary_group = (
            age_analysis_data.copy()
        )

    else:

        temporary_group = (
            age_analysis_data.loc[
                age_analysis_data["group"]
                == group_name
            ].copy()
        )


    pearson_r, pearson_p, pearson_n = (
        safe_pearson(
            temporary_group["age"],
            temporary_group[
                "percentage_fixations_ge_500ms"
            ]
        )
    )


    spearman_rho, spearman_p, spearman_n = (
        safe_spearman(
            temporary_group["age"],
            temporary_group[
                "percentage_fixations_ge_500ms"
            ]
        )
    )


    age_correlation_rows.append({
        "group": group_name,
        "method": "Pearson",
        "n_participants": pearson_n,
        "correlation": pearson_r,
        "p_value": pearson_p,
        "outcome": (
            "Percentage of fixations >=500 ms"
        ),
    })


    age_correlation_rows.append({
        "group": group_name,
        "method": "Spearman",
        "n_participants": spearman_n,
        "correlation": spearman_rho,
        "p_value": spearman_p,
        "outcome": (
            "Percentage of fixations >=500 ms"
        ),
    })


age_correlation_results = pd.DataFrame(
    age_correlation_rows
)


print(
    "\nAGE CORRELATIONS WITH THE PERCENTAGE "
    "OF FIXATIONS ≥500 MS"
)

display(
    age_correlation_results
)


# ============================================================
# 13. PARTICIPANT-LEVEL AGE × GROUP REGRESSION
#
# Outcome:
#   Percentage of fixations ≥500 ms
#
# Predictors:
#   age_centered
#   group_PD
#   age_centered × group_PD
#
# Interpretation:
#
#   age_centered
#       Age slope among Controls
#
#   group_PD
#       PD versus Control difference at the mean age
#
#   age_centered_x_group_PD
#       Difference between the PD and Control age slopes
#
# HC3 robust standard errors are used.
# ============================================================

regression_data = age_analysis_data.loc[
    age_analysis_data["group"].isin(
        [
            "Control",
            "PD",
        ]
    )
].copy()


age_mean = (
    regression_data["age"].mean()
)


regression_data["age_centered"] = (
    regression_data["age"]
    - age_mean
)


regression_data["group_PD"] = (
    regression_data["group"]
    == "PD"
).astype(int)


regression_data[
    "age_centered_x_group_PD"
] = (
    regression_data[
        "age_centered"
    ]
    * regression_data[
        "group_PD"
    ]
)


age_group_predictors = pd.DataFrame({
    "Intercept": np.ones(
        len(regression_data)
    ),

    "age_centered": regression_data[
        "age_centered"
    ].to_numpy(),

    "group_PD": regression_data[
        "group_PD"
    ].to_numpy(),

    "age_centered_x_group_PD": regression_data[
        "age_centered_x_group_PD"
    ].to_numpy(),
})


age_group_outcome = (
    regression_data[
        "percentage_fixations_ge_500ms"
    ]
    .to_numpy()
)


if len(regression_data) >= 8:

    age_group_model = sm.OLS(
        age_group_outcome,
        age_group_predictors
    ).fit(
        cov_type="HC3"
    )


    age_group_confidence_intervals = (
        age_group_model.conf_int(
            alpha=0.05
        )
    )


    age_group_regression_results = pd.DataFrame({
        "term": (
            age_group_model.params.index
        ),

        "beta_percentage_points": (
            age_group_model.params.values
        ),

        "robust_standard_error": (
            age_group_model.bse.values
        ),

        "t_value": (
            age_group_model.tvalues.values
        ),

        "p_value": (
            age_group_model.pvalues.values
        ),

        "ci_95_lower": (
            age_group_confidence_intervals[
                0
            ].values
        ),

        "ci_95_upper": (
            age_group_confidence_intervals[
                1
            ].values
        ),
    })


    print(
        "\nPARTICIPANT-LEVEL AGE × GROUP REGRESSION"
    )

    print(
        f"Age was centered at "
        f"{age_mean:.2f} years."
    )

    print(
        "\nInterpretation:"
        "\n  age_centered:"
        "\n      Age slope in Controls."
        "\n  group_PD:"
        "\n      PD-Control difference at the "
        f"mean age of {age_mean:.2f} years."
        "\n  age_centered_x_group_PD:"
        "\n      Difference between PD and "
        "Control age slopes."
    )

    display(
        age_group_regression_results
    )

    print(
        "\nFull age × group regression summary:"
    )

    print(
        age_group_model.summary()
    )

else:

    warnings.warn(
        "Too few participants with matched ages "
        "for the age-by-group regression."
    )

    age_group_regression_results = pd.DataFrame(
        columns=[
            "term",
            "beta_percentage_points",
            "robust_standard_error",
            "t_value",
            "p_value",
            "ci_95_lower",
            "ci_95_upper",
        ]
    )


# ============================================================
# 14. AGE-ADJUSTED GROUP REGRESSION WITHOUT INTERACTION
#
# This simpler model estimates the PD-Control difference
# after adjusting for age:
#
#   percentage ≥500 ms ~ age + group
# ============================================================

age_adjusted_predictors = pd.DataFrame({
    "Intercept": np.ones(
        len(regression_data)
    ),

    "age_centered": regression_data[
        "age_centered"
    ].to_numpy(),

    "group_PD": regression_data[
        "group_PD"
    ].to_numpy(),
})


if len(regression_data) >= 6:

    age_adjusted_model = sm.OLS(
        age_group_outcome,
        age_adjusted_predictors
    ).fit(
        cov_type="HC3"
    )


    age_adjusted_confidence_intervals = (
        age_adjusted_model.conf_int(
            alpha=0.05
        )
    )


    age_adjusted_regression_results = pd.DataFrame({
        "term": (
            age_adjusted_model.params.index
        ),

        "beta_percentage_points": (
            age_adjusted_model.params.values
        ),

        "robust_standard_error": (
            age_adjusted_model.bse.values
        ),

        "t_value": (
            age_adjusted_model.tvalues.values
        ),

        "p_value": (
            age_adjusted_model.pvalues.values
        ),

        "ci_95_lower": (
            age_adjusted_confidence_intervals[
                0
            ].values
        ),

        "ci_95_upper": (
            age_adjusted_confidence_intervals[
                1
            ].values
        ),
    })


    print(
        "\nAGE-ADJUSTED PD VERSUS CONTROL REGRESSION"
    )

    print(
        "The group_PD coefficient is the estimated "
        "PD-Control difference in percentage points "
        "after adjustment for age."
    )

    display(
        age_adjusted_regression_results
    )

    print(
        "\nFull age-adjusted regression summary:"
    )

    print(
        age_adjusted_model.summary()
    )

else:

    age_adjusted_regression_results = pd.DataFrame(
        columns=[
            "term",
            "beta_percentage_points",
            "robust_standard_error",
            "t_value",
            "p_value",
            "ci_95_lower",
            "ci_95_upper",
        ]
    )


# ============================================================
# 15. SAVE ALL TABLES
# ============================================================

participant_summary_path = (
    AGE_OUTPUT_DIR
    / "participant_500ms_age_summary.csv"
)

group_summary_path = (
    AGE_OUTPUT_DIR
    / "group_500ms_summary.csv"
)

group_comparison_path = (
    AGE_OUTPUT_DIR
    / "group_comparison_500ms.csv"
)

correlation_path = (
    AGE_OUTPUT_DIR
    / "age_correlation_results.csv"
)

age_group_regression_path = (
    AGE_OUTPUT_DIR
    / "age_group_interaction_regression_hc3.csv"
)

age_adjusted_regression_path = (
    AGE_OUTPUT_DIR
    / "age_adjusted_group_regression_hc3.csv"
)


analysis_500ms.to_csv(
    participant_summary_path,
    index=False
)

group_summary_500ms.to_csv(
    group_summary_path,
    index=False
)

group_comparison_500ms.to_csv(
    group_comparison_path,
    index=False
)

age_correlation_results.to_csv(
    correlation_path,
    index=False
)

age_group_regression_results.to_csv(
    age_group_regression_path,
    index=False
)

age_adjusted_regression_results.to_csv(
    age_adjusted_regression_path,
    index=False
)


print(
    "\nSaved tables:"
)

print(participant_summary_path)
print(group_summary_path)
print(group_comparison_path)
print(correlation_path)
print(age_group_regression_path)
print(age_adjusted_regression_path)


# ============================================================
# 16. FIGURE 1:
# PARTICIPANT-LEVEL PERCENTAGE OF FIXATIONS ≥500 MS
# ============================================================

participant_plot = (
    analysis_500ms
    .sort_values(
        "percentage_fixations_ge_500ms",
        ascending=True
    )
    .copy()
)


figure_height = max(
    6,
    0.36 * len(
        participant_plot
    ) + 2
)


plt.figure(
    figsize=(
        11,
        figure_height
    )
)


bar_positions = np.arange(
    len(participant_plot)
)


plt.barh(
    bar_positions,
    participant_plot[
        "percentage_fixations_ge_500ms"
    ]
)


plt.yticks(
    bar_positions,
    [
        f"{subject} ({group})"
        for subject, group in zip(
            participant_plot[
                "subject"
            ],
            participant_plot[
                "group"
            ]
        )
    ]
)


maximum_percentage = (
    participant_plot[
        "percentage_fixations_ge_500ms"
    ].max()
)


if not np.isfinite(
    maximum_percentage
):

    maximum_percentage = 1.0


text_offset = max(
    0.05,
    0.015 * maximum_percentage
)


for plot_position, (_, row) in enumerate(
    participant_plot.iterrows()
):

    plt.text(
        row[
            "percentage_fixations_ge_500ms"
        ] + text_offset,

        plot_position,

        (
            f"{int(row['n_fixations_ge_500ms'])}"
            f"/{int(row['n_total_fixations'])}"
        ),

        va="center",
        fontsize=8
    )


plt.xlabel(
    "Percentage of strictly in-trial "
    "FIXA events ≥500 ms"
)

plt.ylabel(
    "Participant"
)

plt.title(
    "Participant-level prevalence of ≥500-ms fixations\n"
    "Labels show ≥500-ms count / total fixation count"
)


plt.xlim(
    left=0,
    right=(
        maximum_percentage
        + max(
            1.0,
            0.20 * maximum_percentage
        )
    )
)


save_current_figure(
    "participant_500ms_percentage.png"
)


# ============================================================
# 17. FIGURE 2:
# PD VERSUS CONTROL PARTICIPANT-LEVEL PROPORTIONS
# ============================================================

control_plot_values = analysis_500ms.loc[
    analysis_500ms["group"] == "Control",
    "percentage_fixations_ge_500ms"
].dropna().to_numpy()


pd_plot_values = analysis_500ms.loc[
    analysis_500ms["group"] == "PD",
    "percentage_fixations_ge_500ms"
].dropna().to_numpy()


plt.figure(
    figsize=(
        7,
        6
    )
)


plt.boxplot(
    [
        control_plot_values,
        pd_plot_values,
    ],
    tick_labels=[
        "Control",
        "PD",
    ],
    showfliers=False
)


random_generator = np.random.default_rng(
    12345
)


for group_position, group_values in enumerate(
    [
        control_plot_values,
        pd_plot_values,
    ],
    start=1
):

    jitter = random_generator.normal(
        loc=0,
        scale=0.035,
        size=len(group_values)
    )

    plt.scatter(
        np.full(
            len(group_values),
            group_position
        ) + jitter,
        group_values,
        alpha=0.85,
        s=55
    )


if welch_result is not None:

    group_plot_title = (
        "Participant-level percentage of "
        "fixations ≥500 ms\n"
        f"Welch p={welch_result.pvalue:.4g}, "
        f"Hedges' g={hedges_g_value:.3f}"
    )

else:

    group_plot_title = (
        "Participant-level percentage of "
        "fixations ≥500 ms"
    )


plt.ylabel(
    "Percentage of strictly in-trial "
    "FIXA events ≥500 ms"
)

plt.xlabel(
    "Group"
)

plt.title(
    group_plot_title
)


save_current_figure(
    "pd_control_500ms_percentage.png"
)


# ============================================================
# 18. FIGURE 3:
# AGE VERSUS PERCENTAGE OF FIXATIONS ≥500 MS
# ============================================================

plt.figure(
    figsize=(
        10,
        7
    )
)


for group_name in [
    "Control",
    "PD",
]:

    group_data = age_analysis_data.loc[
        age_analysis_data["group"]
        == group_name
    ].copy()

    if group_data.empty:
        continue

    plt.scatter(
        group_data["age"],
        group_data[
            "percentage_fixations_ge_500ms"
        ],
        s=75,
        alpha=0.85,
        label=group_name
    )


    # Add separate linear trend lines
    if (
        len(group_data) >= 2
        and group_data["age"].nunique() >= 2
    ):

        slope, intercept = np.polyfit(
            group_data["age"],
            group_data[
                "percentage_fixations_ge_500ms"
            ],
            deg=1
        )

        age_line = np.linspace(
            group_data["age"].min(),
            group_data["age"].max(),
            100
        )

        outcome_line = (
            intercept
            + slope * age_line
        )

        plt.plot(
            age_line,
            outcome_line,
            linewidth=2
        )


# Participant labels
for _, row in age_analysis_data.iterrows():

    plt.annotate(
        row["subject"],

        (
            row["age"],
            row[
                "percentage_fixations_ge_500ms"
            ]
        ),

        xytext=(
            4,
            4
        ),

        textcoords="offset points",
        fontsize=8,
        alpha=0.8
    )


overall_spearman = (
    age_correlation_results.loc[
        (
            age_correlation_results[
                "group"
            ] == "All"
        )
        & (
            age_correlation_results[
                "method"
            ] == "Spearman"
        )
    ]
)


if not overall_spearman.empty:

    overall_rho = overall_spearman[
        "correlation"
    ].iloc[0]

    overall_spearman_p = overall_spearman[
        "p_value"
    ].iloc[0]

    age_plot_statistics = (
        f"Overall Spearman ρ="
        f"{overall_rho:.3f}, "
        f"p={overall_spearman_p:.4g}"
    )

else:

    age_plot_statistics = ""


plt.xlabel(
    "Participant age (years)"
)

plt.ylabel(
    "Percentage of strictly in-trial "
    "FIXA events ≥500 ms"
)

plt.title(
    "Age versus prevalence of ≥500-ms fixations\n"
    + age_plot_statistics
)

plt.legend()


save_current_figure(
    "age_vs_500ms_fixation_percentage.png"
)


# ============================================================
# 19. FIGURE 4:
# AGE VERSUS RAW NUMBER OF FIXATIONS ≥500 MS
#
# This is descriptive only.
# Raw counts depend on the total number of fixations contributed.
# ============================================================

plt.figure(
    figsize=(
        10,
        7
    )
)


for group_name in [
    "Control",
    "PD",
]:

    group_data = age_analysis_data.loc[
        age_analysis_data["group"]
        == group_name
    ].copy()

    if group_data.empty:
        continue

    plt.scatter(
        group_data["age"],
        group_data[
            "n_fixations_ge_500ms"
        ],
        s=75,
        alpha=0.85,
        label=group_name
    )


for _, row in age_analysis_data.iterrows():

    plt.annotate(
        row["subject"],

        (
            row["age"],
            row[
                "n_fixations_ge_500ms"
            ]
        ),

        xytext=(
            4,
            4
        ),

        textcoords="offset points",
        fontsize=8,
        alpha=0.8
    )


plt.xlabel(
    "Participant age (years)"
)

plt.ylabel(
    "Raw number of fixations ≥500 ms"
)

plt.title(
    "Age versus raw count of ≥500-ms fixations\n"
    "Raw counts depend on total fixation count"
)

plt.legend()


save_current_figure(
    "age_vs_raw_number_500ms_fixations.png"
)


# ============================================================
# 20. FINAL CONSOLE SUMMARY
# ============================================================

total_fixations = int(
    analysis_500ms[
        "n_total_fixations"
    ].sum()
)


total_500ms_fixations = int(
    analysis_500ms[
        "n_fixations_ge_500ms"
    ].sum()
)


overall_event_percentage = (
    100.0
    * total_500ms_fixations
    / total_fixations
)


print(
    "\n"
    + "=" * 75
)

print(
    "FINAL ≥500-MS FIXATION AND AGE ANALYSIS"
)

print(
    "=" * 75
)


print(
    f"Participants in fixation dataset: "
    f"{analysis_500ms['subject'].nunique()}"
)


print(
    f"Participants with matched ages: "
    f"{age_analysis_data['subject'].nunique()}"
)


print(
    f"All strictly in-trial fixations: "
    f"{total_fixations:,}"
)


print(
    f"Fixations ≥"
    f"{LONG_FIXATION_CUTOFF_MS:.0f} ms: "
    f"{total_500ms_fixations:,}"
)


print(
    f"Overall event-level percentage ≥500 ms: "
    f"{overall_event_percentage:.2f}%"
)


print(
    "\nImportant interpretation:"
)

print(
    "• The raw number of ≥500-ms fixations depends on "
    "the total number of fixations contributed by each subject."
)

print(
    "• The main between-subject outcome is therefore the "
    "percentage of each participant's fixations lasting ≥500 ms."
)

print(
    "• Correlations and regressions use one row per participant, "
    "rather than treating individual fixation events as "
    "independent observations."
)

print(
    "• The age × group interaction tests whether the relationship "
    "between age and long-fixation percentage differs between "
    "PD and Control participants."
)

print(
    "\nAll outputs were saved in:"
)

print(
    AGE_OUTPUT_DIR
)
```

```python
# ============================================================
# S189: SIMPLE ONE-BLOCK HARMONIC-OSCILLATOR PILOT
#
# PURPOSE
# -------
# 1. Load REMoDNaV SACC events for s189.
# 2. Find one block with enough clean 4–6° saccades.
# 3. Reconstruct RAW calibrated sample-level gaze from the EDF.
# 4. Extract -20 to +120 ms around each saccade onset.
# 5. Project 2D gaze onto the saccade direction.
# 6. Calculate one median trajectory.
# 7. Fit the constant harmonic-oscillator model:
#
#       observed pupil trajectory = x_eye(t) + y_iris(t)
#
#       y_iris'' + gamma*y_iris' + k*y_iris = -x_eye''
#
#    Fixed:
#       beta = 1
#
#    Estimated:
#       A, mu, gamma, k
#
# This is only a model-validation pilot.
# It does NOT perform PD-versus-Control statistics.
# ============================================================


# ============================================================
# 0. IMPORTS
# ============================================================
# ============================================================
# INSTALL EYELINKIO DEPENDENCY
# Run this before the harmonic-model pilot cell
# ============================================================

import sys
import subprocess
import importlib

# Remove any incomplete or conflicting installation
subprocess.run(
    [
        sys.executable,
        "-m",
        "pip",
        "uninstall",
        "-y",
        "eyelinkio",
    ],
    check=False
)

# Install the same custom fork used by your REMoDNaV pipeline
try:
    subprocess.run(
        [
            sys.executable,
            "-m",
            "pip",
            "install",
            "-q",
            "--no-deps",
            "--force-reinstall",
            "git+https://github.com/pandeychetanya/"
            "eyelinkio.git@chetanyaedits",
        ],
        check=True
    )

    print("Installed the custom eyelinkio fork.")

except subprocess.CalledProcessError:

    print(
        "Custom fork installation failed. "
        "Installing the public eyelinkio package instead."
    )

    subprocess.run(
        [
            sys.executable,
            "-m",
            "pip",
            "install",
            "-q",
            "--force-reinstall",
            "eyelinkio",
        ],
        check=True
    )

importlib.invalidate_caches()

# Verification
import eyelinkio

print("eyelinkio imported successfully.")
print("Location:", eyelinkio.__file__)

from pathlib import Path
import sys
import importlib
import re
import warnings
import traceback

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.integrate import solve_ivp
from scipy.optimize import least_squares
from scipy.special import gamma as gamma_function
from scipy.special import hyp1f1

from IPython.display import display


# ============================================================
# 1. SETTINGS
# ============================================================

SUBJECT_ID = "s189"

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/steer_search_PD"
)

DATA_ROOT = (
    PROJECT_ROOT
    / "steer-search-PD"
)

SUBJECT_ROOT = (
    DATA_ROOT
    / SUBJECT_ID
)

BEHAV_DIR = SUBJECT_ROOT / "behav"
EYE_DIR = SUBJECT_ROOT / "eye"

# Main expected REMoDNaV output from your supplied code
REMODNAV_CANDIDATES = [
    PROJECT_ROOT
    / "remodnav_outputs_calib"
    / "all_saccades_s189_remodnav.csv",

    PROJECT_ROOT
    / "all_remodnav"
    / "all_remodnav_events_s189.csv",
]

OUT_DIR = (
    PROJECT_ROOT
    / "harmonic_model_simple"
    / "s189_one_block_pilot"
)

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# Candidate saccade range
MIN_AMPLITUDE_DEG = 4.0
MAX_AMPLITUDE_DEG = 6.0

# Extracted time window
PRE_ONSET_MS = 20
POST_ONSET_MS = 120

# Pre-saccadic interval used to estimate the starting position
BASELINE_START_MS = -20
BASELINE_END_MS = -5

# Desired number of trajectories in the median
MIN_DESIRED_SACCADES = 20
MAX_SACCADES = 1000000

# If no block contains 20, permit a smaller diagnostic pilot
MIN_FALLBACK_SACCADES = 8

# Apply the same 10-ms blink dilation used by REMoDNaV
BLINK_DILATION_MS = 10.0

# Multistart optimization
N_OPTIMIZATION_STARTS = 24
RANDOM_SEED = 12345

# The paper fixes beta to 1 in the parsimonious model
BETA = 1.0

# Parameter bounds based on the ranges described in the paper.
#
# Time is represented in milliseconds:
#   A      approximately deg / ms^(beta+1)
#   gamma  approximately 1 / ms
#   k      approximately 1 / ms^2
PARAMETER_NAMES = [
    "A",
    "mu",
    "gamma",
    "k",
]

LOWER_BOUNDS = np.array([
    0.020,   # A
    1.000,   # mu
    0.050,   # gamma
    0.010,   # k
])

UPPER_BOUNDS = np.array([
    0.080,   # A
    4.000,   # mu
    0.200,   # gamma
    0.100,   # k
])


print("Output folder:")
print(OUT_DIR)


# ============================================================
# 2. IMPORT EXISTING EYEPROCESS HELPERS
# ============================================================

REPO_DIR = Path(
    "/content/drive/MyDrive/eyeprocess"
)

if not REPO_DIR.exists():
    raise FileNotFoundError(
        f"Could not find the eyeprocess repository:\n{REPO_DIR}"
    )

for path in [
    REPO_DIR,
    REPO_DIR.parent,
]:
    if str(path) not in sys.path:
        sys.path.insert(
            0,
            str(path)
        )

importlib.invalidate_caches()


from eyeprocess.utils.psychopy import el2deg

from eyeprocess.utils.main_sequence_helpers import (
    read_edf_like_pyedfread,
    load_behav_json,
    build_calib_params,
    parse_trials_from_messages,
    get_affine_for_trial,
)


print(
    "\nSuccessfully imported existing "
    "eyeprocess helper functions."
)


# ============================================================
# 3. SMALL GENERAL HELPERS
# ============================================================

def natural_sort_key(value):
    """
    Sort text containing numbers naturally.
    """

    parts = re.split(
        r"(\d+)",
        str(value)
    )

    return [
        int(part)
        if part.isdigit()
        else part.lower()
        for part in parts
    ]


def first_existing_path(paths):
    """
    Return the first existing path.
    """

    for path in paths:
        if path.exists():
            return path

    return None


def save_current_figure(filename):
    """
    Save, display and close the current figure.
    """

    path = OUT_DIR / filename

    plt.tight_layout()

    plt.savefig(
        path,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()

    print(f"Saved: {path}")


# ============================================================
# 4. BLOCK-MEAN CALIBRATION
#
# The lower-level operations use your imported helpers.
#
# This short wrapper is included because the block-mean
# calibrate_block() function is currently defined in your
# notebook, not in main_sequence_helpers.py.
# ============================================================

def calibrate_complete_block_mean(
    samples,
    trials_df,
    calib_params,
    behav
):
    """
    Convert left-eye gaze to degrees and apply one mean affine
    transformation to the complete EDF block.

    Each unique calibration session used within the block
    contributes once to the mean.
    """

    required_sample_columns = {
        "time",
        "gx_left",
        "gy_left",
    }

    missing_sample_columns = (
        required_sample_columns
        - set(samples.columns)
    )

    if missing_sample_columns:
        raise KeyError(
            "Missing required sample columns: "
            f"{sorted(missing_sample_columns)}"
        )


    required_trial_columns = {
        "trial",
        "image_onset_ms",
        "response_ms",
    }

    missing_trial_columns = (
        required_trial_columns
        - set(trials_df.columns)
    )

    if missing_trial_columns:
        raise KeyError(
            "Missing required trial columns: "
            f"{sorted(missing_trial_columns)}"
        )


    samples = (
        samples
        .copy()
        .sort_values("time")
        .reset_index(drop=True)
    )


    time_ms = pd.to_numeric(
        samples["time"],
        errors="coerce"
    ).to_numpy(dtype=float)


    x_pix = pd.to_numeric(
        samples["gx_left"],
        errors="coerce"
    ).to_numpy(dtype=float)


    y_pix = pd.to_numeric(
        samples["gy_left"],
        errors="coerce"
    ).to_numpy(dtype=float)


    monitor_resolution = (
        behav
        .get("config", {})
        .get(
            "monitor_resolution",
            [1920, 1080]
        )
    )

    screen_width = float(
        monitor_resolution[0]
    )

    screen_height = float(
        monitor_resolution[1]
    )


    invalid = (
        ~np.isfinite(x_pix)
        | ~np.isfinite(y_pix)
        | (
            np.abs(x_pix)
            > 10 * screen_width
        )
        | (
            np.abs(y_pix)
            > 10 * screen_height
        )
    )


    # Pupil area is not required.
    # This only uses it when a valid pupil column happens to exist.
    for pupil_column in [
        "pa_left",
        "pupil_left",
        "ps_left",
        "pupil",
    ]:

        if pupil_column not in samples.columns:
            continue

        pupil = pd.to_numeric(
            samples[pupil_column],
            errors="coerce"
        ).to_numpy(dtype=float)

        invalid |= (
            ~np.isfinite(pupil)
            | (pupil <= 0)
        )

        break


    x_pix[invalid] = np.nan
    y_pix[invalid] = np.nan


    # Reuse your existing pixel-to-degree function
    x_deg, y_deg = el2deg(
        x_pix,
        y_pix
    )

    x_deg = np.asarray(
        x_deg,
        dtype=float
    )

    y_deg = np.asarray(
        y_deg,
        dtype=float
    )


    valid_trials = (
        trials_df
        .copy()
    )

    valid_trials["trial"] = pd.to_numeric(
        valid_trials["trial"],
        errors="coerce"
    )

    valid_trials["image_onset_ms"] = pd.to_numeric(
        valid_trials["image_onset_ms"],
        errors="coerce"
    )

    valid_trials["response_ms"] = pd.to_numeric(
        valid_trials["response_ms"],
        errors="coerce"
    )

    valid_trials = (
        valid_trials
        .dropna(
            subset=[
                "trial",
                "image_onset_ms",
                "response_ms",
            ]
        )
        .sort_values("image_onset_ms")
        .drop_duplicates("trial")
        .reset_index(drop=True)
    )


    if valid_trials.empty:
        raise RuntimeError(
            "No valid IMAGE_ONSET-to-RESPONSE trials "
            "were found in this block."
        )


    session_affines = {}

    for trial_number in (
        valid_trials["trial"]
        .astype(int)
        .to_numpy()
    ):

        calibration_id, affine = (
            get_affine_for_trial(
                calib_params,
                int(trial_number)
            )
        )

        session_key = str(
            calibration_id
        )

        A_matrix = np.asarray(
            affine.A,
            dtype=float
        )

        b_vector = np.asarray(
            affine.b,
            dtype=float
        ).reshape(-1)


        if A_matrix.shape != (2, 2):
            raise ValueError(
                f"Calibration {calibration_id} "
                f"has A shape {A_matrix.shape}."
            )

        if b_vector.shape != (2,):
            raise ValueError(
                f"Calibration {calibration_id} "
                f"has b shape {b_vector.shape}."
            )


        if session_key not in session_affines:

            session_affines[
                session_key
            ] = (
                A_matrix.copy(),
                b_vector.copy()
            )

        else:

            previous_A, previous_b = (
                session_affines[
                    session_key
                ]
            )

            if not (
                np.allclose(
                    A_matrix,
                    previous_A
                )
                and np.allclose(
                    b_vector,
                    previous_b
                )
            ):
                raise RuntimeError(
                    "The same calibration session returned "
                    "different affine parameters."
                )


    if not session_affines:
        raise RuntimeError(
            "No usable calibration sessions were found."
        )


    ordered_session_keys = sorted(
        session_affines.keys(),
        key=natural_sort_key
    )


    A_stack = np.stack(
        [
            session_affines[key][0]
            for key in ordered_session_keys
        ],
        axis=0
    )


    b_stack = np.stack(
        [
            session_affines[key][1]
            for key in ordered_session_keys
        ],
        axis=0
    )


    mean_A = np.mean(
        A_stack,
        axis=0
    )

    mean_b = np.mean(
        b_stack,
        axis=0
    )


    x_calibrated = np.full(
        len(samples),
        np.nan,
        dtype=float
    )

    y_calibrated = np.full(
        len(samples),
        np.nan,
        dtype=float
    )


    valid_raw_gaze = (
        np.isfinite(x_deg)
        & np.isfinite(y_deg)
    )


    if valid_raw_gaze.any():

        points = np.column_stack([
            x_deg[valid_raw_gaze],
            y_deg[valid_raw_gaze],
        ])

        corrected_points = (
            points @ mean_A.T
            + mean_b
        )

        x_calibrated[
            valid_raw_gaze
        ] = corrected_points[:, 0]

        y_calibrated[
            valid_raw_gaze
        ] = corrected_points[:, 1]


    return {
        "samples": samples,
        "time_ms": time_ms,
        "x_calibrated": x_calibrated,
        "y_calibrated": y_calibrated,
        "valid_trials": valid_trials,
        "mean_A": mean_A,
        "mean_b": mean_b,
        "calibration_sessions": ordered_session_keys,
    }


# ============================================================
# 5. LOAD AND PREPARE ONE EDF BLOCK
# ============================================================

def load_calibrated_block(
    block_id
):
    """
    Load one behavior/EDF pair, apply the mean calibration,
    and mask blink intervals with 10-ms dilation.
    """

    behav_path = (
        BEHAV_DIR
        / f"{block_id}.json"
    )

    edf_path = (
        EYE_DIR
        / f"{block_id}.edf"
    )


    if not behav_path.exists():
        raise FileNotFoundError(
            f"Missing behavioral JSON:\n{behav_path}"
        )

    if not edf_path.exists():
        raise FileNotFoundError(
            f"Missing EDF:\n{edf_path}"
        )


    behav = load_behav_json(
        behav_path
    )

    samples, eyelink_events, messages = (
        read_edf_like_pyedfread(
            edf_path
        )
    )

    trials_df = parse_trials_from_messages(
        messages
    )

    calibration_parameters = (
        build_calib_params(
            behav
        )
    )


    calibrated = calibrate_complete_block_mean(
        samples=samples,
        trials_df=trials_df,
        calib_params=calibration_parameters,
        behav=behav
    )


    time_ms = calibrated[
        "time_ms"
    ].copy()

    x_calibrated = calibrated[
        "x_calibrated"
    ].copy()

    y_calibrated = calibrated[
        "y_calibrated"
    ].copy()


    # Synchronize missing-data masks
    invalid = (
        ~np.isfinite(x_calibrated)
        | ~np.isfinite(y_calibrated)
    )

    x_calibrated[invalid] = np.nan
    y_calibrated[invalid] = np.nan


    # Blink masking with the same 10-ms dilation
    # used in the REMoDNaV preprocessing.
    number_of_blinks = 0

    if (
        eyelink_events is not None
        and not eyelink_events.empty
        and {
            "type",
            "start",
            "end",
        }.issubset(
            eyelink_events.columns
        )
    ):

        blink_rows = eyelink_events.loc[
            eyelink_events["type"]
            .astype(str)
            .str.lower()
            .eq("blink")
        ]


        for _, blink_row in blink_rows.iterrows():

            blink_start = pd.to_numeric(
                blink_row["start"],
                errors="coerce"
            )

            blink_end = pd.to_numeric(
                blink_row["end"],
                errors="coerce"
            )

            if not (
                np.isfinite(blink_start)
                and np.isfinite(blink_end)
            ):
                continue


            blink_start -= BLINK_DILATION_MS
            blink_end += BLINK_DILATION_MS


            blink_mask = (
                (time_ms >= blink_start)
                & (time_ms <= blink_end)
            )


            if blink_mask.any():

                x_calibrated[
                    blink_mask
                ] = np.nan

                y_calibrated[
                    blink_mask
                ] = np.nan

                number_of_blinks += 1


    positive_time_differences = np.diff(
        time_ms
    )

    positive_time_differences = (
        positive_time_differences[
            np.isfinite(
                positive_time_differences
            )
            & (
                positive_time_differences > 0
            )
        ]
    )


    if not len(
        positive_time_differences
    ):
        raise RuntimeError(
            "Could not infer the EDF sampling interval."
        )


    median_sample_interval_ms = float(
        np.median(
            positive_time_differences
        )
    )

    sampling_rate_hz = (
        1000.0
        / median_sample_interval_ms
    )


    print(
        f"Loaded block {block_id} | "
        f"fs={sampling_rate_hz:.2f} Hz | "
        f"blink intervals={number_of_blinks} | "
        f"calibration sessions="
        f"{calibrated['calibration_sessions']}"
    )


    return {
        "block_id": block_id,
        "behav_path": behav_path,
        "edf_path": edf_path,
        "time_ms": time_ms,
        "x_calibrated": x_calibrated,
        "y_calibrated": y_calibrated,
        "sampling_rate_hz": sampling_rate_hz,
        "median_sample_interval_ms": (
            median_sample_interval_ms
        ),
        "mean_A": calibrated["mean_A"],
        "mean_b": calibrated["mean_b"],
        "calibration_sessions": calibrated[
            "calibration_sessions"
        ],
    }


# ============================================================
# 6. LOAD REMODNAV SACCADE TABLE
# ============================================================

remodnav_path = first_existing_path(
    REMODNAV_CANDIDATES
)


if remodnav_path is None:

    raise FileNotFoundError(
        "Could not find the s189 REMoDNaV event file.\n\n"
        "Checked:\n"
        + "\n".join(
            str(path)
            for path in REMODNAV_CANDIDATES
        )
    )


print("\nREMoDNaV input:")
print(remodnav_path)


remodnav_events = pd.read_csv(
    remodnav_path,
    low_memory=False
)


if "subject" in remodnav_events.columns:

    remodnav_events["subject"] = (
        remodnav_events["subject"]
        .astype("string")
        .str.strip()
        .str.lower()
    )

    remodnav_events = remodnav_events.loc[
        remodnav_events["subject"]
        == SUBJECT_ID
    ].copy()


if "label" not in remodnav_events.columns:

    # The canonical all_saccades CSV may contain only SACC rows.
    remodnav_events["label"] = "SACC"


remodnav_events["label"] = (
    remodnav_events["label"]
    .astype("string")
    .str.strip()
    .str.upper()
)


# This pilot uses major SACC events only
saccades = remodnav_events.loc[
    remodnav_events["label"]
    == "SACC"
].copy()


required_saccade_columns = [
    "block_id",
    "start_abs_ms",
    "end_abs_ms",
]


missing_saccade_columns = [
    column
    for column in required_saccade_columns
    if column not in saccades.columns
]


if missing_saccade_columns:

    raise ValueError(
        "Required REMoDNaV columns are missing:\n"
        + "\n".join(
            missing_saccade_columns
        )
    )


# Identify amplitude column
if "amplitude_deg" in saccades.columns:
    amplitude_column = "amplitude_deg"

elif "amp" in saccades.columns:
    amplitude_column = "amp"

else:
    raise ValueError(
        "Neither amplitude_deg nor amp was found."
    )


numeric_columns = [
    amplitude_column,
    "start_abs_ms",
    "end_abs_ms",
    "image_onset_ms",
    "response_ms",
    "start_x",
    "start_y",
    "end_x",
    "end_y",
    "dx_deg",
    "dy_deg",
    "trial",
    "saccade_index",
]


for column in numeric_columns:

    if column in saccades.columns:

        saccades[column] = pd.to_numeric(
            saccades[column],
            errors="coerce"
        )


saccades = (
    saccades
    .dropna(
        subset=[
            "block_id",
            "start_abs_ms",
            amplitude_column,
        ]
    )
    .sort_values(
        [
            "block_id",
            "start_abs_ms",
        ]
    )
    .reset_index(drop=True)
)


# Compute the next SACC onset using ALL SACC events,
# not only the 4–6° candidates.
saccades[
    "next_saccade_start_abs_ms"
] = (
    saccades
    .groupby(
        "block_id",
        dropna=False
    )["start_abs_ms"]
    .shift(-1)
)


candidate_saccades = saccades.loc[
    (
        saccades[amplitude_column]
        >= MIN_AMPLITUDE_DEG
    )
    & (
        saccades[amplitude_column]
        < MAX_AMPLITUDE_DEG
    )
].copy()


print(
    f"\nSACC events in "
    f"{MIN_AMPLITUDE_DEG:.0f}–"
    f"{MAX_AMPLITUDE_DEG:.0f}° range: "
    f"{len(candidate_saccades):,}"
)


if candidate_saccades.empty:
    raise RuntimeError(
        "No saccades were found in the selected "
        "amplitude range."
    )


# ============================================================
# 7. DEFINE COMMON RELATIVE-TIME GRID
# ============================================================

RELATIVE_TIME_MS = np.arange(
    -PRE_ONSET_MS,
    POST_ONSET_MS + 1,
    1.0
)


BASELINE_MASK = (
    (RELATIVE_TIME_MS >= BASELINE_START_MS)
    & (RELATIVE_TIME_MS <= BASELINE_END_MS)
)


FIT_MASK = (
    RELATIVE_TIME_MS >= 0
)


# ============================================================
# 8. EXTRACT ONE SACCADE TRAJECTORY
# ============================================================

def obtain_direction_vector(
    saccade_row
):
    """
    Obtain the event's direction vector in calibrated coordinates.
    """

    dx = np.nan
    dy = np.nan


    if (
        "dx_deg" in saccade_row.index
        and "dy_deg" in saccade_row.index
    ):

        dx = pd.to_numeric(
            saccade_row["dx_deg"],
            errors="coerce"
        )

        dy = pd.to_numeric(
            saccade_row["dy_deg"],
            errors="coerce"
        )


    if not (
        np.isfinite(dx)
        and np.isfinite(dy)
    ):

        required = [
            "start_x",
            "start_y",
            "end_x",
            "end_y",
        ]

        if not all(
            column in saccade_row.index
            for column in required
        ):
            return None


        start_x = pd.to_numeric(
            saccade_row["start_x"],
            errors="coerce"
        )

        start_y = pd.to_numeric(
            saccade_row["start_y"],
            errors="coerce"
        )

        end_x = pd.to_numeric(
            saccade_row["end_x"],
            errors="coerce"
        )

        end_y = pd.to_numeric(
            saccade_row["end_y"],
            errors="coerce"
        )


        dx = end_x - start_x
        dy = end_y - start_y


    vector_length = np.hypot(
        dx,
        dy
    )


    if (
        not np.isfinite(vector_length)
        or vector_length <= 0
    ):
        return None


    return np.array([
        dx / vector_length,
        dy / vector_length,
    ])


def extract_projected_trajectory(
    saccade_row,
    block_data
):
    """
    Extract raw calibrated 2D gaze and project it onto
    the REMoDNaV saccade direction.
    """

    onset_ms = float(
        saccade_row["start_abs_ms"]
    )


    window_start_ms = (
        onset_ms
        - PRE_ONSET_MS
    )

    window_end_ms = (
        onset_ms
        + POST_ONSET_MS
    )


    # Require the complete fitting window to remain
    # within IMAGE_ONSET -> RESPONSE when those columns exist.
    if (
        "image_onset_ms" in saccade_row.index
        and np.isfinite(
            pd.to_numeric(
                saccade_row["image_onset_ms"],
                errors="coerce"
            )
        )
    ):

        image_onset_ms = float(
            saccade_row[
                "image_onset_ms"
            ]
        )

        if window_start_ms < image_onset_ms:

            return None, (
                "pre-onset window crosses "
                "image onset"
            )


    if (
        "response_ms" in saccade_row.index
        and np.isfinite(
            pd.to_numeric(
                saccade_row["response_ms"],
                errors="coerce"
            )
        )
    ):

        response_ms = float(
            saccade_row[
                "response_ms"
            ]
        )

        if window_end_ms > response_ms:

            return None, (
                "fit window crosses response"
            )


    next_saccade_start = pd.to_numeric(
        saccade_row[
            "next_saccade_start_abs_ms"
        ],
        errors="coerce"
    )


    if (
        np.isfinite(next_saccade_start)
        and next_saccade_start
        <= window_end_ms
    ):

        return None, (
            "another SACC begins within "
            "the fit window"
        )


    time_ms = block_data[
        "time_ms"
    ]

    x_calibrated = block_data[
        "x_calibrated"
    ]

    y_calibrated = block_data[
        "y_calibrated"
    ]


    # Add one sample of padding around the interpolation range.
    sample_mask = (
        (time_ms >= window_start_ms - 2)
        & (time_ms <= window_end_ms + 2)
    )


    selected_time = time_ms[
        sample_mask
    ]

    selected_x = x_calibrated[
        sample_mask
    ]

    selected_y = y_calibrated[
        sample_mask
    ]


    if len(selected_time) < 100:

        return None, (
            "too few EDF samples in window"
        )


    if (
        selected_time[0] > window_start_ms
        or selected_time[-1] < window_end_ms
    ):

        return None, (
            "window outside EDF recording"
        )


    finite = (
        np.isfinite(selected_time)
        & np.isfinite(selected_x)
        & np.isfinite(selected_y)
    )


    if not finite.all():

        return None, (
            "blink, NaN or invalid gaze "
            "inside window"
        )


    # Remove any duplicate timestamps before interpolation.
    unique_time_mask = np.concatenate([
        [True],
        np.diff(selected_time) > 0,
    ])


    selected_time = selected_time[
        unique_time_mask
    ]

    selected_x = selected_x[
        unique_time_mask
    ]

    selected_y = selected_y[
        unique_time_mask
    ]


    time_differences = np.diff(
        selected_time
    )


    maximum_permitted_gap = (
        1.75
        * block_data[
            "median_sample_interval_ms"
        ]
    )


    if (
        len(time_differences)
        and np.nanmax(time_differences)
        > maximum_permitted_gap
    ):

        return None, (
            "unexpected sampling gap "
            "inside window"
        )


    target_time = (
        onset_ms
        + RELATIVE_TIME_MS
    )


    interpolated_x = np.interp(
        target_time,
        selected_time,
        selected_x
    )

    interpolated_y = np.interp(
        target_time,
        selected_time,
        selected_y
    )


    direction_vector = obtain_direction_vector(
        saccade_row
    )


    if direction_vector is None:

        return None, (
            "could not determine "
            "saccade direction"
        )


    ux, uy = direction_vector


    baseline_x = np.median(
        interpolated_x[
            BASELINE_MASK
        ]
    )

    baseline_y = np.median(
        interpolated_y[
            BASELINE_MASK
        ]
    )


    centered_x = (
        interpolated_x
        - baseline_x
    )

    centered_y = (
        interpolated_y
        - baseline_y
    )


    # Movement parallel to the primary saccade direction
    progress = (
        centered_x * ux
        + centered_y * uy
    )


    # Movement perpendicular to the primary direction
    orthogonal = (
        -centered_x * uy
        + centered_y * ux
    )


    # The model sets position at detected onset to zero.
    onset_index = int(
        np.flatnonzero(
            RELATIVE_TIME_MS == 0
        )[0]
    )


    progress = (
        progress
        - progress[onset_index]
    )

    orthogonal = (
        orthogonal
        - orthogonal[onset_index]
    )


    amplitude_deg = float(
        saccade_row[
            amplitude_column
        ]
    )


    late_progress = np.median(
        progress[
            RELATIVE_TIME_MS
            .astype(float)
            .__ge__(100)
        ]
    )


    # Very broad check for a direction/projection failure.
    if not (
        0.40 * amplitude_deg
        <= late_progress
        <= 1.60 * amplitude_deg
    ):

        return None, (
            "late projected position inconsistent "
            "with REMoDNaV amplitude"
        )


    result = {
        "relative_time_ms": (
            RELATIVE_TIME_MS.copy()
        ),
        "progress_deg": progress,
        "orthogonal_deg": orthogonal,
        "amplitude_deg": amplitude_deg,
        "onset_abs_ms": onset_ms,
        "maximum_absolute_orthogonal_deg": (
            float(
                np.max(
                    np.abs(
                        orthogonal[
                            FIT_MASK
                        ]
                    )
                )
            )
        ),
        "late_progress_deg": float(
            late_progress
        ),
    }


    return result, None


# ============================================================
# 9. SCAN BLOCKS UNTIL A SUITABLE PILOT BLOCK IS FOUND
# ============================================================

block_ids = sorted(
    candidate_saccades[
        "block_id"
    ]
    .dropna()
    .astype(str)
    .unique(),
    key=natural_sort_key
)


selected_block_id = None
selected_block_data = None
selected_event_rows = None
selected_trajectory_results = None

all_exclusion_rows = []

best_fallback = None


for block_id in block_ids:

    print(
        "\n"
        + "=" * 75
    )

    print(
        f"Checking block: {block_id}"
    )


    block_candidate_rows = (
        candidate_saccades.loc[
            candidate_saccades[
                "block_id"
            ].astype(str)
            == str(block_id)
        ]
        .copy()
        .reset_index(drop=False)
        .rename(
            columns={
                "index": "original_event_row"
            }
        )
    )


    if block_candidate_rows.empty:
        continue


    try:

        block_data = load_calibrated_block(
            block_id
        )

    except Exception as error:

        print(
            f"Could not load block: "
            f"{type(error).__name__}: {error}"
        )

        continue


    accepted_rows = []
    accepted_trajectories = []


    for _, event_row in (
        block_candidate_rows.iterrows()
    ):

        trajectory, exclusion_reason = (
            extract_projected_trajectory(
                saccade_row=event_row,
                block_data=block_data
            )
        )


        if exclusion_reason is not None:

            exclusion_record = {
                "block_id": block_id,
                "original_event_row": (
                    event_row[
                        "original_event_row"
                    ]
                ),
                "trial": event_row.get(
                    "trial",
                    np.nan
                ),
                "saccade_index": event_row.get(
                    "saccade_index",
                    np.nan
                ),
                "start_abs_ms": event_row[
                    "start_abs_ms"
                ],
                "amplitude_deg": event_row[
                    amplitude_column
                ],
                "exclusion_reason": (
                    exclusion_reason
                ),
            }

            all_exclusion_rows.append(
                exclusion_record
            )

            continue


        accepted_rows.append(
            event_row
        )

        accepted_trajectories.append(
            trajectory
        )


    print(
        f"Candidates in block: "
        f"{len(block_candidate_rows)}"
    )

    print(
        f"Accepted clean trajectories: "
        f"{len(accepted_trajectories)}"
    )


    current_result = {
        "block_id": block_id,
        "block_data": block_data,
        "event_rows": pd.DataFrame(
            accepted_rows
        ),
        "trajectories": (
            accepted_trajectories
        ),
    }


    if (
        best_fallback is None
        or len(
            accepted_trajectories
        )
        > len(
            best_fallback[
                "trajectories"
            ]
        )
    ):

        best_fallback = current_result


    if (
        len(accepted_trajectories)
        >= MIN_DESIRED_SACCADES
    ):

        selected_block_id = block_id
        selected_block_data = block_data

        selected_event_rows = pd.DataFrame(
            accepted_rows
        )

        selected_trajectory_results = (
            accepted_trajectories
        )

        break


# Use the best available block when no block reaches 20.
if selected_block_id is None:

    if (
        best_fallback is None
        or len(
            best_fallback[
                "trajectories"
            ]
        )
        < MIN_FALLBACK_SACCADES
    ):

        maximum_found = (
            0
            if best_fallback is None
            else len(
                best_fallback[
                    "trajectories"
                ]
            )
        )

        raise RuntimeError(
            "No block contained enough clean 4–6° "
            "saccades for the pilot.\n\n"
            f"Maximum accepted in one block: "
            f"{maximum_found}\n"
            f"Minimum fallback required: "
            f"{MIN_FALLBACK_SACCADES}"
        )


    warnings.warn(
        "No block reached the desired 20 clean saccades. "
        "Using the block with the largest available number "
        "for this diagnostic pilot."
    )


    selected_block_id = (
        best_fallback[
            "block_id"
        ]
    )

    selected_block_data = (
        best_fallback[
            "block_data"
        ]
    )

    selected_event_rows = (
        best_fallback[
            "event_rows"
        ]
    )

    selected_trajectory_results = (
        best_fallback[
            "trajectories"
        ]
    )


print(
    "\n"
    + "=" * 75
)

print(
    "SELECTED PILOT BLOCK"
)

print(
    "=" * 75
)

print(
    f"Block: {selected_block_id}"
)

print(
    f"Clean trajectories available: "
    f"{len(selected_trajectory_results)}"
)


# ============================================================
# 10. RANDOMLY SELECT AT MOST 50 SACCADES
# ============================================================

random_generator = np.random.default_rng(
    RANDOM_SEED
)


number_to_use = min(
    MAX_SACCADES,
    len(
        selected_trajectory_results
    )
)


selected_indices = np.sort(
    random_generator.choice(
        len(
            selected_trajectory_results
        ),
        size=number_to_use,
        replace=False
    )
)


selected_trajectory_results = [
    selected_trajectory_results[index]
    for index in selected_indices
]


selected_event_rows = (
    selected_event_rows
    .iloc[
        selected_indices
    ]
    .reset_index(drop=True)
)


print(
    f"Trajectories used in median: "
    f"{number_to_use}"
)


# ============================================================
# 11. CREATE PARTICIPANT/BLOCK MEDIAN TRAJECTORY
# ============================================================

progress_matrix = np.vstack([
    trajectory[
        "progress_deg"
    ]
    for trajectory in (
        selected_trajectory_results
    )
])


orthogonal_matrix = np.vstack([
    trajectory[
        "orthogonal_deg"
    ]
    for trajectory in (
        selected_trajectory_results
    )
])


median_progress = np.median(
    progress_matrix,
    axis=0
)


median_orthogonal = np.median(
    orthogonal_matrix,
    axis=0
)


selected_amplitudes = np.array([
    trajectory[
        "amplitude_deg"
    ]
    for trajectory in (
        selected_trajectory_results
    )
])


# xm is treated as the measured saccade amplitude.
XM_DEG = float(
    np.median(
        selected_amplitudes
    )
)


print(
    f"Median REMoDNaV amplitude, xm: "
    f"{XM_DEG:.4f}°"
)

print(
    f"Median observed position at 120 ms: "
    f"{median_progress[-1]:.4f}°"
)


# ============================================================
# 12. CONSTANT HARMONIC-OSCILLATOR MODEL
#
# Eyeball velocity:
#
#   x'(t) = A * t^beta * exp(-(t/tau)^mu)
#
# Eyeball position:
#
#   x(t) = A*t^(beta+1)/(beta+1)
#          * 1F1(
#               (beta+1)/mu,
#               (beta+mu+1)/mu,
#               -(t/tau)^mu
#            )
#
# The value of tau is determined by xm.
#
# Iris/pupil-relative motion:
#
#   y'' + gamma*y' + k*y = -x''
#
# Observed video-based trajectory:
#
#   z(t) = x(t) + y(t)
# ============================================================

def calculate_tau_ms(
    A,
    mu,
    x_m,
    beta=BETA
):
    """
    Compute tau so the asymptotic eyeball displacement is x_m.
    """

    numerator = (
        (beta + 1.0)
        * x_m
    )

    denominator = (
        A
        * gamma_function(
            (
                beta
                + mu
                + 1.0
            )
            / mu
        )
    )


    if (
        numerator <= 0
        or denominator <= 0
    ):
        return np.nan


    tau = (
        numerator
        / denominator
    ) ** (
        1.0
        / (
            beta + 1.0
        )
    )


    return float(tau)


def eyeball_position_deg(
    time_ms,
    A,
    mu,
    x_m,
    beta=BETA
):
    """
    Latent eyeball position x(t).
    """

    time_ms = np.asarray(
        time_ms,
        dtype=float
    )


    tau = calculate_tau_ms(
        A=A,
        mu=mu,
        x_m=x_m,
        beta=beta
    )


    if not np.isfinite(tau):
        return np.full_like(
            time_ms,
            np.nan
        )


    first_argument = (
        beta + 1.0
    ) / mu

    second_argument = (
        beta
        + mu
        + 1.0
    ) / mu

    hypergeometric_argument = -(
        time_ms
        / tau
    ) ** mu


    position = (
        A
        * time_ms ** (
            beta + 1.0
        )
        / (
            beta + 1.0
        )
        * hyp1f1(
            first_argument,
            second_argument,
            hypergeometric_argument
        )
    )


    return np.asarray(
        position,
        dtype=float
    )


def eyeball_velocity_deg_per_ms(
    time_ms,
    A,
    mu,
    x_m,
    beta=BETA
):
    """
    Latent eyeball velocity x'(t).
    """

    time_ms = np.asarray(
        time_ms,
        dtype=float
    )


    tau = calculate_tau_ms(
        A=A,
        mu=mu,
        x_m=x_m,
        beta=beta
    )


    velocity = (
        A
        * time_ms ** beta
        * np.exp(
            -(
                time_ms
                / tau
            ) ** mu
        )
    )


    return np.asarray(
        velocity,
        dtype=float
    )


def eyeball_acceleration_deg_per_ms2(
    time_ms,
    A,
    mu,
    x_m,
    beta=BETA
):
    """
    Latent eyeball acceleration x''(t).
    """

    time_ms = np.asarray(
        time_ms,
        dtype=float
    )


    tau = calculate_tau_ms(
        A=A,
        mu=mu,
        x_m=x_m,
        beta=beta
    )


    exponential_term = np.exp(
        -(
            time_ms
            / tau
        ) ** mu
    )


    first_term = (
        beta
        * time_ms ** (
            beta - 1.0
        )
    )


    second_term = (
        mu
        / tau ** mu
        * time_ms ** (
            beta
            + mu
            - 1.0
        )
    )


    acceleration = (
        A
        * exponential_term
        * (
            first_term
            - second_term
        )
    )


    return np.asarray(
        acceleration,
        dtype=float
    )


def predict_constant_model(
    time_ms,
    A,
    mu,
    gamma_value,
    k_value,
    x_m
):
    """
    Predict:
        x_eye(t)
        y_iris(t)
        z_observed(t) = x_eye(t) + y_iris(t)
    """

    time_ms = np.asarray(
        time_ms,
        dtype=float
    )


    if (
        len(time_ms) == 0
        or np.any(time_ms < 0)
    ):
        raise ValueError(
            "Model time must begin at zero "
            "and contain no negative values."
        )


    x_eye = eyeball_position_deg(
        time_ms=time_ms,
        A=A,
        mu=mu,
        x_m=x_m
    )


    def oscillator_system(
        current_time,
        state
    ):

        y_iris = state[0]
        y_iris_velocity = state[1]


        x_eye_acceleration = float(
            eyeball_acceleration_deg_per_ms2(
                np.array([
                    current_time
                ]),
                A=A,
                mu=mu,
                x_m=x_m
            )[0]
        )


        y_iris_acceleration = (
            -gamma_value
            * y_iris_velocity
            - k_value
            * y_iris
            - x_eye_acceleration
        )


        return [
            y_iris_velocity,
            y_iris_acceleration,
        ]


    solution = solve_ivp(
        fun=oscillator_system,
        t_span=(
            float(time_ms[0]),
            float(time_ms[-1])
        ),
        y0=[
            0.0,
            0.0,
        ],
        t_eval=time_ms,
        method="RK45",
        rtol=1e-6,
        atol=1e-8,
    )


    if (
        not solution.success
        or solution.y.shape[1]
        != len(time_ms)
    ):

        raise RuntimeError(
            "ODE solution failed: "
            f"{solution.message}"
        )


    y_iris = solution.y[0]

    predicted_observed = (
        x_eye
        + y_iris
    )


    return {
        "x_eye_deg": x_eye,
        "y_iris_deg": y_iris,
        "predicted_observed_deg": (
            predicted_observed
        ),
    }


# ============================================================
# 13. FIT 0–120 MS OF THE MEDIAN TRAJECTORY
# ============================================================

fit_time_ms = RELATIVE_TIME_MS[
    FIT_MASK
].astype(float)


observed_fit_deg = median_progress[
    FIT_MASK
].astype(float)


# Guarantee exact zero at detected onset.
observed_fit_deg = (
    observed_fit_deg
    - observed_fit_deg[0]
)


def model_residuals(
    parameter_vector
):
    """
    Residuals used by scipy.optimize.least_squares.
    """

    A, mu, gamma_value, k_value = (
        parameter_vector
    )


    try:

        prediction = predict_constant_model(
            time_ms=fit_time_ms,
            A=A,
            mu=mu,
            gamma_value=gamma_value,
            k_value=k_value,
            x_m=XM_DEG
        )

        residuals = (
            prediction[
                "predicted_observed_deg"
            ]
            - observed_fit_deg
        )


        if not np.isfinite(
            residuals
        ).all():

            return np.full(
                len(observed_fit_deg),
                1e6
            )


        return residuals


    except Exception:

        return np.full(
            len(observed_fit_deg),
            1e6
        )


rng = np.random.default_rng(
    RANDOM_SEED
)


starting_values = [
    (
        LOWER_BOUNDS
        + UPPER_BOUNDS
    ) / 2.0
]


for _ in range(
    N_OPTIMIZATION_STARTS - 1
):

    starting_values.append(
        rng.uniform(
            LOWER_BOUNDS,
            UPPER_BOUNDS
        )
    )


optimization_rows = []
successful_fits = []


print(
    "\nFitting constant harmonic model..."
)


for start_number, initial_values in enumerate(
    starting_values,
    start=1
):

    try:

        fitted = least_squares(
            fun=model_residuals,
            x0=initial_values,
            bounds=(
                LOWER_BOUNDS,
                UPPER_BOUNDS
            ),
            method="trf",
            loss="linear",
            x_scale="jac",
            max_nfev=500,
            ftol=1e-10,
            xtol=1e-10,
            gtol=1e-10,
        )


        residual_vector = model_residuals(
            fitted.x
        )


        residual_sum_of_squares = float(
            np.sum(
                residual_vector ** 2
            )
        )


        fit_record = {
            "start_number": start_number,
            "success": bool(
                fitted.success
            ),
            "message": str(
                fitted.message
            ),
            "n_function_evaluations": int(
                fitted.nfev
            ),
            "rss": residual_sum_of_squares,
            "A": fitted.x[0],
            "mu": fitted.x[1],
            "gamma": fitted.x[2],
            "k": fitted.x[3],
        }


        optimization_rows.append(
            fit_record
        )


        if (
            fitted.success
            and np.isfinite(
                residual_sum_of_squares
            )
        ):

            successful_fits.append(
                (
                    residual_sum_of_squares,
                    fitted
                )
            )


    except Exception as error:

        optimization_rows.append({
            "start_number": start_number,
            "success": False,
            "message": (
                f"{type(error).__name__}: "
                f"{error}"
            ),
            "n_function_evaluations": np.nan,
            "rss": np.nan,
            "A": np.nan,
            "mu": np.nan,
            "gamma": np.nan,
            "k": np.nan,
        })


optimization_results = pd.DataFrame(
    optimization_rows
)


if not successful_fits:

    raise RuntimeError(
        "No optimization start produced a "
        "successful model fit."
    )


successful_fits.sort(
    key=lambda item: item[0]
)


best_rss, best_fit = (
    successful_fits[0]
)


BEST_A = float(
    best_fit.x[0]
)

BEST_MU = float(
    best_fit.x[1]
)

BEST_GAMMA = float(
    best_fit.x[2]
)

BEST_K = float(
    best_fit.x[3]
)


best_prediction = predict_constant_model(
    time_ms=fit_time_ms,
    A=BEST_A,
    mu=BEST_MU,
    gamma_value=BEST_GAMMA,
    k_value=BEST_K,
    x_m=XM_DEG
)


predicted_observed_deg = best_prediction[
    "predicted_observed_deg"
]

inferred_eyeball_deg = best_prediction[
    "x_eye_deg"
]

inferred_iris_deg = best_prediction[
    "y_iris_deg"
]


residual_deg = (
    observed_fit_deg
    - predicted_observed_deg
)


rmse_deg = float(
    np.sqrt(
        np.mean(
            residual_deg ** 2
        )
    )
)


total_sum_of_squares = float(
    np.sum(
        (
            observed_fit_deg
            - np.mean(
                observed_fit_deg
            )
        ) ** 2
    )
)


if total_sum_of_squares > 0:

    r_squared = (
        1.0
        - best_rss
        / total_sum_of_squares
    )

else:

    r_squared = np.nan


n_observations = len(
    observed_fit_deg
)

n_parameters = 4


bic = (
    n_observations
    * np.log(
        best_rss
        / n_observations
    )
    + n_parameters
    * np.log(
        n_observations
    )
)


tau_ms = calculate_tau_ms(
    A=BEST_A,
    mu=BEST_MU,
    x_m=XM_DEG
)


natural_angular_frequency_per_ms = (
    np.sqrt(
        BEST_K
    )
)


natural_frequency_hz = (
    natural_angular_frequency_per_ms
    / (
        2.0 * np.pi
    )
    * 1000.0
)


damping_ratio = (
    BEST_GAMMA
    / (
        2.0
        * np.sqrt(
            BEST_K
        )
    )
)


eyeball_velocity_deg_s = (
    1000.0
    * eyeball_velocity_deg_per_ms(
        time_ms=fit_time_ms,
        A=BEST_A,
        mu=BEST_MU,
        x_m=XM_DEG
    )
)


inferred_peak_velocity_deg_s = float(
    np.max(
        eyeball_velocity_deg_s
    )
)


above_98_percent = np.flatnonzero(
    inferred_eyeball_deg
    >= (
        0.98
        * XM_DEG
    )
)


if len(
    above_98_percent
):

    inferred_duration_98_ms = float(
        fit_time_ms[
            above_98_percent[0]
        ]
    )

else:

    inferred_duration_98_ms = np.nan


# Check whether parameters are close to their bounds.
parameter_values = np.array([
    BEST_A,
    BEST_MU,
    BEST_GAMMA,
    BEST_K,
])


parameter_ranges = (
    UPPER_BOUNDS
    - LOWER_BOUNDS
)


near_lower_bound = (
    parameter_values
    - LOWER_BOUNDS
) <= (
    0.01
    * parameter_ranges
)


near_upper_bound = (
    UPPER_BOUNDS
    - parameter_values
) <= (
    0.01
    * parameter_ranges
)


parameters_near_bounds = [
    PARAMETER_NAMES[index]
    for index in range(
        len(PARAMETER_NAMES)
    )
    if (
        near_lower_bound[index]
        or near_upper_bound[index]
    )
]


# ============================================================
# 14. DISPLAY FIT RESULTS
# ============================================================

fit_summary = pd.DataFrame([{
    "subject": SUBJECT_ID,
    "block_id": selected_block_id,
    "n_saccades_used": number_to_use,
    "amplitude_range_min_deg": MIN_AMPLITUDE_DEG,
    "amplitude_range_max_deg": MAX_AMPLITUDE_DEG,
    "median_remodnav_amplitude_xm_deg": XM_DEG,
    "beta_fixed": BETA,
    "A": BEST_A,
    "mu": BEST_MU,
    "gamma_per_ms": BEST_GAMMA,
    "k_per_ms_squared": BEST_K,
    "tau_ms": tau_ms,
    "natural_angular_frequency_per_ms": (
        natural_angular_frequency_per_ms
    ),
    "natural_frequency_hz": (
        natural_frequency_hz
    ),
    "damping_ratio": damping_ratio,
    "rmse_deg": rmse_deg,
    "r_squared": r_squared,
    "rss": best_rss,
    "bic": bic,
    "inferred_eyeball_peak_velocity_deg_s": (
        inferred_peak_velocity_deg_s
    ),
    "inferred_eyeball_duration_98_ms": (
        inferred_duration_98_ms
    ),
    "parameters_near_bounds": (
        "; ".join(
            parameters_near_bounds
        )
        if parameters_near_bounds
        else ""
    ),
}])


print(
    "\nBEST CONSTANT-MODEL FIT"
)

display(
    fit_summary.T
)


if parameters_near_bounds:

    warnings.warn(
        "One or more parameters are close to the "
        "optimization bounds: "
        + ", ".join(
            parameters_near_bounds
        )
        + ". Treat this as a pilot fit requiring inspection."
    )


print(
    "\nBest five multistart solutions:"
)

display(
    optimization_results
    .sort_values(
        "rss"
    )
    .head()
)


# ============================================================
# 15. SAVE TABLES
# ============================================================

selected_saccades_path = (
    OUT_DIR
    / "selected_saccades.csv"
)

excluded_saccades_path = (
    OUT_DIR
    / "excluded_saccades.csv"
)

individual_trajectories_path = (
    OUT_DIR
    / "individual_projected_trajectories.csv"
)

median_trajectory_path = (
    OUT_DIR
    / "median_trajectory_and_model_fit.csv"
)

fit_summary_path = (
    OUT_DIR
    / "constant_model_fit.csv"
)

optimization_results_path = (
    OUT_DIR
    / "multistart_optimization_results.csv"
)


selected_event_rows.to_csv(
    selected_saccades_path,
    index=False
)


pd.DataFrame(
    all_exclusion_rows
).to_csv(
    excluded_saccades_path,
    index=False
)


individual_trajectory_rows = []


for trajectory_number, trajectory in enumerate(
    selected_trajectory_results,
    start=1
):

    for time_value, progress_value, orthogonal_value in zip(
        trajectory[
            "relative_time_ms"
        ],
        trajectory[
            "progress_deg"
        ],
        trajectory[
            "orthogonal_deg"
        ]
    ):

        individual_trajectory_rows.append({
            "trajectory_number": (
                trajectory_number
            ),
            "relative_time_ms": (
                time_value
            ),
            "progress_deg": (
                progress_value
            ),
            "orthogonal_deg": (
                orthogonal_value
            ),
            "remodnav_amplitude_deg": (
                trajectory[
                    "amplitude_deg"
                ]
            ),
            "onset_abs_ms": (
                trajectory[
                    "onset_abs_ms"
                ]
            ),
        })


individual_trajectories = pd.DataFrame(
    individual_trajectory_rows
)


individual_trajectories.to_csv(
    individual_trajectories_path,
    index=False
)


median_output = pd.DataFrame({
    "relative_time_ms": (
        RELATIVE_TIME_MS
    ),
    "median_observed_progress_deg": (
        median_progress
    ),
    "median_orthogonal_deg": (
        median_orthogonal
    ),
})


median_output[
    "used_for_model_fit"
] = FIT_MASK


median_output[
    "model_predicted_observed_deg"
] = np.nan

median_output[
    "model_inferred_eyeball_deg"
] = np.nan

median_output[
    "model_inferred_iris_relative_deg"
] = np.nan

median_output[
    "model_residual_observed_minus_predicted_deg"
] = np.nan


median_output.loc[
    FIT_MASK,
    "model_predicted_observed_deg"
] = predicted_observed_deg

median_output.loc[
    FIT_MASK,
    "model_inferred_eyeball_deg"
] = inferred_eyeball_deg

median_output.loc[
    FIT_MASK,
    "model_inferred_iris_relative_deg"
] = inferred_iris_deg

median_output.loc[
    FIT_MASK,
    "model_residual_observed_minus_predicted_deg"
] = residual_deg


median_output.to_csv(
    median_trajectory_path,
    index=False
)


fit_summary.to_csv(
    fit_summary_path,
    index=False
)


optimization_results.to_csv(
    optimization_results_path,
    index=False
)


print("\nSaved tables:")
print(selected_saccades_path)
print(excluded_saccades_path)
print(individual_trajectories_path)
print(median_trajectory_path)
print(fit_summary_path)
print(optimization_results_path)


# ============================================================
# 16. FIGURE 1:
# INDIVIDUAL PROJECTED TRAJECTORIES AND THEIR MEDIAN
# ============================================================

plt.figure(
    figsize=(10, 6)
)


for trajectory in (
    selected_trajectory_results
):

    plt.plot(
        trajectory[
            "relative_time_ms"
        ],
        trajectory[
            "progress_deg"
        ],
        alpha=0.20,
        linewidth=0.8
    )


plt.plot(
    RELATIVE_TIME_MS,
    median_progress,
    linewidth=3,
    label="Median trajectory"
)


plt.axvline(
    0,
    linestyle="--",
    linewidth=1,
    label="REMoDNaV onset"
)


plt.xlabel(
    "Time from REMoDNaV saccade onset (ms)"
)

plt.ylabel(
    "Projected gaze position (degrees)"
)

plt.title(
    f"{SUBJECT_ID}, block {selected_block_id}: "
    f"{number_to_use} clean "
    f"{MIN_AMPLITUDE_DEG:.0f}–"
    f"{MAX_AMPLITUDE_DEG:.0f}° saccades"
)

plt.legend()


save_current_figure(
    "individual_trajectory_overlay.png"
)


# ============================================================
# 17. FIGURE 2:
# OBSERVED VERSUS MODEL-PREDICTED TRAJECTORY
# ============================================================

plt.figure(
    figsize=(10, 6)
)


plt.plot(
    fit_time_ms,
    observed_fit_deg,
    linewidth=3,
    label="Observed median pupil-based trajectory"
)


plt.plot(
    fit_time_ms,
    predicted_observed_deg,
    linewidth=2,
    label="Model-predicted pupil trajectory"
)


plt.plot(
    fit_time_ms,
    inferred_eyeball_deg,
    linewidth=2,
    linestyle="--",
    label="Model-inferred eyeball trajectory"
)


plt.xlabel(
    "Time from saccade onset (ms)"
)

plt.ylabel(
    "Position change (degrees)"
)

plt.title(
    "Constant harmonic-oscillator pilot fit\n"
    f"RMSE={rmse_deg:.4f}°, "
    f"R²={r_squared:.4f}"
)

plt.legend()


save_current_figure(
    "observed_vs_harmonic_model_fit.png"
)


# ============================================================
# 18. FIGURE 3:
# MODEL-INFERRED RELATIVE IRIS/PUPIL MOVEMENT
# ============================================================

plt.figure(
    figsize=(10, 5)
)


plt.plot(
    fit_time_ms,
    inferred_iris_deg,
    linewidth=2
)


plt.axhline(
    0,
    linestyle="--",
    linewidth=1
)


plt.xlabel(
    "Time from saccade onset (ms)"
)

plt.ylabel(
    "Inferred relative pupil/iris movement (degrees)"
)

plt.title(
    "Model-inferred oscillatory component"
)


save_current_figure(
    "model_inferred_iris_component.png"
)


# ============================================================
# 19. FIGURE 4:
# MODEL RESIDUALS
# ============================================================

plt.figure(
    figsize=(10, 5)
)


plt.plot(
    fit_time_ms,
    residual_deg,
    linewidth=2
)


plt.axhline(
    0,
    linestyle="--",
    linewidth=1
)


plt.xlabel(
    "Time from saccade onset (ms)"
)

plt.ylabel(
    "Observed − predicted position (degrees)"
)

plt.title(
    "Constant harmonic-model residuals"
)


save_current_figure(
    "harmonic_model_residuals.png"
)


# ============================================================
# 20. FINAL SUMMARY
# ============================================================

print(
    "\n"
    + "=" * 75
)

print(
    "S189 HARMONIC-MODEL PILOT COMPLETE"
)

print(
    "=" * 75
)


print(
    f"Selected block: "
    f"{selected_block_id}"
)

print(
    f"Clean saccades used: "
    f"{number_to_use}"
)

print(
    f"Median amplitude, xm: "
    f"{XM_DEG:.3f}°"
)

print(
    f"A:     {BEST_A:.6f}"
)

print(
    f"mu:    {BEST_MU:.6f}"
)

print(
    f"gamma: {BEST_GAMMA:.6f} per ms"
)

print(
    f"k:     {BEST_K:.6f} per ms²"
)

print(
    f"RMSE:  {rmse_deg:.5f}°"
)

print(
    f"R²:    {r_squared:.5f}"
)

print(
    f"Inferred eyeball peak velocity: "
    f"{inferred_peak_velocity_deg_s:.2f} deg/s"
)

print(
    f"Inferred 98% eyeball duration: "
    f"{inferred_duration_98_ms:.2f} ms"
)

print(
    "\nInspect these figures first:"
)

print(
    OUT_DIR
    / "individual_trajectory_overlay.png"
)

print(
    OUT_DIR
    / "observed_vs_harmonic_model_fit.png"
)

print(
    OUT_DIR
    / "harmonic_model_residuals.png"
)

print(
    "\nAll outputs saved in:"
)

print(
    OUT_DIR
)
```

```python
# ============================================================
# REPRESENTATIVE INDIVIDUAL-SACCADE HARMONIC MODEL
# PAPER-STYLE xm: FULL FIXATION-TO-FIXATION AMPLITUDE
#
# Run AFTER the earlier s189 one-block pilot cell.
#
# PAPER-STYLE xm
# --------------
# xm is fixed as the 2D distance between:
#
#   1. the pre-saccadic fixation center
#   2. the post-saccadic fixation center
#
# It is NOT defined from:
#   - the peak overshoot,
#   - the REMoDNaV SACC endpoint,
#   - or the late projected one-dimensional position.
#
# Since classify_isp=False was used in the REMoDNaV pipeline,
# explicit fixation events are unavailable. Fixation centers are
# therefore estimated directly from stable raw calibrated gaze:
#
#   pre fixation:  -20 to -5 ms
#   post fixation: most stable 20-ms window from 70 to 120 ms
#
# The model itself is fitted from 0 to 120 ms, with observed
# position set to zero at detected REMoDNaV onset.
# ============================================================


# ============================================================
# 0. IMPORTS
# ============================================================

from pathlib import Path
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.optimize import least_squares
from IPython.display import display


# ============================================================
# 1. CHECK THAT THE EARLIER PILOT CELL HAS RUN
# ============================================================

required_objects = [
    "selected_event_rows",
    "selected_block_data",
    "selected_block_id",
    "RELATIVE_TIME_MS",
    "FIT_MASK",
    "LOWER_BOUNDS",
    "UPPER_BOUNDS",
    "PARAMETER_NAMES",
    "BETA",
    "predict_constant_model",
    "calculate_tau_ms",
    "eyeball_velocity_deg_per_ms",
    "OUT_DIR",
]

missing_objects = [
    name
    for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise RuntimeError(
        "Run the earlier s189 one-block harmonic-model pilot "
        "cell first.\n\nMissing objects:\n"
        + "\n".join(missing_objects)
    )

if not isinstance(selected_event_rows, pd.DataFrame):
    raise TypeError(
        "selected_event_rows must be a pandas DataFrame."
    )

if selected_event_rows.empty:
    raise RuntimeError(
        "selected_event_rows contains no clean saccades."
    )


# Use a separate folder so the earlier stable-endpoint-xm
# results are preserved.
PAPER_XM_OUT_DIR = (
    Path(OUT_DIR)
    / "representative_single_saccade_paper_xm"
)

PAPER_XM_OUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

print("Paper-style xm output folder:")
print(PAPER_XM_OUT_DIR)


# ============================================================
# 2. SETTINGS
# ============================================================

# Pre-saccadic fixation-center interval.
PRE_FIXATION_START_MS = -20.0
PRE_FIXATION_END_MS = -5.0

# Search interval for the post-saccadic fixation center.
POST_FIXATION_SEARCH_START_MS = 70.0
POST_FIXATION_SEARCH_END_MS = 120.0

# Width of each candidate post-fixation window.
POST_FIXATION_WINDOW_WIDTH_MS = 20.0

# Stability thresholds used for selecting a representative
# trajectory. These do not define xm; they only assess whether
# the two estimated fixation centers are credible.
MAX_PRE_FIXATION_RADIAL_MAD_DEG = 0.15
MAX_POST_FIXATION_RADIAL_MAD_DEG = 0.15

MAX_PRE_FIXATION_SPEED_DEG_S = 10.0
MAX_POST_FIXATION_SPEED_DEG_S = 10.0

# At a correct onset, gaze should still be close to the
# pre-saccadic fixation center.
MAX_ONSET_TO_PREFIX_DISTANCE_DEG = 0.20

# Broad consistency check against REMoDNaV amplitude.
MIN_XM_TO_REMODNAV_RATIO = 0.70
MAX_XM_TO_REMODNAV_RATIO = 1.30

# Number of multistart optimization runs.
N_SINGLE_SACCADE_STARTS = 40
SINGLE_RANDOM_SEED = 20260803

# Interpolation tolerance for unexpected timestamp gaps.
MAX_SAMPLE_GAP_MULTIPLIER = 1.75


# ============================================================
# 3. SMALL HELPERS
# ============================================================

def get_numeric_value(mapping, possible_names, default=np.nan):
    """
    Return the first finite numeric value found under one of
    possible_names.
    """

    for name in possible_names:

        if name not in mapping:
            continue

        value = pd.to_numeric(
            mapping[name],
            errors="coerce"
        )

        if np.isfinite(value):
            return float(value)

    return default


def get_remodnav_amplitude(event_row):
    """
    Read the REMoDNaV amplitude for comparison only.
    This value is not used as the model's xm.
    """

    return get_numeric_value(
        event_row,
        [
            "amplitude_deg",
            "amp",
        ]
    )


def radial_median_deviation(x, y, center_x, center_y):
    """
    Median radial distance from a two-dimensional center.
    """

    radial_distances = np.hypot(
        x - center_x,
        y - center_y
    )

    return float(
        np.median(radial_distances)
    )


def estimate_2d_speed_deg_s(time_ms, x_deg, y_deg):
    """
    Estimate the magnitude of linear 2D drift in a window.
    """

    if len(time_ms) < 3:
        return np.nan

    slope_x_deg_ms = np.polyfit(
        time_ms,
        x_deg,
        deg=1
    )[0]

    slope_y_deg_ms = np.polyfit(
        time_ms,
        y_deg,
        deg=1
    )[0]

    return float(
        1000.0
        * np.hypot(
            slope_x_deg_ms,
            slope_y_deg_ms
        )
    )


# ============================================================
# 4. INTERPOLATE RAW CALIBRATED 2D GAZE
# ============================================================

def interpolate_calibrated_xy(
    onset_abs_ms,
    relative_time_ms,
    block_data,
):
    """
    Interpolate raw calibrated 2D gaze to the requested
    relative-time grid.

    The gaze values come from selected_block_data, which was
    reconstructed directly from the EDF and block-mean affine
    calibration in the earlier pilot cell.
    """

    relative_time_ms = np.asarray(
        relative_time_ms,
        dtype=float
    )

    target_abs_time_ms = (
        float(onset_abs_ms)
        + relative_time_ms
    )

    block_time_ms = np.asarray(
        block_data["time_ms"],
        dtype=float
    )

    block_x_deg = np.asarray(
        block_data["x_calibrated"],
        dtype=float
    )

    block_y_deg = np.asarray(
        block_data["y_calibrated"],
        dtype=float
    )

    requested_start_ms = float(
        np.min(target_abs_time_ms)
    )

    requested_end_ms = float(
        np.max(target_abs_time_ms)
    )

    # Include a small amount of padding around the interpolation
    # interval.
    sample_mask = (
        (block_time_ms >= requested_start_ms - 2.0)
        & (block_time_ms <= requested_end_ms + 2.0)
    )

    time_selected = block_time_ms[
        sample_mask
    ]

    x_selected = block_x_deg[
        sample_mask
    ]

    y_selected = block_y_deg[
        sample_mask
    ]

    if len(time_selected) < 20:
        return None, "too few raw samples"

    if (
        time_selected[0] > requested_start_ms
        or time_selected[-1] < requested_end_ms
    ):
        return None, "requested window crosses EDF boundary"

    finite = (
        np.isfinite(time_selected)
        & np.isfinite(x_selected)
        & np.isfinite(y_selected)
    )

    if not finite.all():
        return None, "blink, NaN or invalid gaze in window"

    # Remove duplicate or non-increasing timestamps.
    keep_unique = np.concatenate([
        [True],
        np.diff(time_selected) > 0,
    ])

    time_selected = time_selected[
        keep_unique
    ]

    x_selected = x_selected[
        keep_unique
    ]

    y_selected = y_selected[
        keep_unique
    ]

    sample_differences_ms = np.diff(
        time_selected
    )

    typical_interval_ms = float(
        block_data[
            "median_sample_interval_ms"
        ]
    )

    if len(sample_differences_ms):

        maximum_allowed_gap_ms = (
            MAX_SAMPLE_GAP_MULTIPLIER
            * typical_interval_ms
        )

        if (
            np.nanmax(sample_differences_ms)
            > maximum_allowed_gap_ms
        ):
            return None, "unexpected timestamp gap in window"

    interpolated_x_deg = np.interp(
        target_abs_time_ms,
        time_selected,
        x_selected
    )

    interpolated_y_deg = np.interp(
        target_abs_time_ms,
        time_selected,
        y_selected
    )

    return {
        "relative_time_ms": relative_time_ms,
        "absolute_time_ms": target_abs_time_ms,
        "x_deg": interpolated_x_deg,
        "y_deg": interpolated_y_deg,
    }, None


# ============================================================
# 5. PAPER-STYLE FIXATION-TO-FIXATION xm
# ============================================================

def estimate_paper_style_xm(
    event_row,
    block_data,
):
    """
    Estimate paper-style Full saccade amplitude:

        xm = Euclidean distance between the pre-saccadic and
             post-saccadic fixation centers.

    The post-saccadic center comes from the most stable 20-ms
    two-dimensional window between 70 and 120 ms.
    """

    onset_abs_ms = get_numeric_value(
        event_row,
        ["start_abs_ms"]
    )

    if not np.isfinite(onset_abs_ms):
        return None, "missing start_abs_ms"

    interpolated, interpolation_error = (
        interpolate_calibrated_xy(
            onset_abs_ms=onset_abs_ms,
            relative_time_ms=RELATIVE_TIME_MS,
            block_data=block_data,
        )
    )

    if interpolation_error is not None:
        return None, interpolation_error

    relative_time_ms = interpolated[
        "relative_time_ms"
    ]

    x_deg = interpolated[
        "x_deg"
    ]

    y_deg = interpolated[
        "y_deg"
    ]


    # --------------------------------------------------------
    # Pre-saccadic fixation center
    # --------------------------------------------------------

    pre_fixation_mask = (
        (relative_time_ms >= PRE_FIXATION_START_MS)
        & (relative_time_ms <= PRE_FIXATION_END_MS)
    )

    if pre_fixation_mask.sum() < 8:
        return None, "too few pre-fixation samples"

    pre_x_values = x_deg[
        pre_fixation_mask
    ]

    pre_y_values = y_deg[
        pre_fixation_mask
    ]

    pre_time_values = relative_time_ms[
        pre_fixation_mask
    ]

    pre_fixation_x_deg = float(
        np.median(pre_x_values)
    )

    pre_fixation_y_deg = float(
        np.median(pre_y_values)
    )

    pre_fixation_radial_mad_deg = (
        radial_median_deviation(
            pre_x_values,
            pre_y_values,
            pre_fixation_x_deg,
            pre_fixation_y_deg,
        )
    )

    pre_fixation_speed_deg_s = (
        estimate_2d_speed_deg_s(
            pre_time_values,
            pre_x_values,
            pre_y_values,
        )
    )


    # --------------------------------------------------------
    # Position at detected REMoDNaV onset
    # --------------------------------------------------------

    onset_indices = np.flatnonzero(
        np.isclose(
            relative_time_ms,
            0.0
        )
    )

    if len(onset_indices) != 1:
        return None, "relative-time grid has no unique t=0"

    onset_index = int(
        onset_indices[0]
    )

    onset_x_deg = float(
        x_deg[onset_index]
    )

    onset_y_deg = float(
        y_deg[onset_index]
    )

    onset_to_prefix_distance_deg = float(
        np.hypot(
            onset_x_deg - pre_fixation_x_deg,
            onset_y_deg - pre_fixation_y_deg,
        )
    )


    # --------------------------------------------------------
    # Search for the most stable post-saccadic fixation window
    # --------------------------------------------------------

    post_window_rows = []

    latest_post_window_start_ms = (
        POST_FIXATION_SEARCH_END_MS
        - POST_FIXATION_WINDOW_WIDTH_MS
    )

    for window_start_ms in np.arange(
        POST_FIXATION_SEARCH_START_MS,
        latest_post_window_start_ms + 1.0,
        1.0,
    ):

        window_end_ms = (
            window_start_ms
            + POST_FIXATION_WINDOW_WIDTH_MS
        )

        post_window_mask = (
            (relative_time_ms >= window_start_ms)
            & (relative_time_ms <= window_end_ms)
        )

        if post_window_mask.sum() < 10:
            continue

        post_time_values = relative_time_ms[
            post_window_mask
        ]

        post_x_values = x_deg[
            post_window_mask
        ]

        post_y_values = y_deg[
            post_window_mask
        ]

        post_center_x_deg = float(
            np.median(post_x_values)
        )

        post_center_y_deg = float(
            np.median(post_y_values)
        )

        post_radial_mad_deg = (
            radial_median_deviation(
                post_x_values,
                post_y_values,
                post_center_x_deg,
                post_center_y_deg,
            )
        )

        post_speed_deg_s = (
            estimate_2d_speed_deg_s(
                post_time_values,
                post_x_values,
                post_y_values,
            )
        )

        # Approximate total drift over the window.
        post_total_drift_deg = (
            abs(post_speed_deg_s)
            * POST_FIXATION_WINDOW_WIDTH_MS
            / 1000.0
        )

        post_stability_score = (
            post_radial_mad_deg
            + post_total_drift_deg
        )

        post_window_rows.append({
            "window_start_ms": window_start_ms,
            "window_end_ms": window_end_ms,
            "post_fixation_x_deg": post_center_x_deg,
            "post_fixation_y_deg": post_center_y_deg,
            "post_fixation_radial_mad_deg": (
                post_radial_mad_deg
            ),
            "post_fixation_speed_deg_s": (
                post_speed_deg_s
            ),
            "post_fixation_stability_score": (
                post_stability_score
            ),
        })

    if not post_window_rows:
        return None, "no usable post-fixation window"

    post_window_table = pd.DataFrame(
        post_window_rows
    ).sort_values(
        "post_fixation_stability_score",
        ascending=True
    )

    best_post_window = (
        post_window_table.iloc[0]
    )

    post_fixation_x_deg = float(
        best_post_window[
            "post_fixation_x_deg"
        ]
    )

    post_fixation_y_deg = float(
        best_post_window[
            "post_fixation_y_deg"
        ]
    )

    post_window_start_ms = float(
        best_post_window[
            "window_start_ms"
        ]
    )

    post_window_end_ms = float(
        best_post_window[
            "window_end_ms"
        ]
    )

    post_fixation_radial_mad_deg = float(
        best_post_window[
            "post_fixation_radial_mad_deg"
        ]
    )

    post_fixation_speed_deg_s = float(
        best_post_window[
            "post_fixation_speed_deg_s"
        ]
    )

    post_fixation_stability_score = float(
        best_post_window[
            "post_fixation_stability_score"
        ]
    )


    # --------------------------------------------------------
    # PAPER-STYLE FULL AMPLITUDE
    # --------------------------------------------------------

    fixation_dx_deg = (
        post_fixation_x_deg
        - pre_fixation_x_deg
    )

    fixation_dy_deg = (
        post_fixation_y_deg
        - pre_fixation_y_deg
    )

    paper_xm_deg = float(
        np.hypot(
            fixation_dx_deg,
            fixation_dy_deg,
        )
    )

    if (
        not np.isfinite(paper_xm_deg)
        or paper_xm_deg <= 0
    ):
        return None, "invalid fixation-to-fixation amplitude"

    direction_x = (
        fixation_dx_deg
        / paper_xm_deg
    )

    direction_y = (
        fixation_dy_deg
        / paper_xm_deg
    )


    # --------------------------------------------------------
    # Convert raw 2D gaze to 1D position change using the
    # fixation-to-fixation direction vector.
    #
    # The observed trajectory is set to zero at REMoDNaV onset,
    # matching the paper's fitting convention.
    # --------------------------------------------------------

    centered_at_onset_x = (
        x_deg
        - onset_x_deg
    )

    centered_at_onset_y = (
        y_deg
        - onset_y_deg
    )

    progress_deg = (
        centered_at_onset_x
        * direction_x
        + centered_at_onset_y
        * direction_y
    )

    orthogonal_deg = (
        -centered_at_onset_x
        * direction_y
        + centered_at_onset_y
        * direction_x
    )

    observed_postfix_position_from_onset_deg = float(
        (
            post_fixation_x_deg
            - onset_x_deg
        )
        * direction_x
        + (
            post_fixation_y_deg
            - onset_y_deg
        )
        * direction_y
    )


    # --------------------------------------------------------
    # Quality-control comparisons
    # --------------------------------------------------------

    remodnav_amplitude_deg = (
        get_remodnav_amplitude(
            event_row
        )
    )

    if (
        np.isfinite(remodnav_amplitude_deg)
        and remodnav_amplitude_deg > 0
    ):

        xm_to_remodnav_ratio = (
            paper_xm_deg
            / remodnav_amplitude_deg
        )

        amplitude_consistent = (
            xm_to_remodnav_ratio
            >= MIN_XM_TO_REMODNAV_RATIO
            and xm_to_remodnav_ratio
            <= MAX_XM_TO_REMODNAV_RATIO
        )

    else:

        xm_to_remodnav_ratio = np.nan
        amplitude_consistent = True

    passes_fixation_qc = (
        pre_fixation_radial_mad_deg
        <= MAX_PRE_FIXATION_RADIAL_MAD_DEG

        and post_fixation_radial_mad_deg
        <= MAX_POST_FIXATION_RADIAL_MAD_DEG

        and abs(
            pre_fixation_speed_deg_s
        )
        <= MAX_PRE_FIXATION_SPEED_DEG_S

        and abs(
            post_fixation_speed_deg_s
        )
        <= MAX_POST_FIXATION_SPEED_DEG_S

        and onset_to_prefix_distance_deg
        <= MAX_ONSET_TO_PREFIX_DISTANCE_DEG

        and amplitude_consistent
    )

    post_fixation_mask = (
        (relative_time_ms >= post_window_start_ms)
        & (relative_time_ms <= post_window_end_ms)
    )

    return {
        "relative_time_ms": relative_time_ms,
        "absolute_time_ms": interpolated[
            "absolute_time_ms"
        ],
        "x_deg": x_deg,
        "y_deg": y_deg,

        "pre_fixation_mask": pre_fixation_mask,
        "post_fixation_mask": post_fixation_mask,

        "pre_fixation_x_deg": pre_fixation_x_deg,
        "pre_fixation_y_deg": pre_fixation_y_deg,
        "pre_fixation_radial_mad_deg": (
            pre_fixation_radial_mad_deg
        ),
        "pre_fixation_speed_deg_s": (
            pre_fixation_speed_deg_s
        ),

        "post_fixation_x_deg": post_fixation_x_deg,
        "post_fixation_y_deg": post_fixation_y_deg,
        "post_fixation_radial_mad_deg": (
            post_fixation_radial_mad_deg
        ),
        "post_fixation_speed_deg_s": (
            post_fixation_speed_deg_s
        ),
        "post_fixation_stability_score": (
            post_fixation_stability_score
        ),

        "post_window_start_ms": post_window_start_ms,
        "post_window_end_ms": post_window_end_ms,

        "onset_x_deg": onset_x_deg,
        "onset_y_deg": onset_y_deg,
        "onset_to_prefix_distance_deg": (
            onset_to_prefix_distance_deg
        ),

        "fixation_dx_deg": fixation_dx_deg,
        "fixation_dy_deg": fixation_dy_deg,
        "direction_x": direction_x,
        "direction_y": direction_y,

        # This is the value supplied to the harmonic model.
        "paper_xm_deg": paper_xm_deg,

        "observed_postfix_position_from_onset_deg": (
            observed_postfix_position_from_onset_deg
        ),

        "remodnav_amplitude_deg": (
            remodnav_amplitude_deg
        ),
        "xm_to_remodnav_ratio": (
            xm_to_remodnav_ratio
        ),

        "progress_deg": progress_deg,
        "orthogonal_deg": orthogonal_deg,

        "passes_fixation_qc": bool(
            passes_fixation_qc
        ),
    }, None


# ============================================================
# 6. RECONSTRUCT PAPER-STYLE TRAJECTORIES FOR ALL CLEAN EVENTS
# ============================================================

clean_event_rows = (
    selected_event_rows
    .reset_index(drop=True)
    .copy()
)

paper_results_by_index = {}
paper_ranking_rows = []


for trajectory_index, event_row in (
    clean_event_rows.iterrows()
):

    result, exclusion_reason = (
        estimate_paper_style_xm(
            event_row=event_row,
            block_data=selected_block_data,
        )
    )

    trial_value = get_numeric_value(
        event_row,
        ["trial"]
    )

    saccade_index_value = get_numeric_value(
        event_row,
        ["saccade_index"]
    )

    start_abs_ms_value = get_numeric_value(
        event_row,
        ["start_abs_ms"]
    )

    if exclusion_reason is not None:

        paper_ranking_rows.append({
            "trajectory_index": trajectory_index,
            "trial": trial_value,
            "saccade_index": saccade_index_value,
            "start_abs_ms": start_abs_ms_value,
            "paper_xm_available": False,
            "passes_fixation_qc": False,
            "exclusion_reason": exclusion_reason,
        })

        continue

    paper_results_by_index[
        trajectory_index
    ] = result

    paper_ranking_rows.append({
        "trajectory_index": trajectory_index,
        "trial": trial_value,
        "saccade_index": saccade_index_value,
        "start_abs_ms": start_abs_ms_value,

        "paper_xm_available": True,
        "passes_fixation_qc": result[
            "passes_fixation_qc"
        ],
        "exclusion_reason": "",

        "remodnav_amplitude_deg": result[
            "remodnav_amplitude_deg"
        ],
        "paper_xm_deg": result[
            "paper_xm_deg"
        ],
        "xm_to_remodnav_ratio": result[
            "xm_to_remodnav_ratio"
        ],

        "pre_fixation_x_deg": result[
            "pre_fixation_x_deg"
        ],
        "pre_fixation_y_deg": result[
            "pre_fixation_y_deg"
        ],
        "pre_fixation_radial_mad_deg": result[
            "pre_fixation_radial_mad_deg"
        ],
        "pre_fixation_speed_deg_s": result[
            "pre_fixation_speed_deg_s"
        ],

        "post_fixation_x_deg": result[
            "post_fixation_x_deg"
        ],
        "post_fixation_y_deg": result[
            "post_fixation_y_deg"
        ],
        "post_fixation_radial_mad_deg": result[
            "post_fixation_radial_mad_deg"
        ],
        "post_fixation_speed_deg_s": result[
            "post_fixation_speed_deg_s"
        ],

        "post_window_start_ms": result[
            "post_window_start_ms"
        ],
        "post_window_end_ms": result[
            "post_window_end_ms"
        ],

        "onset_to_prefix_distance_deg": result[
            "onset_to_prefix_distance_deg"
        ],
    })


paper_ranking = pd.DataFrame(
    paper_ranking_rows
)

valid_paper_indices = sorted(
    paper_results_by_index.keys()
)

if not valid_paper_indices:
    raise RuntimeError(
        "No clean saccade produced a usable "
        "fixation-to-fixation amplitude."
    )

print(
    "\nPaper-style xm successfully estimated for "
    f"{len(valid_paper_indices)}/{len(clean_event_rows)} "
    "clean saccades."
)


# ============================================================
# 7. CREATE THE PAPER-PROJECTED MEDIAN FOR REPRESENTATIVE
#    SACCADE SELECTION
# ============================================================

paper_progress_matrix = np.vstack([
    paper_results_by_index[index][
        "progress_deg"
    ]
    for index in valid_paper_indices
])

paper_progress_median = np.median(
    paper_progress_matrix,
    axis=0
)

selection_fit_mask = (
    np.asarray(RELATIVE_TIME_MS)
    >= 0
)

shape_rmse_by_index = {}

for trajectory_index in valid_paper_indices:

    trajectory_progress = (
        paper_results_by_index[
            trajectory_index
        ]["progress_deg"]
    )

    shape_rmse = float(
        np.sqrt(
            np.mean(
                (
                    trajectory_progress[
                        selection_fit_mask
                    ]
                    - paper_progress_median[
                        selection_fit_mask
                    ]
                ) ** 2
            )
        )
    )

    shape_rmse_by_index[
        trajectory_index
    ] = shape_rmse


paper_ranking[
    "shape_rmse_to_paper_median_deg"
] = paper_ranking[
    "trajectory_index"
].map(
    shape_rmse_by_index
)


# Prefer trajectories passing all fixation QC criteria.
representative_candidates = (
    paper_ranking.loc[
        paper_ranking[
            "paper_xm_available"
        ].eq(True)
        & paper_ranking[
            "passes_fixation_qc"
        ].eq(True)
    ]
    .copy()
)

if not representative_candidates.empty:

    representative_candidates = (
        representative_candidates
        .sort_values(
            [
                "shape_rmse_to_paper_median_deg",
                "post_fixation_radial_mad_deg",
                "onset_to_prefix_distance_deg",
            ]
        )
    )

    representative_row = (
        representative_candidates.iloc[0]
    )

    selection_method = (
        "Closest to the median of paper-projected clean "
        "trajectories among saccades passing fixation QC"
    )

else:

    warnings.warn(
        "No trajectory passed every fixation QC threshold. "
        "Using the valid trajectory closest to the "
        "paper-projected median."
    )

    representative_candidates = (
        paper_ranking.loc[
            paper_ranking[
                "paper_xm_available"
            ].eq(True)
        ]
        .copy()
        .sort_values(
            "shape_rmse_to_paper_median_deg"
        )
    )

    representative_row = (
        representative_candidates.iloc[0]
    )

    selection_method = (
        "Closest to the median of paper-projected clean "
        "trajectories; fixation QC fallback"
    )


REPRESENTATIVE_INDEX = int(
    representative_row[
        "trajectory_index"
    ]
)

representative_event_row = (
    clean_event_rows
    .iloc[
        REPRESENTATIVE_INDEX
    ]
    .copy()
)

representative_result = (
    paper_results_by_index[
        REPRESENTATIVE_INDEX
    ]
)


# ============================================================
# 8. EXTRACT THE REPRESENTATIVE PAPER-STYLE TRAJECTORY
# ============================================================

representative_progress_deg = np.asarray(
    representative_result[
        "progress_deg"
    ],
    dtype=float
)

representative_orthogonal_deg = np.asarray(
    representative_result[
        "orthogonal_deg"
    ],
    dtype=float
)

representative_x_deg = np.asarray(
    representative_result[
        "x_deg"
    ],
    dtype=float
)

representative_y_deg = np.asarray(
    representative_result[
        "y_deg"
    ],
    dtype=float
)

pre_fixation_mask = np.asarray(
    representative_result[
        "pre_fixation_mask"
    ],
    dtype=bool
)

post_fixation_mask = np.asarray(
    representative_result[
        "post_fixation_mask"
    ],
    dtype=bool
)


# PAPER-STYLE xm USED BY THE MODEL
SINGLE_XM_DEG = float(
    representative_result[
        "paper_xm_deg"
    ]
)


print(
    "\n"
    + "=" * 78
)

print(
    "REPRESENTATIVE SACCADE WITH PAPER-STYLE xm"
)

print(
    "=" * 78
)

print(
    f"Selection method: {selection_method}"
)

print(
    f"Trajectory index: {REPRESENTATIVE_INDEX}"
)

print(
    f"Trial: "
    f"{representative_row.get('trial', np.nan)}"
)

print(
    f"REMoDNaV within-trial saccade index: "
    f"{representative_row.get('saccade_index', np.nan)}"
)

print(
    "RMSE to paper-projected median: "
    f"{representative_row['shape_rmse_to_paper_median_deg']:.4f}°"
)

print(
    "REMoDNaV amplitude: "
    f"{representative_result['remodnav_amplitude_deg']:.4f}°"
)

print(
    "Paper-style fixation-to-fixation xm: "
    f"{SINGLE_XM_DEG:.4f}°"
)

print(
    "Pre-fixation center: "
    f"({representative_result['pre_fixation_x_deg']:.4f}, "
    f"{representative_result['pre_fixation_y_deg']:.4f})°"
)

print(
    "Post-fixation center: "
    f"({representative_result['post_fixation_x_deg']:.4f}, "
    f"{representative_result['post_fixation_y_deg']:.4f})°"
)

print(
    "Post-fixation window: "
    f"{representative_result['post_window_start_ms']:.0f}–"
    f"{representative_result['post_window_end_ms']:.0f} ms"
)

print(
    "Onset-to-pre-fixation distance: "
    f"{representative_result['onset_to_prefix_distance_deg']:.4f}°"
)

print(
    "Passed fixation QC: "
    f"{representative_result['passes_fixation_qc']}"
)


# ============================================================
# 9. FIT THE CONSTANT HARMONIC MODEL
# ============================================================

single_fit_time_ms = np.asarray(
    RELATIVE_TIME_MS[
        FIT_MASK
    ],
    dtype=float
)

single_observed_deg = np.asarray(
    representative_progress_deg[
        FIT_MASK
    ],
    dtype=float
)

# Numerical safeguard: t=0 must equal exactly zero.
single_observed_deg = (
    single_observed_deg
    - single_observed_deg[0]
)


def single_saccade_residuals(
    parameter_vector
):
    """
    Residuals for one paper-style fixation-to-fixation
    amplitude fit.
    """

    A_value = float(
        parameter_vector[0]
    )

    mu_value = float(
        parameter_vector[1]
    )

    gamma_value = float(
        parameter_vector[2]
    )

    k_value = float(
        parameter_vector[3]
    )

    try:

        prediction = predict_constant_model(
            time_ms=single_fit_time_ms,
            A=A_value,
            mu=mu_value,
            gamma_value=gamma_value,
            k_value=k_value,
            x_m=SINGLE_XM_DEG,
        )

        predicted_observed = np.asarray(
            prediction[
                "predicted_observed_deg"
            ],
            dtype=float
        )

        residuals = (
            predicted_observed
            - single_observed_deg
        )

        if not np.isfinite(
            residuals
        ).all():

            return np.full(
                len(single_observed_deg),
                1e6
            )

        return residuals

    except Exception:

        return np.full(
            len(single_observed_deg),
            1e6
        )


single_rng = np.random.default_rng(
    SINGLE_RANDOM_SEED
)

single_starting_values = [
    (
        LOWER_BOUNDS
        + UPPER_BOUNDS
    )
    / 2.0
]

for _ in range(
    N_SINGLE_SACCADE_STARTS - 1
):

    single_starting_values.append(
        single_rng.uniform(
            LOWER_BOUNDS,
            UPPER_BOUNDS
        )
    )


single_optimization_rows = []
single_successful_fits = []

print(
    "\nFitting constant model using paper-style xm..."
)


for start_number, initial_values in enumerate(
    single_starting_values,
    start=1
):

    try:

        fitted_result = least_squares(
            fun=single_saccade_residuals,
            x0=initial_values,
            bounds=(
                LOWER_BOUNDS,
                UPPER_BOUNDS
            ),
            method="trf",
            loss="linear",
            x_scale="jac",
            max_nfev=1000,
            ftol=1e-11,
            xtol=1e-11,
            gtol=1e-11,
        )

        residual_vector = (
            single_saccade_residuals(
                fitted_result.x
            )
        )

        rss_value = float(
            np.sum(
                residual_vector ** 2
            )
        )

        optimization_row = {
            "start_number": start_number,
            "success": bool(
                fitted_result.success
            ),
            "message": str(
                fitted_result.message
            ),
            "n_function_evaluations": int(
                fitted_result.nfev
            ),
            "rss": rss_value,
            "A": fitted_result.x[0],
            "mu": fitted_result.x[1],
            "gamma": fitted_result.x[2],
            "k": fitted_result.x[3],
        }

        single_optimization_rows.append(
            optimization_row
        )

        if (
            fitted_result.success
            and np.isfinite(rss_value)
        ):

            single_successful_fits.append(
                (
                    rss_value,
                    fitted_result
                )
            )

    except Exception as error:

        single_optimization_rows.append({
            "start_number": start_number,
            "success": False,
            "message": (
                f"{type(error).__name__}: {error}"
            ),
            "n_function_evaluations": np.nan,
            "rss": np.nan,
            "A": np.nan,
            "mu": np.nan,
            "gamma": np.nan,
            "k": np.nan,
        })


single_optimization_results = pd.DataFrame(
    single_optimization_rows
)

if not single_successful_fits:
    raise RuntimeError(
        "No multistart optimization run succeeded."
    )

single_successful_fits.sort(
    key=lambda item: item[0]
)

SINGLE_BEST_RSS, single_best_fit = (
    single_successful_fits[0]
)

SINGLE_BEST_A = float(
    single_best_fit.x[0]
)

SINGLE_BEST_MU = float(
    single_best_fit.x[1]
)

SINGLE_BEST_GAMMA = float(
    single_best_fit.x[2]
)

SINGLE_BEST_K = float(
    single_best_fit.x[3]
)


# ============================================================
# 10. BEST-FIT PREDICTIONS
# ============================================================

single_best_prediction = predict_constant_model(
    time_ms=single_fit_time_ms,
    A=SINGLE_BEST_A,
    mu=SINGLE_BEST_MU,
    gamma_value=SINGLE_BEST_GAMMA,
    k_value=SINGLE_BEST_K,
    x_m=SINGLE_XM_DEG,
)

single_predicted_pupil_deg = np.asarray(
    single_best_prediction[
        "predicted_observed_deg"
    ],
    dtype=float
)

single_inferred_eyeball_deg = np.asarray(
    single_best_prediction[
        "x_eye_deg"
    ],
    dtype=float
)

single_inferred_iris_deg = np.asarray(
    single_best_prediction[
        "y_iris_deg"
    ],
    dtype=float
)

single_residual_deg = (
    single_observed_deg
    - single_predicted_pupil_deg
)

single_rmse_deg = float(
    np.sqrt(
        np.mean(
            single_residual_deg ** 2
        )
    )
)

single_total_sum_squares = float(
    np.sum(
        (
            single_observed_deg
            - np.mean(
                single_observed_deg
            )
        ) ** 2
    )
)

if single_total_sum_squares > 0:

    single_r_squared = (
        1.0
        - SINGLE_BEST_RSS
        / single_total_sum_squares
    )

else:

    single_r_squared = np.nan


single_n_observations = len(
    single_observed_deg
)

single_bic = (
    single_n_observations
    * np.log(
        max(
            SINGLE_BEST_RSS
            / single_n_observations,
            np.finfo(float).tiny
        )
    )
    + 4
    * np.log(
        single_n_observations
    )
)

single_tau_ms = calculate_tau_ms(
    A=SINGLE_BEST_A,
    mu=SINGLE_BEST_MU,
    x_m=SINGLE_XM_DEG,
)

single_natural_angular_frequency_per_ms = (
    np.sqrt(
        SINGLE_BEST_K
    )
)

single_natural_frequency_hz = (
    single_natural_angular_frequency_per_ms
    / (
        2.0
        * np.pi
    )
    * 1000.0
)

single_damping_ratio = (
    SINGLE_BEST_GAMMA
    / (
        2.0
        * np.sqrt(
            SINGLE_BEST_K
        )
    )
)

single_eyeball_velocity_deg_s = (
    1000.0
    * eyeball_velocity_deg_per_ms(
        time_ms=single_fit_time_ms,
        A=SINGLE_BEST_A,
        mu=SINGLE_BEST_MU,
        x_m=SINGLE_XM_DEG,
    )
)

single_peak_velocity_deg_s = float(
    np.max(
        single_eyeball_velocity_deg_s
    )
)

single_98_percent_indices = np.flatnonzero(
    single_inferred_eyeball_deg
    >= (
        0.98
        * SINGLE_XM_DEG
    )
)

if len(single_98_percent_indices):

    single_duration_98_ms = float(
        single_fit_time_ms[
            single_98_percent_indices[0]
        ]
    )

else:

    single_duration_98_ms = np.nan


# ============================================================
# 11. BOUND CHECK
# ============================================================

single_parameter_values = np.array([
    SINGLE_BEST_A,
    SINGLE_BEST_MU,
    SINGLE_BEST_GAMMA,
    SINGLE_BEST_K,
])

single_parameter_ranges = (
    UPPER_BOUNDS
    - LOWER_BOUNDS
)

single_near_lower = (
    single_parameter_values
    - LOWER_BOUNDS
) <= (
    0.01
    * single_parameter_ranges
)

single_near_upper = (
    UPPER_BOUNDS
    - single_parameter_values
) <= (
    0.01
    * single_parameter_ranges
)

single_parameters_near_bounds = [
    PARAMETER_NAMES[index]
    for index in range(
        len(PARAMETER_NAMES)
    )
    if (
        single_near_lower[index]
        or single_near_upper[index]
    )
]


# ============================================================
# 12. SUMMARY TABLE
# ============================================================

single_fit_summary = pd.DataFrame([{
    "subject": (
        SUBJECT_ID
        if "SUBJECT_ID" in globals()
        else "s189"
    ),

    "block_id": selected_block_id,

    "representative_trajectory_index": (
        REPRESENTATIVE_INDEX
    ),

    "trial": get_numeric_value(
        representative_event_row,
        ["trial"]
    ),

    "remodnav_saccade_index_within_trial": (
        get_numeric_value(
            representative_event_row,
            ["saccade_index"]
        )
    ),

    "start_abs_ms": get_numeric_value(
        representative_event_row,
        ["start_abs_ms"]
    ),

    "selection_method": selection_method,

    "shape_rmse_to_paper_median_deg": (
        representative_row[
            "shape_rmse_to_paper_median_deg"
        ]
    ),

    "xm_definition": (
        "2D Euclidean distance between estimated "
        "pre- and post-saccadic fixation centers"
    ),

    "remodnav_amplitude_deg": (
        representative_result[
            "remodnav_amplitude_deg"
        ]
    ),

    "paper_xm_deg": SINGLE_XM_DEG,

    "xm_to_remodnav_ratio": (
        representative_result[
            "xm_to_remodnav_ratio"
        ]
    ),

    "pre_fixation_start_ms": (
        PRE_FIXATION_START_MS
    ),

    "pre_fixation_end_ms": (
        PRE_FIXATION_END_MS
    ),

    "pre_fixation_x_deg": (
        representative_result[
            "pre_fixation_x_deg"
        ]
    ),

    "pre_fixation_y_deg": (
        representative_result[
            "pre_fixation_y_deg"
        ]
    ),

    "pre_fixation_radial_mad_deg": (
        representative_result[
            "pre_fixation_radial_mad_deg"
        ]
    ),

    "pre_fixation_speed_deg_s": (
        representative_result[
            "pre_fixation_speed_deg_s"
        ]
    ),

    "post_fixation_window_start_ms": (
        representative_result[
            "post_window_start_ms"
        ]
    ),

    "post_fixation_window_end_ms": (
        representative_result[
            "post_window_end_ms"
        ]
    ),

    "post_fixation_x_deg": (
        representative_result[
            "post_fixation_x_deg"
        ]
    ),

    "post_fixation_y_deg": (
        representative_result[
            "post_fixation_y_deg"
        ]
    ),

    "post_fixation_radial_mad_deg": (
        representative_result[
            "post_fixation_radial_mad_deg"
        ]
    ),

    "post_fixation_speed_deg_s": (
        representative_result[
            "post_fixation_speed_deg_s"
        ]
    ),

    "onset_x_deg": (
        representative_result[
            "onset_x_deg"
        ]
    ),

    "onset_y_deg": (
        representative_result[
            "onset_y_deg"
        ]
    ),

    "onset_to_prefix_distance_deg": (
        representative_result[
            "onset_to_prefix_distance_deg"
        ]
    ),

    "passed_fixation_qc": (
        representative_result[
            "passes_fixation_qc"
        ]
    ),

    "beta_fixed": BETA,

    "A": SINGLE_BEST_A,
    "mu": SINGLE_BEST_MU,
    "gamma_per_ms": SINGLE_BEST_GAMMA,
    "k_per_ms_squared": SINGLE_BEST_K,

    "tau_ms": single_tau_ms,

    "natural_angular_frequency_per_ms": (
        single_natural_angular_frequency_per_ms
    ),

    "natural_frequency_hz": (
        single_natural_frequency_hz
    ),

    "damping_ratio": (
        single_damping_ratio
    ),

    "rmse_deg": single_rmse_deg,
    "r_squared": single_r_squared,
    "rss": SINGLE_BEST_RSS,
    "bic": single_bic,

    "inferred_eyeball_peak_velocity_deg_s": (
        single_peak_velocity_deg_s
    ),

    "inferred_eyeball_duration_98_ms": (
        single_duration_98_ms
    ),

    "parameters_near_bounds": (
        "; ".join(
            single_parameters_near_bounds
        )
        if single_parameters_near_bounds
        else ""
    ),
}])


print(
    "\n"
    + "=" * 78
)

print(
    "PAPER-STYLE FIXATION-TO-FIXATION xm FIT"
)

print(
    "=" * 78
)

display(
    single_fit_summary.T
)

print(
    "\nBest five multistart solutions:"
)

display(
    single_optimization_results
    .sort_values("rss")
    .head()
)

if single_parameters_near_bounds:

    warnings.warn(
        "The following fitted parameters remain close "
        "to their bounds: "
        + ", ".join(
            single_parameters_near_bounds
        )
    )


# ============================================================
# 13. SAVE TABLES
# ============================================================

paper_ranking_path = (
    PAPER_XM_OUT_DIR
    / "paper_xm_clean_saccade_ranking.csv"
)

representative_event_path = (
    PAPER_XM_OUT_DIR
    / "representative_saccade_event_paper_xm.csv"
)

single_trace_path = (
    PAPER_XM_OUT_DIR
    / "representative_saccade_trace_paper_xm.csv"
)

single_fit_path = (
    PAPER_XM_OUT_DIR
    / "representative_saccade_constant_model_fit_paper_xm.csv"
)

single_optimization_path = (
    PAPER_XM_OUT_DIR
    / "representative_saccade_multistart_paper_xm.csv"
)


paper_ranking.to_csv(
    paper_ranking_path,
    index=False
)


representative_event_output = pd.DataFrame([
    representative_event_row.to_dict()
])

for column, value in {
    "paper_xm_deg": SINGLE_XM_DEG,

    "pre_fixation_x_deg": (
        representative_result[
            "pre_fixation_x_deg"
        ]
    ),

    "pre_fixation_y_deg": (
        representative_result[
            "pre_fixation_y_deg"
        ]
    ),

    "post_fixation_x_deg": (
        representative_result[
            "post_fixation_x_deg"
        ]
    ),

    "post_fixation_y_deg": (
        representative_result[
            "post_fixation_y_deg"
        ]
    ),

    "post_fixation_window_start_ms": (
        representative_result[
            "post_window_start_ms"
        ]
    ),

    "post_fixation_window_end_ms": (
        representative_result[
            "post_window_end_ms"
        ]
    ),

    "onset_to_prefix_distance_deg": (
        representative_result[
            "onset_to_prefix_distance_deg"
        ]
    ),
}.items():

    representative_event_output[
        column
    ] = value


representative_event_output.to_csv(
    representative_event_path,
    index=False
)


single_trace_output = pd.DataFrame({
    "relative_time_ms": (
        RELATIVE_TIME_MS
    ),

    "absolute_time_ms": (
        representative_result[
            "absolute_time_ms"
        ]
    ),

    "raw_calibrated_x_deg": (
        representative_x_deg
    ),

    "raw_calibrated_y_deg": (
        representative_y_deg
    ),

    "paper_direction_projected_progress_deg": (
        representative_progress_deg
    ),

    "paper_direction_orthogonal_deg": (
        representative_orthogonal_deg
    ),

    "in_pre_fixation_window": (
        pre_fixation_mask
    ),

    "in_post_fixation_window": (
        post_fixation_mask
    ),

    "used_for_model_fit": (
        FIT_MASK
    ),
})


single_trace_output[
    "model_predicted_pupil_deg"
] = np.nan

single_trace_output[
    "model_inferred_eyeball_deg"
] = np.nan

single_trace_output[
    "model_inferred_iris_relative_deg"
] = np.nan

single_trace_output[
    "residual_observed_minus_predicted_deg"
] = np.nan


single_trace_output.loc[
    FIT_MASK,
    "model_predicted_pupil_deg"
] = single_predicted_pupil_deg

single_trace_output.loc[
    FIT_MASK,
    "model_inferred_eyeball_deg"
] = single_inferred_eyeball_deg

single_trace_output.loc[
    FIT_MASK,
    "model_inferred_iris_relative_deg"
] = single_inferred_iris_deg

single_trace_output.loc[
    FIT_MASK,
    "residual_observed_minus_predicted_deg"
] = single_residual_deg


single_trace_output.to_csv(
    single_trace_path,
    index=False
)

single_fit_summary.to_csv(
    single_fit_path,
    index=False
)

single_optimization_results.to_csv(
    single_optimization_path,
    index=False
)


print("\nSaved tables:")
print(paper_ranking_path)
print(representative_event_path)
print(single_trace_path)
print(single_fit_path)
print(single_optimization_path)


# ============================================================
# 14. FIGURE 1:
# 2D FIXATION-TO-FIXATION GEOMETRY
# ============================================================

plt.figure(
    figsize=(7, 7)
)

plt.plot(
    representative_x_deg,
    representative_y_deg,
    linewidth=1.5,
    label="Raw calibrated gaze"
)

plt.scatter(
    representative_result[
        "pre_fixation_x_deg"
    ],
    representative_result[
        "pre_fixation_y_deg"
    ],
    s=90,
    marker="o",
    label="Pre-fixation center"
)

plt.scatter(
    representative_result[
        "post_fixation_x_deg"
    ],
    representative_result[
        "post_fixation_y_deg"
    ],
    s=90,
    marker="s",
    label="Post-fixation center"
)

plt.scatter(
    representative_result[
        "onset_x_deg"
    ],
    representative_result[
        "onset_y_deg"
    ],
    s=80,
    marker="x",
    label="REMoDNaV onset position"
)

plt.plot(
    [
        representative_result[
            "pre_fixation_x_deg"
        ],
        representative_result[
            "post_fixation_x_deg"
        ],
    ],
    [
        representative_result[
            "pre_fixation_y_deg"
        ],
        representative_result[
            "post_fixation_y_deg"
        ],
    ],
    linestyle="--",
    linewidth=2,
    label=(
        f"Full amplitude xm = "
        f"{SINGLE_XM_DEG:.3f}°"
    )
)

plt.xlabel(
    "Horizontal calibrated gaze (degrees)"
)

plt.ylabel(
    "Vertical calibrated gaze (degrees)"
)

plt.title(
    "Paper-style fixation-to-fixation amplitude"
)

plt.gca().set_aspect(
    "equal",
    adjustable="datalim"
)

plt.legend()
plt.tight_layout()

fixation_geometry_figure = (
    PAPER_XM_OUT_DIR
    / "paper_xm_fixation_geometry.png"
)

plt.savefig(
    fixation_geometry_figure,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print(
    f"Saved: {fixation_geometry_figure}"
)


# ============================================================
# 15. FIGURE 2:
# REPRESENTATIVE TRAJECTORY VS PAPER-PROJECTED MEDIAN
# ============================================================

plt.figure(
    figsize=(10, 6)
)

plt.plot(
    RELATIVE_TIME_MS,
    paper_progress_median,
    linewidth=2,
    linestyle="--",
    label="Median of paper-projected clean trajectories"
)

plt.plot(
    RELATIVE_TIME_MS,
    representative_progress_deg,
    linewidth=2,
    label="Selected representative saccade"
)

plt.axvspan(
    PRE_FIXATION_START_MS,
    PRE_FIXATION_END_MS,
    alpha=0.12,
    label="Pre-fixation-center window"
)

plt.axvspan(
    representative_result[
        "post_window_start_ms"
    ],
    representative_result[
        "post_window_end_ms"
    ],
    alpha=0.12,
    label="Post-fixation-center window"
)

plt.axhline(
    SINGLE_XM_DEG,
    linestyle=":",
    linewidth=1.5,
    label=f"Full amplitude xm = {SINGLE_XM_DEG:.3f}°"
)

plt.axvline(
    0,
    linestyle="--",
    linewidth=1
)

plt.xlabel(
    "Time from REMoDNaV onset (ms)"
)

plt.ylabel(
    "Position change along fixation-to-fixation direction (degrees)"
)

plt.title(
    "Representative clean saccade using paper-style projection"
)

plt.legend()
plt.tight_layout()

selection_figure = (
    PAPER_XM_OUT_DIR
    / "representative_selection_paper_xm.png"
)

plt.savefig(
    selection_figure,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print(
    f"Saved: {selection_figure}"
)


# ============================================================
# 16. FIGURE 3:
# OBSERVED, PREDICTED PUPIL AND INFERRED EYEBALL
# ============================================================

plt.figure(
    figsize=(10, 6)
)

plt.plot(
    single_fit_time_ms,
    single_observed_deg,
    linewidth=2.5,
    label="Observed pupil/gaze trajectory"
)

plt.plot(
    single_fit_time_ms,
    single_predicted_pupil_deg,
    linewidth=2,
    label="Model-predicted pupil trajectory"
)

plt.plot(
    single_fit_time_ms,
    single_inferred_eyeball_deg,
    linewidth=2,
    linestyle="--",
    label="Model-inferred eyeball trajectory"
)

plt.axhline(
    SINGLE_XM_DEG,
    linestyle=":",
    linewidth=1.5,
    label=(
        f"Fixation-to-fixation xm = "
        f"{SINGLE_XM_DEG:.3f}°"
    )
)

plt.xlabel(
    "Time from saccade onset (ms)"
)

plt.ylabel(
    "Position change (degrees)"
)

plt.title(
    "Constant harmonic model with paper-style xm\n"
    f"RMSE={single_rmse_deg:.4f}°, "
    f"R²={single_r_squared:.4f}"
)

plt.legend()
plt.tight_layout()

fit_figure = (
    PAPER_XM_OUT_DIR
    / "observed_predicted_eyeball_paper_xm.png"
)

plt.savefig(
    fit_figure,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print(
    f"Saved: {fit_figure}"
)


# ============================================================
# 17. FIGURE 4:
# RESIDUALS
# ============================================================

plt.figure(
    figsize=(10, 5)
)

plt.plot(
    single_fit_time_ms,
    single_residual_deg,
    linewidth=2
)

plt.axhline(
    0,
    linestyle="--",
    linewidth=1
)

plt.xlabel(
    "Time from saccade onset (ms)"
)

plt.ylabel(
    "Observed − predicted position (degrees)"
)

plt.title(
    "Residuals: paper-style fixation-to-fixation xm"
)

plt.tight_layout()

residual_figure = (
    PAPER_XM_OUT_DIR
    / "residuals_paper_xm.png"
)

plt.savefig(
    residual_figure,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print(
    f"Saved: {residual_figure}"
)


# ============================================================
# 18. FIGURE 5:
# INFERRED RELATIVE PUPIL/IRIS MOVEMENT
# ============================================================

plt.figure(
    figsize=(10, 5)
)

plt.plot(
    single_fit_time_ms,
    single_inferred_iris_deg,
    linewidth=2
)

plt.axhline(
    0,
    linestyle="--",
    linewidth=1
)

plt.xlabel(
    "Time from saccade onset (ms)"
)

plt.ylabel(
    "Inferred pupil/iris movement relative to eyeball (degrees)"
)

plt.title(
    "Model-inferred oscillatory component"
)

plt.tight_layout()

iris_figure = (
    PAPER_XM_OUT_DIR
    / "inferred_iris_component_paper_xm.png"
)

plt.savefig(
    iris_figure,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print(
    f"Saved: {iris_figure}"
)


# ============================================================
# 19. FINAL SUMMARY
# ============================================================

print(
    "\n"
    + "=" * 78
)

print(
    "PAPER-STYLE xm INDIVIDUAL-SACCADE ANALYSIS COMPLETE"
)

print(
    "=" * 78
)

print(
    f"Representative trajectory index: "
    f"{REPRESENTATIVE_INDEX}"
)

print(
    f"Trial: "
    f"{get_numeric_value(representative_event_row, ['trial'])}"
)

print(
    "REMoDNaV within-trial saccade index: "
    f"{get_numeric_value(representative_event_row, ['saccade_index'])}"
)

print(
    f"REMoDNaV amplitude: "
    f"{representative_result['remodnav_amplitude_deg']:.3f}°"
)

print(
    "Paper-style fixation-to-fixation xm: "
    f"{SINGLE_XM_DEG:.3f}°"
)

print(
    "Pre-fixation center: "
    f"({representative_result['pre_fixation_x_deg']:.3f}, "
    f"{representative_result['pre_fixation_y_deg']:.3f})°"
)

print(
    "Post-fixation center: "
    f"({representative_result['post_fixation_x_deg']:.3f}, "
    f"{representative_result['post_fixation_y_deg']:.3f})°"
)

print(
    "Post-fixation window: "
    f"{representative_result['post_window_start_ms']:.0f}–"
    f"{representative_result['post_window_end_ms']:.0f} ms"
)

print(
    "Onset-to-pre-fixation distance: "
    f"{representative_result['onset_to_prefix_distance_deg']:.3f}°"
)

print(
    f"A:     {SINGLE_BEST_A:.6f}"
)

print(
    f"mu:    {SINGLE_BEST_MU:.6f}"
)

print(
    f"gamma: {SINGLE_BEST_GAMMA:.6f} per ms"
)

print(
    f"k:     {SINGLE_BEST_K:.6f} per ms²"
)

print(
    f"RMSE:  {single_rmse_deg:.5f}°"
)

print(
    f"R²:    {single_r_squared:.5f}"
)

print(
    "Inferred eyeball peak velocity: "
    f"{single_peak_velocity_deg_s:.2f} deg/s"
)

print(
    "Inferred eyeball 98% duration: "
    f"{single_duration_98_ms:.2f} ms"
)

print(
    "Parameters near bounds: "
    + (
        ", ".join(
            single_parameters_near_bounds
        )
        if single_parameters_near_bounds
        else "None"
    )
)

print(
    "\nAll results saved in:"
)

print(
    PAPER_XM_OUT_DIR
)
```

```python
# ============================================================
# FOLLOW-UP REMoDNaV GRAPHS: 4, 5, 6, 7, 8, AND 11
#
# Input:
#   /content/drive/MyDrive/steer_search_PD/all_remodnav/
#       all_remodnav_events_<subject>.csv
#
# Output:
#   /content/drive/MyDrive/steer_search_PD/all_remodnav/
#       specific_followup_graphs/
#
# Definitions used here:
#   Saccade       = SACC or ISAC
#   Fixation      = FIXA
#   High PSO      = HPSO or IHPS immediately following a saccade
#   Eligible PSO  = saccade ends at least one PSO window before RESPONSE
#
# To reproduce a prior analysis that counted only major saccades,
# change SACCADE_LABELS to {"SACC"} near the top.
# ============================================================

# -------------------------
# 0. Mount Google Drive
# -------------------------
from google.colab import drive
drive.mount("/content/drive", force_remount=True)

# -------------------------
# 1. Imports
# -------------------------
from pathlib import Path
import re
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.stats import spearmanr, ttest_ind
from IPython.display import display


# ============================================================
# 2. SETTINGS
# ============================================================
EVENT_DIR = Path(
    "/content/drive/MyDrive/steer_search_PD/all_remodnav"
)

FIG_DIR = EVENT_DIR / "specific_followup_graphs"
FIG_DIR.mkdir(parents=True, exist_ok=True)

GROUP_ORDER = ["Control", "PD"]

# REMoDNaV event labels used in the requested analyses.
SACCADE_LABELS = {"SACC", "ISAC"}
FIXATION_LABELS = {"FIXA"}
HIGH_PSO_LABELS = {"HPSO", "IHPS"}

# The reference REMoDNaV run used a 40-ms maximum PSO duration.
# If requested_pso_ms exists in the CSVs, the code automatically uses
# its median instead.
DEFAULT_PSO_WINDOW_MS = 40.0
AUTO_USE_FILE_PSO_WINDOW = True

# A high PSO must be the immediately following event. This optional timing
# check prevents pairing events separated by a long blank interval.
PSO_PAIR_MIN_GAP_MS = -2.0
PSO_PAIR_MAX_GAP_MS = 10.0

# Participant-balanced curve settings.
N_BOOT = 2000
RANDOM_SEED = 20260804
MIN_SUBJECTS_PER_GROUP_BIN = 5
MIN_TRIALS_PER_SUBJECT_BIN = 3
MIN_SACCADES_PER_SUBJECT_BIN = 10

# Plot limits/settings.
N_VIEWING_DURATION_BINS = 10
N_AMPLITUDE_BINS = 8
N_VELOCITY_BINS = 8
MAX_FIXATION_ORDER = 12
SURVIVAL_MAX_MS = 1000.0
SURVIVAL_N_THRESHOLDS = 151
LABEL_ALL_PARTICIPANTS = True

# First-saccade latency QC. Set APPLY_FIRST_LATENCY_QC=False to retain
# every positive first-saccade latency.
APPLY_FIRST_LATENCY_QC = False
FIRST_LATENCY_MIN_MS = 80.0
FIRST_LATENCY_MAX_MS = 1000.0

# Image format.
SAVE_DPI = 240

assert EVENT_DIR.exists(), f"Missing REMoDNaV folder: {EVENT_DIR}"

rng = np.random.default_rng(RANDOM_SEED)

cycle_colors = plt.rcParams["axes.prop_cycle"].by_key()["color"]
GROUP_COLORS = {
    group: cycle_colors[index]
    for index, group in enumerate(GROUP_ORDER)
}


# ============================================================
# 3. SMALL HELPERS
# ============================================================
def infer_group(subject):
    subject = str(subject).lower()
    if subject.startswith("p"):
        return "PD"
    if subject.startswith("s"):
        return "Control"
    return "Unknown"


def subject_from_filename(path):
    match = re.search(r"all_remodnav_events_(.+)\.csv$", path.name)
    return match.group(1) if match else path.stem


def save_figure(fig, filename):
    output_path = FIG_DIR / filename
    fig.savefig(output_path, dpi=SAVE_DPI, bbox_inches="tight")
    print("Saved:", output_path)
    return output_path


def quantile_bin_edges(series, n_bins):
    values = pd.to_numeric(series, errors="coerce").to_numpy(float)
    values = values[np.isfinite(values)]

    if len(values) < 10:
        raise RuntimeError("Too few finite values to form bins.")

    edges = np.quantile(values, np.linspace(0.0, 1.0, n_bins + 1))
    edges = np.unique(edges)

    if len(edges) < 3:
        raise RuntimeError("The variable does not contain enough unique values.")

    # Guarantee that minimum and maximum values fall inside the bins.
    edges[0] = np.nextafter(edges[0], -np.inf)
    edges[-1] = np.nextafter(edges[-1], np.inf)
    return edges


def matrix_statistic(matrix, statistic="mean"):
    with warnings.catch_warnings():
        warnings.simplefilter("ignore", category=RuntimeWarning)
        if statistic == "mean":
            return np.nanmean(matrix, axis=0)
        if statistic == "median":
            return np.nanmedian(matrix, axis=0)
    raise ValueError(f"Unknown statistic: {statistic}")


def bootstrap_matrix_ci(matrix, statistic="mean", n_boot=N_BOOT):
    """
    Bootstrap rows (participants), preserving the complete curve from each
    sampled participant. Columns may contain NaN when a participant has
    insufficient observations in a particular bin.
    """
    matrix = np.asarray(matrix, dtype=float)

    if matrix.ndim != 2 or matrix.shape[0] == 0:
        n_columns = matrix.shape[1] if matrix.ndim == 2 else 0
        return (
            np.full(n_columns, np.nan),
            np.full(n_columns, np.nan),
            np.full(n_columns, np.nan),
        )

    estimate = matrix_statistic(matrix, statistic=statistic)
    boot = np.full((n_boot, matrix.shape[1]), np.nan)

    for boot_index in range(n_boot):
        sampled_rows = rng.integers(0, matrix.shape[0], size=matrix.shape[0])
        boot[boot_index] = matrix_statistic(
            matrix[sampled_rows],
            statistic=statistic,
        )

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", category=RuntimeWarning)
        lower = np.nanpercentile(boot, 2.5, axis=0)
        upper = np.nanpercentile(boot, 97.5, axis=0)

    return estimate, lower, upper


def hedges_g_pd_minus_control(control_values, pd_values):
    """Hedges' g where positive means PD > Control."""
    control_values = np.asarray(control_values, dtype=float)
    pd_values = np.asarray(pd_values, dtype=float)
    control_values = control_values[np.isfinite(control_values)]
    pd_values = pd_values[np.isfinite(pd_values)]

    n_control = len(control_values)
    n_pd = len(pd_values)

    if n_control < 2 or n_pd < 2:
        return np.nan

    var_control = np.var(control_values, ddof=1)
    var_pd = np.var(pd_values, ddof=1)

    pooled_var = (
        (n_control - 1) * var_control
        + (n_pd - 1) * var_pd
    ) / (n_control + n_pd - 2)

    if not np.isfinite(pooled_var) or pooled_var <= 0:
        return np.nan

    pooled_sd = np.sqrt(pooled_var)
    cohen_d = (
        np.mean(pd_values) - np.mean(control_values)
    ) / pooled_sd

    correction = 1.0 - 3.0 / (4.0 * (n_control + n_pd) - 9.0)
    return correction * cohen_d


def format_p(p_value):
    if not np.isfinite(p_value):
        return "NA"
    if p_value < 0.001:
        return "<.001"
    return f"{p_value:.3f}".replace("0.", ".")


def participant_boxplot(
    dataframe,
    value_column,
    ylabel,
    title,
    filename,
):
    plot_data = dataframe[
        ["subject", "group", value_column]
    ].dropna().copy()

    values_by_group = {
        group: plot_data.loc[
            plot_data["group"].eq(group), value_column
        ].to_numpy(float)
        for group in GROUP_ORDER
    }

    control_values = values_by_group["Control"]
    pd_values = values_by_group["PD"]

    if len(control_values) >= 2 and len(pd_values) >= 2:
        welch = ttest_ind(
            control_values,
            pd_values,
            equal_var=False,
            nan_policy="omit",
        )
        p_value = float(welch.pvalue)
        g_value = hedges_g_pd_minus_control(control_values, pd_values)
    else:
        p_value = np.nan
        g_value = np.nan

    fig, ax = plt.subplots(figsize=(7.4, 6.0))

    ax.boxplot(
        [values_by_group[group] for group in GROUP_ORDER],
        positions=[1, 2],
        widths=0.52,
        showfliers=False,
        labels=GROUP_ORDER,
    )

    for group_position, group in enumerate(GROUP_ORDER, start=1):
        group_rows = plot_data.loc[
            plot_data["group"].eq(group)
        ].sort_values("subject")

        x_values = rng.normal(
            loc=float(group_position),
            scale=0.045,
            size=len(group_rows),
        )

        ax.scatter(
            x_values,
            group_rows[value_column],
            s=48,
            alpha=0.85,
            color=GROUP_COLORS[group],
            zorder=3,
        )

        if LABEL_ALL_PARTICIPANTS:
            for x_value, (_, row) in zip(x_values, group_rows.iterrows()):
                ax.annotate(
                    str(row["subject"]),
                    (x_value, row[value_column]),
                    xytext=(4, 4),
                    textcoords="offset points",
                    fontsize=7.5,
                    alpha=0.8,
                )

        if len(group_rows):
            ax.scatter(
                [group_position],
                [group_rows[value_column].mean()],
                marker="D",
                s=55,
                color=GROUP_COLORS[group],
                edgecolor="black",
                linewidth=0.6,
                zorder=4,
            )

    subtitle = (
        f"Welch p={format_p(p_value)}, "
        f"Hedges' g (PD-Control)={g_value:.3f}"
        if np.isfinite(g_value)
        else f"Welch p={format_p(p_value)}"
    )

    ax.set_title(f"{title}\n{subtitle}")
    ax.set_ylabel(ylabel)
    ax.set_xlabel("")
    ax.grid(axis="y", alpha=0.25)

    fig.tight_layout()
    save_figure(fig, filename)
    plt.show()
    plt.close(fig)

    return {
        "metric": value_column,
        "n_control": len(control_values),
        "n_pd": len(pd_values),
        "welch_p": p_value,
        "hedges_g_pd_minus_control": g_value,
    }


def continuous_binned_summary(
    dataframe,
    x_column,
    y_column,
    edges,
    min_observations_per_subject_bin,
    statistic="mean",
):
    """Participant-balanced curve for a continuous y variable."""
    data = dataframe[
        ["subject", "group", x_column, y_column]
    ].copy()

    data[x_column] = pd.to_numeric(data[x_column], errors="coerce")
    data[y_column] = pd.to_numeric(data[y_column], errors="coerce")
    data = data.dropna()

    data["bin"] = pd.cut(
        data[x_column],
        bins=edges,
        labels=False,
        include_lowest=True,
    )
    data = data.dropna(subset=["bin"])
    data["bin"] = data["bin"].astype(int)

    bin_numbers = np.arange(len(edges) - 1)
    bin_x = data.groupby("bin")[x_column].median().reindex(bin_numbers)

    subject_bin = (
        data.groupby(["subject", "group", "bin"], as_index=False)
        .agg(
            subject_value=(y_column, "mean"),
            n_observations=(y_column, "size"),
        )
    )

    subject_bin.loc[
        subject_bin["n_observations"]
        < min_observations_per_subject_bin,
        "subject_value",
    ] = np.nan

    rows = []

    for group in GROUP_ORDER:
        group_subjects = sorted(
            data.loc[data["group"].eq(group), "subject"].unique()
        )

        matrix = (
            subject_bin.loc[subject_bin["group"].eq(group)]
            .pivot(index="subject", columns="bin", values="subject_value")
            .reindex(index=group_subjects, columns=bin_numbers)
            .to_numpy(float)
        )

        estimate, lower, upper = bootstrap_matrix_ci(
            matrix,
            statistic=statistic,
        )
        n_subjects = np.sum(np.isfinite(matrix), axis=0)

        raw_group = data.loc[data["group"].eq(group)]
        n_events = (
            raw_group.groupby("bin").size().reindex(bin_numbers, fill_value=0)
        )

        for bin_number in bin_numbers:
            rows.append({
                "group": group,
                "bin": int(bin_number),
                "x": bin_x.loc[bin_number],
                "estimate": estimate[bin_number],
                "ci_lower": lower[bin_number],
                "ci_upper": upper[bin_number],
                "n_subjects": int(n_subjects[bin_number]),
                "n_observations": int(n_events.loc[bin_number]),
                "bin_left": edges[bin_number],
                "bin_right": edges[bin_number + 1],
            })

    return pd.DataFrame(rows), subject_bin


def binary_binned_summary(
    dataframe,
    x_column,
    outcome_column,
    edges,
    min_observations_per_subject_bin,
):
    """Participant-balanced probability curve for a binary outcome."""
    data = dataframe[
        ["subject", "group", x_column, outcome_column]
    ].copy()

    data[x_column] = pd.to_numeric(data[x_column], errors="coerce")
    data[outcome_column] = pd.to_numeric(
        data[outcome_column], errors="coerce"
    )
    data = data.dropna()
    data = data.loc[data[outcome_column].isin([0, 1])].copy()

    data["bin"] = pd.cut(
        data[x_column],
        bins=edges,
        labels=False,
        include_lowest=True,
    )
    data = data.dropna(subset=["bin"])
    data["bin"] = data["bin"].astype(int)

    bin_numbers = np.arange(len(edges) - 1)
    bin_x = data.groupby("bin")[x_column].median().reindex(bin_numbers)

    subject_bin = (
        data.groupby(["subject", "group", "bin"], as_index=False)
        .agg(
            n_high=(outcome_column, "sum"),
            n_saccades=(outcome_column, "size"),
        )
    )
    subject_bin["probability"] = (
        subject_bin["n_high"] / subject_bin["n_saccades"]
    )
    subject_bin.loc[
        subject_bin["n_saccades"]
        < min_observations_per_subject_bin,
        "probability",
    ] = np.nan

    rows = []

    for group in GROUP_ORDER:
        group_subjects = sorted(
            data.loc[data["group"].eq(group), "subject"].unique()
        )

        matrix = (
            subject_bin.loc[subject_bin["group"].eq(group)]
            .pivot(index="subject", columns="bin", values="probability")
            .reindex(index=group_subjects, columns=bin_numbers)
            .to_numpy(float)
        )

        estimate, lower, upper = bootstrap_matrix_ci(
            matrix,
            statistic="mean",
        )
        n_subjects = np.sum(np.isfinite(matrix), axis=0)

        raw_group = data.loc[data["group"].eq(group)]
        raw_counts = (
            raw_group.groupby("bin")[outcome_column]
            .agg(["sum", "size"])
            .reindex(bin_numbers, fill_value=0)
        )

        for bin_number in bin_numbers:
            rows.append({
                "group": group,
                "bin": int(bin_number),
                "x": bin_x.loc[bin_number],
                "probability": estimate[bin_number],
                "ci_lower": lower[bin_number],
                "ci_upper": upper[bin_number],
                "n_subjects": int(n_subjects[bin_number]),
                "n_high_events": int(raw_counts.loc[bin_number, "sum"]),
                "n_saccades": int(raw_counts.loc[bin_number, "size"]),
                "bin_left": edges[bin_number],
                "bin_right": edges[bin_number + 1],
            })

    return pd.DataFrame(rows), subject_bin


def draw_continuous_curve(
    ax,
    summary,
    y_column,
    ylabel,
    title,
    min_subjects=MIN_SUBJECTS_PER_GROUP_BIN,
):
    for group in GROUP_ORDER:
        group_summary = summary.loc[
            summary["group"].eq(group)
            & summary["n_subjects"].ge(min_subjects)
        ].sort_values("x")

        if group_summary.empty:
            continue

        line = ax.plot(
            group_summary["x"],
            group_summary[y_column],
            marker="o",
            linewidth=2,
            label=group,
            color=GROUP_COLORS[group],
        )[0]

        ax.fill_between(
            group_summary["x"],
            group_summary["ci_lower"],
            group_summary["ci_upper"],
            alpha=0.18,
            color=line.get_color(),
        )

    ax.set_title(title)
    ax.set_ylabel(ylabel)
    ax.grid(alpha=0.25)
    ax.legend()


# ============================================================
# 4. LOAD ALL PER-SUBJECT REMoDNaV EVENT FILES
# ============================================================
event_files = sorted(EVENT_DIR.glob("all_remodnav_events_*.csv"))

if not event_files:
    raise FileNotFoundError(
        "No all_remodnav_events_<subject>.csv files found in "
        f"{EVENT_DIR}"
    )

wanted_columns = {
    "group", "subject", "block_id", "block_number",
    "event_index_block", "event_index_trial", "label", "trial",
    "trial_assignment_status", "image", "stim_id", "rt_ms",
    "start_abs_ms", "end_abs_ms", "onset_ms_from_image",
    "offset_ms_from_image", "duration_ms", "amplitude_deg",
    "peak_velocity_deg_s", "image_onset_ms", "response_ms",
    "requested_pso_ms", "direction_deg", "direction_8",
}

frames = []

for path in event_files:
    file_subject = subject_from_filename(path)

    dataframe = pd.read_csv(
        path,
        usecols=lambda column: column in wanted_columns,
        low_memory=False,
    )

    if "subject" not in dataframe.columns:
        dataframe["subject"] = file_subject
    else:
        dataframe["subject"] = dataframe["subject"].fillna(file_subject)

    if "group" not in dataframe.columns:
        dataframe["group"] = dataframe["subject"].map(infer_group)
    else:
        missing_group = (
            dataframe["group"].isna()
            | dataframe["group"].astype(str).eq("")
            | dataframe["group"].astype(str).eq("Unknown")
        )
        dataframe.loc[missing_group, "group"] = (
            dataframe.loc[missing_group, "subject"].map(infer_group)
        )

    frames.append(dataframe)

all_events = pd.concat(frames, ignore_index=True)

all_events["label"] = (
    all_events["label"].astype(str).str.strip().str.upper()
)

numeric_columns = [
    "block_number", "event_index_block", "event_index_trial", "trial",
    "rt_ms", "start_abs_ms", "end_abs_ms", "onset_ms_from_image",
    "offset_ms_from_image", "duration_ms", "amplitude_deg",
    "peak_velocity_deg_s", "image_onset_ms", "response_ms",
    "requested_pso_ms", "direction_deg",
]

for column in numeric_columns:
    if column in all_events.columns:
        all_events[column] = pd.to_numeric(
            all_events[column], errors="coerce"
        )
    else:
        all_events[column] = np.nan

for column in [
    "block_id", "trial_assignment_status", "image", "stim_id",
    "direction_8",
]:
    if column not in all_events.columns:
        all_events[column] = np.nan

# Keep only events fully contained inside IMAGE_ONSET -> RESPONSE.
trial_events = all_events.loc[all_events["trial"].notna()].copy()

if "trial_assignment_status" in trial_events.columns:
    complete_mask = trial_events["trial_assignment_status"].eq(
        "complete_within_trial"
    )
    if complete_mask.any():
        trial_events = trial_events.loc[complete_mask].copy()

trial_keys = ["subject", "block_id", "trial"]
sort_columns = [
    "subject", "block_id", "trial", "start_abs_ms", "event_index_block"
]
trial_events = trial_events.sort_values(sort_columns).reset_index(drop=True)

print("=" * 100)
print("LOADED REMoDNaV EVENTS")
print("=" * 100)
print(f"Files: {len(event_files)}")
print(f"Subjects: {trial_events['subject'].nunique()}")
print(f"Trial-contained events: {len(trial_events):,}")
print("Groups:")
display(
    trial_events[["subject", "group"]]
    .drop_duplicates()
    .groupby("group")
    .size()
    .rename("n_subjects")
    .reset_index()
)
print("Event labels:")
display(
    trial_events["label"].value_counts().rename_axis("label").reset_index(
        name="n_events"
    )
)


# ============================================================
# 5. BUILD DERIVED TRIAL, FIXATION, AND SACCADE TABLES
# ============================================================
# -------------------------
# 5A. One row per trial
# -------------------------
trial_table = (
    trial_events.groupby(trial_keys, as_index=False)
    .agg(
        group=("group", "first"),
        image=("image", "first"),
        stim_id=("stim_id", "first"),
        rt_ms=("rt_ms", "first"),
        image_onset_ms=("image_onset_ms", "first"),
        response_ms=("response_ms", "first"),
    )
)

trial_saccade_counts = (
    trial_events.assign(
        is_saccade=trial_events["label"].isin(SACCADE_LABELS).astype(int)
    )
    .groupby(trial_keys, as_index=False)["is_saccade"]
    .sum()
    .rename(columns={"is_saccade": "n_saccades"})
)

trial_fixation_counts = (
    trial_events.assign(
        is_fixation=trial_events["label"].isin(FIXATION_LABELS).astype(int)
    )
    .groupby(trial_keys, as_index=False)["is_fixation"]
    .sum()
    .rename(columns={"is_fixation": "n_fixations"})
)

trial_table = (
    trial_table
    .merge(trial_saccade_counts, on=trial_keys, how="left")
    .merge(trial_fixation_counts, on=trial_keys, how="left")
)

trial_table["n_saccades"] = trial_table["n_saccades"].fillna(0).astype(int)
trial_table["n_fixations"] = trial_table["n_fixations"].fillna(0).astype(int)

fallback_rt = trial_table["response_ms"] - trial_table["image_onset_ms"]
trial_table["rt_ms"] = trial_table["rt_ms"].where(
    np.isfinite(trial_table["rt_ms"]),
    fallback_rt,
)
trial_table["viewing_duration_s"] = trial_table["rt_ms"] / 1000.0
trial_table = trial_table.loc[
    np.isfinite(trial_table["viewing_duration_s"])
    & trial_table["viewing_duration_s"].gt(0)
].copy()
trial_table["saccade_frequency_hz"] = (
    trial_table["n_saccades"] / trial_table["viewing_duration_s"]
)

# -------------------------
# 5B. Fixations and fixation order
# -------------------------
fixations = trial_events.loc[
    trial_events["label"].isin(FIXATION_LABELS)
].copy()
fixations = fixations.loc[
    np.isfinite(fixations["duration_ms"])
    & fixations["duration_ms"].gt(0)
].copy()
fixations = fixations.sort_values(sort_columns).reset_index(drop=True)
fixations["fixation_order"] = (
    fixations.groupby(trial_keys).cumcount() + 1
)
fixations["n_fixations_in_trial"] = (
    fixations.groupby(trial_keys)["duration_ms"].transform("size")
)
fixations["normalized_fixation_progress"] = (
    fixations["fixation_order"] / fixations["n_fixations_in_trial"]
)

# -------------------------
# 5C. Event sequence, saccades, and high-PSO pairing
# -------------------------
sequence = trial_events.copy().sort_values(sort_columns).reset_index(drop=True)
sequence["next_label"] = (
    sequence.groupby(trial_keys)["label"].shift(-1)
)
sequence["next_start_abs_ms"] = (
    sequence.groupby(trial_keys)["start_abs_ms"].shift(-1)
)
sequence["next_end_abs_ms"] = (
    sequence.groupby(trial_keys)["end_abs_ms"].shift(-1)
)

saccades = sequence.loc[
    sequence["label"].isin(SACCADE_LABELS)
].copy()
saccades["saccade_order"] = (
    saccades.groupby(trial_keys).cumcount() + 1
)

# Response fallback where needed.
saccade_response_ms = saccades["response_ms"].where(
    np.isfinite(saccades["response_ms"]),
    saccades["image_onset_ms"] + saccades["rt_ms"],
)
saccades["time_from_saccade_end_to_response_ms"] = (
    saccade_response_ms - saccades["end_abs_ms"]
)

if AUTO_USE_FILE_PSO_WINDOW:
    available_pso_windows = pd.to_numeric(
        all_events["requested_pso_ms"], errors="coerce"
    )
    available_pso_windows = available_pso_windows[
        np.isfinite(available_pso_windows)
        & available_pso_windows.gt(0)
    ]
    if len(available_pso_windows):
        pso_window_ms = float(available_pso_windows.median())
    else:
        pso_window_ms = DEFAULT_PSO_WINDOW_MS
else:
    pso_window_ms = DEFAULT_PSO_WINDOW_MS

saccades["pso_event_gap_ms"] = (
    saccades["next_start_abs_ms"] - saccades["end_abs_ms"]
)

saccades["eligible_for_high_pso"] = (
    np.isfinite(saccades["time_from_saccade_end_to_response_ms"])
    & saccades["time_from_saccade_end_to_response_ms"].ge(pso_window_ms)
)

saccades["has_high_pso"] = (
    saccades["next_label"].isin(HIGH_PSO_LABELS)
    & saccades["pso_event_gap_ms"].between(
        PSO_PAIR_MIN_GAP_MS,
        PSO_PAIR_MAX_GAP_MS,
        inclusive="both",
    )
).astype(int)

eligible_saccades = saccades.loc[
    saccades["eligible_for_high_pso"]
].copy()

# -------------------------
# 5D. First saccade per trial
# -------------------------
first_saccades = (
    saccades.sort_values(sort_columns)
    .groupby(trial_keys, group_keys=False)
    .head(1)
    .copy()
)

first_saccades["first_saccade_latency_ms"] = (
    first_saccades["onset_ms_from_image"]
)

latency_fallback = (
    first_saccades["start_abs_ms"]
    - first_saccades["image_onset_ms"]
)
first_saccades["first_saccade_latency_ms"] = (
    first_saccades["first_saccade_latency_ms"].where(
        np.isfinite(first_saccades["first_saccade_latency_ms"]),
        latency_fallback,
    )
)

first_saccades_raw_n = len(first_saccades)
first_saccades = first_saccades.loc[
    np.isfinite(first_saccades["first_saccade_latency_ms"])
    & first_saccades["first_saccade_latency_ms"].gt(0)
].copy()

if APPLY_FIRST_LATENCY_QC:
    before_qc = len(first_saccades)
    first_saccades = first_saccades.loc[
        first_saccades["first_saccade_latency_ms"].between(
            FIRST_LATENCY_MIN_MS,
            FIRST_LATENCY_MAX_MS,
            inclusive="both",
        )
    ].copy()
    print(
        "First-saccade latency QC: "
        f"kept {len(first_saccades):,}/{before_qc:,} positive-latency trials "
        f"inside {FIRST_LATENCY_MIN_MS:g}-{FIRST_LATENCY_MAX_MS:g} ms."
    )

# -------------------------
# 5E. Final fixation per trial and response-confirmation interval
# -------------------------
final_fixations = (
    fixations.sort_values(sort_columns)
    .groupby(trial_keys, group_keys=False)
    .tail(1)
    .copy()
)

final_response_ms = final_fixations["response_ms"].where(
    np.isfinite(final_fixations["response_ms"]),
    final_fixations["image_onset_ms"] + final_fixations["rt_ms"],
)
final_fixations["final_fixation_to_response_ms"] = (
    final_response_ms - final_fixations["start_abs_ms"]
)
final_fixations["final_fixation_end_to_response_ms"] = (
    final_response_ms - final_fixations["end_abs_ms"]
)

final_fixations = final_fixations.loc[
    np.isfinite(final_fixations["final_fixation_to_response_ms"])
    & final_fixations["final_fixation_to_response_ms"].ge(0)
    & final_fixations["final_fixation_to_response_ms"].le(
        final_fixations["rt_ms"] + 1.0
    )
].copy()

print("\nDerived-table counts:")
print(f"  Trials: {len(trial_table):,}")
print(f"  FIXA events: {len(fixations):,}")
print(f"  SACC/ISAC events: {len(saccades):,}")
print(f"  PSO-eligible saccades: {len(eligible_saccades):,}")
print(f"  High-PSO paired saccades: {eligible_saccades['has_high_pso'].sum():,}")
print(f"  PSO window used: {pso_window_ms:g} ms")

# Save reusable derived tables.
trial_table.to_csv(FIG_DIR / "derived_trial_table.csv", index=False)
fixations.to_csv(FIG_DIR / "derived_fixation_events.csv", index=False)
eligible_saccades.to_csv(
    FIG_DIR / "derived_saccades_with_high_pso.csv", index=False
)
first_saccades.to_csv(
    FIG_DIR / "derived_first_saccades.csv", index=False
)
final_fixations.to_csv(
    FIG_DIR / "derived_final_fixations.csv", index=False
)


# ============================================================
# GRAPH 4A
# Participant median fixation duration vs mean saccade frequency
# ============================================================
subject_fixation_duration = (
    fixations.groupby(["subject", "group"], as_index=False)
    .agg(median_fixation_duration_ms=("duration_ms", "median"))
)

subject_saccade_frequency = (
    trial_table.groupby(["subject", "group"], as_index=False)
    .agg(
        mean_saccade_frequency_hz=("saccade_frequency_hz", "mean"),
        n_trials=("trial", "size"),
    )
)

subject_graph4 = subject_fixation_duration.merge(
    subject_saccade_frequency,
    on=["subject", "group"],
    how="inner",
)
subject_graph4.to_csv(
    FIG_DIR / "04A_subject_fixation_duration_vs_saccade_frequency.csv",
    index=False,
)

fig, ax = plt.subplots(figsize=(8.0, 6.2))
correlation_text = []

for group in GROUP_ORDER:
    group_data = subject_graph4.loc[
        subject_graph4["group"].eq(group)
    ].copy()

    ax.scatter(
        group_data["median_fixation_duration_ms"],
        group_data["mean_saccade_frequency_hz"],
        s=58,
        alpha=0.85,
        label=group,
        color=GROUP_COLORS[group],
    )

    if LABEL_ALL_PARTICIPANTS:
        for _, row in group_data.iterrows():
            ax.annotate(
                str(row["subject"]),
                (
                    row["median_fixation_duration_ms"],
                    row["mean_saccade_frequency_hz"],
                ),
                xytext=(4, 4),
                textcoords="offset points",
                fontsize=7.5,
            )

    if len(group_data) >= 3:
        x = group_data["median_fixation_duration_ms"].to_numpy(float)
        y = group_data["mean_saccade_frequency_hz"].to_numpy(float)
        slope, intercept = np.polyfit(x, y, 1)
        x_line = np.linspace(x.min(), x.max(), 100)
        ax.plot(
            x_line,
            intercept + slope * x_line,
            linewidth=2,
            color=GROUP_COLORS[group],
        )

        rho, p_value = spearmanr(x, y)
        correlation_text.append(
            f"{group}: rho={rho:.2f}, p={format_p(p_value)}"
        )

ax.set_title(
    "Median fixation duration vs saccade frequency\n"
    + " | ".join(correlation_text)
)
ax.set_xlabel("Participant median fixation duration (ms)")
ax.set_ylabel("Participant mean saccade frequency (saccades/s)")
ax.grid(alpha=0.25)
ax.legend()
fig.tight_layout()
save_figure(
    fig,
    "04A_fixation_duration_vs_saccade_frequency.png",
)
plt.show()
plt.close(fig)


# ============================================================
# GRAPH 4B
# Saccade count vs image-viewing duration
# ============================================================
viewing_edges = quantile_bin_edges(
    trial_table["viewing_duration_s"],
    N_VIEWING_DURATION_BINS,
)

viewing_summary, viewing_subject_bins = continuous_binned_summary(
    dataframe=trial_table,
    x_column="viewing_duration_s",
    y_column="n_saccades",
    edges=viewing_edges,
    min_observations_per_subject_bin=MIN_TRIALS_PER_SUBJECT_BIN,
    statistic="mean",
)

viewing_summary.to_csv(
    FIG_DIR / "04B_saccade_count_vs_viewing_duration_summary.csv",
    index=False,
)
viewing_subject_bins.to_csv(
    FIG_DIR / "04B_saccade_count_vs_viewing_duration_subject_bins.csv",
    index=False,
)

fig, ax = plt.subplots(figsize=(8.0, 6.2))
draw_continuous_curve(
    ax=ax,
    summary=viewing_summary,
    y_column="estimate",
    ylabel="Mean saccade count per trial",
    title=(
        "Saccade count vs image-viewing duration\n"
        "Participant-balanced means with 95% bootstrap CIs"
    ),
)
ax.set_xlabel("Image-viewing duration (s)")
fig.tight_layout()
save_figure(
    fig,
    "04B_saccade_count_vs_viewing_duration.png",
)
plt.show()
plt.close(fig)


# ============================================================
# GRAPH 5
# Fixation-duration survival curve
# ============================================================
survival_upper = min(
    SURVIVAL_MAX_MS,
    float(fixations["duration_ms"].quantile(0.995)),
)
survival_upper = max(survival_upper, 300.0)
survival_thresholds = np.linspace(
    0.0,
    survival_upper,
    SURVIVAL_N_THRESHOLDS,
)

survival_rows = []
fig, ax = plt.subplots(figsize=(8.0, 6.2))

for group in GROUP_ORDER:
    group_data = fixations.loc[fixations["group"].eq(group)]
    group_subjects = sorted(group_data["subject"].unique())

    subject_curves = []
    for subject in group_subjects:
        durations = group_data.loc[
            group_data["subject"].eq(subject), "duration_ms"
        ].to_numpy(float)
        durations = durations[np.isfinite(durations) & (durations > 0)]
        if len(durations) == 0:
            continue
        subject_curves.append(
            np.mean(
                durations[:, None] > survival_thresholds[None, :],
                axis=0,
            )
        )

    matrix = np.asarray(subject_curves, dtype=float)
    estimate, lower, upper = bootstrap_matrix_ci(
        matrix,
        statistic="mean",
    )

    line = ax.plot(
        survival_thresholds,
        estimate,
        linewidth=2,
        label=f"{group} (n={len(matrix)})",
        color=GROUP_COLORS[group],
    )[0]
    ax.fill_between(
        survival_thresholds,
        lower,
        upper,
        alpha=0.18,
        color=line.get_color(),
    )

    for threshold, est, low, high in zip(
        survival_thresholds, estimate, lower, upper
    ):
        survival_rows.append({
            "group": group,
            "duration_threshold_ms": threshold,
            "survival_probability": est,
            "ci_lower": low,
            "ci_upper": high,
            "n_subjects": len(matrix),
        })

ax.set_title(
    "Fixation-duration survival curves\n"
    "Probability that a FIXA event exceeds each duration"
)
ax.set_xlabel("Fixation-duration threshold (ms)")
ax.set_ylabel("P(fixation duration > threshold)")
ax.set_ylim(-0.01, 1.01)
ax.grid(alpha=0.25)
ax.legend()
fig.tight_layout()
save_figure(fig, "05_fixation_duration_survival_curve.png")
plt.show()
plt.close(fig)

pd.DataFrame(survival_rows).to_csv(
    FIG_DIR / "05_fixation_duration_survival_curve.csv",
    index=False,
)


# ============================================================
# GRAPH 6
# Fixation duration vs fixation order
# ============================================================
fixation_order_subject = (
    fixations.loc[fixations["fixation_order"].le(MAX_FIXATION_ORDER)]
    .groupby(["subject", "group", "fixation_order"], as_index=False)
    .agg(
        subject_median_duration_ms=("duration_ms", "median"),
        n_fixations=("duration_ms", "size"),
    )
)

order_numbers = np.arange(1, MAX_FIXATION_ORDER + 1)
fixation_order_rows = []

fig, ax = plt.subplots(figsize=(8.0, 6.2))

for group in GROUP_ORDER:
    group_subjects = sorted(
        fixation_order_subject.loc[
            fixation_order_subject["group"].eq(group), "subject"
        ].unique()
    )

    matrix = (
        fixation_order_subject.loc[
            fixation_order_subject["group"].eq(group)
        ]
        .pivot(
            index="subject",
            columns="fixation_order",
            values="subject_median_duration_ms",
        )
        .reindex(index=group_subjects, columns=order_numbers)
        .to_numpy(float)
    )

    estimate, lower, upper = bootstrap_matrix_ci(
        matrix,
        statistic="median",
    )
    n_subjects = np.sum(np.isfinite(matrix), axis=0)
    keep = n_subjects >= MIN_SUBJECTS_PER_GROUP_BIN

    line = ax.plot(
        order_numbers[keep],
        estimate[keep],
        marker="o",
        linewidth=2,
        label=group,
        color=GROUP_COLORS[group],
    )[0]
    ax.fill_between(
        order_numbers[keep],
        lower[keep],
        upper[keep],
        alpha=0.18,
        color=line.get_color(),
    )

    for order, est, low, high, n_subject in zip(
        order_numbers, estimate, lower, upper, n_subjects
    ):
        fixation_order_rows.append({
            "group": group,
            "fixation_order": int(order),
            "group_median_of_subject_medians_ms": est,
            "ci_lower": low,
            "ci_upper": high,
            "n_subjects": int(n_subject),
        })

ax.set_title(
    "Fixation duration across search order\n"
    "Median across participant medians with 95% bootstrap CIs"
)
ax.set_xlabel("Fixation order within trial")
ax.set_ylabel("Fixation duration (ms)")
ax.set_xticks(order_numbers)
ax.grid(alpha=0.25)
ax.legend()
fig.tight_layout()
save_figure(fig, "06_fixation_duration_vs_fixation_order.png")
plt.show()
plt.close(fig)

pd.DataFrame(fixation_order_rows).to_csv(
    FIG_DIR / "06_fixation_duration_vs_fixation_order_summary.csv",
    index=False,
)
fixation_order_subject.to_csv(
    FIG_DIR / "06_fixation_duration_vs_fixation_order_subject_values.csv",
    index=False,
)


# ============================================================
# GRAPH 7
# High-PSO probability vs saccade amplitude
# ============================================================
amplitude_data = eligible_saccades.loc[
    np.isfinite(eligible_saccades["amplitude_deg"])
    & eligible_saccades["amplitude_deg"].gt(0)
].copy()

amplitude_edges = quantile_bin_edges(
    amplitude_data["amplitude_deg"],
    N_AMPLITUDE_BINS,
)

amplitude_pso_summary, amplitude_pso_subject_bins = binary_binned_summary(
    dataframe=amplitude_data,
    x_column="amplitude_deg",
    outcome_column="has_high_pso",
    edges=amplitude_edges,
    min_observations_per_subject_bin=MIN_SACCADES_PER_SUBJECT_BIN,
)

amplitude_pso_summary.to_csv(
    FIG_DIR / "07_high_PSO_probability_vs_amplitude_summary.csv",
    index=False,
)
amplitude_pso_subject_bins.to_csv(
    FIG_DIR / "07_high_PSO_probability_vs_amplitude_subject_bins.csv",
    index=False,
)

fig, ax = plt.subplots(figsize=(8.0, 6.2))
draw_continuous_curve(
    ax=ax,
    summary=amplitude_pso_summary,
    y_column="probability",
    ylabel="High-PSO probability per eligible saccade",
    title=(
        "High-PSO probability vs saccade amplitude\n"
        "HPSO/IHPS immediately following SACC/ISAC"
    ),
)
ax.set_xlabel("Saccade amplitude (deg; median within quantile bin)")
ax.set_ylim(bottom=0)
fig.tight_layout()
save_figure(fig, "07_high_PSO_probability_vs_amplitude.png")
plt.show()
plt.close(fig)


# ============================================================
# GRAPH 8A
# High-PSO probability vs peak velocity
# ============================================================
velocity_data = eligible_saccades.loc[
    np.isfinite(eligible_saccades["peak_velocity_deg_s"])
    & eligible_saccades["peak_velocity_deg_s"].gt(0)
].copy()

velocity_edges = quantile_bin_edges(
    velocity_data["peak_velocity_deg_s"],
    N_VELOCITY_BINS,
)

velocity_pso_summary, velocity_pso_subject_bins = binary_binned_summary(
    dataframe=velocity_data,
    x_column="peak_velocity_deg_s",
    outcome_column="has_high_pso",
    edges=velocity_edges,
    min_observations_per_subject_bin=MIN_SACCADES_PER_SUBJECT_BIN,
)

velocity_pso_summary.to_csv(
    FIG_DIR / "08A_high_PSO_probability_vs_peak_velocity_summary.csv",
    index=False,
)
velocity_pso_subject_bins.to_csv(
    FIG_DIR / "08A_high_PSO_probability_vs_peak_velocity_subject_bins.csv",
    index=False,
)

fig, ax = plt.subplots(figsize=(8.0, 6.2))
draw_continuous_curve(
    ax=ax,
    summary=velocity_pso_summary,
    y_column="probability",
    ylabel="High-PSO probability per eligible saccade",
    title="High-PSO probability vs saccade peak velocity",
)
ax.set_xlabel("Peak velocity (deg/s; median within quantile bin)")
ax.set_ylim(bottom=0)
fig.tight_layout()
save_figure(fig, "08A_high_PSO_probability_vs_peak_velocity.png")
plt.show()
plt.close(fig)


# ============================================================
# GRAPH 8B
# Peak velocity vs high-PSO probability, stratified by amplitude
# ============================================================
stratified_data = eligible_saccades.loc[
    np.isfinite(eligible_saccades["amplitude_deg"])
    & eligible_saccades["amplitude_deg"].gt(0)
    & np.isfinite(eligible_saccades["peak_velocity_deg_s"])
    & eligible_saccades["peak_velocity_deg_s"].gt(0)
].copy()

amp_tertiles = stratified_data["amplitude_deg"].quantile([1/3, 2/3])
amp_q1 = float(amp_tertiles.iloc[0])
amp_q2 = float(amp_tertiles.iloc[1])

stratified_data["amplitude_stratum"] = pd.cut(
    stratified_data["amplitude_deg"],
    bins=[-np.inf, amp_q1, amp_q2, np.inf],
    labels=[
        f"Small (<= {amp_q1:.2f} deg)",
        f"Medium ({amp_q1:.2f}-{amp_q2:.2f} deg)",
        f"Large (> {amp_q2:.2f} deg)",
    ],
    include_lowest=True,
)

stratified_summaries = []
fig, axes = plt.subplots(1, 3, figsize=(17.0, 5.2), sharey=True)

for ax, amplitude_stratum in zip(
    axes,
    stratified_data["amplitude_stratum"].cat.categories,
):
    stratum_data = stratified_data.loc[
        stratified_data["amplitude_stratum"].eq(amplitude_stratum)
    ].copy()

    stratum_edges = quantile_bin_edges(
        stratum_data["peak_velocity_deg_s"],
        max(5, N_VELOCITY_BINS - 2),
    )

    stratum_summary, _ = binary_binned_summary(
        dataframe=stratum_data,
        x_column="peak_velocity_deg_s",
        outcome_column="has_high_pso",
        edges=stratum_edges,
        min_observations_per_subject_bin=MIN_SACCADES_PER_SUBJECT_BIN,
    )
    stratum_summary["amplitude_stratum"] = str(amplitude_stratum)
    stratified_summaries.append(stratum_summary)

    draw_continuous_curve(
        ax=ax,
        summary=stratum_summary,
        y_column="probability",
        ylabel=(
            "High-PSO probability"
            if ax is axes[0]
            else ""
        ),
        title=str(amplitude_stratum),
    )
    ax.set_xlabel("Peak velocity (deg/s)")
    ax.set_ylim(bottom=0)

fig.suptitle(
    "High-PSO probability vs peak velocity within amplitude strata",
    y=1.02,
)
fig.tight_layout()
save_figure(
    fig,
    "08B_high_PSO_probability_vs_peak_velocity_by_amplitude.png",
)
plt.show()
plt.close(fig)

pd.concat(stratified_summaries, ignore_index=True).to_csv(
    FIG_DIR
    / "08B_high_PSO_probability_vs_peak_velocity_by_amplitude_summary.csv",
    index=False,
)


# ============================================================
# GRAPH 11A
# First-saccade latency
# ============================================================
first_latency_subject = (
    first_saccades.groupby(["subject", "group"], as_index=False)
    .agg(
        median_first_saccade_latency_ms=(
            "first_saccade_latency_ms", "median"
        ),
        n_trials=("trial", "size"),
    )
)
first_latency_subject.to_csv(
    FIG_DIR / "11A_first_saccade_latency_subject_values.csv",
    index=False,
)

stats_11a = participant_boxplot(
    dataframe=first_latency_subject,
    value_column="median_first_saccade_latency_ms",
    ylabel="Participant median first-saccade latency (ms)",
    title="Search initiation: first-saccade latency",
    filename="11A_first_saccade_latency.png",
)


# ============================================================
# GRAPH 11B
# Final-fixation onset to response
# ============================================================
final_response_subject = (
    final_fixations.groupby(["subject", "group"], as_index=False)
    .agg(
        median_final_fixation_to_response_ms=(
            "final_fixation_to_response_ms", "median"
        ),
        n_trials=("trial", "size"),
    )
)
final_response_subject.to_csv(
    FIG_DIR / "11B_final_fixation_to_response_subject_values.csv",
    index=False,
)

stats_11b = participant_boxplot(
    dataframe=final_response_subject,
    value_column="median_final_fixation_to_response_ms",
    ylabel="Participant median final-fixation onset to response (ms)",
    title="Response confirmation: final fixation onset to response",
    filename="11B_final_fixation_to_response.png",
)


# ============================================================
# 12. SAVE A COMPACT RUN SUMMARY
# ============================================================
run_summary = pd.DataFrame([
    {
        "n_event_files": len(event_files),
        "n_subjects": trial_events["subject"].nunique(),
        "n_trials": len(trial_table),
        "n_fixations": len(fixations),
        "n_saccades": len(saccades),
        "n_pso_eligible_saccades": len(eligible_saccades),
        "n_high_pso_saccades": int(eligible_saccades["has_high_pso"].sum()),
        "pso_window_ms": pso_window_ms,
        "saccade_labels": ",".join(sorted(SACCADE_LABELS)),
        "high_pso_labels": ",".join(sorted(HIGH_PSO_LABELS)),
        "first_latency_qc_applied": APPLY_FIRST_LATENCY_QC,
        "first_latency_min_ms": FIRST_LATENCY_MIN_MS,
        "first_latency_max_ms": FIRST_LATENCY_MAX_MS,
        "graph_11A_welch_p": stats_11a["welch_p"],
        "graph_11A_hedges_g_pd_minus_control": (
            stats_11a["hedges_g_pd_minus_control"]
        ),
        "graph_11B_welch_p": stats_11b["welch_p"],
        "graph_11B_hedges_g_pd_minus_control": (
            stats_11b["hedges_g_pd_minus_control"]
        ),
    }
])
run_summary.to_csv(FIG_DIR / "followup_graphs_run_summary.csv", index=False)

print("\n" + "=" * 100)
print("ALL REQUESTED FOLLOW-UP GRAPHS ARE COMPLETE")
print("=" * 100)
print("Output folder:")
print(FIG_DIR)
print("\nMain PNG files:")
for path in sorted(FIG_DIR.glob("*.png")):
    print(" ", path.name)
print("\nRun summary:")
display(run_summary)
```

```python
# ============================================================
# DIRECTLY SELECT RANDOM FIXA EVENTS FROM ALL REMoDNaV CSV FILES
#
# Selects:
#   - 3 random short FIXA rows from the lowest 25% of durations
#   - 3 random long FIXA rows with duration > 500 ms
#
# IMPORTANT:
#   FIXA events are selected DIRECTLY from:
#       all_remodnav_events_<subject>.csv
#
#   This code does NOT use:
#       - the existing `fixations` dataframe
#       - derived_fixation_events.csv
#       - any newly derived fixation segmentation
#
#   EDF files are reopened only to retrieve the sample-by-sample
#   gaze position inside each already-labelled FIXA interval.
# ============================================================


# ============================================================
# 0. MOUNT GOOGLE DRIVE
# ============================================================
from google.colab import drive

drive.mount(
    "/content/drive",
    force_remount=True,
)


# ============================================================
# 1. IMPORTS
# ============================================================
from pathlib import Path
import sys
import re
import importlib
import subprocess
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.signal import savgol_filter
from IPython.display import display


# ============================================================
# 2. OPTIONAL DEPENDENCY SETUP
# ============================================================
# Install eyelinkio only when it is unavailable.
try:
    import eyelinkio

except ImportError:
    subprocess.run(
        [
            sys.executable,
            "-m",
            "pip",
            "install",
            "-q",
            "--no-deps",
            "--force-reinstall",
            (
                "git+https://github.com/"
                "pandeychetanya/eyelinkio.git@chetanyaedits"
            ),
        ],
        check=True,
    )


# ============================================================
# 3. SETTINGS
# ============================================================
EVENT_DIR = Path(
    "/content/drive/MyDrive/steer_search_PD/all_remodnav"
)

DATA_ROOT = Path(
    "/content/drive/MyDrive/steer_search_PD/steer-search-PD"
)

REPO_DIR = Path(
    "/content/drive/MyDrive/eyeprocess"
)

OUTPUT_DIR = (
    EVENT_DIR
    / "specific_followup_graphs"
    / "random_direct_FIXA_traces"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


# ------------------------------------------------------------
# PARTICIPANT SELECTION
# ------------------------------------------------------------
# None:
#   use directly labelled FIXA events from all participants.
#
# Examples:
#   SUBJECT_IDS = ["p007"]
#   SUBJECT_IDS = ["p007", "p008", "s189"]
#
SUBJECT_IDS = None


# ------------------------------------------------------------
# RANDOM FIXATION SELECTION
# ------------------------------------------------------------
N_SHORT = 3
N_LONG = 3

# Short pool:
# lowest 25% of directly labelled FIXA durations.
SHORT_QUANTILE = 0.25

# Long pool:
# directly labelled FIXA events strictly longer than 500 ms.
LONG_FIXATION_MIN_MS = 500.0

# Reproducible random selection.
RANDOM_SEED = None

# Prefer a different participant for each selected fixation
# within the short and long categories.
PREFER_DIFFERENT_SUBJECTS = True


# ------------------------------------------------------------
# EVENT FILTERING
# ------------------------------------------------------------
# True:
#   only FIXA rows fully contained inside IMAGE_ONSET -> RESPONSE.
#
# False:
#   include every directly labelled FIXA event, including events
#   outside trials.
#
ONLY_COMPLETE_TRIAL_FIXATIONS = True


# ------------------------------------------------------------
# TRACE DISPLAY
# ------------------------------------------------------------
# Time displayed before and after the directly labelled FIXA interval.
CONTEXT_BEFORE_MS = 100.0
CONTEXT_AFTER_MS = 100.0

# Smooth the sample-level trace only for visualization.
SMOOTH_FOR_DISPLAY = True
SMOOTH_WINDOW_MS = 5.0
SMOOTH_POLYORDER = 2

SAVE_DPI = 240


assert EVENT_DIR.exists(), (
    f"Missing REMoDNaV event directory:\n{EVENT_DIR}"
)

assert DATA_ROOT.exists(), (
    f"Missing participant-data directory:\n{DATA_ROOT}"
)

assert REPO_DIR.exists(), (
    f"Missing eyeprocess repository:\n{REPO_DIR}"
)


# ============================================================
# 4. IMPORT EYEPROCESS HELPERS
# ============================================================
for path in (
    REPO_DIR,
    REPO_DIR.parent,
):
    if str(path) not in sys.path:
        sys.path.insert(
            0,
            str(path),
        )

importlib.invalidate_caches()

from eyeprocess.utils.psychopy import el2deg

from eyeprocess.utils.main_sequence_helpers import (
    read_edf_like_pyedfread,
    load_behav_json,
)


# ============================================================
# 5. SMALL HELPERS
# ============================================================
def infer_group(subject):
    """Infer PD or Control group from the subject identifier."""
    subject = str(subject).lower()

    if subject.startswith("p"):
        return "PD"

    if subject.startswith("s"):
        return "Control"

    return "Unknown"


def subject_from_filename(path):
    """Extract the subject ID from an all-event CSV filename."""
    match = re.search(
        r"all_remodnav_events_(.+)\.csv$",
        path.name,
    )

    if match:
        return match.group(1)

    return path.stem


def valid_filename(value, fallback):
    """Use a saved filename when valid; otherwise use a fallback."""
    if value is None:
        return fallback

    try:
        if pd.isna(value):
            return fallback
    except Exception:
        pass

    value = str(value).strip()

    if value == "" or value.lower() == "nan":
        return fallback

    return value


def contiguous_true_runs(mask):
    """Return half-open ranges for consecutive True values."""
    mask = np.asarray(
        mask,
        dtype=bool,
    )

    if mask.size == 0:
        return []

    difference = np.diff(
        mask.astype(np.int8)
    )

    starts = (
        np.flatnonzero(difference == 1)
        + 1
    )

    stops = (
        np.flatnonzero(difference == -1)
        + 1
    )

    if mask[0]:
        starts = np.r_[0, starts]

    if mask[-1]:
        stops = np.r_[stops, len(mask)]

    return [
        (int(start), int(stop))
        for start, stop in zip(starts, stops)
    ]


def smooth_finite_segments(
    values,
    requested_window,
    polyorder=2,
):
    """
    Apply Savitzky-Golay smoothing separately to each finite segment.

    NaN intervals remain NaN and are not interpolated across.
    """
    values = np.asarray(
        values,
        dtype=float,
    )

    output = values.copy()
    finite_mask = np.isfinite(values)

    for start, stop in contiguous_true_runs(finite_mask):

        segment_length = stop - start

        if segment_length < 5:
            continue

        window = min(
            int(requested_window),
            segment_length,
        )

        if window % 2 == 0:
            window -= 1

        minimum_window = polyorder + 2

        if minimum_window % 2 == 0:
            minimum_window += 1

        if window < minimum_window:
            continue

        try:
            output[start:stop] = savgol_filter(
                values[start:stop],
                window_length=window,
                polyorder=min(
                    polyorder,
                    window - 1,
                ),
                mode="interp",
            )

        except Exception as error:
            warnings.warn(
                "Smoothing failed for one finite segment: "
                f"{type(error).__name__}: {error}"
            )

    return output


# ============================================================
# 6. FIND THE REQUESTED PER-SUBJECT ALL-EVENT FILES
# ============================================================
all_event_files = sorted(
    EVENT_DIR.glob(
        "all_remodnav_events_*.csv"
    )
)

if not all_event_files:
    raise FileNotFoundError(
        "No files matching "
        "all_remodnav_events_<subject>.csv were found in:\n"
        f"{EVENT_DIR}"
    )


if SUBJECT_IDS is not None:

    requested_subjects = {
        str(subject).strip()
        for subject in SUBJECT_IDS
    }

    event_files = [
        path
        for path in all_event_files
        if subject_from_filename(path)
        in requested_subjects
    ]

    found_subjects = {
        subject_from_filename(path)
        for path in event_files
    }

    missing_subjects = (
        requested_subjects
        - found_subjects
    )

    if missing_subjects:
        raise FileNotFoundError(
            "No all-event CSV was found for:\n"
            f"{sorted(missing_subjects)}"
        )

else:
    event_files = all_event_files


print("=" * 100)
print("DIRECT REMoDNaV FIXA SELECTION")
print("=" * 100)

print(
    f"All-event files selected: "
    f"{len(event_files)}"
)

print(
    "Participants:\n"
    + ", ".join(
        subject_from_filename(path)
        for path in event_files
    )
)


# ============================================================
# 7. LOAD FIXA ROWS DIRECTLY FROM EACH ALL-EVENT CSV
# ============================================================
wanted_columns = {
    "group",
    "subject",
    "block_id",
    "block_number",
    "trial",
    "event_index_block",
    "event_index_trial",
    "label",
    "trial_assignment_status",
    "duration_ms",
    "start_abs_ms",
    "end_abs_ms",
    "start_x",
    "start_y",
    "end_x",
    "end_y",
    "image",
    "stim_id",
    "edf_file",
    "behav_file",
    "sampling_rate_hz",
    "mean_affine_A00",
    "mean_affine_A01",
    "mean_affine_A10",
    "mean_affine_A11",
    "mean_affine_b0",
    "mean_affine_b1",
}


fixation_frames = []

for event_path in event_files:

    file_subject = subject_from_filename(
        event_path
    )

    dataframe = pd.read_csv(
        event_path,
        usecols=lambda column: (
            column in wanted_columns
        ),
        low_memory=False,
    )

    if "subject" not in dataframe.columns:
        dataframe["subject"] = file_subject

    else:
        dataframe["subject"] = (
            dataframe["subject"]
            .fillna(file_subject)
            .astype(str)
            .str.strip()
        )

    if "group" not in dataframe.columns:
        dataframe["group"] = (
            dataframe["subject"]
            .map(infer_group)
        )

    else:
        missing_group = (
            dataframe["group"].isna()
            | dataframe["group"].astype(str).str.strip().eq("")
            | dataframe["group"].astype(str).str.strip().eq("Unknown")
        )

        dataframe.loc[
            missing_group,
            "group",
        ] = dataframe.loc[
            missing_group,
            "subject",
        ].map(infer_group)

    if "label" not in dataframe.columns:
        raise KeyError(
            f"The file does not contain a REMoDNaV label column:\n"
            f"{event_path}"
        )

    dataframe["label"] = (
        dataframe["label"]
        .astype(str)
        .str.strip()
        .str.upper()
    )

    # ---------------------------------------------------------
    # DIRECT FILTER:
    # Retain only rows that REMoDNaV labelled FIXA.
    # ---------------------------------------------------------
    dataframe = dataframe.loc[
        dataframe["label"].eq("FIXA")
    ].copy()

    if (
        ONLY_COMPLETE_TRIAL_FIXATIONS
        and "trial_assignment_status"
        in dataframe.columns
    ):
        dataframe = dataframe.loc[
            dataframe[
                "trial_assignment_status"
            ].eq("complete_within_trial")
        ].copy()

    dataframe["source_event_csv"] = str(
        event_path
    )

    fixation_frames.append(
        dataframe
    )


if not fixation_frames:
    raise RuntimeError(
        "No event CSV files produced a fixation table."
    )


direct_fixa = pd.concat(
    fixation_frames,
    ignore_index=True,
)


# Numeric conversion.
numeric_columns = [
    "block_number",
    "trial",
    "event_index_block",
    "event_index_trial",
    "duration_ms",
    "start_abs_ms",
    "end_abs_ms",
    "sampling_rate_hz",
    "mean_affine_A00",
    "mean_affine_A01",
    "mean_affine_A10",
    "mean_affine_A11",
    "mean_affine_b0",
    "mean_affine_b1",
]

for column in numeric_columns:

    if column not in direct_fixa.columns:
        direct_fixa[column] = np.nan

    direct_fixa[column] = pd.to_numeric(
        direct_fixa[column],
        errors="coerce",
    )


direct_fixa = direct_fixa.loc[
    np.isfinite(direct_fixa["duration_ms"])
    & direct_fixa["duration_ms"].gt(0)
    & np.isfinite(direct_fixa["start_abs_ms"])
    & np.isfinite(direct_fixa["end_abs_ms"])
    & direct_fixa["end_abs_ms"].gt(
        direct_fixa["start_abs_ms"]
    )
].copy()


if direct_fixa.empty:
    raise RuntimeError(
        "No valid directly labelled FIXA rows were found."
    )


print("\nDirect FIXA counts:")

display(
    direct_fixa.groupby(
        ["group", "subject"]
    )
    .size()
    .rename("n_FIXA")
    .reset_index()
)

print(
    f"\nTotal directly labelled FIXA rows: "
    f"{len(direct_fixa):,}"
)


# ============================================================
# 8. CREATE SHORT AND LONG DIRECT-FIXA POOLS
# ============================================================
# Short fixations remain defined using the bottom duration quartile.
short_cutoff_ms = float(
    direct_fixa["duration_ms"].quantile(
        SHORT_QUANTILE
    )
)

short_pool = direct_fixa.loc[
    direct_fixa["duration_ms"].le(
        short_cutoff_ms
    )
].copy()


# Long fixations are now defined using a fixed threshold:
# strictly greater than 500 ms.
long_pool = direct_fixa.loc[
    direct_fixa["duration_ms"].gt(
        LONG_FIXATION_MIN_MS
    )
].copy()


print("\n" + "=" * 100)
print("DIRECT FIXA DURATION POOLS")
print("=" * 100)

print(
    f"Short fixation definition: "
    f"duration <= {short_cutoff_ms:.2f} ms "
    f"(bottom {SHORT_QUANTILE:.0%})"
)

print(
    f"Long fixation definition: "
    f"duration > {LONG_FIXATION_MIN_MS:.2f} ms"
)

print(
    f"Short pool: {len(short_pool):,} FIXA rows "
    f"from {short_pool['subject'].nunique()} participants"
)

print(
    f"Long pool: {len(long_pool):,} FIXA rows "
    f"from {long_pool['subject'].nunique()} participants"
)


if len(short_pool) < N_SHORT:
    raise RuntimeError(
        f"The short-fixation pool contains only "
        f"{len(short_pool)} events, but {N_SHORT} were requested."
    )

if len(long_pool) < N_LONG:
    raise RuntimeError(
        f"The long-fixation pool contains only "
        f"{len(long_pool)} events above "
        f"{LONG_FIXATION_MIN_MS:g} ms, but {N_LONG} were requested."
    )


# ============================================================
# 9. RANDOMLY CHOOSE THREE DIRECT FIXA ROWS FROM EACH POOL
# ============================================================
random_generator = np.random.default_rng(
    RANDOM_SEED
)


def random_event_sample(
    pool,
    n_events,
    category,
):
    """
    Randomly select fixation events.

    When possible, randomly choose different participants and then
    randomly choose one fixation from each chosen participant.
    """
    if len(pool) < n_events:
        raise RuntimeError(
            f"Only {len(pool)} {category} FIXA rows were found, "
            f"but {n_events} were requested."
        )

    available_subjects = (
        pool["subject"]
        .dropna()
        .astype(str)
        .unique()
    )

    if (
        PREFER_DIFFERENT_SUBJECTS
        and len(available_subjects) >= n_events
    ):
        chosen_subjects = random_generator.choice(
            available_subjects,
            size=n_events,
            replace=False,
        )

        selected_rows = []

        for subject in chosen_subjects:

            subject_pool = pool.loc[
                pool["subject"].eq(subject)
            ]

            chosen_index = random_generator.choice(
                subject_pool.index.to_numpy(),
                size=1,
                replace=False,
            )[0]

            selected_rows.append(
                pool.loc[chosen_index]
            )

        selected = pd.DataFrame(
            selected_rows
        ).reset_index(drop=True)

    else:
        chosen_indices = random_generator.choice(
            pool.index.to_numpy(),
            size=n_events,
            replace=False,
        )

        selected = (
            pool.loc[chosen_indices]
            .copy()
            .reset_index(drop=True)
        )

    selected["fixation_category"] = category

    return selected


selected_short = random_event_sample(
    pool=short_pool,
    n_events=N_SHORT,
    category="Short",
)

selected_long = random_event_sample(
    pool=long_pool,
    n_events=N_LONG,
    category="Long",
)


selected_events = pd.concat(
    [
        selected_short,
        selected_long,
    ],
    ignore_index=True,
)


selected_events["category_order"] = (
    selected_events["fixation_category"]
    .map({
        "Short": 0,
        "Long": 1,
    })
)


selected_events = (
    selected_events
    .sort_values(
        [
            "category_order",
            "duration_ms",
        ]
    )
    .reset_index(drop=True)
)


selected_events["example_number"] = (
    selected_events.groupby(
        "fixation_category"
    ).cumcount()
    + 1
)


selected_events["trace_id"] = (
    selected_events[
        "fixation_category"
    ].str.lower()
    + "_"
    + selected_events[
        "example_number"
    ].astype(str)
)


print("\nRandomly selected direct FIXA rows:")

display(
    selected_events[
        [
            "trace_id",
            "fixation_category",
            "subject",
            "group",
            "block_id",
            "trial",
            "event_index_block",
            "duration_ms",
            "start_abs_ms",
            "end_abs_ms",
        ]
    ]
)


# Verify that every selected long fixation exceeds 500 ms.
selected_long_check = selected_events.loc[
    selected_events["fixation_category"].eq("Long")
]

assert selected_long_check["duration_ms"].gt(
    LONG_FIXATION_MIN_MS
).all(), (
    "At least one selected long fixation does not exceed "
    f"{LONG_FIXATION_MIN_MS:g} ms."
)


# ============================================================
# 10. LOAD AND CALIBRATE THE EDF BLOCK FOR EACH CHOSEN FIXA
# ============================================================
block_cache = {}


def load_calibrated_block(event_row):
    """
    Reopen the EDF block and apply the exact mean affine saved
    in the directly selected REMoDNaV FIXA row.
    """
    subject = str(
        event_row["subject"]
    )

    block_id = str(
        event_row["block_id"]
    )

    cache_key = (
        subject,
        block_id,
    )

    if cache_key in block_cache:
        return block_cache[cache_key]

    edf_filename = valid_filename(
        event_row.get("edf_file"),
        f"{block_id}.edf",
    )

    behav_filename = valid_filename(
        event_row.get("behav_file"),
        f"{block_id}.json",
    )

    edf_path = (
        DATA_ROOT
        / subject
        / "eye"
        / edf_filename
    )

    behav_path = (
        DATA_ROOT
        / subject
        / "behav"
        / behav_filename
    )

    # Fallback to filenames based on block_id.
    if not edf_path.exists():
        edf_path = (
            DATA_ROOT
            / subject
            / "eye"
            / f"{block_id}.edf"
        )

    if not behav_path.exists():
        behav_path = (
            DATA_ROOT
            / subject
            / "behav"
            / f"{block_id}.json"
        )

    if not edf_path.exists():
        raise FileNotFoundError(
            f"Missing EDF file:\n{edf_path}"
        )

    if not behav_path.exists():
        raise FileNotFoundError(
            f"Missing behavioral file:\n{behav_path}"
        )

    print(
        f"Loading raw gaze samples: "
        f"{subject} | {block_id}"
    )

    behavior = load_behav_json(
        behav_path
    )

    samples, edf_events, messages = (
        read_edf_like_pyedfread(
            edf_path
        )
    )

    required_sample_columns = {
        "time",
        "gx_left",
        "gy_left",
    }

    missing_columns = (
        required_sample_columns
        - set(samples.columns)
    )

    if missing_columns:
        raise KeyError(
            f"EDF samples are missing columns: "
            f"{sorted(missing_columns)}"
        )

    samples = (
        samples
        .sort_values("time")
        .reset_index(drop=True)
    )

    time_ms = pd.to_numeric(
        samples["time"],
        errors="coerce",
    ).to_numpy(float)

    x_pixels = pd.to_numeric(
        samples["gx_left"],
        errors="coerce",
    ).to_numpy(float)

    y_pixels = pd.to_numeric(
        samples["gy_left"],
        errors="coerce",
    ).to_numpy(float)


    monitor_resolution = (
        behavior
        .get("config", {})
        .get(
            "monitor_resolution",
            [1920, 1080],
        )
    )

    screen_width = float(
        monitor_resolution[0]
    )

    screen_height = float(
        monitor_resolution[1]
    )


    invalid_samples = (
        ~np.isfinite(time_ms)
        | ~np.isfinite(x_pixels)
        | ~np.isfinite(y_pixels)
        | (np.abs(x_pixels) > 10 * screen_width)
        | (np.abs(y_pixels) > 10 * screen_height)
    )


    # Apply the same pupil-validity check used during processing.
    for pupil_column in (
        "pa_left",
        "pupil_left",
        "ps_left",
        "pupil",
    ):
        if pupil_column in samples.columns:

            pupil = pd.to_numeric(
                samples[pupil_column],
                errors="coerce",
            ).to_numpy(float)

            invalid_samples |= (
                ~np.isfinite(pupil)
                | (pupil <= 0)
            )

            break


    # Mask EDF blink intervals.
    if (
        edf_events is not None
        and not edf_events.empty
        and {
            "type",
            "start",
            "end",
        }.issubset(edf_events.columns)
    ):

        blink_events = edf_events.loc[
            edf_events["type"]
            .astype(str)
            .str.lower()
            .eq("blink")
        ]

        for _, blink in blink_events.iterrows():

            blink_start = pd.to_numeric(
                blink["start"],
                errors="coerce",
            )

            blink_end = pd.to_numeric(
                blink["end"],
                errors="coerce",
            )

            if (
                not np.isfinite(blink_start)
                or not np.isfinite(blink_end)
            ):
                continue

            invalid_samples |= (
                (time_ms >= blink_start)
                & (time_ms <= blink_end)
            )


    x_pixels[invalid_samples] = np.nan
    y_pixels[invalid_samples] = np.nan


    # Convert gaze from pixels to degrees.
    x_degrees, y_degrees = el2deg(
        x_pixels,
        y_pixels,
    )

    x_degrees = np.asarray(
        x_degrees,
        dtype=float,
    )

    y_degrees = np.asarray(
        y_degrees,
        dtype=float,
    )


    # Read the exact block-mean affine saved in the FIXA row.
    affine_values = pd.to_numeric(
        pd.Series([
            event_row["mean_affine_A00"],
            event_row["mean_affine_A01"],
            event_row["mean_affine_A10"],
            event_row["mean_affine_A11"],
            event_row["mean_affine_b0"],
            event_row["mean_affine_b1"],
        ]),
        errors="coerce",
    ).to_numpy(float)


    if not np.isfinite(affine_values).all():
        raise RuntimeError(
            "The selected FIXA row does not contain a valid "
            f"saved mean affine: {subject} | {block_id}"
        )


    mean_A = np.array(
        [
            [
                affine_values[0],
                affine_values[1],
            ],
            [
                affine_values[2],
                affine_values[3],
            ],
        ],
        dtype=float,
    )

    mean_b = np.array(
        [
            affine_values[4],
            affine_values[5],
        ],
        dtype=float,
    )


    x_calibrated = np.full(
        len(x_degrees),
        np.nan,
        dtype=float,
    )

    y_calibrated = np.full(
        len(y_degrees),
        np.nan,
        dtype=float,
    )


    valid_gaze = (
        np.isfinite(x_degrees)
        & np.isfinite(y_degrees)
    )


    if valid_gaze.any():

        raw_points = np.column_stack(
            [
                x_degrees[valid_gaze],
                y_degrees[valid_gaze],
            ]
        )

        corrected_points = (
            raw_points
            @ mean_A.T
            + mean_b
        )

        x_calibrated[valid_gaze] = (
            corrected_points[:, 0]
        )

        y_calibrated[valid_gaze] = (
            corrected_points[:, 1]
        )


    time_differences = np.diff(
        time_ms
    )

    time_differences = time_differences[
        np.isfinite(time_differences)
        & (time_differences > 0)
    ]

    if len(time_differences) == 0:
        raise RuntimeError(
            f"Could not infer sampling rate for {edf_path}"
        )


    sampling_rate_hz = (
        1000.0
        / np.median(time_differences)
    )


    smoothing_samples = max(
        5,
        int(
            round(
                SMOOTH_WINDOW_MS
                / 1000.0
                * sampling_rate_hz
            )
        ),
    )

    if smoothing_samples % 2 == 0:
        smoothing_samples += 1


    if SMOOTH_FOR_DISPLAY:

        x_display = smooth_finite_segments(
            x_calibrated,
            requested_window=smoothing_samples,
            polyorder=SMOOTH_POLYORDER,
        )

        y_display = smooth_finite_segments(
            y_calibrated,
            requested_window=smoothing_samples,
            polyorder=SMOOTH_POLYORDER,
        )

    else:
        x_display = x_calibrated.copy()
        y_display = y_calibrated.copy()


    block_data = {
        "time_ms": time_ms,
        "x_raw_deg": x_calibrated,
        "y_raw_deg": y_calibrated,
        "x_display_deg": x_display,
        "y_display_deg": y_display,
        "sampling_rate_hz": sampling_rate_hz,
        "edf_path": edf_path,
    }

    block_cache[cache_key] = block_data

    return block_data


# ============================================================
# 11. EXTRACT THE SIX DIRECT FIXA TRACE WINDOWS
# ============================================================
trace_objects = []
trace_dataframes = []


for _, event_row in selected_events.iterrows():

    block_data = load_calibrated_block(
        event_row
    )

    fixation_start_ms = float(
        event_row["start_abs_ms"]
    )

    fixation_end_ms = float(
        event_row["end_abs_ms"]
    )

    direct_fixation_duration_ms = float(
        event_row["duration_ms"]
    )


    time_ms = block_data["time_ms"]

    context_mask = (
        (time_ms >= (
            fixation_start_ms
            - CONTEXT_BEFORE_MS
        ))
        & (time_ms <= (
            fixation_end_ms
            + CONTEXT_AFTER_MS
        ))
    )

    fixation_mask = (
        (time_ms >= fixation_start_ms)
        & (time_ms <= fixation_end_ms)
    )


    if context_mask.sum() < 5:
        raise RuntimeError(
            "Too few samples were found for "
            f"{event_row['trace_id']}."
        )


    fixation_center_x = np.nanmedian(
        block_data["x_display_deg"][
            fixation_mask
        ]
    )

    fixation_center_y = np.nanmedian(
        block_data["y_display_deg"][
            fixation_mask
        ]
    )


    if (
        not np.isfinite(fixation_center_x)
        or not np.isfinite(fixation_center_y)
    ):
        raise RuntimeError(
            "No finite gaze position exists within "
            f"{event_row['trace_id']}."
        )


    trace_dataframe = pd.DataFrame({
        "trace_id": event_row["trace_id"],
        "fixation_category": event_row[
            "fixation_category"
        ],
        "subject": event_row["subject"],
        "group": event_row["group"],
        "block_id": event_row["block_id"],
        "trial": event_row["trial"],
        "event_index_block": event_row[
            "event_index_block"
        ],
        "fixation_duration_ms": (
            direct_fixation_duration_ms
        ),
        "time_abs_ms": time_ms[
            context_mask
        ],
        "time_from_fixation_onset_ms": (
            time_ms[context_mask]
            - fixation_start_ms
        ),
        "inside_direct_FIXA": (
            (time_ms[context_mask] >= fixation_start_ms)
            & (time_ms[context_mask] <= fixation_end_ms)
        ),
        "x_raw_relative_deg": (
            block_data["x_raw_deg"][
                context_mask
            ]
            - fixation_center_x
        ),
        "y_raw_relative_deg": (
            block_data["y_raw_deg"][
                context_mask
            ]
            - fixation_center_y
        ),
        "x_display_relative_deg": (
            block_data["x_display_deg"][
                context_mask
            ]
            - fixation_center_x
        ),
        "y_display_relative_deg": (
            block_data["y_display_deg"][
                context_mask
            ]
            - fixation_center_y
        ),
    })


    trace_dataframes.append(
        trace_dataframe
    )

    trace_objects.append({
        "metadata": event_row.to_dict(),
        "trace": trace_dataframe,
    })


all_trace_samples = pd.concat(
    trace_dataframes,
    ignore_index=True,
)


# ============================================================
# 12. SAVE SELECTED DIRECT FIXA ROWS AND TRACE SAMPLES
# ============================================================
selection_csv_path = (
    OUTPUT_DIR
    / "random_direct_FIXA_selection_long_above_500ms.csv"
)

selected_events.to_csv(
    selection_csv_path,
    index=False,
)


trace_csv_path = (
    OUTPUT_DIR
    / "random_direct_FIXA_trace_samples_long_above_500ms.csv"
)

all_trace_samples.to_csv(
    trace_csv_path,
    index=False,
)


print("\nSaved direct FIXA selection:")
print(selection_csv_path)

print("\nSaved trace samples:")
print(trace_csv_path)


# ============================================================
# 13. FIND A SHARED Y-AXIS LIMIT
# ============================================================
inside_values = all_trace_samples.loc[
    all_trace_samples[
        "inside_direct_FIXA"
    ],
    [
        "x_display_relative_deg",
        "y_display_relative_deg",
    ],
].to_numpy(float)


inside_values = np.abs(
    inside_values[
        np.isfinite(inside_values)
    ]
)


if len(inside_values):
    shared_limit = float(
        np.percentile(
            inside_values,
            99,
        )
    )

else:
    shared_limit = 0.25


shared_limit = max(
    0.10,
    shared_limit * 1.25,
)

# Prevent one extremely noisy trace from making all panels unreadable.
shared_limit = min(
    shared_limit,
    3.0,
)


# ============================================================
# 14. PLOT THREE SHORT AND THREE LONG DIRECT FIXA TRACES
# ============================================================
fig, axes = plt.subplots(
    nrows=2,
    ncols=3,
    figsize=(16, 8.5),
    sharey=True,
)

categories = [
    "Short",
    "Long",
]


for row_index, category in enumerate(
    categories
):

    category_traces = [
        item
        for item in trace_objects
        if item["metadata"][
            "fixation_category"
        ] == category
    ]

    category_traces = sorted(
        category_traces,
        key=lambda item: (
            item["metadata"]["duration_ms"]
        ),
    )


    for column_index, trace_object in enumerate(
        category_traces
    ):

        axis = axes[
            row_index,
            column_index,
        ]

        metadata = trace_object[
            "metadata"
        ]

        trace = trace_object[
            "trace"
        ]


        # Raw calibrated gaze shown faintly.
        axis.plot(
            trace[
                "time_from_fixation_onset_ms"
            ],
            trace[
                "x_raw_relative_deg"
            ],
            linewidth=0.65,
            alpha=0.22,
        )

        axis.plot(
            trace[
                "time_from_fixation_onset_ms"
            ],
            trace[
                "y_raw_relative_deg"
            ],
            linewidth=0.65,
            alpha=0.22,
        )


        # Smoothed display traces.
        axis.plot(
            trace[
                "time_from_fixation_onset_ms"
            ],
            trace[
                "x_display_relative_deg"
            ],
            linewidth=1.8,
            label="Horizontal gaze",
        )

        axis.plot(
            trace[
                "time_from_fixation_onset_ms"
            ],
            trace[
                "y_display_relative_deg"
            ],
            linewidth=1.8,
            label="Vertical gaze",
        )


        direct_duration = float(
            metadata["duration_ms"]
        )


        # Exact interval from the directly selected FIXA event row.
        axis.axvspan(
            0,
            direct_duration,
            alpha=0.12,
            label="Direct REMoDNaV FIXA interval",
        )

        axis.axvline(
            0,
            linestyle="--",
            linewidth=1,
        )

        axis.axvline(
            direct_duration,
            linestyle="--",
            linewidth=1,
        )

        axis.axhline(
            0,
            linewidth=0.8,
            alpha=0.5,
        )


        axis.set_xlim(
            -CONTEXT_BEFORE_MS,
            direct_duration
            + CONTEXT_AFTER_MS,
        )

        axis.set_ylim(
            -shared_limit,
            shared_limit,
        )


        trial_value = metadata.get(
            "trial",
            np.nan,
        )

        trial_text = (
            str(int(trial_value))
            if np.isfinite(trial_value)
            else "NA"
        )


        category_definition = (
            f"bottom {SHORT_QUANTILE:.0%}"
            if category == "Short"
            else f"> {LONG_FIXATION_MIN_MS:g} ms"
        )


        axis.set_title(
            f"{category} {column_index + 1}: "
            f"{metadata['subject']} "
            f"({metadata['group']})\n"
            f"FIXA duration = "
            f"{direct_duration:.1f} ms | "
            f"{category_definition} | "
            f"trial {trial_text}"
        )

        axis.set_xlabel(
            "Time from direct FIXA onset (ms)"
        )

        if column_index == 0:
            axis.set_ylabel(
                "Gaze displacement from fixation center (deg)"
            )

        axis.grid(
            alpha=0.25
        )


handles, labels = axes[0, 0].get_legend_handles_labels()

unique_legend = {}

for handle, label in zip(
    handles,
    labels,
):
    unique_legend[label] = handle


fig.legend(
    unique_legend.values(),
    unique_legend.keys(),
    loc="upper center",
    ncol=3,
    frameon=True,
)


fig.suptitle(
    "Random REMoDNaV FIXA traces selected directly "
    "from all-event CSV files\n"
    "Top: short FIXA events from bottom quartile | "
    "Bottom: FIXA events longer than 500 ms",
    fontsize=15,
    y=1.03,
)


fig.tight_layout()


figure_path = (
    OUTPUT_DIR
    / "random_3_short_3_long_above_500ms_DIRECT_FIXA_traces.png"
)

fig.savefig(
    figure_path,
    dpi=SAVE_DPI,
    bbox_inches="tight",
)

plt.show()
plt.close(fig)


# ============================================================
# 15. FINAL OUTPUT
# ============================================================
print("\n" + "=" * 100)
print("DIRECT FIXA TRACE PLOTTING COMPLETE")
print("=" * 100)

print(
    "Every selected event came directly from a row whose "
    "all-event CSV label was FIXA."
)

print(
    f"\nShort fixation definition: "
    f"duration <= {short_cutoff_ms:.2f} ms "
    f"(bottom {SHORT_QUANTILE:.0%})"
)

print(
    f"Long fixation definition: "
    f"duration > {LONG_FIXATION_MIN_MS:.2f} ms"
)

print("\nSaved figure:")
print(figure_path)

print("\nChosen FIXA events:")

display(
    selected_events[
        [
            "trace_id",
            "fixation_category",
            "subject",
            "group",
            "block_id",
            "trial",
            "event_index_block",
            "duration_ms",
            "source_event_csv",
        ]
    ]
)
```

```python
# ============================================================
# COUNT FIXATIONS LONGER THAN 500 ms
#
# Reads:
#   /content/drive/MyDrive/steer_search_PD/main_sequence_outputs/
#       all_fixations_p001.csv
#       all_fixations_p002.csv
#       all_fixations_s178.csv
#       ...
#
# Produces:
#   1. Per-participant counts
#   2. Group-level counts
#   3. Overall counts
#   4. CSV summary files
#
# Long fixation definition:
#   duration_ms > 500
# ============================================================


# ============================================================
# 0. MOUNT GOOGLE DRIVE
# ============================================================
from google.colab import drive

drive.mount(
    "/content/drive",
    force_remount=True,
)


# ============================================================
# 1. IMPORTS
# ============================================================
from pathlib import Path
import re

import numpy as np
import pandas as pd
from IPython.display import display


# ============================================================
# 2. SETTINGS
# ============================================================
MAIN_SEQUENCE_DIR = Path(
    "/content/drive/MyDrive/steer_search_PD/main_sequence_outputs"
)

LONG_FIXATION_THRESHOLD_MS = 500.0

OUTPUT_DIR = (
    MAIN_SEQUENCE_DIR
    / "long_fixation_summary"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True,
)

assert MAIN_SEQUENCE_DIR.exists(), (
    f"Folder not found:\n{MAIN_SEQUENCE_DIR}"
)


# ============================================================
# 3. SMALL HELPERS
# ============================================================
def subject_from_filename(file_path):
    """
    Extract the participant ID from a filename such as:
        all_fixations_p001.csv
        all_fixations_s178.csv
    """
    match = re.fullmatch(
        r"all_fixations_([ps]\d+)\.csv",
        file_path.name,
        flags=re.IGNORECASE,
    )

    if match:
        return match.group(1).lower()

    return None


def infer_group(subject_id):
    """Infer group from participant ID."""
    subject_id = str(subject_id).lower()

    if subject_id.startswith("p"):
        return "PD"

    if subject_id.startswith("s"):
        return "Control"

    return "Unknown"


# ============================================================
# 4. FIND ALL PER-PARTICIPANT FIXATION FILES
# ============================================================
candidate_files = sorted(
    MAIN_SEQUENCE_DIR.glob("all_fixations_*.csv")
)

# Keep only files with standard participant names such as
# all_fixations_p001.csv or all_fixations_s178.csv.
fixation_files = [
    file_path
    for file_path in candidate_files
    if subject_from_filename(file_path) is not None
]

if not fixation_files:
    raise FileNotFoundError(
        "No participant fixation files were found matching:\n"
        f"{MAIN_SEQUENCE_DIR}/all_fixations_p###.csv\n"
        f"{MAIN_SEQUENCE_DIR}/all_fixations_s###.csv"
    )


print("=" * 100)
print("FIXATION FILES FOUND")
print("=" * 100)

print(f"Number of files: {len(fixation_files)}")

for file_path in fixation_files:
    print(" ", file_path.name)


# ============================================================
# 5. READ EACH FILE AND COUNT LONG FIXATIONS
# ============================================================
participant_rows = []
all_fixation_frames = []


for file_path in fixation_files:

    file_subject = subject_from_filename(
        file_path
    )

    dataframe = pd.read_csv(
        file_path,
        low_memory=False,
    )

    if "duration_ms" not in dataframe.columns:
        print(
            f"WARNING: Skipping {file_path.name}; "
            "duration_ms column is missing."
        )
        continue

    # Use subject_id from the file when available.
    if "subject_id" in dataframe.columns:

        available_subjects = (
            dataframe["subject_id"]
            .dropna()
            .astype(str)
            .str.strip()
        )

        if len(available_subjects):
            subject_id = available_subjects.iloc[0]
        else:
            subject_id = file_subject

    else:
        subject_id = file_subject


    # Use group from the file when available.
    if "group" in dataframe.columns:

        available_groups = (
            dataframe["group"]
            .dropna()
            .astype(str)
            .str.strip()
        )

        if len(available_groups):
            group = available_groups.iloc[0]
        else:
            group = infer_group(subject_id)

    else:
        group = infer_group(subject_id)


    # Convert duration to numeric.
    dataframe["duration_ms"] = pd.to_numeric(
        dataframe["duration_ms"],
        errors="coerce",
    )


    # Every row should represent one fixation.
    n_fixation_rows = len(
        dataframe
    )

    # Rows containing a usable duration value.
    valid_duration_mask = (
        np.isfinite(dataframe["duration_ms"])
        & dataframe["duration_ms"].gt(0)
    )

    n_valid_duration_fixations = int(
        valid_duration_mask.sum()
    )

    n_missing_or_invalid_duration = int(
        n_fixation_rows
        - n_valid_duration_fixations
    )


    # Strictly greater than 500 ms.
    long_fixation_mask = (
        valid_duration_mask
        & dataframe["duration_ms"].gt(
            LONG_FIXATION_THRESHOLD_MS
        )
    )

    n_fixations_above_500ms = int(
        long_fixation_mask.sum()
    )


    # Percentage using all fixation rows as denominator.
    percentage_of_all_fixations = (
        100.0
        * n_fixations_above_500ms
        / n_fixation_rows
        if n_fixation_rows > 0
        else np.nan
    )


    # Percentage using only fixations with valid duration values.
    percentage_of_valid_fixations = (
        100.0
        * n_fixations_above_500ms
        / n_valid_duration_fixations
        if n_valid_duration_fixations > 0
        else np.nan
    )


    participant_rows.append({
        "subject_id": subject_id,
        "group": group,
        "source_file": file_path.name,
        "total_fixation_rows": n_fixation_rows,
        "fixations_with_valid_duration": (
            n_valid_duration_fixations
        ),
        "missing_or_invalid_duration": (
            n_missing_or_invalid_duration
        ),
        "fixations_above_500ms": (
            n_fixations_above_500ms
        ),
        "percentage_above_500ms_of_all_fixations": (
            percentage_of_all_fixations
        ),
        "percentage_above_500ms_of_valid_fixations": (
            percentage_of_valid_fixations
        ),
    })


    # Keep a combined event-level table for optional inspection.
    dataframe = dataframe.copy()

    dataframe["subject_id"] = subject_id
    dataframe["group"] = group

    dataframe["is_above_500ms"] = (
        long_fixation_mask.astype(int)
    )

    dataframe["source_file"] = (
        file_path.name
    )

    all_fixation_frames.append(
        dataframe
    )


if not participant_rows:
    raise RuntimeError(
        "No usable fixation files remained after checking "
        "for the duration_ms column."
    )


# ============================================================
# 6. PER-PARTICIPANT SUMMARY
# ============================================================
participant_summary = pd.DataFrame(
    participant_rows
)


# Natural ordering: PD participants followed by controls,
# with numerical ordering within group.
participant_summary["subject_number"] = (
    participant_summary["subject_id"]
    .astype(str)
    .str.extract(r"(\d+)", expand=False)
)

participant_summary["subject_number"] = pd.to_numeric(
    participant_summary["subject_number"],
    errors="coerce",
)

participant_summary["group_order"] = (
    participant_summary["group"]
    .map({
        "PD": 0,
        "Control": 1,
    })
    .fillna(2)
)

participant_summary = (
    participant_summary
    .sort_values(
        [
            "group_order",
            "subject_number",
            "subject_id",
        ]
    )
    .drop(
        columns=[
            "group_order",
            "subject_number",
        ]
    )
    .reset_index(drop=True)
)


print("\n" + "=" * 100)
print(
    f"PER-PARTICIPANT FIXATIONS ABOVE "
    f"{LONG_FIXATION_THRESHOLD_MS:g} ms"
)
print("=" * 100)

display_columns = [
    "subject_id",
    "group",
    "total_fixation_rows",
    "fixations_above_500ms",
    "percentage_above_500ms_of_all_fixations",
    "missing_or_invalid_duration",
]

display(
    participant_summary[
        display_columns
    ].style.format({
        "percentage_above_500ms_of_all_fixations": "{:.2f}%",
    })
)


# ============================================================
# 7. GROUP-LEVEL SUMMARY
# ============================================================
group_summary = (
    participant_summary
    .groupby(
        "group",
        as_index=False,
    )
    .agg(
        n_subjects=(
            "subject_id",
            "nunique",
        ),
        total_fixation_rows=(
            "total_fixation_rows",
            "sum",
        ),
        fixations_with_valid_duration=(
            "fixations_with_valid_duration",
            "sum",
        ),
        missing_or_invalid_duration=(
            "missing_or_invalid_duration",
            "sum",
        ),
        fixations_above_500ms=(
            "fixations_above_500ms",
            "sum",
        ),
        median_subject_percentage_above_500ms=(
            "percentage_above_500ms_of_all_fixations",
            "median",
        ),
        mean_subject_percentage_above_500ms=(
            "percentage_above_500ms_of_all_fixations",
            "mean",
        ),
    )
)


group_summary["pooled_percentage_above_500ms"] = (
    100.0
    * group_summary["fixations_above_500ms"]
    / group_summary["total_fixation_rows"]
)


print("\n" + "=" * 100)
print("GROUP-LEVEL SUMMARY")
print("=" * 100)

display(
    group_summary.style.format({
        "pooled_percentage_above_500ms": "{:.2f}%",
        "median_subject_percentage_above_500ms": "{:.2f}%",
        "mean_subject_percentage_above_500ms": "{:.2f}%",
    })
)


# ============================================================
# 8. OVERALL SUMMARY
# ============================================================
overall_total_fixations = int(
    participant_summary[
        "total_fixation_rows"
    ].sum()
)

overall_valid_fixations = int(
    participant_summary[
        "fixations_with_valid_duration"
    ].sum()
)

overall_invalid_fixations = int(
    participant_summary[
        "missing_or_invalid_duration"
    ].sum()
)

overall_long_fixations = int(
    participant_summary[
        "fixations_above_500ms"
    ].sum()
)


overall_percentage_all = (
    100.0
    * overall_long_fixations
    / overall_total_fixations
    if overall_total_fixations > 0
    else np.nan
)

overall_percentage_valid = (
    100.0
    * overall_long_fixations
    / overall_valid_fixations
    if overall_valid_fixations > 0
    else np.nan
)


overall_summary = pd.DataFrame([
    {
        "n_subjects": (
            participant_summary["subject_id"].nunique()
        ),
        "total_fixation_rows": (
            overall_total_fixations
        ),
        "fixations_with_valid_duration": (
            overall_valid_fixations
        ),
        "missing_or_invalid_duration": (
            overall_invalid_fixations
        ),
        "fixations_above_500ms": (
            overall_long_fixations
        ),
        "percentage_above_500ms_of_all_fixations": (
            overall_percentage_all
        ),
        "percentage_above_500ms_of_valid_fixations": (
            overall_percentage_valid
        ),
    }
])


print("\n" + "=" * 100)
print("OVERALL RESULT")
print("=" * 100)

print(
    f"Fixations longer than "
    f"{LONG_FIXATION_THRESHOLD_MS:g} ms: "
    f"{overall_long_fixations:,} "
    f"out of {overall_total_fixations:,} total fixation rows"
)

print(
    f"Percentage: "
    f"{overall_percentage_all:.2f}%"
)

if overall_invalid_fixations > 0:
    print(
        f"\nNote: {overall_invalid_fixations:,} fixation rows "
        "had missing, zero, negative, or nonnumeric durations."
    )

display(
    overall_summary.style.format({
        "percentage_above_500ms_of_all_fixations": "{:.2f}%",
        "percentage_above_500ms_of_valid_fixations": "{:.2f}%",
    })
)


# ============================================================
# 9. COMBINE AND SAVE EVENT-LEVEL DATA
# ============================================================
all_fixations = pd.concat(
    all_fixation_frames,
    ignore_index=True,
)


participant_output_path = (
    OUTPUT_DIR
    / "fixations_above_500ms_by_subject.csv"
)

group_output_path = (
    OUTPUT_DIR
    / "fixations_above_500ms_by_group.csv"
)

overall_output_path = (
    OUTPUT_DIR
    / "fixations_above_500ms_overall.csv"
)

long_fixations_output_path = (
    OUTPUT_DIR
    / "all_fixation_events_above_500ms.csv"
)


participant_summary.to_csv(
    participant_output_path,
    index=False,
)

group_summary.to_csv(
    group_output_path,
    index=False,
)

overall_summary.to_csv(
    overall_output_path,
    index=False,
)

all_fixations.loc[
    all_fixations["is_above_500ms"].eq(1)
].to_csv(
    long_fixations_output_path,
    index=False,
)


# ============================================================
# 10. FINAL OUTPUT PATHS
# ============================================================
print("\n" + "=" * 100)
print("SUMMARY FILES SAVED")
print("=" * 100)

print("\nParticipant summary:")
print(participant_output_path)

print("\nGroup summary:")
print(group_output_path)

print("\nOverall summary:")
print(overall_output_path)

print("\nAll individual fixation rows above 500 ms:")
print(long_fixations_output_path)
```
